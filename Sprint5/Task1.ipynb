<h1 style="color:DodgerBlue">HashSet<T></h1>



`HashSet<T>` в C# — это коллекция, которая хранит уникальные элементы без определенного порядка. Она полезна, когда нужно быстро проверять наличие элемента или добавлять только уникальные элементы. Давайте рассмотрим, как с ней работать на примере классов `Transport`, `Car`, и `Bike`.

### Создание классов Transport, Car, и Bike

Сначала определим базовый класс `Transport` и его наследники `Car` и `Bike`:

```csharp
public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public override bool Equals(object obj)
    {
        // Простая проверка равенства базируется на модели и производителе
        if (obj is Transport transport)
        {
            return Model == transport.Model && Manufacturer == transport.Manufacturer;
        }
        return false;
    }

    public override int GetHashCode()
    {
        // Создание хеш-кода на основе модели и производителя
        return HashCode.Combine(Model, Manufacturer);
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Model: {Model}, Manufacturer: {Manufacturer}");
    }
}

public class Car : Transport
{
    public int NumberOfDoors { get; set; }

    public Car(string model, string manufacturer, int numberOfDoors)
        : base(model, manufacturer)
    {
        NumberOfDoors = numberOfDoors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Number of Doors: {NumberOfDoors}");
    }
}

public class Bike : Transport
{
    public bool HasCarrier { get; set; }

    public Bike(string model, string manufacturer, bool hasCarrier)
        : base(model, manufacturer)
    {
        HasCarrier = hasCarrier;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Has Carrier: {HasCarrier}");
    }
}
```

### Работа с HashSet<T>

Теперь рассмотрим, как использовать `HashSet<T>` для хранения объектов `Transport`.

```csharp
using System;
using System.Collections.Generic;


// Создаем HashSet для уникальных объектов транспортных средств
HashSet<Transport> transportSet = new HashSet<Transport>();

// Добавляем объекты Car и Bike в HashSet
var car1 = new Car("Model S", "Tesla", 4);
var car2 = new Car("Model 3", "Tesla", 4);
 var bike1 = new Bike("Mountain Explorer", "Giant", true);

transportSet.Add(car1);
transportSet.Add(car2);
transportSet.Add(bike1);

// Попробуем добавить повторяющийся объект
var duplicateCar = new Car("Model S", "Tesla", 4);
bool isAdded = transportSet.Add(duplicateCar);

Console.WriteLine($"Was duplicate car added? {isAdded}"); // false

// Перебираем и выводим информацию обо всех уникальных транспортных средствах
foreach (var transport in transportSet)
{
    transport.DisplayInfo();
    Console.WriteLine();
}

```

### Объяснения и комментарии:

- **Уникальность объектов**: `HashSet<T>` не допускает дублирования. Если вы пытаетесь добавить элемент, который уже присутствует (определяется методом `Equals` и `GetHashCode`), он не добавляется.

- **Методы `Equals` и `GetHashCode`**: Эти методы переопределены для корректной работы с `HashSet`. `Equals` определяет, когда два объекта считаются равными, а `GetHashCode` возвращает хеш-код, используемый для быстрого поиска элемента.

- **Быстрая проверка на наличие**: Благодаря хешированию, `HashSet` позволяет быстро проверять, существует ли элемент в коллекции (`O(1)` — амортизированное время).

- **Отсутствие порядка**: Элементы в `HashSet` не имеют определенного порядка, в отличие от `List<T>`.

`HashSet<T>` идеально подходит для сценариев, где требуется исключительно уникальность элементов и быстрое выполнение операций поиска.

````
using System;
using System.Collections.Generic;

// ========== КЛАССЫ ТРАНСПОРТНЫХ СРЕДСТВ ==========

public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }
    public int Year { get; set; }

    public Transport(string model, string manufacturer, int year)
    {
        Model = model;
        Manufacturer = manufacturer;
        Year = year;
    }

    public override bool Equals(object obj)
    {
        if (obj is Transport transport)
        {
            return Model == transport.Model && 
                   Manufacturer == transport.Manufacturer && 
                   Year == transport.Year;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Model, Manufacturer, Year);
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Модель: {Model}, Производитель: {Manufacturer}, Год: {Year}");
    }
}

public class Car : Transport
{
    public int NumberOfDoors { get; set; }
    public string BodyType { get; set; }
    public int Horsepower { get; set; }

    public Car(string model, string manufacturer, int year, int numberOfDoors, string bodyType, int horsepower)
        : base(model, manufacturer, year)
    {
        NumberOfDoors = numberOfDoors;
        BodyType = bodyType;
        Horsepower = horsepower;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип: Автомобиль, Дверей: {NumberOfDoors}, Кузов: {BodyType}, Мощность: {Horsepower} л.с.");
    }

    public override bool Equals(object obj)
    {
        if (obj is Car car)
        {
            return base.Equals(obj) && 
                   NumberOfDoors == car.NumberOfDoors && 
                   BodyType == car.BodyType && 
                   Horsepower == car.Horsepower;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(base.GetHashCode(), NumberOfDoors, BodyType, Horsepower);
    }
}

public class Bike : Transport
{
    public bool HasSidecar { get; set; }
    public string EngineType { get; set; }
    public int Horsepower { get; set; }

    public Bike(string model, string manufacturer, int year, bool hasSidecar, string engineType, int horsepower)
        : base(model, manufacturer, year)
    {
        HasSidecar = hasSidecar;
        EngineType = engineType;
        Horsepower = horsepower;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип: Мотоцикл, Двигатель: {EngineType}, Коляска: {(HasSidecar ? "Есть" : "Нет")}, Мощность: {Horsepower} л.с.");
    }

    public override bool Equals(object obj)
    {
        if (obj is Bike bike)
        {
            return base.Equals(obj) && 
                   HasSidecar == bike.HasSidecar && 
                   EngineType == bike.EngineType && 
                   Horsepower == bike.Horsepower;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(base.GetHashCode(), HasSidecar, EngineType, Horsepower);
    }
}

public class Truck : Transport
{
    public double LoadCapacity { get; set; }
    public int Axles { get; set; }

    public Truck(string model, string manufacturer, int year, double loadCapacity, int axles)
        : base(model, manufacturer, year)
    {
        LoadCapacity = loadCapacity;
        Axles = axles;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип: Грузовик, Грузоподъемность: {LoadCapacity} т, Оси: {Axles}");
    }

    public override bool Equals(object obj)
    {
        if (obj is Truck truck)
        {
            return base.Equals(obj) && 
                   LoadCapacity == truck.LoadCapacity && 
                   Axles == truck.Axles;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(base.GetHashCode(), LoadCapacity, Axles);
    }
}

// ========== ДЕМОНСТРАЦИЯ РАБОТЫ С HASHSET<T> ДЛЯ TRANSPORT ==========
Console.WriteLine("========== HASHSET<T>: ТРАНСПОРТНЫЕ СРЕДСТВА ==========\n");

// Создаем HashSet для уникальных объектов транспортных средств
HashSet<Transport> transportSet = new HashSet<Transport>();

// Создаем объекты
var car1 = new Car("Model S", "Tesla", 2023, 4, "Седан", 1020);
var car2 = new Car("Model 3", "Tesla", 2023, 4, "Седан", 450);
var car3 = new Car("X5", "BMW", 2024, 5, "Внедорожник", 340);
var bike1 = new Bike("Street Glide", "Harley-Davidson", 2022, false, "V-Twin", 95);
var bike2 = new Bike("R1", "Yamaha", 2024, false, "Inline-4", 200);
var truck1 = new Truck("FH16", "Volvo", 2023, 25.5, 3);
var truck2 = new Truck("Actros", "Mercedes", 2022, 30, 4);

// Добавляем объекты в HashSet
Console.WriteLine("=== 1. Добавление объектов в HashSet ===\n");

Console.WriteLine("Добавляем автомобили:");
Console.WriteLine($"  Добавлен {car1.Manufacturer} {car1.Model}: {transportSet.Add(car1)}");
Console.WriteLine($"  Добавлен {car2.Manufacturer} {car2.Model}: {transportSet.Add(car2)}");
Console.WriteLine($"  Добавлен {car3.Manufacturer} {car3.Model}: {transportSet.Add(car3)}");

Console.WriteLine("\nДобавляем мотоциклы:");
Console.WriteLine($"  Добавлен {bike1.Manufacturer} {bike1.Model}: {transportSet.Add(bike1)}");
Console.WriteLine($"  Добавлен {bike2.Manufacturer} {bike2.Model}: {transportSet.Add(bike2)}");

Console.WriteLine("\nДобавляем грузовики:");
Console.WriteLine($"  Добавлен {truck1.Manufacturer} {truck1.Model}: {transportSet.Add(truck1)}");
Console.WriteLine($"  Добавлен {truck2.Manufacturer} {truck2.Model}: {transportSet.Add(truck2)}");

// Демонстрация уникальности - попытка добавить дубликаты
Console.WriteLine("\n=== 2. Проверка уникальности (добавление дубликатов) ===\n");

var duplicateCar = new Car("Model S", "Tesla", 2023, 4, "Седан", 1020);
bool isAdded = transportSet.Add(duplicateCar);
Console.WriteLine($"Попытка добавить дубликат {duplicateCar.Manufacturer} {duplicateCar.Model}: {(isAdded ? "Добавлен" : "НЕ добавлен (уже существует)")}");

var duplicateBike = new Bike("R1", "Yamaha", 2024, false, "Inline-4", 200);
isAdded = transportSet.Add(duplicateBike);
Console.WriteLine($"Попытка добавить дубликат {duplicateBike.Manufacturer} {duplicateBike.Model}: {(isAdded ? "Добавлен" : "НЕ добавлен (уже существует)")}");

var differentCar = new Car("Model S Plaid", "Tesla", 2024, 4, "Седан", 1020);
isAdded = transportSet.Add(differentCar);
Console.WriteLine($"Попытка добавить новый {differentCar.Manufacturer} {differentCar.Model}: {(isAdded ? "Добавлен" : "НЕ добавлен")}");

// Вывод всех уникальных транспортных средств
Console.WriteLine("\n=== 3. Все уникальные транспортные средства в HashSet ===\n");
Console.WriteLine($"Всего элементов в HashSet: {transportSet.Count}\n");

foreach (var transport in transportSet)
{
    transport.DisplayInfo();
    Console.WriteLine();
}

// Проверка наличия элементов
Console.WriteLine("=== 4. Проверка наличия элементов ===\n");

bool containsCar = transportSet.Contains(car1);
Console.WriteLine($"Содержит {car1.Manufacturer} {car1.Model}: {containsCar}");

bool containsDuplicate = transportSet.Contains(duplicateCar);
Console.WriteLine($"Содержит дубликат {duplicateCar.Manufacturer} {duplicateCar.Model}: {containsDuplicate}");

bool containsNewCar = transportSet.Contains(differentCar);
Console.WriteLine($"Содержит {differentCar.Manufacturer} {differentCar.Model}: {containsNewCar}");

// Удаление элементов
Console.WriteLine("\n=== 5. Удаление элементов ===\n");

bool removed = transportSet.Remove(car2);
Console.WriteLine($"Удален {car2.Manufacturer} {car2.Model}: {removed}");
Console.WriteLine($"Количество после удаления: {transportSet.Count}");

removed = transportSet.Remove(truck2);
Console.WriteLine($"Удален {truck2.Manufacturer} {truck2.Model}: {removed}");
Console.WriteLine($"Количество после удаления: {transportSet.Count}");

// Операции с множествами
Console.WriteLine("\n=== 6. Операции с множествами ===\n");

HashSet<Transport> sportVehicles = new HashSet<Transport>
{
    new Car("Model S Plaid", "Tesla", 2024, 4, "Седан", 1020),
    new Bike("R1", "Yamaha", 2024, false, "Inline-4", 200),
    new Car("Mustang GT", "Ford", 2023, 2, "Купе", 450)
};

Console.WriteLine("Спортивные автомобили:");
foreach (var v in sportVehicles)
{
    Console.WriteLine($"  {v.Manufacturer} {v.Model}");
}

// Объединение множеств
transportSet.UnionWith(sportVehicles);
Console.WriteLine($"\nПосле объединения со спортивными: {transportSet.Count} элементов");

// Пересечение множеств
transportSet.IntersectWith(sportVehicles);
Console.WriteLine($"\nПосле пересечения со спортивными: {transportSet.Count} элементов");

foreach (var v in transportSet)
{
    Console.WriteLine($"  {v.Manufacturer} {v.Model}");
}

Console.WriteLine("\n========== ДЕМОНСТРАЦИЯ TRANSPORT ЗАВЕРШЕНА ==========\n");
````

````
using System;
using System.Collections.Generic;

// ========== КЛАССЫ ЖИВОТНЫХ ==========

public abstract class Animal
{
    public string Name { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }

    public Animal(string name, int age, double weight)
    {
        Name = name;
        Age = age;
        Weight = weight;
    }

    public override bool Equals(object obj)
    {
        if (obj is Animal animal)
        {
            return Name == animal.Name && 
                   Age == animal.Age && 
                   Weight == animal.Weight;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Age, Weight);
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Имя: {Name}, Возраст: {Age} лет, Вес: {Weight} кг");
    }

    public abstract void MakeSound();
}

public class Dog : Animal
{
    public string Breed { get; set; }
    public bool IsTrained { get; set; }
    public string OwnerName { get; set; }

    public Dog(string name, int age, double weight, string breed, bool isTrained, string ownerName)
        : base(name, age, weight)
    {
        Breed = breed;
        IsTrained = isTrained;
        OwnerName = ownerName;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип: Собака, Порода: {Breed}, Дрессирована: {(IsTrained ? "Да" : "Нет")}, Хозяин: {OwnerName}");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} лает: Гав-гав!");
    }

    public override bool Equals(object obj)
    {
        if (obj is Dog dog)
        {
            return base.Equals(obj) && 
                   Breed == dog.Breed && 
                   IsTrained == dog.IsTrained && 
                   OwnerName == dog.OwnerName;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(base.GetHashCode(), Breed, IsTrained, OwnerName);
    }
}

public class Cat : Animal
{
    public string Color { get; set; }
    public bool IsIndoor { get; set; }
    public string OwnerName { get; set; }

    public Cat(string name, int age, double weight, string color, bool isIndoor, string ownerName)
        : base(name, age, weight)
    {
        Color = color;
        IsIndoor = isIndoor;
        OwnerName = ownerName;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип: Кошка, Окрас: {Color}, Домашняя: {(IsIndoor ? "Да" : "Нет")}, Хозяин: {OwnerName}");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} мяукает: Мяу-мяу!");
    }

    public override bool Equals(object obj)
    {
        if (obj is Cat cat)
        {
            return base.Equals(obj) && 
                   Color == cat.Color && 
                   IsIndoor == cat.IsIndoor && 
                   OwnerName == cat.OwnerName;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(base.GetHashCode(), Color, IsIndoor, OwnerName);
    }
}

public class Bird : Animal
{
    public double WingSpan { get; set; }
    public bool CanFly { get; set; }
    public string Habitat { get; set; }

    public Bird(string name, int age, double weight, double wingSpan, bool canFly, string habitat)
        : base(name, age, weight)
    {
        WingSpan = wingSpan;
        CanFly = canFly;
        Habitat = habitat;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип: Птица, Размах крыльев: {WingSpan} м, Летает: {(CanFly ? "Да" : "Нет")}, Среда обитания: {Habitat}");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} чирикает: Чик-чирик!");
    }

    public override bool Equals(object obj)
    {
        if (obj is Bird bird)
        {
            return base.Equals(obj) && 
                   WingSpan == bird.WingSpan && 
                   CanFly == bird.CanFly && 
                   Habitat == bird.Habitat;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(base.GetHashCode(), WingSpan, CanFly, Habitat);
    }
}

public class Fish : Animal
{
    public string WaterType { get; set; }
    public double MaxDepth { get; set; }
    public string Habitat { get; set; }

    public Fish(string name, int age, double weight, string waterType, double maxDepth, string habitat)
        : base(name, age, weight)
    {
        WaterType = waterType;
        MaxDepth = maxDepth;
        Habitat = habitat;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип: Рыба, Тип воды: {WaterType}, Макс. глубина: {MaxDepth} м, Среда обитания: {Habitat}");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} издает булькающие звуки: Буль-буль!");
    }

    public override bool Equals(object obj)
    {
        if (obj is Fish fish)
        {
            return base.Equals(obj) && 
                   WaterType == fish.WaterType && 
                   MaxDepth == fish.MaxDepth && 
                   Habitat == fish.Habitat;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(base.GetHashCode(), WaterType, MaxDepth, Habitat);
    }
}

// ========== ДЕМОНСТРАЦИЯ РАБОТЫ С HASHSET<T> ДЛЯ ANIMAL ==========
class AnimalHashSetDemo
{
    public static void Run()
    {
        Console.WriteLine("========== HASHSET<T>: ЖИВОТНЫЕ ==========\n");

        // Создаем HashSet для уникальных объектов животных
        HashSet<Animal> animalSet = new HashSet<Animal>();

        // Создаем объекты
        var dog1 = new Dog("Рекс", 3, 25.5, "Овчарка", true, "Иван");
        var dog2 = new Dog("Шарик", 4, 20.0, "Дворняга", false, "Петр");
        var dog3 = new Dog("Лайка", 2, 18.5, "Хаски", true, "Сергей");
        
        var cat1 = new Cat("Муся", 2, 4.2, "серый", true, "Элина");
        var cat2 = new Cat("Васька", 3, 3.8, "черный", false, "Анна");
        var cat3 = new Cat("Барсик", 4, 5.0, "рыжий", true, "Мария");
        
        var bird1 = new Bird("Кеша", 1, 0.35, 0.5, true, "лес");
        var bird2 = new Bird("Пингвин", 5, 30.0, 0.8, false, "Антарктида");
        var bird3 = new Bird("Орел", 4, 6.0, 2.2, true, "горы");
        
        var fish1 = new Fish("Немо", 2, 0.15, "Соленая", 50, "океан");
        var fish2 = new Fish("Золотая", 1, 0.1, "Пресная", 2, "пруд");
        var fish3 = new Fish("Акула", 8, 500.0, "Соленая", 1000, "океан");

        // Добавляем объекты в HashSet
        Console.WriteLine("=== 1. Добавление объектов в HashSet ===\n");
        
        Console.WriteLine("Добавляем собак:");
        Console.WriteLine($"  Добавлен {dog1.Name}: {animalSet.Add(dog1)}");
        Console.WriteLine($"  Добавлен {dog2.Name}: {animalSet.Add(dog2)}");
        Console.WriteLine($"  Добавлен {dog3.Name}: {animalSet.Add(dog3)}");
        
        Console.WriteLine("\nДобавляем кошек:");
        Console.WriteLine($"  Добавлена {cat1.Name}: {animalSet.Add(cat1)}");
        Console.WriteLine($"  Добавлен {cat2.Name}: {animalSet.Add(cat2)}");
        Console.WriteLine($"  Добавлен {cat3.Name}: {animalSet.Add(cat3)}");
        
        Console.WriteLine("\nДобавляем птиц:");
        Console.WriteLine($"  Добавлен {bird1.Name}: {animalSet.Add(bird1)}");
        Console.WriteLine($"  Добавлен {bird2.Name}: {animalSet.Add(bird2)}");
        Console.WriteLine($"  Добавлен {bird3.Name}: {animalSet.Add(bird3)}");
        
        Console.WriteLine("\nДобавляем рыб:");
        Console.WriteLine($"  Добавлен {fish1.Name}: {animalSet.Add(fish1)}");
        Console.WriteLine($"  Добавлена {fish2.Name}: {animalSet.Add(fish2)}");
        Console.WriteLine($"  Добавлена {fish3.Name}: {animalSet.Add(fish3)}");

        // Демонстрация уникальности - попытка добавить дубликаты
        Console.WriteLine("\n=== 2. Проверка уникальности (добавление дубликатов) ===\n");
        
        var duplicateDog = new Dog("Рекс", 3, 25.5, "Овчарка", true, "Иван");
        bool isAdded = animalSet.Add(duplicateDog);
        Console.WriteLine($"Попытка добавить дубликат собаки {duplicateDog.Name}: {(isAdded ? "Добавлен" : "НЕ добавлен (уже существует)")}");
        
        var duplicateCat = new Cat("Муся", 2, 4.2, "серый", true, "Элина");
        isAdded = animalSet.Add(duplicateCat);
        Console.WriteLine($"Попытка добавить дубликат кошки {duplicateCat.Name}: {(isAdded ? "Добавлена" : "НЕ добавлена (уже существует)")}");
        
        var duplicateBird = new Bird("Кеша", 1, 0.35, 0.5, true, "лес");
        isAdded = animalSet.Add(duplicateBird);
        Console.WriteLine($"Попытка добавить дубликат птицы {duplicateBird.Name}: {(isAdded ? "Добавлен" : "НЕ добавлен (уже существует)")}");
        
        var newCat = new Cat("Снежок", 1, 3.2, "белый", true, "Ольга");
        isAdded = animalSet.Add(newCat);
        Console.WriteLine($"\nПопытка добавить новую кошку {newCat.Name}: {(isAdded ? "Добавлена" : "НЕ добавлена")}");

        // Вывод всех уникальных животных
        Console.WriteLine("\n=== 3. Все уникальные животные в HashSet ===\n");
        Console.WriteLine($"Всего элементов в HashSet: {animalSet.Count}\n");
        
        foreach (var animal in animalSet)
        {
            animal.DisplayInfo();
            animal.MakeSound();
            Console.WriteLine();
        }

        // Проверка наличия элементов
        Console.WriteLine("=== 4. Проверка наличия элементов ===\n");
        
        bool containsDog = animalSet.Contains(dog1);
        Console.WriteLine($"Содержит собаку {dog1.Name}: {containsDog}");
        
        bool containsCat = animalSet.Contains(cat1);
        Console.WriteLine($"Содержит кошку {cat1.Name}: {containsCat}");
        
        bool containsDuplicate = animalSet.Contains(duplicateDog);
        Console.WriteLine($"Содержит дубликат собаки {duplicateDog.Name}: {containsDuplicate}");
        
        bool containsNewCat = animalSet.Contains(newCat);
        Console.WriteLine($"Содержит новую кошку {newCat.Name}: {containsNewCat}");

        // Удаление элементов
        Console.WriteLine("\n=== 5. Удаление элементов ===\n");
        
        bool removed = animalSet.Remove(dog2);
        Console.WriteLine($"Удалена собака {dog2.Name}: {removed}");
        Console.WriteLine($"Количество после удаления: {animalSet.Count}");
        
        removed = animalSet.Remove(bird2);
        Console.WriteLine($"Удалена птица {bird2.Name}: {removed}");
        Console.WriteLine($"Количество после удаления: {animalSet.Count}");
        
        removed = animalSet.Remove(fish2);
        Console.WriteLine($"Удалена рыба {fish2.Name}: {removed}");
        Console.WriteLine($"Количество после удаления: {animalSet.Count}");

        // Операции с множествами
        Console.WriteLine("\n=== 6. Операции с множествами ===\n");
        
        HashSet<Animal> trainedAnimals = new HashSet<Animal>
        {
            new Dog("Рекс", 3, 25.5, "Овчарка", true, "Иван"),
            new Dog("Лайка", 2, 18.5, "Хаски", true, "Сергей"),
            new Cat("Муся", 2, 4.2, "серый", true, "Элина")
        };
        
        Console.WriteLine("Дрессированные животные:");
        foreach (var a in trainedAnimals)
        {
            Console.WriteLine($"  {a.Name} ({a.GetType().Name})");
        }
        
        // Пересечение множеств
        var intersection = new HashSet<Animal>(animalSet);
        intersection.IntersectWith(trainedAnimals);
        Console.WriteLine($"\nПересечение (дрессированные животные в основной коллекции): {intersection.Count}");
        foreach (var a in intersection)
        {
            Console.WriteLine($"  {a.Name} ({a.GetType().Name})");
        }
        
        // Объединение множеств
        HashSet<Animal> wildAnimals = new HashSet<Animal>
        {
            new Bird("Орел", 4, 6.0, 2.2, true, "горы"),
            new Fish("Акула", 8, 500.0, "Соленая", 1000, "океан"),
            new Bird("Пингвин", 5, 30.0, 0.8, false, "Антарктида")
        };
        
        var union = new HashSet<Animal>(animalSet);
        union.UnionWith(wildAnimals);
        Console.WriteLine($"\nОбъединение с дикими животными: {union.Count} элементов");
        
        // Разность множеств
        var difference = new HashSet<Animal>(animalSet);
        difference.ExceptWith(trainedAnimals);
        Console.WriteLine($"\nРазность (животные без дрессированных): {difference.Count}");
        foreach (var a in difference)
        {
            Console.WriteLine($"  {a.Name} ({a.GetType().Name})");
        }

        // Подмножество
        bool isSubset = trainedAnimals.IsSubsetOf(animalSet);
        Console.WriteLine($"\nДрессированные животные являются подмножеством основной коллекции: {isSubset}");

        Console.WriteLine("\n========== ДЕМОНСТРАЦИЯ ANIMAL ЗАВЕРШЕНА ==========");
    }
}

// ========== ЗАПУСК ДЕМОНСТРАЦИИ ==========
AnimalHashSetDemo.Run();
````
