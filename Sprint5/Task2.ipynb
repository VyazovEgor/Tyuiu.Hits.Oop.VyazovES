<h1 style="color:DodgerBlue">Делегаты, лямбда-выражения и события<T></h1>



Делегаты, лямбда-выражения и события являются важными концепциями в C#, которые позволяют работать с методами как с объектами, писать краткий код и управлять реакциями на события. Рассмотрим их на примере гипотетических классов `Transport`, `Car`, и `Bike`.

### Делегаты

**Делегат** — это тип безопасного указателя на метод, похожий на функцию в C или C++. Делегаты в C# используются для передачи методов как параметров, а также для создания обратных вызовов.

**Пример:**

Создадим делегат, который может ссылаться на методы, выводящие информацию о транспортном средстве:

```csharp
public delegate void DisplayDelegate();

public class Transport

    public string Model { get; set; }

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Model: {Model}, Manufacturer: {Manufacturer}");
    }
}
```

### Лямбда-выражения

**Лямбда-выражения** позволяют вам писать анонимные методы в более краткой форме. Это часто используется в LINQ и асинхронном программировании.

**Пример:**

```csharp
using System;
using System.Collections.Generic;
using System.Linq;


 
List<Transport> transports = new List<Transport>
{
    new Car("Model S", "Tesla", 4),
    new Bike("Speedster", "Giant", true),
    new Car("Model 3", "Tesla", 4),
};

// Используем лямбда-выражение для фильтрации транспортных средств
var teslaCars = transports.OfType<Car>().Where(c => c.Manufacturer == "Tesla");

foreach (var car in teslaCars)
{
    car.DisplayInfo();
}


```

### События

**События** используются для оповещения подписчиков о случившихся действиях, предоставляя более высокоуровневый способ взаимодействия между различными частями кода.

**Пример:**

```csharp
public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public delegate void TransportHandler(string message);
    public event TransportHandler TransportUpdated;

    public void UpdateModel(string newModel)
    {
        Model = newModel;
        // Вызываем событие после обновления модели
        TransportUpdated?.Invoke($"Model updated to {newModel}");
    }
}



Transport transport = new Transport { Model = "Old Model", Manufacturer = "Generic" };
// Подписываем метод на событие
transport.TransportUpdated += OnTransportUpdated;
// Обновляем модель; это вызовет событие
transport.UpdateModel("New Model");


static void OnTransportUpdated(string message)
{
    Console.WriteLine(message);
}
```

### Объяснения и комментарии:

- **Делегаты**: Тип данных, который можно использовать для хранения ссылок на методы. Полезно для обратных вызовов и передачи методов.

- **Лямбда-выражения**: Более краткая запись анонимных методов, часто используется в LINQ для фильтрации, проекций и агрегаций.

- **События**: Механизм уведомления объектов об изменениях состояния или других действиях. Позволяют подписываться на события в одной части программы и реакцию на них в другой.

Эти механизмы делают программу гибкой и более структурированной, обеспечивая четкое отделение логики и упрощая повторное использование кода.

````
using System;
using System.Collections.Generic;
using System.Linq;

// ========== КЛАССЫ ТРАНСПОРТНЫХ СРЕДСТВ ==========

// Делегаты
public delegate void TransportEventHandler(string message);
public delegate void SpeedChangeHandler(int newSpeed);
public delegate void DisplayHandler();

public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }
    public int Year { get; set; }
    protected int currentSpeed;
    
    // События
    public event TransportEventHandler TransportUpdated;
    public event SpeedChangeHandler SpeedChanged;
    public event TransportEventHandler TransportStarted;
    public event TransportEventHandler TransportStopped;
    
    public Transport(string model, string manufacturer, int year)
    {
        Model = model;
        Manufacturer = manufacturer;
        Year = year;
        currentSpeed = 0;
    }
    
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Модель: {Model}, Производитель: {Manufacturer}, Год: {Year}");
    }
    
    public virtual void Start()
    {
        currentSpeed = 0;
        TransportStarted?.Invoke($"{Manufacturer} {Model} запущен");
    }
    
    public virtual void Stop()
    {
        int oldSpeed = currentSpeed;
        currentSpeed = 0;
        TransportStopped?.Invoke($"{Manufacturer} {Model} остановлен");
        if (oldSpeed > 0)
            SpeedChanged?.Invoke(currentSpeed);
    }
    
    public void UpdateModel(string newModel)
    {
        string oldModel = Model;
        Model = newModel;
        TransportUpdated?.Invoke($"Модель изменена с {oldModel} на {newModel}");
    }
    
    public void Accelerate(int amount)
    {
        int oldSpeed = currentSpeed;
        currentSpeed += amount;
        SpeedChanged?.Invoke(currentSpeed);
        Console.WriteLine($"{Manufacturer} {Model} разогнался с {oldSpeed} до {currentSpeed} км/ч");
    }
    
    public int GetCurrentSpeed() => currentSpeed;
}

public class Car : Transport
{
    public int NumberOfDoors { get; set; }
    public string BodyType { get; set; }
    public int Horsepower { get; set; }
    
    public Car(string model, string manufacturer, int year, int numberOfDoors, string bodyType, int horsepower)
        : base(model, manufacturer, year)
    {
        NumberOfDoors = numberOfDoors;
        BodyType = bodyType;
        Horsepower = horsepower;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип: Автомобиль, Дверей: {NumberOfDoors}, Кузов: {BodyType}, Мощность: {Horsepower} л.с.");
    }
    
    public override void Start()
    {
        base.Start();
        Console.WriteLine($"  Двигатель автомобиля запущен, мощность {Horsepower} л.с.");
    }
    
    public void Honk()
    {
        Console.WriteLine($"{Manufacturer} {Model} сигналит: Бип-бип!");
    }
}

public class Bike : Transport
{
    public bool HasSidecar { get; set; }
    public string EngineType { get; set; }
    public int Horsepower { get; set; }
    
    public Bike(string model, string manufacturer, int year, bool hasSidecar, string engineType, int horsepower)
        : base(model, manufacturer, year)
    {
        HasSidecar = hasSidecar;
        EngineType = engineType;
        Horsepower = horsepower;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип: Мотоцикл, Двигатель: {EngineType}, Коляска: {(HasSidecar ? "Есть" : "Нет")}, Мощность: {Horsepower} л.с.");
    }
    
    public override void Start()
    {
        base.Start();
        Console.WriteLine($"  Двигатель мотоцикла {EngineType} запущен, мощность {Horsepower} л.с.");
    }
    
    public void Wheelie()
    {
        if (currentSpeed > 20)
            Console.WriteLine($"{Manufacturer} {Model} едет на заднем колесе!");
        else
            Console.WriteLine($"Нужно разогнаться быстрее для вили! Текущая скорость: {currentSpeed} км/ч");
    }
}

public class Truck : Transport
{
    public double LoadCapacity { get; set; }
    public int Axles { get; set; }
    
    public Truck(string model, string manufacturer, int year, double loadCapacity, int axles)
        : base(model, manufacturer, year)
    {
        LoadCapacity = loadCapacity;
        Axles = axles;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип: Грузовик, Грузоподъемность: {LoadCapacity} т, Оси: {Axles}");
    }
    
    public override void Start()
    {
        base.Start();
        Console.WriteLine($"  Дизельный двигатель грузовика запущен");
    }
    
    public void LoadCargo(double weight)
    {
        if (weight <= LoadCapacity)
            Console.WriteLine($"Грузовик {Manufacturer} {Model} загружен на {weight} тонн.");
        else
            Console.WriteLine($"Грузовик {Manufacturer} {Model} перегружен! Максимум {LoadCapacity} тонн.");
    }
}

// ========== ДЕМОНСТРАЦИЯ ДЛЯ TRANSPORT ==========
Console.WriteLine("========== ДЕЛЕГАТЫ, ЛЯМБДА-ВЫРАЖЕНИЯ И СОБЫТИЯ: ТРАНСПОРТ ==========\n");

// ========== 1. ДЕЛЕГАТЫ ==========
Console.WriteLine("=== 1. ДЕЛЕГАТЫ ===\n");

Car car = new Car("Model S", "Tesla", 2023, 4, "Седан", 1020);
Bike bike = new Bike("Street Glide", "Harley-Davidson", 2022, false, "V-Twin", 95);
Truck truck = new Truck("FH16", "Volvo", 2023, 25.5, 3);

// Создание делегатов
DisplayHandler displayCar = car.DisplayInfo;
DisplayHandler displayBike = bike.DisplayInfo;
DisplayHandler displayTruck = truck.DisplayInfo;

Console.WriteLine("Вызов методов через делегаты:");
displayCar();
Console.WriteLine();
displayBike();
Console.WriteLine();
displayTruck();

// Многоадресный делегат
Console.WriteLine("\n--- Многоадресный делегат ---");
DisplayHandler multiDelegate = displayCar + displayBike + displayTruck;
multiDelegate();

// ========== 2. ЛЯМБДА-ВЫРАЖЕНИЯ ==========
Console.WriteLine("\n=== 2. ЛЯМБДА-ВЫРАЖЕНИЯ ===\n");

List<Transport> transports = new List<Transport>
{
    new Car("Model S", "Tesla", 2023, 4, "Седан", 1020),
    new Car("Model 3", "Tesla", 2023, 4, "Седан", 450),
    new Bike("Street Glide", "Harley-Davidson", 2022, false, "V-Twin", 95),
    new Car("X5", "BMW", 2024, 5, "Внедорожник", 340),
    new Bike("R1", "Yamaha", 2024, false, "Inline-4", 200),
    new Truck("FH16", "Volvo", 2023, 25.5, 3),
    new Car("Mustang", "Ford", 2023, 2, "Купе", 450),
    new Truck("Actros", "Mercedes", 2022, 30, 4)
};

// Фильтрация с помощью лямбда-выражений
Console.WriteLine("Автомобили Tesla:");
var teslaCars = transports.OfType<Car>().Where(c => c.Manufacturer == "Tesla");
foreach (var carItem in teslaCars)
{
    carItem.DisplayInfo();
}

Console.WriteLine("\nМощные автомобили (>400 л.с.):");
var powerfulCars = transports.OfType<Car>().Where(c => c.Horsepower > 400);
foreach (var carItem in powerfulCars)
{
    Console.WriteLine($"  {carItem.Manufacturer} {carItem.Model} - {carItem.Horsepower} л.с.");
}

Console.WriteLine("\nМотоциклы с мощностью > 150 л.с.:");
var powerfulBikes = transports.OfType<Bike>().Where(b => b.Horsepower > 150);
foreach (var bikeItem in powerfulBikes)
{
    Console.WriteLine($"  {bikeItem.Manufacturer} {bikeItem.Model} - {bikeItem.Horsepower} л.с.");
}

// Сортировка с помощью лямбда-выражений
Console.WriteLine("\nТранспорт по году выпуска (от новых к старым):");
var sortedByYear = transports.OrderByDescending(t => t.Year);
foreach (var t in sortedByYear)
{
    Console.WriteLine($"  {t.Manufacturer} {t.Model} - {t.Year} год");
}

// Проекция с помощью лямбда-выражений
Console.WriteLine("\nСписок производителей:");
var manufacturers = transports.Select(t => t.Manufacturer).Distinct();
foreach (var m in manufacturers)
{
    Console.WriteLine($"  {m}");
}

// Агрегация с помощью лямбда-выражений
int totalHorsepower = transports.OfType<Car>().Sum(c => c.Horsepower) + 
                        transports.OfType<Bike>().Sum(b => b.Horsepower);
Console.WriteLine($"\nСуммарная мощность всех автомобилей и мотоциклов: {totalHorsepower} л.с.");

double avgYear = transports.Average(t => t.Year);
Console.WriteLine($"Средний год выпуска: {avgYear:F0}");

// ========== 3. СОБЫТИЯ ==========
Console.WriteLine("\n=== 3. СОБЫТИЯ ===\n");

Car eventCar = new Car("Model S", "Tesla", 2023, 4, "Седан", 1020);

// Подписка на события с помощью лямбда-выражений
eventCar.TransportStarted += (message) => Console.WriteLine($"[СОБЫТИЕ] Запуск: {message}");
eventCar.TransportStopped += (message) => Console.WriteLine($"[СОБЫТИЕ] Остановка: {message}");
eventCar.SpeedChanged += (newSpeed) => Console.WriteLine($"[СОБЫТИЕ] Скорость изменена: {newSpeed} км/ч");
eventCar.TransportUpdated += (message) => Console.WriteLine($"[СОБЫТИЕ] Обновление: {message}");

Console.WriteLine("Демонстрация событий:");
eventCar.Start();
eventCar.Accelerate(30);
eventCar.Accelerate(20);
eventCar.UpdateModel("Model S Plaid");
eventCar.Stop();

// Демонстрация с мотоциклом
Console.WriteLine("\n--- События для мотоцикла ---");
Bike eventBike = new Bike("R1", "Yamaha", 2024, false, "Inline-4", 200);

eventBike.TransportStarted += (msg) => Console.WriteLine($"Мотоцикл: {msg}");
eventBike.SpeedChanged += (speed) => Console.WriteLine($"Мотоцикл разогнался до {speed} км/ч");

eventBike.Start();
eventBike.Accelerate(50);
eventBike.Wheelie();

// ========== 4. ИСПОЛЬЗОВАНИЕ ДЕЛЕГАТОВ В КАЧЕСТВЕ ПАРАМЕТРОВ ==========
Console.WriteLine("\n=== 4. ДЕЛЕГАТЫ КАК ПАРАМЕТРЫ МЕТОДОВ ===\n");

void ProcessTransport(Transport transport, DisplayHandler displayAction)
{
    Console.WriteLine($"Обработка: {transport.Manufacturer} {transport.Model}");
    displayAction();
}

ProcessTransport(car, car.DisplayInfo);
ProcessTransport(bike, bike.DisplayInfo);

// Использование лямбда-выражения в качестве делегата
ProcessTransport(truck, () => Console.WriteLine($"Грузовик {truck.Manufacturer} {truck.Model} готов к работе"));

// ========== 5. ФУНКЦИОНАЛЬНЫЕ ДЕЛЕГАТЫ ==========
Console.WriteLine("\n=== 5. ФУНКЦИОНАЛЬНЫЕ ДЕЛЕГАТЫ (Func, Action, Predicate) ===\n");

// Func - возвращает значение
Func<Transport, string> getDescription = t => $"{t.Manufacturer} {t.Model} ({t.Year})";
Console.WriteLine($"Описание: {getDescription(car)}");

// Action - не возвращает значение
Action<Transport> printShortInfo = t => Console.WriteLine($"  {t.Manufacturer} {t.Model}");
Console.WriteLine("Краткая информация о транспорте:");
transports.Take(3).ToList().ForEach(printShortInfo);

// Predicate - возвращает bool
Predicate<Transport> isTesla = t => t.Manufacturer == "Tesla";
bool hasTesla = transports.Exists(t => isTesla(t));
Console.WriteLine($"\nЕсть ли автомобили Tesla в коллекции: {hasTesla}");

Console.WriteLine("\n========== ДЕМОНСТРАЦИЯ TRANSPORT ЗАВЕРШЕНА ==========\n");
````

````
using System;
using System.Collections.Generic;
using System.Linq;

// ========== КЛАССЫ ЖИВОТНЫХ ==========

// Делегаты
public delegate void AnimalEventHandler(string message);
public delegate void HealthChangeHandler(int energyLevel);
public delegate void SoundHandler(string sound);

public abstract class Animal
{
    public string Name { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }
    protected int energyLevel;
    
    // События
    public event AnimalEventHandler AnimalFed;
    public event AnimalEventHandler AnimalSlept;
    public event HealthChangeHandler EnergyChanged;
    public event SoundHandler AnimalSound;
    public event AnimalEventHandler AnimalMoved;
    
    public Animal(string name, int age, double weight)
    {
        Name = name;
        Age = age;
        Weight = weight;
        energyLevel = 100;
    }
    
    // Защищенные методы для вызова событий
    protected virtual void OnAnimalFed(string message)
    {
        AnimalFed?.Invoke(message);
    }
    
    protected virtual void OnAnimalSlept(string message)
    {
        AnimalSlept?.Invoke(message);
    }
    
    protected virtual void OnEnergyChanged(int energyLevel)
    {
        EnergyChanged?.Invoke(energyLevel);
    }
    
    protected virtual void OnAnimalSound(string sound)
    {
        AnimalSound?.Invoke(sound);
    }
    
    protected virtual void OnAnimalMoved(string message)
    {
        AnimalMoved?.Invoke(message);
    }
    
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Имя: {Name}, Возраст: {Age} лет, Вес: {Weight} кг, Энергия: {energyLevel}%");
    }
    
    public abstract void MakeSound();
    
    public virtual void Move()
    {
        if (energyLevel >= 10)
        {
            int oldEnergy = energyLevel;
            energyLevel -= 10;
            OnAnimalMoved($"{Name} двигается");
            OnEnergyChanged(energyLevel);
            Console.WriteLine($"{Name} двигается. Энергия: {energyLevel}%");
        }
        else
        {
            Console.WriteLine($"{Name} слишком устал(а), чтобы двигаться!");
        }
    }
    
    public virtual void Eat()
    {
        int oldEnergy = energyLevel;
        energyLevel = Math.Min(100, energyLevel + 20);
        OnAnimalFed($"{Name} покормлен(а)");
        OnEnergyChanged(energyLevel);
        Console.WriteLine($"{Name} поел(а). Энергия: {energyLevel}%");
    }
    
    public virtual void Sleep()
    {
        int oldEnergy = energyLevel;
        energyLevel = 100;
        OnAnimalSlept($"{Name} выспался(ась)");
        OnEnergyChanged(energyLevel);
        Console.WriteLine($"{Name} выспался(ась). Энергия: {energyLevel}%");
    }
    
    public int GetEnergyLevel() => energyLevel;
    
    public void SetName(string newName)
    {
        string oldName = Name;
        Name = newName;
        OnAnimalFed($"{oldName} переименован(а) в {newName}");
    }
}

public class Dog : Animal
{
    public string Breed { get; set; }
    public bool IsTrained { get; set; }
    public string OwnerName { get; set; }
    
    public Dog(string name, int age, double weight, string breed, bool isTrained, string ownerName)
        : base(name, age, weight)
    {
        Breed = breed;
        IsTrained = isTrained;
        OwnerName = ownerName;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип: Собака, Порода: {Breed}, Дрессирована: {(IsTrained ? "Да" : "Нет")}, Хозяин: {OwnerName}");
    }
    
    public override void MakeSound()
    {
        string sound = "Гав-гав!";
        OnAnimalSound(sound);
        Console.WriteLine($"{Name} лает: {sound}");
    }
    
    public void Fetch()
    {
        if (energyLevel >= 15)
        {
            energyLevel -= 15;
            Console.WriteLine($"{Name} приносит мяч! Энергия: {energyLevel}%");
            OnEnergyChanged(energyLevel);
        }
        else
        {
            Console.WriteLine($"{Name} слишком устал(а), чтобы играть!");
        }
    }
    
    public void PerformCommand(string command)
    {
        if (IsTrained)
            Console.WriteLine($"{Name} выполняет команду: {command}");
        else
            Console.WriteLine($"{Name} еще не обучен(а)");
    }
}

public class Cat : Animal
{
    public string Color { get; set; }
    public bool IsIndoor { get; set; }
    public string OwnerName { get; set; }
    
    public Cat(string name, int age, double weight, string color, bool isIndoor, string ownerName)
        : base(name, age, weight)
    {
        Color = color;
        IsIndoor = isIndoor;
        OwnerName = ownerName;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип: Кошка, Окрас: {Color}, Домашняя: {(IsIndoor ? "Да" : "Нет")}, Хозяин: {OwnerName}");
    }
    
    public override void MakeSound()
    {
        string sound = "Мяу-мяу!";
        OnAnimalSound(sound);
        Console.WriteLine($"{Name} мяукает: {sound}");
    }
    
    public override void Move()
    {
        if (energyLevel >= 5)
        {
            int oldEnergy = energyLevel;
            energyLevel -= 5;
            OnAnimalMoved($"{Name} грациозно крадется");
            OnEnergyChanged(energyLevel);
            Console.WriteLine($"{Name} грациозно крадется. Энергия: {energyLevel}%");
        }
        else
        {
            Console.WriteLine($"{Name} слишком устала, чтобы двигаться!");
        }
    }
    
    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет: Мур-мур-мур...");
    }
    
    public void Scratch()
    {
        Console.WriteLine($"{Name} точит когти о когтеточку.");
    }
}

public class Bird : Animal
{
    public double WingSpan { get; set; }
    public bool CanFly { get; set; }
    public string Habitat { get; set; }
    
    public Bird(string name, int age, double weight, double wingSpan, bool canFly, string habitat)
        : base(name, age, weight)
    {
        WingSpan = wingSpan;
        CanFly = canFly;
        Habitat = habitat;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип: Птица, Размах крыльев: {WingSpan} м, Летает: {(CanFly ? "Да" : "Нет")}, Среда обитания: {Habitat}");
    }
    
    public override void MakeSound()
    {
        string sound = "Чик-чирик!";
        OnAnimalSound(sound);
        Console.WriteLine($"{Name} чирикает: {sound}");
    }
    
    public override void Move()
    {
        if (CanFly && energyLevel >= 15)
        {
            int oldEnergy = energyLevel;
            energyLevel -= 15;
            OnAnimalMoved($"{Name} летит");
            OnEnergyChanged(energyLevel);
            Console.WriteLine($"{Name} летит с размахом крыльев {WingSpan} м. Энергия: {energyLevel}%");
        }
        else if (!CanFly && energyLevel >= 5)
        {
            int oldEnergy = energyLevel;
            energyLevel -= 5;
            OnAnimalMoved($"{Name} ходит");
            OnEnergyChanged(energyLevel);
            Console.WriteLine($"{Name} ходит по земле. Энергия: {energyLevel}%");
        }
        else
        {
            Console.WriteLine($"{Name} слишком устала, чтобы двигаться!");
        }
    }
    
    public void BuildNest()
    {
        Console.WriteLine($"{Name} строит гнездо в {Habitat}.");
    }
}

public class Fish : Animal
{
    public string WaterType { get; set; }
    public double MaxDepth { get; set; }
    public string Habitat { get; set; }
    
    public Fish(string name, int age, double weight, string waterType, double maxDepth, string habitat)
        : base(name, age, weight)
    {
        WaterType = waterType;
        MaxDepth = maxDepth;
        Habitat = habitat;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип: Рыба, Тип воды: {WaterType}, Макс. глубина: {MaxDepth} м, Среда обитания: {Habitat}");
    }
    
    public override void MakeSound()
    {
        string sound = "Буль-буль!";
        OnAnimalSound(sound);
        Console.WriteLine($"{Name} издает булькающие звуки: {sound}");
    }
    
    public override void Move()
    {
        if (energyLevel >= 5)
        {
            int oldEnergy = energyLevel;
            energyLevel -= 5;
            OnAnimalMoved($"{Name} плавает");
            OnEnergyChanged(energyLevel);
            Console.WriteLine($"{Name} плавает в {WaterType} воде. Энергия: {energyLevel}%");
        }
        else
        {
            Console.WriteLine($"{Name} слишком устала, чтобы плавать!");
        }
    }
    
    public void Dive()
    {
        Console.WriteLine($"{Name} ныряет на глубину {MaxDepth} м!");
    }
}

// ========== ДЕМОНСТРАЦИЯ ДЛЯ ANIMAL ==========
class AnimalDelegateDemo
{
    public static void Run()
    {
        Console.WriteLine("========== ДЕЛЕГАТЫ, ЛЯМБДА-ВЫРАЖЕНИЯ И СОБЫТИЯ: ЖИВОТНЫЕ ==========\n");
        
        // ========== 1. ДЕЛЕГАТЫ ==========
        Console.WriteLine("=== 1. ДЕЛЕГАТЫ ===\n");
        
        Dog dog = new Dog("Рекс", 3, 25.5, "Овчарка", true, "Иван");
        Cat cat = new Cat("Муся", 2, 4.2, "серый", true, "Элина");
        Bird bird = new Bird("Кеша", 1, 0.35, 0.5, true, "лес");
        Fish fish = new Fish("Немо", 2, 0.15, "Соленая", 50, "океан");
        
        // Создание делегатов
        Action displayDog = () => dog.DisplayInfo();
        Action displayCat = () => cat.DisplayInfo();
        Action displayBird = () => bird.DisplayInfo();
        Action displayFish = () => fish.DisplayInfo();
        
        Console.WriteLine("Вызов методов через делегаты:");
        displayDog();
        Console.WriteLine();
        displayCat();
        Console.WriteLine();
        displayBird();
        Console.WriteLine();
        displayFish();
        
        // Многоадресный делегат
        Console.WriteLine("\n--- Многоадресный делегат ---");
        Action multiDelegate = displayDog + displayCat + displayBird + displayFish;
        multiDelegate();
        
        // ========== 2. ЛЯМБДА-ВЫРАЖЕНИЯ ==========
        Console.WriteLine("\n=== 2. ЛЯМБДА-ВЫРАЖЕНИЯ ===\n");
        
        List<Animal> animals = new List<Animal>
        {
            new Dog("Рекс", 3, 25.5, "Овчарка", true, "Иван"),
            new Dog("Шарик", 4, 20.0, "Дворняга", false, "Петр"),
            new Cat("Муся", 2, 4.2, "серый", true, "Элина"),
            new Cat("Васька", 3, 3.8, "черный", false, "Анна"),
            new Bird("Кеша", 1, 0.35, 0.5, true, "лес"),
            new Bird("Пингвин", 5, 30.0, 0.8, false, "Антарктида"),
            new Fish("Немо", 2, 0.15, "Соленая", 50, "океан"),
            new Fish("Золотая", 1, 0.1, "Пресная", 2, "пруд")
        };
        
        // Фильтрация с помощью лямбда-выражений
        Console.WriteLine("Дрессированные собаки:");
        var trainedDogs = animals.OfType<Dog>().Where(d => d.IsTrained);
        foreach (var d in trainedDogs)
        {
            Console.WriteLine($"  {d.Name} ({d.Breed})");
        }
        
        Console.WriteLine("\nДомашние кошки:");
        var indoorCats = animals.OfType<Cat>().Where(c => c.IsIndoor);
        foreach (var c in indoorCats)
        {
            Console.WriteLine($"  {c.Name} ({c.Color})");
        }
        
        Console.WriteLine("\nЛетающие птицы:");
        var flyingBirds = animals.OfType<Bird>().Where(b => b.CanFly);
        foreach (var b in flyingBirds)
        {
            Console.WriteLine($"  {b.Name} (размах крыльев: {b.WingSpan} м)");
        }
        
        // Сортировка с помощью лямбда-выражений
        Console.WriteLine("\nЖивотные по возрасту (от младших к старшим):");
        var sortedByAge = animals.OrderBy(a => a.Age);
        foreach (var a in sortedByAge)
        {
            Console.WriteLine($"  {a.Name} ({a.GetType().Name}) - {a.Age} лет");
        }
        
        Console.WriteLine("\nЖивотные по весу (от тяжелых к легким):");
        var sortedByWeight = animals.OrderByDescending(a => a.Weight);
        foreach (var a in sortedByWeight)
        {
            Console.WriteLine($"  {a.Name} ({a.GetType().Name}) - {a.Weight} кг");
        }
        
        // Проекция с помощью лямбда-выражений
        Console.WriteLine("\nСписок всех имен животных:");
        var names = animals.Select(a => a.Name);
        foreach (var n in names)
        {
            Console.WriteLine($"  {n}");
        }
        
        // Агрегация с помощью лямбда-выражений
        double avgWeight = animals.Average(a => a.Weight);
        Console.WriteLine($"\nСредний вес всех животных: {avgWeight:F1} кг");
        
        double avgAge = animals.Average(a => a.Age);
        Console.WriteLine($"Средний возраст всех животных: {avgAge:F1} лет");
        
        var heaviestAnimal = animals.OrderByDescending(a => a.Weight).First();
        Console.WriteLine($"Самое тяжелое животное: {heaviestAnimal.Name} ({heaviestAnimal.Weight} кг)");
        
        // ========== 3. СОБЫТИЯ ==========
        Console.WriteLine("\n=== 3. СОБЫТИЯ ===\n");
        
        Dog eventDog = new Dog("Рекс", 3, 25.5, "Овчарка", true, "Иван");
        
        // Подписка на события с помощью лямбда-выражений
        eventDog.AnimalFed += (message) => Console.WriteLine($"[СОБЫТИЕ] Кормление: {message}");
        eventDog.AnimalSlept += (message) => Console.WriteLine($"[СОБЫТИЕ] Сон: {message}");
        eventDog.EnergyChanged += (energy) => Console.WriteLine($"[СОБЫТИЕ] Энергия изменилась: {energy}%");
        eventDog.AnimalSound += (sound) => Console.WriteLine($"[СОБЫТИЕ] Звук: {sound}");
        eventDog.AnimalMoved += (message) => Console.WriteLine($"[СОБЫТИЕ] Движение: {message}");
        
        Console.WriteLine("Демонстрация событий для собаки:");
        eventDog.MakeSound();
        eventDog.Move();
        eventDog.Eat();
        eventDog.Move();
        eventDog.Fetch();
        eventDog.Sleep();
        
        // Демонстрация для кошки
        Console.WriteLine("\n--- События для кошки ---");
        Cat eventCat = new Cat("Муся", 2, 4.2, "серый", true, "Элина");
        
        eventCat.AnimalSound += (sound) => Console.WriteLine($"Кошка издает звук: {sound}");
        eventCat.EnergyChanged += (energy) => Console.WriteLine($"Энергия кошки: {energy}%");
        
        eventCat.MakeSound();
        eventCat.Move();
        eventCat.Purr();
        eventCat.Eat();
        
        // ========== 4. ИСПОЛЬЗОВАНИЕ ДЕЛЕГАТОВ В КАЧЕСТВЕ ПАРАМЕТРОВ ==========
        Console.WriteLine("\n=== 4. ДЕЛЕГАТЫ КАК ПАРАМЕТРЫ МЕТОДОВ ===\n");
        
        void ProcessAnimal(Animal animal, Action displayAction)
        {
            Console.WriteLine($"Обработка: {animal.Name}");
            displayAction();
        }
        
        ProcessAnimal(dog, () => dog.DisplayInfo());
        ProcessAnimal(cat, () => cat.DisplayInfo());
        ProcessAnimal(bird, () => bird.DisplayInfo());
        
        // ========== 5. ФУНКЦИОНАЛЬНЫЕ ДЕЛЕГАТЫ ==========
        Console.WriteLine("\n=== 5. ФУНКЦИОНАЛЬНЫЕ ДЕЛЕГАТЫ (Func, Action, Predicate) ===\n");
        
        // Func - возвращает значение
        Func<Animal, string> getDescription = a => $"{a.Name} ({a.GetType().Name}), {a.Age} лет, {a.Weight} кг";
        Console.WriteLine($"Описание: {getDescription(dog)}");
        Console.WriteLine($"Описание: {getDescription(cat)}");
        
        // Action - не возвращает значение
        Action<Animal> printShortInfo = a => Console.WriteLine($"  {a.Name} - {a.GetType().Name}");
        Console.WriteLine("Краткая информация о животных:");
        animals.Take(5).ToList().ForEach(printShortInfo);
        
        // Predicate - возвращает bool
        Predicate<Animal> isHeavy = a => a.Weight > 10;
        bool hasHeavyAnimal = animals.Exists(a => isHeavy(a));
        Console.WriteLine($"\nЕсть ли тяжелые животные (>10 кг): {hasHeavyAnimal}");
        
        Predicate<Animal> isOld = a => a.Age > 4;
        bool hasOldAnimal = animals.Exists(a => isOld(a));
        Console.WriteLine($"Есть ли старые животные (>4 лет): {hasOldAnimal}");
        
        // ========== 6. ДОПОЛНИТЕЛЬНАЯ ДЕМОНСТРАЦИЯ ==========
        Console.WriteLine("\n=== 6. ДОПОЛНИТЕЛЬНАЯ ДЕМОНСТРАЦИЯ ===\n");
        
        // Группировка с помощью лямбда-выражений
        var groupedByType = animals.GroupBy(a => a.GetType().Name);
        Console.WriteLine("Группировка животных по типу:");
        foreach (var group in groupedByType)
        {
            Console.WriteLine($"  {group.Key}: {group.Count()} особей");
            foreach (var a in group)
            {
                Console.WriteLine($"    - {a.Name}");
            }
        }
        
        // Поиск с помощью лямбда-выражений
        var specificCat = animals.OfType<Cat>().FirstOrDefault(c => c.Name == "Муся");
        if (specificCat != null)
        {
            Console.WriteLine($"\nНайдена кошка Муся:");
            specificCat.DisplayInfo();
        }
        
        Console.WriteLine("\n========== ДЕМОНСТРАЦИЯ ANIMAL ЗАВЕРШЕНА ==========");
    }
}

// ========== ЗАПУСК ДЕМОНСТРАЦИИ ==========

Console.WriteLine("\n" + new string('=', 100) + "\n");
AnimalDelegateDemo.Run();
````
