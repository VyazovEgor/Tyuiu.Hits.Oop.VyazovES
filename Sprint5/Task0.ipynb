<h1 style="color:DodgerBlue">Коллекции в .NET </h1>



Коллекции в .NET - это структуры данных, которые позволяют организовать и управлять наборами объектов. Они предоставляют эффективные способы хранения, доступа и манипулирования данными. Вот основные типы коллекций в .NET:

### 1. Статические коллекции

Статические коллекции являются неизменяемыми и могут содержать только один экземпляр заданного типа данных.

- `System.Collections.Generic.EmptyCollection<T>`
- `System.Collections.Generic.Singleton<T>`
- `System.Collections.ObjectModel.ReadOnlyCollection<T>`

### 2. Неупорядоченные коллекции

Неупорядоченные коллекции не гарантируют порядок элементов.

- `System.Collections.Generic.List<T>` - динамический массив
- `System.Collections.Generic.Dictionary<TKey,TValue>` - словарь (ассоциативный массив)
- `System.Collections.Generic.HashSet<T>` - множество уникальных элементов
- `System.Collections.Generic.SortedSet<T>` - отсортированный набор элементов

### 3. Сортированные коллекции

Сортированные коллекции поддерживают упорядоченный доступ и могут быть отсортированы по определенному критерию.

- `System.Collections.Generic.SortedDictionary<TKey,TValue>`
- `System.Collections.Generic.SortedSet<T>`
- `System.Collections.Generic.SortedList<TKey,TValue>`

### 4. Множества

Множества содержат только уникальные элементы.

- `System.Collections.Generic.HashSet<T>`
- `System.Collections.Generic.ISet<T>` - интерфейс для неупорядоченных множеств

### 5. Ключевые коллекции

Ключевые коллекции используются для хранения пар ключ-значение.

- `System.Collections.Generic.Dictionary<TKey,TValue>`
- `System.Collections.Generic.KeyedCollection<TKey,TElement>`
- `System.Collections.Generic.Lookup<TKey,TValue>`

### 6. Коллекции с ограниченным размером

Эти коллекции имеют фиксированный или максимальный размер.

- `System.Collections.Generic.Stack<T>` - стек
- `System.Collections.Generic.Queue<T>` - очередь
- `System.Collections.Generic.ConcurrentQueue<T>` - потокобезопасная очередь

### 7. Потокобезопасные коллекции

Потокобезопасные коллекции обеспечивают безопасность доступа из нескольких потоков.

- `System.Collections.Concurrent.ConcurrentDictionary<TKey,TValue>`
- `System.Collections.Concurrent.ConcurrentBag<T>`
- `System.Collections.Concurrent.BlockingCollection<T>`

### 8. Специализированные коллекции

Специализированные коллекции предназначены для конкретных задач.

- `System.Collections.Generic.BCLPortabilityLibrary.Collections.Generic.MultiMap<TKey,TKey2,TValue>`
- `System.Collections.Generic.BCLPortabilityLibrary.Collections.Generic.MultiDictionary<TKey,TValue>`

Это основные типы коллекций в .NET Framework и .NET Core. Выбор конкретной коллекции зависит от требований вашего проекта, таких как необходимость сортировки, потокобезопасности или специфики использования данных.

Работа с `List<T>` и `Dictionary<TKey, TValue>` в C# позволяет эффективно управлять коллекциями объектов различных типов. Давайте рассмотрим, как это можно сделать на примере классов `Transport`, `Car`, и `Bike`.

### Класс Transport и его производные

Начнем с базового класса `Transport`, который будет основой для классов `Car` и `Bike`.

```csharp
public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Модель: {Model}, Производитель: {Manufacturer}");
    }
}

public class Car : Transport
{
    public int NumberOfDoors { get; set; }

    public Car(string model, string manufacturer, int numberOfDoors)
        : base(model, manufacturer)
    {
        NumberOfDoors = numberOfDoors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Количество дверей: {NumberOfDoors}");
    }
}

public class Bike : Transport
{
    public bool HasCarrier { get; set; }

    public Bike(string model, string manufacturer, bool hasCarrier)
        : base(model, manufacturer)
    {
        HasCarrier = hasCarrier;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Имеет наситель: {HasCarrier}");
    }
}
```

### Работа с List<T>

Теперь давайте посмотрим, как работать с `List<T>` для хранения коллекции объектов транспорта.

```csharp
using System;
using System.Collections.Generic;


// Создаем список Transport
List<Transport> transportList = new List<Transport>();

// Добавляем объекты Car и Bike в список
transportList.Add(new Car("Model S", "Tesla", 4));
transportList.Add(new Bike("Mountain Explorer", "Giant", true));

// Перебираем и выводим информацию обо всех транспортных средствах
foreach (var transport in transportList)
{
    transport.DisplayInfo();
    Console.WriteLine();
}

```

### Работа с Dictionary<TKey, TValue>

Теперь давайте создадим коллекцию, которая будет хранить объекты `Transport`, используя `Dictionary<TKey, TValue>`, где ключом будет, например, регистрационный номер транспорта.

```csharp
using System;
using System.Collections.Generic;

// Создаем словарь с ключом типа string и значением типа Transport
Dictionary<string, Transport> transportDictionary = new Dictionary<string, Transport>();

// Добавляем объекты с уникальными ключами
transportDictionary.Add("TSL123", new Car("Model S", "Tesla", 4));
transportDictionary.Add("BIK456", new Bike("Mountain Explorer", "Giant", true));

// Поиск и вывод информации о транспорте по ключу
if (transportDictionary.TryGetValue("TSL123", out Transport foundTransport))
{
    Console.WriteLine("Найденый траснпорт:");
    foundTransport.DisplayInfo();
}
else
{
    Console.WriteLine("Транспорт не найден");
}

```

### Объяснения и примеры использования:

- **List<T>**: Используется, когда порядок элементов важен и возможны дубликаты. Вы можете добавлять, удалять, итерировать и изменять элементы по индексу.

- **Dictionary<TKey,TValue>**: Подходит, когда нужно организовать элементы по уникальным ключам для быстрого доступа. Это удобно для случаев, когда вы хотите находить элемент по ключу, как регистрационный номер в примере.

Эти структуры данных предоставляют мощные и гибкие возможности для управления коллекциями объектов, каждая из которых имеет свои преимущества в определённых ситуациях.

````
using System;
using System.Collections.Generic;
using System.Linq;

// ========== КЛАССЫ ТРАНСПОРТНЫХ СРЕДСТВ ==========

// Базовый класс Transport
public abstract class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }
    public int Year { get; set; }
    protected int currentSpeed;
    
    public Transport(string model, string manufacturer, int year)
    {
        Model = model;
        Manufacturer = manufacturer;
        Year = year;
        currentSpeed = 0;
    }
    
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Модель: {Model}, Производитель: {Manufacturer}, Год: {Year}");
    }
    
    public virtual void Start()
    {
        Console.WriteLine($"{Manufacturer} {Model} запущен.");
    }
    
    public virtual void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Manufacturer} {Model} остановлен.");
    }
    
    public abstract void Honk();
}

// Класс Car
public class Car : Transport
{
    public int NumberOfDoors { get; set; }
    public string BodyType { get; set; }
    public int Horsepower { get; set; }
    
    public Car(string model, string manufacturer, int year, int numberOfDoors, string bodyType, int horsepower) 
        : base(model, manufacturer, year)
    {
        NumberOfDoors = numberOfDoors;
        BodyType = bodyType;
        Horsepower = horsepower;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип: Автомобиль, Дверей: {NumberOfDoors}, Кузов: {BodyType}, Мощность: {Horsepower} л.с.");
    }
    
    public override void Honk()
    {
        Console.WriteLine($"{Manufacturer} {Model} сигналит: Бип-бип!");
    }
    
    public void Accelerate(int amount)
    {
        currentSpeed += amount;
        Console.WriteLine($"{Manufacturer} {Model} разгоняется до {currentSpeed} км/ч.");
    }
    
    public void Brake(int amount)
    {
        currentSpeed = Math.Max(0, currentSpeed - amount);
        Console.WriteLine($"{Manufacturer} {Model} тормозит до {currentSpeed} км/ч.");
    }
}

// Класс Bike (Мотоцикл)
public class Bike : Transport
{
    public bool HasSidecar { get; set; }
    public string EngineType { get; set; }
    public int Horsepower { get; set; }
    
    public Bike(string model, string manufacturer, int year, bool hasSidecar, string engineType, int horsepower) 
        : base(model, manufacturer, year)
    {
        HasSidecar = hasSidecar;
        EngineType = engineType;
        Horsepower = horsepower;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип: Мотоцикл, Двигатель: {EngineType}, Коляска: {(HasSidecar ? "Есть" : "Нет")}, Мощность: {Horsepower} л.с.");
    }
    
    public override void Honk()
    {
        Console.WriteLine($"{Manufacturer} {Model} сигналит: Бип!");
    }
    
    public void Wheelie()
    {
        if (currentSpeed > 20)
            Console.WriteLine($"{Manufacturer} {Model} едет на заднем колесе!");
        else
            Console.WriteLine($"Нужно разогнаться быстрее для вили! Текущая скорость: {currentSpeed} км/ч");
    }
}

// Класс Truck (Грузовик)
public class Truck : Transport
{
    public double LoadCapacity { get; set; }
    public int Axles { get; set; }
    
    public Truck(string model, string manufacturer, int year, double loadCapacity, int axles) 
        : base(model, manufacturer, year)
    {
        LoadCapacity = loadCapacity;
        Axles = axles;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип: Грузовик, Грузоподъемность: {LoadCapacity} т, Оси: {Axles}");
    }
    
    public override void Honk()
    {
        Console.WriteLine($"{Manufacturer} {Model} гудит: БИИИП-БИИИП!");
    }
    
    public void LoadCargo(double weight)
    {
        if (weight <= LoadCapacity)
            Console.WriteLine($"Грузовик {Manufacturer} {Model} загружен на {weight} тонн.");
        else
            Console.WriteLine($"Грузовик {Manufacturer} {Model} перегружен! Максимум {LoadCapacity} тонн.");
    }
}

// ========== ДЕМОНСТРАЦИЯ РАБОТЫ С КОЛЛЕКЦИЯМИ ДЛЯ TRANSPORT ==========
class TransportCollectionDemo
{
    public static void Run()
    {
        Console.WriteLine("========== КОЛЛЕКЦИИ: ТРАНСПОРТНЫЕ СРЕДСТВА ==========\n");
        
        // ========== 1. РАБОТА С LIST<TRANSPORT> ==========
        Console.WriteLine("=== 1. List<Transport> ===\n");
        
        List<Transport> transportList = new List<Transport>();
        
        // Добавляем объекты в список
        transportList.Add(new Car("Model S", "Tesla", 2023, 4, "Седан", 1020));
        transportList.Add(new Bike("Street Glide", "Harley-Davidson", 2022, false, "V-Twin", 95));
        transportList.Add(new Truck("FH16", "Volvo", 2023, 25.5, 3));
        transportList.Add(new Car("X5", "BMW", 2024, 5, "Внедорожник", 340));
        transportList.Add(new Bike("R1", "Yamaha", 2024, false, "Inline-4", 200));
        transportList.Add(new Car("Mustang", "Ford", 2023, 2, "Купе", 450));
        transportList.Add(new Truck("Actros", "Mercedes", 2022, 30, 4));
        
        Console.WriteLine("Все транспортные средства в списке:");
        foreach (var transport in transportList)
        {
            transport.DisplayInfo();
            transport.Start();
            transport.Honk();
            Console.WriteLine();
        }
        
        // 1.1 Фильтрация по типу
        Console.WriteLine("--- 1.1 Фильтрация по типу ---\n");
        
        var cars = transportList.OfType<Car>().ToList();
        Console.WriteLine($"Всего автомобилей: {cars.Count}");
        foreach (var car in cars)
        {
            Console.WriteLine($"  {car.Manufacturer} {car.Model} ({car.Horsepower} л.с.)");
        }
        
        var bikes = transportList.OfType<Bike>().ToList();
        Console.WriteLine($"\nВсего мотоциклов: {bikes.Count}");
        foreach (var bike in bikes)
        {
            Console.WriteLine($"  {bike.Manufacturer} {bike.Model} ({bike.Horsepower} л.с.)");
        }
        
        var trucks = transportList.OfType<Truck>().ToList();
        Console.WriteLine($"\nВсего грузовиков: {trucks.Count}");
        
        // 1.2 Поиск и фильтрация по свойствам
        Console.WriteLine("\n--- 1.2 Поиск и фильтрация по свойствам ---\n");
        
        var powerfulCars = cars.Where(c => c.Horsepower > 400).ToList();
        Console.WriteLine($"Мощные автомобили (>400 л.с.): {powerfulCars.Count}");
        foreach (var car in powerfulCars)
        {
            Console.WriteLine($"  {car.Manufacturer} {car.Model} - {car.Horsepower} л.с.");
        }
        
        var bmwCars = cars.Where(c => c.Manufacturer == "BMW").ToList();
        Console.WriteLine($"\nАвтомобили BMW: {bmwCars.Count}");
        
        var heavyTrucks = trucks.Where(t => t.LoadCapacity > 25).ToList();
        Console.WriteLine($"\nТяжелые грузовики (>25 т): {heavyTrucks.Count}");
        
        // 1.3 Сортировка
        Console.WriteLine("\n--- 1.3 Сортировка ---\n");
        
        var sortedByYear = transportList.OrderBy(t => t.Year).ToList();
        Console.WriteLine("Транспорт по году выпуска (от старых к новым):");
        foreach (var t in sortedByYear)
        {
            Console.WriteLine($"  {t.Manufacturer} {t.Model} - {t.Year} год");
        }
        
        var sortedByManufacturer = transportList.OrderBy(t => t.Manufacturer).ToList();
        Console.WriteLine("\nТранспорт по производителю:");
        foreach (var t in sortedByManufacturer)
        {
            Console.WriteLine($"  {t.Manufacturer} - {t.Model}");
        }
        
        // 1.4 Поиск первого элемента
        Console.WriteLine("\n--- 1.4 Поиск ---\n");
        
        var firstCar = transportList.OfType<Car>().FirstOrDefault();
        if (firstCar != null)
        {
            Console.WriteLine($"Первый автомобиль в списке: {firstCar.Manufacturer} {firstCar.Model}");
        }
        
        var teslaCar = transportList.OfType<Car>().FirstOrDefault(c => c.Manufacturer == "Tesla");
        if (teslaCar != null)
        {
            Console.WriteLine($"Найден Tesla: {teslaCar.Manufacturer} {teslaCar.Model}");
        }
        
        // 1.5 Удаление элементов
        Console.WriteLine("\n--- 1.5 Удаление элементов ---\n");
        
        transportList.RemoveAt(0);
        Console.WriteLine($"После удаления первого элемента осталось: {transportList.Count}");
        
        transportList.RemoveAll(t => t is Truck && ((Truck)t).LoadCapacity > 28);
        Console.WriteLine($"После удаления тяжелых грузовиков осталось: {transportList.Count}");
        
        // ========== 2. РАБОТА С DICTIONARY<STRING, TRANSPORT> ==========
        Console.WriteLine("\n=== 2. Dictionary<string, Transport> ===\n");
        
        Dictionary<string, Transport> transportDictionary = new Dictionary<string, Transport>();
        
        // Добавляем объекты с уникальными ключами (регистрационные номера)
        transportDictionary.Add("A123BC", new Car("Model S", "Tesla", 2023, 4, "Седан", 1020));
        transportDictionary.Add("B456CD", new Bike("Street Glide", "Harley-Davidson", 2022, false, "V-Twin", 95));
        transportDictionary.Add("C789DE", new Truck("FH16", "Volvo", 2023, 25.5, 3));
        transportDictionary.Add("D012EF", new Car("X5", "BMW", 2024, 5, "Внедорожник", 340));
        transportDictionary.Add("E345FG", new Bike("R1", "Yamaha", 2024, false, "Inline-4", 200));
        transportDictionary.Add("F678GH", new Car("Mustang", "Ford", 2023, 2, "Купе", 450));
        transportDictionary.Add("G901HI", new Truck("Actros", "Mercedes", 2022, 30, 4));
        
        Console.WriteLine("Все транспортные средства в словаре:");
        foreach (var kvp in transportDictionary)
        {
            Console.WriteLine($"Рег. номер: {kvp.Key}");
            kvp.Value.DisplayInfo();
            Console.WriteLine();
        }
        
        // 2.1 Поиск по ключу
        Console.WriteLine("--- 2.1 Поиск по ключу ---\n");
        
        string searchPlate = "D012EF";
        if (transportDictionary.TryGetValue(searchPlate, out Transport foundTransport))
        {
            Console.WriteLine($"Найден транспорт по номеру {searchPlate}:");
            foundTransport.DisplayInfo();
            
            if (foundTransport is Car foundCar)
            {
                foundCar.Accelerate(50);
                foundCar.Brake(20);
            }
        }
        else
        {
            Console.WriteLine($"Транспорт с номером {searchPlate} не найден");
        }
        
        // 2.2 Поиск по значению
        Console.WriteLine("\n--- 2.2 Поиск по значению ---\n");
        
        var bmwEntries = transportDictionary.Where(kvp => kvp.Value is Car car && car.Manufacturer == "BMW").ToList();
        Console.WriteLine($"Автомобили BMW в словаре: {bmwEntries.Count}");
        foreach (var entry in bmwEntries)
        {
            Console.WriteLine($"  {entry.Key}: {((Car)entry.Value).Manufacturer} {((Car)entry.Value).Model}");
        }
        
        // 2.3 Обновление элемента
        Console.WriteLine("\n--- 2.3 Обновление элемента ---\n");
        
        if (transportDictionary.ContainsKey("A123BC"))
        {
            Console.WriteLine("Обновляем информацию об автомобиле Tesla:");
            transportDictionary["A123BC"] = new Car("Model S Plaid", "Tesla", 2024, 4, "Седан", 1020);
            transportDictionary["A123BC"].DisplayInfo();
        }
        
        // 2.4 Удаление элемента
        Console.WriteLine("\n--- 2.4 Удаление элемента ---\n");
        
        transportDictionary.Remove("C789DE");
        Console.WriteLine("После удаления грузовика с номером C789DE:");
        foreach (var kvp in transportDictionary)
        {
            Console.WriteLine($"  {kvp.Key}: {kvp.Value.Manufacturer} {kvp.Value.Model}");
        }
        
        // 2.5 Проверка наличия ключа
        Console.WriteLine("\n--- 2.5 Проверка наличия ключа ---\n");
        
        string checkPlate = "Z999ZZ";
        if (transportDictionary.ContainsKey(checkPlate))
            Console.WriteLine($"Номер {checkPlate} существует");
        else
            Console.WriteLine($"Номер {checkPlate} не найден в словаре");
        
        // ========== 3. СТАТИСТИКА ==========
        Console.WriteLine("\n=== 3. СТАТИСТИКА ===\n");
        
        Console.WriteLine($"Всего транспортных средств в List: {transportList.Count}");
        Console.WriteLine($"Всего транспортных средств в Dictionary: {transportDictionary.Count}");
        
        double avgYear = transportList.Average(t => t.Year);
        Console.WriteLine($"Средний год выпуска: {avgYear:F0}");
        
        int totalHorsepower = transportList.OfType<Car>().Sum(c => c.Horsepower) + 
                              transportList.OfType<Bike>().Sum(b => b.Horsepower);
        Console.WriteLine($"Суммарная мощность всех автомобилей и мотоциклов: {totalHorsepower} л.с.");
        
        var maxHorsepowerCar = transportList.OfType<Car>().OrderByDescending(c => c.Horsepower).FirstOrDefault();
        if (maxHorsepowerCar != null)
        {
            Console.WriteLine($"Самый мощный автомобиль: {maxHorsepowerCar.Manufacturer} {maxHorsepowerCar.Model} ({maxHorsepowerCar.Horsepower} л.с.)");
        }
        
        Console.WriteLine("\n========== ДЕМОНСТРАЦИЯ TRANSPORT ЗАВЕРШЕНА ==========\n");
    }
}


TransportCollectionDemo.Run();
````

````
using System;
using System.Collections.Generic;
using System.Linq;

// ========== КЛАССЫ ЖИВОТНЫХ ==========

// Базовый класс Animal
public abstract class Animal
{
    public string Name { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }
    protected int energyLevel;
    
    public Animal(string name, int age, double weight)
    {
        Name = name;
        Age = age;
        Weight = weight;
        energyLevel = 100;
    }
    
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Имя: {Name}, Возраст: {Age} лет, Вес: {Weight} кг, Энергия: {energyLevel}%");
    }
    
    public abstract void MakeSound();
    public abstract void Move();
    
    public virtual void Eat()
    {
        energyLevel = Math.Min(100, energyLevel + 20);
        Console.WriteLine($"{Name} поел. Энергия: {energyLevel}%");
    }
    
    public virtual void Sleep()
    {
        energyLevel = 100;
        Console.WriteLine($"{Name} выспался. Энергия: {energyLevel}%");
    }
}

// Класс Dog
public class Dog : Animal
{
    public string Breed { get; set; }
    public bool IsTrained { get; set; }
    public string OwnerName { get; set; }
    
    public Dog(string name, int age, double weight, string breed, bool isTrained, string ownerName) 
        : base(name, age, weight)
    {
        Breed = breed;
        IsTrained = isTrained;
        OwnerName = ownerName;
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"Собака {Name} лает: Гав-гав!");
    }
    
    public override void Move()
    {
        if (energyLevel >= 10)
        {
            energyLevel -= 10;
            Console.WriteLine($"Собака {Name} бегает на четырех лапах. Энергия: {energyLevel}%");
        }
        else
        {
            Console.WriteLine($"Собака {Name} слишком устала, чтобы двигаться!");
        }
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Порода: {Breed}, Дрессирована: {(IsTrained ? "Да" : "Нет")}, Хозяин: {OwnerName}");
    }
    
    public void Fetch()
    {
        Console.WriteLine($"{Name} приносит мяч!");
    }
    
    public void PerformCommand(string command)
    {
        if (IsTrained)
            Console.WriteLine($"{Name} выполняет команду: {command}");
        else
            Console.WriteLine($"{Name} еще не обучен(а)");
    }
}

// Класс Cat
public class Cat : Animal
{
    public string Color { get; set; }
    public bool IsIndoor { get; set; }
    public string OwnerName { get; set; }
    
    public Cat(string name, int age, double weight, string color, bool isIndoor, string ownerName) 
        : base(name, age, weight)
    {
        Color = color;      // цвет "серый"
        IsIndoor = isIndoor;
        OwnerName = ownerName;
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"Кошка {Name} мяукает: Мяу-мяу!");
    }
    
    public override void Move()
    {
        if (energyLevel >= 5)
        {
            energyLevel -= 5;
            Console.WriteLine($"Кошка {Name} грациозно крадется. Энергия: {energyLevel}%");
        }
        else
        {
            Console.WriteLine($"Кошка {Name} слишком устала, чтобы двигаться!");
        }
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Окрас: {Color}, Домашняя: {(IsIndoor ? "Да" : "Нет")}, Хозяин: {OwnerName}");
    }
    
    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет от удовольствия!");
    }
    
    public void Scratch()
    {
        Console.WriteLine($"{Name} точит когти о когтеточку.");
    }
}

// Класс Bird
public class Bird : Animal
{
    public double WingSpan { get; set; }
    public bool CanFly { get; set; }
    public string Habitat { get; set; }
    
    public Bird(string name, int age, double weight, double wingSpan, bool canFly, string habitat) 
        : base(name, age, weight)
    {
        WingSpan = wingSpan;
        CanFly = canFly;
        Habitat = habitat;
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"Птица {Name} чирикает: Чик-чирик!");
    }
    
    public override void Move()
    {
        if (CanFly && energyLevel >= 15)
        {
            energyLevel -= 15;
            Console.WriteLine($"Птица {Name} летит с размахом крыльев {WingSpan} м. Энергия: {energyLevel}%");
        }
        else if (!CanFly && energyLevel >= 5)
        {
            energyLevel -= 5;
            Console.WriteLine($"Птица {Name} ходит по земле. Энергия: {energyLevel}%");
        }
        else
        {
            Console.WriteLine($"Птица {Name} слишком устала, чтобы двигаться!");
        }
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Размах крыльев: {WingSpan} м, Летает: {(CanFly ? "Да" : "Нет")}, Среда обитания: {Habitat}");
    }
    
    public void BuildNest()
    {
        Console.WriteLine($"{Name} строит гнездо в {Habitat}.");
    }
    
    public void Fly()
    {
        if (CanFly)
            Console.WriteLine($"{Name} летит высоко в небе!");
        else
            Console.WriteLine($"{Name} не может летать");
    }
}

// Класс Fish
public class Fish : Animal
{
    public string WaterType { get; set; }
    public double MaxDepth { get; set; }
    public string Habitat { get; set; }
    
    public Fish(string name, int age, double weight, string waterType, double maxDepth, string habitat) 
        : base(name, age, weight)
    {
        WaterType = waterType;
        MaxDepth = maxDepth;
        Habitat = habitat;
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"Рыба {Name} издает булькающие звуки: Буль-буль!");
    }
    
    public override void Move()
    {
        if (energyLevel >= 5)
        {
            energyLevel -= 5;
            Console.WriteLine($"Рыба {Name} плавает в {WaterType} воде. Энергия: {energyLevel}%");
        }
        else
        {
            Console.WriteLine($"Рыба {Name} слишком устала, чтобы плавать!");
        }
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип воды: {WaterType}, Макс. глубина: {MaxDepth} м, Среда обитания: {Habitat}");
    }
    
    public void Swim()
    {
        Console.WriteLine($"{Name} плавает в {WaterType} воде на глубине до {MaxDepth} м.");
    }
    
    public void Dive()
    {
        Console.WriteLine($"{Name} ныряет на глубину {MaxDepth} м!");
    }
}

// ========== ДЕМОНСТРАЦИЯ РАБОТЫ С КОЛЛЕКЦИЯМИ ДЛЯ ANIMAL ==========
class AnimalCollectionDemo
{
    public static void Run()
    {
        Console.WriteLine("========== КОЛЛЕКЦИИ: ЖИВОТНЫЕ ==========\n");
        
        // ========== 1. РАБОТА С LIST<ANIMAL> ==========
        Console.WriteLine("=== 1. List<Animal> ===\n");
        
        List<Animal> animalList = new List<Animal>();
        
        // Добавляем объекты в список
        animalList.Add(new Dog("Рекс", 3, 25.5, "Овчарка", true, "Иван"));
        animalList.Add(new Cat("Муся", 2, 4.2, "серый", true, "Элина"));
        animalList.Add(new Bird("Кеша", 1, 0.35, 0.5, true, "лес"));
        animalList.Add(new Fish("Немо", 2, 0.15, "Соленая", 50, "океан"));
        animalList.Add(new Dog("Шарик", 4, 20.0, "Дворняга", false, "Петр"));
        animalList.Add(new Cat("Васька", 3, 3.8, "черный", false, "Анна"));
        animalList.Add(new Bird("Пингвин", 5, 30.0, 0.8, false, "Антарктида"));
        animalList.Add(new Fish("Золотая", 1, 0.1, "Пресная", 2, "пруд"));
        animalList.Add(new Dog("Лайка", 2, 18.5, "Хаски", true, "Сергей"));
        
        Console.WriteLine("Все животные в списке:");
        foreach (var animal in animalList)
        {
            animal.DisplayInfo();
            animal.MakeSound();
            animal.Move();
            Console.WriteLine();
        }
        
        // 1.1 Фильтрация по типу
        Console.WriteLine("--- 1.1 Фильтрация по типу ---\n");
        
        var dogs = animalList.OfType<Dog>().ToList();
        Console.WriteLine($"Всего собак: {dogs.Count}");
        foreach (var dog in dogs)
        {
            Console.WriteLine($"  {dog.Name} - {dog.Breed}, хозяин: {dog.OwnerName}");
        }
        
        var cats = animalList.OfType<Cat>().ToList();
        Console.WriteLine($"\nВсего кошек: {cats.Count}");
        foreach (var cat in cats)
        {
            Console.WriteLine($"  {cat.Name} - {cat.Color}, хозяин: {cat.OwnerName}");
        }
        
        var birds = animalList.OfType<Bird>().ToList();
        Console.WriteLine($"\nВсего птиц: {birds.Count}");
        
        var fishes = animalList.OfType<Fish>().ToList();
        Console.WriteLine($"\nВсего рыб: {fishes.Count}");
        
        // 1.2 Поиск и фильтрация по свойствам
        Console.WriteLine("\n--- 1.2 Поиск и фильтрация по свойствам ---\n");
        
        // Поиск кошки Муси
        var musya = animalList.OfType<Cat>().FirstOrDefault(c => c.Name == "Муся");
        if (musya != null)
        {
            Console.WriteLine("Найдена кошка Муся:");
            musya.DisplayInfo();
            musya.Purr();
        }
        
        var trainedDogs = dogs.Where(d => d.IsTrained).ToList();
        Console.WriteLine($"\nДрессированные собаки: {trainedDogs.Count}");
        foreach (var dog in trainedDogs)
        {
            Console.WriteLine($"  {dog.Name} - {dog.Breed}");
        }
        
        var indoorCats = cats.Where(c => c.IsIndoor).ToList();
        Console.WriteLine($"\nДомашние кошки: {indoorCats.Count}");
        
        var flyingBirds = birds.Where(b => b.CanFly).ToList();
        Console.WriteLine($"\nЛетающие птицы: {flyingBirds.Count}");
        
        // 1.3 Сортировка
        Console.WriteLine("\n--- 1.3 Сортировка ---\n");
        
        var sortedByAge = animalList.OrderBy(a => a.Age).ToList();
        Console.WriteLine("Животные по возрасту (от младших к старшим):");
        foreach (var a in sortedByAge)
        {
            Console.WriteLine($"  {a.Name} ({a.GetType().Name}) - {a.Age} лет");
        }
        
        var sortedByWeight = animalList.OrderByDescending(a => a.Weight).ToList();
        Console.WriteLine("\nЖивотные по весу (от тяжелых к легким):");
        foreach (var a in sortedByWeight)
        {
            Console.WriteLine($"  {a.Name} ({a.GetType().Name}) - {a.Weight} кг");
        }
        
        var sortedByName = animalList.OrderBy(a => a.Name).ToList();
        Console.WriteLine("\nЖивотные по имени (алфавитный порядок):");
        foreach (var a in sortedByName)
        {
            Console.WriteLine($"  {a.Name}");
        }
        
        // 1.4 Поиск первого элемента
        Console.WriteLine("\n--- 1.4 Поиск ---\n");
        
        var firstDog = animalList.OfType<Dog>().FirstOrDefault();
        if (firstDog != null)
        {
            Console.WriteLine($"Первая собака в списке: {firstDog.Name} ({firstDog.Breed})");
        }
        
        var oldestAnimal = animalList.OrderByDescending(a => a.Age).FirstOrDefault();
        if (oldestAnimal != null)
        {
            Console.WriteLine($"Самое старое животное: {oldestAnimal.Name} ({oldestAnimal.Age} лет)");
        }
        
        // 1.5 Удаление элементов
        Console.WriteLine("\n--- 1.5 Удаление элементов ---\n");
        
        animalList.RemoveAt(0);
        Console.WriteLine($"После удаления первого элемента осталось: {animalList.Count}");
        
        animalList.RemoveAll(a => a is Fish && ((Fish)a).Weight < 0.2);
        Console.WriteLine($"После удаления маленьких рыб осталось: {animalList.Count}");
        
        // ========== 2. РАБОТА С DICTIONARY<STRING, ANIMAL> ==========
        Console.WriteLine("\n=== 2. Dictionary<string, Animal> ===\n");
        
        Dictionary<string, Animal> animalDictionary = new Dictionary<string, Animal>();
        
        // Добавляем объекты с уникальными ключами (клички)
        animalDictionary.Add("Рекс", new Dog("Рекс", 3, 25.5, "Овчарка", true, "Иван"));
        animalDictionary.Add("Муся", new Cat("Муся", 2, 4.2, "серый", true, "Элина"));
        animalDictionary.Add("Кеша", new Bird("Кеша", 1, 0.35, 0.5, true, "лес"));
        animalDictionary.Add("Немо", new Fish("Немо", 2, 0.15, "Соленая", 50, "океан"));
        animalDictionary.Add("Шарик", new Dog("Шарик", 4, 20.0, "Дворняга", false, "Петр"));
        animalDictionary.Add("Васька", new Cat("Васька", 3, 3.8, "черный", false, "Анна"));
        animalDictionary.Add("Пингвин", new Bird("Пингвин", 5, 30.0, 0.8, false, "Антарктида"));
        animalDictionary.Add("Золотая", new Fish("Золотая", 1, 0.1, "Пресная", 2, "пруд"));
        
        Console.WriteLine("Все животные в словаре:");
        foreach (var kvp in animalDictionary)
        {
            Console.WriteLine($"Кличка: {kvp.Key}");
            kvp.Value.DisplayInfo();
            Console.WriteLine();
        }
        
        // 2.1 Поиск по ключу
        Console.WriteLine("--- 2.1 Поиск по ключу ---\n");
        
        string searchName = "Муся";
        if (animalDictionary.TryGetValue(searchName, out Animal foundAnimal))
        {
            Console.WriteLine($"Найдено животное по кличке {searchName}:");
            foundAnimal.DisplayInfo();
            if (foundAnimal is Cat cat)
            {
                cat.Purr();
            }
        }
        else
        {
            Console.WriteLine($"Животное с кличкой {searchName} не найдено");
        }
        
        // 2.2 Поиск по значению
        Console.WriteLine("\n--- 2.2 Поиск по значению ---\n");
        
        var dogsInDict = animalDictionary.Where(kvp => kvp.Value is Dog).ToList();
        Console.WriteLine($"Собаки в словаре: {dogsInDict.Count}");
        foreach (var dog in dogsInDict)
        {
            Console.WriteLine($"  {dog.Key} - {((Dog)dog.Value).Breed}, хозяин: {((Dog)dog.Value).OwnerName}");
        }
        
        var catsInDict = animalDictionary.Where(kvp => kvp.Value is Cat).ToList();
        Console.WriteLine($"\nКошки в словаре: {catsInDict.Count}");
        foreach (var cat in catsInDict)
        {
            Console.WriteLine($"  {cat.Key} - {((Cat)cat.Value).Color}, хозяин: {((Cat)cat.Value).OwnerName}");
        }
        
        // 2.3 Обновление элемента
        Console.WriteLine("\n--- 2.3 Обновление элемента ---\n");
        
        if (animalDictionary.ContainsKey("Муся"))
        {
            Console.WriteLine("Обновляем информацию о кошке Мусе (увеличиваем возраст):");
            animalDictionary["Муся"] = new Cat("Муся", 3, 4.5, "серый", true, "Элина");
            animalDictionary["Муся"].DisplayInfo();
        }
        
        // 2.4 Удаление элемента
        Console.WriteLine("\n--- 2.4 Удаление элемента ---\n");
        
        animalDictionary.Remove("Васька");
        Console.WriteLine("После удаления кошки Васька:");
        foreach (var kvp in animalDictionary)
        {
            Console.WriteLine($"  {kvp.Key}: {kvp.Value.GetType().Name} - {kvp.Value.Name}");
        }
        
        // 2.5 Проверка наличия ключа
        Console.WriteLine("\n--- 2.5 Проверка наличия ключа ---\n");
        
        string checkName = "Мурзик";
        if (animalDictionary.ContainsKey(checkName))
            Console.WriteLine($"Кличка {checkName} существует");
        else
            Console.WriteLine($"Кличка {checkName} не найдена в словаре");
        
        // ========== 3. СТАТИСТИКА ==========
        Console.WriteLine("\n=== 3. СТАТИСТИКА ===\n");
        
        Console.WriteLine($"Всего животных в List: {animalList.Count}");
        Console.WriteLine($"Всего животных в Dictionary: {animalDictionary.Count}");
        
        double avgAge = animalList.Average(a => a.Age);
        Console.WriteLine($"Средний возраст животных: {avgAge:F1} лет");
        
        double avgWeight = animalList.Average(a => a.Weight);
        Console.WriteLine($"Средний вес животных: {avgWeight:F1} кг");
        
        var heaviestAnimal = animalList.OrderByDescending(a => a.Weight).FirstOrDefault();
        if (heaviestAnimal != null)
        {
            Console.WriteLine($"Самое тяжелое животное: {heaviestAnimal.Name} ({heaviestAnimal.Weight} кг)");
        }
        
        var oldestAnimal2 = animalList.OrderByDescending(a => a.Age).FirstOrDefault();
        if (oldestAnimal2 != null)
        {
            Console.WriteLine($"Самое старое животное: {oldestAnimal2.Name} ({oldestAnimal2.Age} лет)");
        }
        
        // Группировка по типу
        var groupedByType = animalList.GroupBy(a => a.GetType().Name);
        Console.WriteLine("\nГруппировка животных по типу:");
        foreach (var group in groupedByType)
        {
            Console.WriteLine($"  {group.Key}: {group.Count()} особей");
            foreach (var animal in group)
            {
                Console.WriteLine($"    - {animal.Name}");
            }
        }
        
        // Информация о кошке Мусе
        Console.WriteLine("\n--- Информация о кошке Мусе ---");
        var musyaFinal = animalList.OfType<Cat>().FirstOrDefault(c => c.Name == "Муся");
        if (musyaFinal != null)
        {
            Console.WriteLine($"Имя: {musyaFinal.Name}");
            Console.WriteLine($"Возраст: {musyaFinal.Age} года");
            Console.WriteLine($"Цвет: {musyaFinal.Color}");
            Console.WriteLine($"Хозяин: {musyaFinal.OwnerName}");
            Console.WriteLine($"Домашняя: {(musyaFinal.IsIndoor ? "Да" : "Нет")}");
            musyaFinal.Purr();
        }
        
        Console.WriteLine("\n========== ДЕМОНСТРАЦИЯ ANIMAL ЗАВЕРШЕНА ==========");
    }
}

// ========== ЗАПУСК ДЕМОНСТРАЦИИ ==========
AnimalCollectionDemo.Run(); 
````
