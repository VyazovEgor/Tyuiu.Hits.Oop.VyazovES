### Перекрытие методов в C#

Перекрытие методов (или переопределение) — это процесс, при котором производный класс предоставляет новую реализацию метода, уже существующего в его базовом классе. Это позволяет изменить или расширить поведение метода в производном классе без изменения исходного метода в базовом классе. Перекрытие методов является ключевой особенностью полиморфизма в объектно-ориентированном программировании.

#### Пример с классами `Vehicle`, `Car` и `Motorcycle`

Давайте рассмотрим пример с классами `Vehicle`, `Car` и `Motorcycle`, где каждый из этих классов имеет метод `Drive()`, который принимает параметр `speed`.

```csharp
// Базовый класс Vehicle с виртуальным методом Drive()
public class Vehicle
{
    public virtual void Drive(int speed)
    {
        Console.WriteLine($"Транспорт движется {speed} km/h.");
    }
}

// Производный класс Car, переопределяющий метод Drive()
public class Car : Vehicle
{
    public override void Drive(int speed)
    {
        Console.WriteLine($"Автомобиль движется {speed} km/h.");
    }
}

// Производный класс Motorcycle, также переопределяющий метод Drive()
public class Motorcycle : Vehicle
{
    public override void Drive(int speed)
    {
        Console.WriteLine($"Мотоцикл движется {speed} km/h.");
    }
}
```

В этом примере, метод `Drive()` в классе `Car` и `Motorcycle` переопределяет метод `Drive()` из базового класса `Vehicle`. Это означает, что при вызове метода `Drive()` на объекте типа `Car` или `Motorcycle`, будет вызвана соответствующая реализация метода в этих классах, а не в классе `Vehicle`.

#### Отличие перекрытия от перегрузки методов

Перегрузка методов — это процесс определения нескольких методов с одним и тем же именем, но с разным количеством или типом параметров. Например:

```csharp
public class ExampleClass
{
    public void Print(int number)
    {
        Console.WriteLine(number);
    }

    public void Print(double number)
    {
        Console.WriteLine(number.ToString());
    }
}
```

В этом случае, у нас есть два метода с именем `Print`, но с разным типом параметра. Это отличается от перекрытия методов, где метод в производном классе полностью заменяет метод в базовом классе.

#### Вызов методов родителя

Иногда в производном классе может потребоваться вызвать метод родителя. Для этого используется ключевое слово `base`. Например, если мы хотим добавить дополнительную логику в метод `Drive()` класса `Car`, сохраняя при этом оригинальное поведение метода `Drive()` из класса `Vehicle`.

```csharp
public class Car : Vehicle
{
    public override void Drive(int speed)
    {
        base.Drive(speed); // Вызов метода Drive() из класса Vehicle
        Console.WriteLine("Добавляем специфическое поведение автомобиля");
    }
}
```

В этом примере, после вызова `base.Drive(speed)`, выполняется дополнительная логика, специфичная для класса `Car`.


### Перегрузка методов в C#

Перегрузка методов — это способность класса определить несколько методов с одним и тем же именем, но с разным количеством или типом параметров. Это позволяет разработчикам предоставлять различные варианты выполнения одного и того же действия в зависимости от входных данных. Перегрузка методов является одной из форм выражения полиморфизма в объектно-ориентированном программировании.

#### Пример с классами `Vehicle`, `Car` и `Motorcycle`

Давайте рассмотрим пример, где класс `Vehicle` содержит перегруженные методы `Drive()`, которые принимают разные типы параметров.

```csharp
public class Vehicle
{
    // Метод Drive(), принимающий int
    public void Drive(int speed)
    {
        Console.WriteLine($"Транспорт движется {speed} km/h.");
    }

    // Метод Drive(), принимающий string
    public void Drive(string mode)
    {
        Console.WriteLine($"Автомобиль движется в {mode} режиме.");
    }

    // Метод Drive(), принимающий bool
    public void Drive(bool isManual)
    {
        if (isManual)
        {
            Console.WriteLine("Автомобиль переключает передачи вручную.");
        }
        else
        {
            Console.WriteLine("Автомобиль оснащен автоматической коробкой передач.");
        }
    }
}
```

В этом примере класс `Vehicle` определяет три метода с именем `Drive()`, каждый из которых принимает разные типы параметров: `int`, `string` и `bool`. Это позволяет вызывать разные версии метода `Drive()` в зависимости от контекста, например:

```csharp
Vehicle myVehicle = new Vehicle();

myVehicle.Drive(60); // Вызов метода Drive() с int параметром
myVehicle.Drive("Eco"); // Вызов метода Drive() с string параметром
myVehicle.Drive(true); // Вызов метода Drive() с bool параметром
```

#### Отличие перегрузки от переопределения

Отличие перегрузки методов от переопределения заключается в том, что перегрузка методов включает определение нескольких методов с одним именем, но с разным набором параметров, тогда как переопределение методов (или перекрытие) означает предоставление новой реализации метода в производном классе, который уже существует в базовом классе.

#### Вызов методов

При вызове перегруженного метода компилятор выбирает подходящую версию метода на основе типов и количества аргументов, переданных в вызове. Если типы и количество аргументов совпадают с несколькими версиями метода, компилятор выберет ту версию, которая была объявлена первой.

### Заключение

Перегрузка методов в C# позволяет классам предоставлять различные варианты выполнения одного и того же действия, адаптируясь к различным условиям использования. Это мощный инструмент для создания гибких и удобных API, позволяющих клиентам класса использовать его функциональность наиболее эффективно.

Перекрытие методов в C# позволяет производным классам изменять или расширять поведение методов базовых классов, что является ключевой особенностью полиморфизма. Это отличается от перегрузки методов, которая предполагает определение нескольких методов с одинаковым именем, но с разным набором параметров.

````
using System;
using System.Collections.Generic;

// ========== БАЗОВЫЙ КЛАСС VEHICLE ==========
public class Vehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    
    public Vehicle(string make, string model)
    {
        Make = make;
        Model = model;
    }
    
    // ========== ПЕРЕГРУЗКА МЕТОДОВ (overload) ==========
    // Перегруженный метод Drive() с параметром int
    public void Drive(int speed)
    {
        Console.WriteLine($"Транспорт движется {speed} км/ч.");
    }
    
    // Перегруженный метод Drive() с параметром string (режим движения)
    public void Drive(string mode)
    {
        Console.WriteLine($"Транспорт движется в {mode} режиме.");
    }
    
    // Перегруженный метод Drive() с параметром bool (механическая/автоматическая)
    public void Drive(bool isManual)
    {
        if (isManual)
        {
            Console.WriteLine("Транспорт оснащен механической коробкой передач. Водитель переключает передачи вручную.");
        }
        else
        {
            Console.WriteLine("Транспорт оснащен автоматической коробкой передач.");
        }
    }
    
    // Перегруженный метод Drive() с двумя параметрами
    public void Drive(int speed, string mode)
    {
        Console.WriteLine($"Транспорт движется со скоростью {speed} км/ч в {mode} режиме.");
    }
    
    // ========== ПЕРЕКРЫТИЕ (override) - виртуальный метод ==========
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Транспортное средство: {Make} {Model}");
    }
    
    public virtual void StartEngine()
    {
        Console.WriteLine("Двигатель запущен.");
    }
    
    public virtual void StopEngine()
    {
        Console.WriteLine("Двигатель остановлен.");
    }
}

// ========== ПРОИЗВОДНЫЙ КЛАСС CAR ==========
public class Car : Vehicle
{
    public int Doors { get; set; }
    public string BodyType { get; set; }
    
    public Car(string make, string model, int doors, string bodyType) : base(make, model)
    {
        Doors = doors;
        BodyType = bodyType;
    }
    
    // ========== ПЕРЕГРУЗКА МЕТОДОВ (свои перегруженные методы) ==========
    // Перегруженный метод Drive() с дополнительным параметром
    public void Drive(int speed, bool withMusic)
    {
        Console.WriteLine($"Автомобиль {Make} {Model} движется {speed} км/ч. Музыка: {(withMusic ? "включена" : "выключена")}");
    }
    
    // Перегруженный метод Drive() для городского/трассового режима
    public void Drive(string terrain, int speed)
    {
        Console.WriteLine($"Автомобиль {Make} {Model} едет по {terrain} со скоростью {speed} км/ч");
    }
    
    // ========== ПЕРЕКРЫТИЕ (override) методов базового класса ==========
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип: Автомобиль, Дверей: {Doors}, Кузов: {BodyType}");
    }
    
    public override void StartEngine()
    {
        Console.WriteLine($"Автомобиль {Make} {Model}: Двигатель запущен! VRRROOOOM!");
    }
    
    public override void StopEngine()
    {
        Console.WriteLine($"Автомобиль {Make} {Model}: Двигатель остановлен. Ключ вынут.");
    }
    
    // Собственный метод
    public void OpenTrunk()
    {
        Console.WriteLine($"Багажник автомобиля {Make} {Model} открыт. Объем: 500 литров");
    }
}

// ========== ПРОИЗВОДНЫЙ КЛАСС MOTORCYCLE ==========
public class Motorcycle : Vehicle
{
    public string EngineType { get; set; }
    public bool HasSidecar { get; set; }
    
    public Motorcycle(string make, string model, string engineType, bool hasSidecar) : base(make, model)
    {
        EngineType = engineType;
        HasSidecar = hasSidecar;
    }
    
    // ========== ПЕРЕГРУЗКА МЕТОДОВ ==========
    // Перегруженный метод Drive() для мотоцикла
    public void Drive(int speed, bool withHelmet)
    {
        Console.WriteLine($"Мотоцикл {Make} {Model} движется {speed} км/ч. Шлем: {(withHelmet ? "надет" : "не надет")}");
    }
    
    // Перегруженный метод для разных типов дорог
    public void Drive(string roadType)
    {
        Console.WriteLine($"Мотоцикл {Make} {Model} едет по {roadType} с ветерком");
    }
    
    // ========== ПЕРЕКРЫТИЕ (override) ==========
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип: Мотоцикл, Двигатель: {EngineType}, Коляска: {(HasSidecar ? "Есть" : "Нет")}");
    }
    
    public override void StartEngine()
    {
        Console.WriteLine($"Мотоцикл {Make} {Model}: Двигатель запущен! BRRRAAAP!");
    }
    
    public override void StopEngine()
    {
        Console.WriteLine($"Мотоцикл {Make} {Model}: Двигатель заглушен.");
    }
    
    // Собственный метод
    public void Wheelie()
    {
        Console.WriteLine($"Мотоцикл {Make} {Model} едет на заднем колесе!");
    }
}

// ========== ПРОИЗВОДНЫЙ КЛАСС TRUCK ==========
public class Truck : Vehicle
{
    public double LoadCapacity { get; set; }
    public int Axles { get; set; }
    
    public Truck(string make, string model, double loadCapacity, int axles) : base(make, model)
    {
        LoadCapacity = loadCapacity;
        Axles = axles;
    }
    
    // ========== ПЕРЕГРУЗКА МЕТОДОВ ==========
    public void Drive(int speed, double cargoWeight)
    {
        if (cargoWeight > LoadCapacity)
            Console.WriteLine($"Грузовик {Make} {Model}: ПЕРЕГРУЗ! Максимум {LoadCapacity} т, загружено {cargoWeight} т");
        else
            Console.WriteLine($"Грузовик {Make} {Model} движется {speed} км/ч с грузом {cargoWeight} т");
    }
    
    // ========== ПЕРЕКРЫТИЕ (override) ==========
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип: Грузовик, Грузоподъемность: {LoadCapacity} т, Оси: {Axles}");
    }
    
    public override void StartEngine()
    {
        Console.WriteLine($"Грузовик {Make} {Model}: Дизельный двигатель запущен! BRRRMMM!");
    }
    
    public override void StopEngine()
    {
        Console.WriteLine($"Грузовик {Make} {Model}: Двигатель заглушен. Стоянка активирована.");
    }
}

// ========== ДЕМОНСТРАЦИЯ ==========
Console.WriteLine("========== ПЕРЕГРУЗКА И ПЕРЕКРЫТИЕ МЕТОДОВ: VEHICLE ==========\n");

// Создание объектов
Car car = new Car("Toyota", "Camry", 4, "Седан");
Motorcycle motorcycle = new Motorcycle("Honda", "CBR600RR", "Inline-4", false);
Truck truck = new Truck("Volvo", "FH16", 25.5, 3);

// ========== ДЕМОНСТРАЦИЯ ПЕРЕГРУЗКИ МЕТОДОВ ==========
Console.WriteLine("=== 1. ПЕРЕГРУЗКА МЕТОДОВ (OVERLOAD) ===\n");

Console.WriteLine("--- Перегрузка в базовом классе Vehicle ---");
Console.Write("Drive(60): "); car.Drive(60);
Console.Write("Drive(\"Eco\"): "); car.Drive("Eco");
Console.Write("Drive(true): "); car.Drive(true);
Console.Write("Drive(80, \"Sport\"): "); car.Drive(80, "Sport");

Console.WriteLine("\n--- Перегрузка в производном классе Car ---");
car.Drive(100, true);
car.Drive("городской", 50);

Console.WriteLine("\n--- Перегрузка в производном классе Motorcycle ---");
motorcycle.Drive(120, true);
motorcycle.Drive("горной");

Console.WriteLine("\n--- Перегрузка в производном классе Truck ---");
truck.Drive(80, 20);
truck.Drive(60, 30); // Перегруз

// ========== ДЕМОНСТРАЦИЯ ПЕРЕКРЫТИЯ МЕТОДОВ ==========
Console.WriteLine("\n=== 2. ПЕРЕКРЫТИЕ МЕТОДОВ (OVERRIDE) ===\n");

Console.WriteLine("--- Вызов переопределенных методов ---");
car.DisplayInfo();
car.StartEngine();
car.StopEngine();
Console.WriteLine();

motorcycle.DisplayInfo();
motorcycle.StartEngine();
motorcycle.StopEngine();
Console.WriteLine();

truck.DisplayInfo();
truck.StartEngine();
truck.StopEngine();

// ========== ДЕМОНСТРАЦИЯ ВЫЗОВА МЕТОДОВ РОДИТЕЛЯ (base) ==========
Console.WriteLine("\n=== 3. ВЫЗОВ МЕТОДОВ РОДИТЕЛЯ (base) ===\n");
Console.WriteLine("В методе DisplayInfo() Car вызван base.DisplayInfo():");
car.DisplayInfo();

// ========== ПОЛИМОРФИЗМ С ПЕРЕКРЫТИЕМ ==========
Console.WriteLine("\n=== 4. ПОЛИМОРФИЗМ (массив Vehicle) ===\n");

Vehicle[] vehicles = new Vehicle[]
{
    new Car("BMW", "X5", 5, "Внедорожник"),
    new Motorcycle("Yamaha", "R1", "V4", false),
    new Truck("MAN", "TGX", 30, 4),
    car,
    motorcycle,
    truck
};

foreach (var v in vehicles)
{
    v.DisplayInfo();
    v.StartEngine();
    Console.WriteLine();
}

Console.WriteLine("========== ДЕМОНСТРАЦИЯ VEHICLE ЗАВЕРШЕНА ==========\n");
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также используйте перекрытия и перегрузки методов.

````
public class Animal
{
    public string Name { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }
    
    public Animal(string name, int age, double weight)
    {
        Name = name;
        Age = age;
        Weight = weight;
    }
    
    // ========== ПЕРЕГРУЗКА МЕТОДОВ (overload) ==========
    // Перегруженный метод MakeSound() с параметром int (громкость)
    public void MakeSound(int volume)
    {
        string volumeLevel = volume switch
        {
            <= 30 => "тихо",
            <= 70 => "средне",
            _ => "громко"
        };
        Console.WriteLine($"{Name} издает звук {volumeLevel} (громкость: {volume} дБ).");
    }
    
    // Перегруженный метод MakeSound() с параметром string (эмоция)
    public void MakeSound(string emotion)
    {
        Console.WriteLine($"{Name} издает звук, выражая {emotion}.");
    }
    
    // Перегруженный метод MakeSound() с двумя параметрами
    public void MakeSound(int volume, string emotion)
    {
        string volumeLevel = volume switch
        {
            <= 30 => "тихо",
            <= 70 => "средне",
            _ => "громко"
        };
        Console.WriteLine($"{Name} {volumeLevel} издает звук, выражая {emotion} (громкость: {volume} дБ).");
    }
    
    // Перегруженный метод Move() с параметром int (скорость)
    public void Move(int speed)
    {
        Console.WriteLine($"{Name} двигается со скоростью {speed} км/ч.");
    }
    
    // Перегруженный метод Move() с параметром string (способ передвижения)
    public void Move(string movementType)
    {
        Console.WriteLine($"{Name} передвигается {movementType}.");
    }
    
    // ========== ПЕРЕКРЫТИЕ (override) - виртуальные методы ==========
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Имя: {Name}, Возраст: {Age} лет, Вес: {Weight} кг");
    }
    
    public virtual void Eat()
    {
        Console.WriteLine($"{Name} ест.");
    }
    
    public virtual void Sleep()
    {
        Console.WriteLine($"{Name} спит.");
    }
}

// ========== ПРОИЗВОДНЫЙ КЛАСС DOG ==========
public class Dog : Animal
{
    public string Breed { get; set; }
    public bool IsTrained { get; set; }
    
    public Dog(string name, int age, double weight, string breed, bool isTrained) 
        : base(name, age, weight)
    {
        Breed = breed;
        IsTrained = isTrained;
    }
    
    // ========== ПЕРЕГРУЗКА МЕТОДОВ (свои перегруженные методы) ==========
    public void MakeSound(string emotion, int times)
    {
        string sound = emotion == "радость" ? "Гав-гав!" : "Р-р-р...";
        Console.Write($"{Name} лает {times} раз: ");
        for (int i = 0; i < times; i++)
        {
            Console.Write(sound + " ");
        }
        Console.WriteLine();
    }
    
    public void Move(int speed, bool isChasing)
    {
        if (isChasing)
            Console.WriteLine($"Собака {Name} мчится за добычей со скоростью {speed} км/ч!");
        else
            Console.WriteLine($"Собака {Name} бежит со скоростью {speed} км/ч.");
    }
    
    // ========== ПЕРЕКРЫТИЕ (override) методов базового класса ==========
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Порода: {Breed}, Дрессирована: {(IsTrained ? "Да" : "Нет")}");
    }
    
    public override void Eat()
    {
        Console.WriteLine($"Собака {Name} с аппетитом ест корм.");
        if (IsTrained)
            Console.WriteLine($"  {Name} ждет команду после еды!");
    }
    
    public override void Sleep()
    {
        Console.WriteLine($"Собака {Name} свернулась калачиком и спит.");
    }
    
    // Собственные методы
    public void Bark()
    {
        Console.WriteLine($"{Name} лает: Гав-гав!");
    }
    
    public void Fetch(string toy)
    {
        Console.WriteLine($"Собака {Name} приносит {toy}");
    }
}

// ========== ПРОИЗВОДНЫЙ КЛАСС CAT ==========
public class Cat : Animal
{
    public string Color { get; set; }
    public bool IsIndoor { get; set; }
    
    public Cat(string name, int age, double weight, string color, bool isIndoor) 
        : base(name, age, weight)
    {
        Color = color;
        IsIndoor = isIndoor;
    }
    
    // ========== ПЕРЕГРУЗКА МЕТОДОВ ==========
    public void MakeSound(int volume, string mood)
    {
        string sound = mood == "довольный" ? "Мур-мур-мур..." : "МЯУ! МЯУ!";
        Console.WriteLine($"Кошка {Name} {sound} (громкость: {volume} дБ)");
    }
    
    public void Move(int speed, bool isHunting)
    {
        if (isHunting)
            Console.WriteLine($"Кошка {Name} крадется к добыче со скоростью {speed} км/ч");
        else
            Console.WriteLine($"Кошка {Name} грациозно идет со скоростью {speed} км/ч");
    }
    
    // ========== ПЕРЕКРЫТИЕ (override) ==========
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Окрас: {Color}, Домашняя: {(IsIndoor ? "Да" : "Нет")}");
    }
    
    public override void Eat()
    {
        Console.WriteLine($"Кошка {Name} изящно пробует рыбу.");
        if (!IsIndoor)
            Console.WriteLine($"  {Name} может принести мышь в благодарность!");
    }
    
    public override void Sleep()
    {
        Console.WriteLine($"Кошка {Name} свернулась клубочком и спит.");
    }
    
    // Собственные методы
    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет: Мур-мур-мур...");
    }
    
    public void Scratch()
    {
        Console.WriteLine($"{Name} точит когти о когтеточку.");
    }
}

// ========== ПРОИЗВОДНЫЙ КЛАСС BIRD ==========
public class Bird : Animal
{
    public double WingSpan { get; set; }
    public bool CanFly { get; set; }
    
    public Bird(string name, int age, double weight, double wingSpan, bool canFly) 
        : base(name, age, weight)
    {
        WingSpan = wingSpan;
        CanFly = canFly;
    }
    
    // ========== ПЕРЕГРУЗКА МЕТОДОВ ==========
    public void MakeSound(string song)
    {
        Console.WriteLine($"Птица {Name} поет: \"{song}\"");
    }
    
    public void Move(int speed, int altitude)
    {
        if (CanFly)
            Console.WriteLine($"Птица {Name} летит на высоте {altitude} м со скоростью {speed} км/ч");
        else
            Console.WriteLine($"Птица {Name} бежит по земле со скоростью {speed} км/ч");
    }
    
    // ========== ПЕРЕКРЫТИЕ (override) ==========
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Размах крыльев: {WingSpan} м, Умеет летать: {(CanFly ? "Да" : "Нет")}");
    }
    
    public override void Eat()
    {
        Console.WriteLine($"Птица {Name} клюет зерна.");
    }
    
    public override void Sleep()
    {
        Console.WriteLine($"Птица {Name} спит, спрятав голову под крыло.");
    }
    
    // Собственные методы
    public void Fly()
    {
        if (CanFly)
            Console.WriteLine($"Птица {Name} летит, расправив крылья на {WingSpan} м");
        else
            Console.WriteLine($"Птица {Name} не может летать");
    }
    
    public void BuildNest()
    {
        Console.WriteLine($"Птица {Name} строит гнездо из веточек.");
    }
}

// ========== ПРОИЗВОДНЫЙ КЛАСС FISH ==========
public class Fish : Animal
{
    public string WaterType { get; set; }
    public double MaxDepth { get; set; }
    
    public Fish(string name, int age, double weight, string waterType, double maxDepth) 
        : base(name, age, weight)
    {
        WaterType = waterType;
        MaxDepth = maxDepth;
    }
    
    // ========== ПЕРЕГРУЗКА МЕТОДОВ ==========
    public void Move(int speed, int depth)
    {
        Console.WriteLine($"Рыба {Name} плывет на глубине {depth} м со скоростью {speed} км/ч");
    }
    
    public void MakeSound(int volume, string emotion)
    {
        Console.WriteLine($"Рыба {Name} издает булькающие звуки (громкость: {volume} дБ), выражая {emotion}");
    }
    
    // ========== ПЕРЕКРЫТИЕ (override) ==========
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип воды: {WaterType}, Макс. глубина: {MaxDepth} м");
    }
    
    public override void Eat()
    {
        Console.WriteLine($"Рыба {Name} проглатывает планктон.");
    }
    
    public override void Sleep()
    {
        Console.WriteLine($"Рыба {Name} спит, неподвижно зависнув в воде.");
    }
    
    // Собственные методы
    public void Swim()
    {
        Console.WriteLine($"Рыба {Name} плавает в {WaterType} воде.");
    }
}

// ========== ДЕМОНСТРАЦИЯ ДЛЯ ANIMAL ==========
class AnimalDemo
{
    public static void Run()
    {
        Console.WriteLine("========== ПЕРЕГРУЗКА И ПЕРЕКРЫТИЕ МЕТОДОВ: ANIMAL ==========\n");
        
        // Создание объектов
        Dog dog = new Dog("Рекс", 3, 25.5, "Овчарка", true);
        Cat cat = new Cat("Муся", 2, 4.2, "Серый", true);
        Bird bird = new Bird("Кеша", 1, 0.35, 0.5, true);
        Fish fish = new Fish("Немо", 2, 0.15, "Соленая", 50);
        
        // ========== ДЕМОНСТРАЦИЯ ПЕРЕГРУЗКИ МЕТОДОВ ==========
        Console.WriteLine("=== 1. ПЕРЕГРУЗКА МЕТОДОВ (OVERLOAD) ===\n");
        
        Console.WriteLine("--- Перегрузка в базовом классе Animal ---");
        Console.Write("MakeSound(30): "); dog.MakeSound(30);
        Console.Write("MakeSound(\"радость\"): "); dog.MakeSound("радость");
        Console.Write("MakeSound(80, \"гнев\"): "); dog.MakeSound(80, "гнев");
        
        Console.WriteLine("\n--- Перегрузка в производном классе Dog ---");
        dog.MakeSound("радость", 3);
        dog.Move(40, true);
        
        Console.WriteLine("\n--- Перегрузка в производном классе Cat ---");
        cat.MakeSound(50, "довольный");
        cat.Move(30, true);
        
        Console.WriteLine("\n--- Перегрузка в производном классе Bird ---");
        bird.MakeSound("Чик-чирик, я веселая птичка!");
        bird.Move(60, 100);
        
        Console.WriteLine("\n--- Перегрузка в производном классе Fish ---");
        fish.Move(20, 15);
        fish.MakeSound(40, "спокойствие");
        
        // ========== ДЕМОНСТРАЦИЯ ПЕРЕКРЫТИЯ МЕТОДОВ ==========
        Console.WriteLine("\n=== 2. ПЕРЕКРЫТИЕ МЕТОДОВ (OVERRIDE) ===\n");
        
        Console.WriteLine("--- Собака ---");
        dog.DisplayInfo();
        dog.Eat();
        dog.Sleep();
        Console.WriteLine();
        
        Console.WriteLine("--- Кошка ---");
        cat.DisplayInfo();
        cat.Eat();
        cat.Sleep();
        Console.WriteLine();
        
        Console.WriteLine("--- Птица ---");
        bird.DisplayInfo();
        bird.Eat();
        bird.Sleep();
        Console.WriteLine();
        
        Console.WriteLine("--- Рыба ---");
        fish.DisplayInfo();
        fish.Eat();
        fish.Sleep();
        
        // ========== ДЕМОНСТРАЦИЯ ВЫЗОВА МЕТОДОВ РОДИТЕЛЯ (base) ==========
        Console.WriteLine("\n=== 3. ВЫЗОВ МЕТОДОВ РОДИТЕЛЯ (base) ===\n");
        Console.WriteLine("В методе DisplayInfo() Dog вызван base.DisplayInfo():");
        dog.DisplayInfo();
        
        // ========== ПОЛИМОРФИЗМ С ПЕРЕКРЫТИЕМ ==========
        Console.WriteLine("\n=== 4. ПОЛИМОРФИЗМ (массив Animal) ===\n");
        
        Animal[] animals = new Animal[]
        {
            new Dog("Шарик", 4, 20, "Дворняга", false),
            new Cat("Васька", 3, 3.8, "Черный", false),
            new Bird("Пингвин", 5, 30, 0.8, false),
            new Fish("Золотая", 1, 0.1, "Пресная", 2),
            dog,
            cat,
            bird,
            fish
        };
        
        foreach (var a in animals)
        {
            a.DisplayInfo();
            a.Eat();
            a.Sleep();
            Console.WriteLine();
        }
        
        // ========== ДЕМОНСТРАЦИЯ СПЕЦИФИЧНЫХ МЕТОДОВ ==========
        Console.WriteLine("=== 5. СПЕЦИФИЧНЫЕ МЕТОДЫ КАЖДОГО КЛАССА ===\n");
        
        dog.Bark();
        dog.Fetch("мяч");
        cat.Purr();
        cat.Scratch();
        bird.Fly();
        bird.BuildNest();
        fish.Swim();
        
        Console.WriteLine("\n========== ДЕМОНСТРАЦИЯ ANIMAL ЗАВЕРШЕНА ==========");
    }
}

// ========== ЗАПУСК ВСЕХ ДЕМОНСТРАЦИЙ ==========

Console.WriteLine("\n" + new string('=', 80) + "\n");
AnimalDemo.Run();
````
