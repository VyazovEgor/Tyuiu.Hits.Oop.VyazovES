```

using System;
using System.Collections.Generic;
using System.Linq;

// ========== БАЗОВЫЙ КЛАСС PERSON ==========
public class Person
{
    // Приватные поля для инкапсуляции
    private string _name;
    private int _age;
    private string _gender;
    private string _email;
    private string _phone;
    
    // Свойства с геттерами и сеттерами (с валидацией)
    public string Name
    {
        get { return _name; }
        set 
        { 
            if (!string.IsNullOrWhiteSpace(value))
                _name = value;
            else
                throw new ArgumentException("Имя не может быть пустым!");
        }
    }
    
    public int Age
    {
        get { return _age; }
        set 
        { 
            if (value >= 0 && value <= 120)
                _age = value;
            else
                throw new ArgumentOutOfRangeException("Возраст должен быть от 0 до 120 лет!");
        }
    }
    
    public string Gender
    {
        get { return _gender; }
        set 
        { 
            if (value == "Мужской" || value == "Женский" || value == "Другой")
                _gender = value;
            else
                throw new ArgumentException("Пол должен быть 'Мужской', 'Женский' или 'Другой'!");
        }
    }
    
    public string Email
    {
        get { return _email; }
        set 
        { 
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Email не может быть пустым!");
            if (value.Contains("@") && value.Contains("."))
                _email = value;
            else
                throw new ArgumentException("Некорректный формат email!");
        }
    }
    
    public string Phone
    {
        get { return _phone; }
        set { _phone = value; }
    }
    
    // Статический атрибут
    public static int TotalPeople { get; private set; } = 0;
    protected static List<Person> allPeople = new List<Person>();
    
    // Конструкторы (перегрузка)
    public Person(string name, int age, string gender)
    {
        Name = name;
        Age = age;
        Gender = gender;
        TotalPeople++;
        allPeople.Add(this);
    }
    
    public Person(string name, int age, string gender, string email) : this(name, age, gender)
    {
        Email = email;
    }
    
    public Person(string name, int age, string gender, string email, string phone) : this(name, age, gender, email)
    {
        Phone = phone;
    }
    
    // ========== ПЕРЕГРУЗКА МЕТОДОВ (OVERLOAD) ==========
    
    // Перегруженный метод GetInfo() без параметров
    public virtual string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Пол: {Gender}";
    }
    
    // Перегруженный метод GetInfo() с параметром detailed
    public virtual string GetInfo(bool detailed)
    {
        if (detailed)
        {
            string emailInfo = string.IsNullOrEmpty(Email) ? "не указан" : Email;
            string phoneInfo = string.IsNullOrEmpty(Phone) ? "не указан" : Phone;
            return $"Имя: {Name}, Возраст: {Age}, Пол: {Gender}, Email: {emailInfo}, Телефон: {phoneInfo}";
        }
        return GetInfo();
    }
    
    // Перегруженный метод GetInfo() с параметром format
    public virtual string GetInfo(string format)
    {
        return format.ToLower() switch
        {
            "short" => $"{Name}, {Age} лет",
            "name" => Name,
            "age" => $"{Age} лет",
            _ => GetInfo()
        };
    }
    
    // Перегруженный метод SayHello() без параметров
    public virtual void SayHello()
    {
        Console.WriteLine($"Привет! Меня зовут {Name}.");
    }
    
    // Перегруженный метод SayHello() с параметром language
    public virtual void SayHello(string language)
    {
        switch (language.ToLower())
        {
            case "english":
                Console.WriteLine($"Hello! My name is {Name}.");
                break;
            case "russian":
                Console.WriteLine($"Привет! Меня зовут {Name}.");
                break;
            case "french":
                Console.WriteLine($"Bonjour! Je m'appelle {Name}.");
                break;
            default:
                SayHello();
                break;
        }
    }
    
    // Перегруженный метод SayHello() с параметром formal
    public virtual void SayHello(bool formal)
    {
        if (formal)
            Console.WriteLine($"Здравствуйте! Меня зовут {Name}. Рад(а) познакомиться!");
        else
            SayHello();
    }
    
    // Перегруженный метод Communicate() с разными параметрами
    public virtual void Communicate(string message)
    {
        Console.WriteLine($"{Name} говорит: {message}");
    }
    
    public virtual void Communicate(Person person, string message)
    {
        Console.WriteLine($"{Name} говорит {person.Name}: {message}");
    }
    
    public virtual void Communicate(string message, int times)
    {
        for (int i = 0; i < times; i++)
        {
            Console.WriteLine($"{Name} говорит: {message} (раз {i + 1})");
        }
    }
    
    // ========== ВИРТУАЛЬНЫЕ МЕТОДЫ ДЛЯ ПЕРЕКРЫТИЯ ==========
    public virtual void Work()
    {
        Console.WriteLine($"{Name} работает.");
    }
    
    public virtual void Study()
    {
        Console.WriteLine($"{Name} учится.");
    }
    
    public virtual void Relax()
    {
        Console.WriteLine($"{Name} отдыхает.");
    }
    
    // Статический метод
    public static void DisplayTotalPeople()
    {
        Console.WriteLine($"\nВсего создано людей: {TotalPeople}");
    }
    
    // Статический метод поиска
    public static Person FindPerson(string name)
    {
        return allPeople.Find(p => p.Name.Equals(name, StringComparison.OrdinalIgnoreCase));
    }
}

// ========== ПРОИЗВОДНЫЙ КЛАСС STUDENT (ПРОСТОЕ НАСЛЕДОВАНИЕ) ==========
public class Student : Person
{
    private string _university;
    private int _course;
    private List<int> _grades = new List<int>();
    private string _major;
    private double _scholarship;
    
    public string University
    {
        get { return _university; }
        set 
        { 
            if (!string.IsNullOrWhiteSpace(value))
                _university = value;
            else
                throw new ArgumentException("Название университета не может быть пустым!");
        }
    }
    
    public int Course
    {
        get { return _course; }
        set 
        { 
            if (value >= 1 && value <= 6)
                _course = value;
            else
                throw new ArgumentOutOfRangeException("Курс должен быть от 1 до 6!");
        }
    }
    
    public string Major
    {
        get { return _major; }
        set { _major = value; }
    }
    
    public double Scholarship
    {
        get { return _scholarship; }
        set 
        { 
            if (value >= 0)
                _scholarship = value;
            else
                throw new ArgumentOutOfRangeException("Стипендия не может быть отрицательной!");
        }
    }
    
    // Конструкторы (перегрузка)
    public Student(string name, int age, string gender, string university, int course, string major) 
        : base(name, age, gender)
    {
        University = university;
        Course = course;
        Major = major;
    }
    
    public Student(string name, int age, string gender, string university, int course, string major, string email) 
        : base(name, age, gender, email)
    {
        University = university;
        Course = course;
        Major = major;
    }
    
    // ========== ПЕРЕГРУЗКА МЕТОДОВ ==========
    public void AddGrade(int grade)
    {
        if (grade >= 2 && grade <= 5)
        {
            _grades.Add(grade);
            Console.WriteLine($"Студенту {Name} добавлена оценка: {grade}");
            UpdateScholarship();
        }
        else
        {
            Console.WriteLine("Оценка должна быть от 2 до 5!");
        }
    }
    
    public void AddGrade(int grade, string subject)
    {
        Console.Write($"Студент {Name} получил {grade} по предмету {subject}: ");
        AddGrade(grade);
    }
    
    public void AddGrade(params int[] grades)
    {
        foreach (var grade in grades)
        {
            AddGrade(grade);
        }
    }
    
    private void UpdateScholarship()
    {
        double avg = GetAverageGrade();
        if (avg >= 4.8)
            Scholarship = 5000;
        else if (avg >= 4.5)
            Scholarship = 4000;
        else if (avg >= 4.0)
            Scholarship = 3000;
        else
            Scholarship = 0;
    }
    
    public double GetAverageGrade()
    {
        if (_grades.Count == 0) return 0;
        return _grades.Average();
    }
    
    public string GetGradeStatus()
    {
        double avg = GetAverageGrade();
        return avg switch
        {
            >= 4.8 => "Отлично",
            >= 4.0 => "Хорошо",
            >= 3.0 => "Удовлетворительно",
            _ => "Требуется улучшение"
        };
    }
    
    // ========== ПЕРЕКРЫТИЕ (OVERRIDE) МЕТОДОВ ==========
    public override void SayHello()
    {
        Console.WriteLine($"Привет! Я студент {Name}, учусь на {Course}-м курсе в {University} по специальности '{Major}'.");
    }
    
    public override void SayHello(string language)
    {
        if (language.ToLower() == "english")
            Console.WriteLine($"Hi! I'm {Name}, a student at {University}, majoring in {Major}.");
        else
            base.SayHello(language);
    }
    
    public override string GetInfo(bool detailed)
    {
        if (detailed)
        {
            return base.GetInfo(detailed) + $", Университет: {University}, Курс: {Course}, " +
                   $"Специальность: {Major}, Средний балл: {GetAverageGrade():F2}, " +
                   $"Статус: {GetGradeStatus()}, Стипендия: {Scholarship:C}";
        }
        return base.GetInfo(detailed);
    }
    
    public override void Work()
    {
        Console.WriteLine($"{Name} подрабатывает во время учебы в {University}.");
    }
    
    public override void Study()
    {
        Console.WriteLine($"{Name} усердно учится на {Course} курсе по специальности {Major}.");
        if (GetAverageGrade() >= 4.5)
            Console.WriteLine($"  Отличная успеваемость! Средний балл: {GetAverageGrade():F2}");
    }
    
    public override void Relax()
    {
        Console.WriteLine($"{Name} отдыхает с друзьями в студенческом кампусе.");
    }
}

// ========== ПРОИЗВОДНЫЙ КЛАСС EMPLOYEE ==========
public class Employee : Person
{
    private string _company;
    private decimal _salary;
    private string _position;
    private int _experience;
    private List<string> _projects = new List<string>();
    
    public string Company
    {
        get { return _company; }
        set 
        { 
            if (!string.IsNullOrWhiteSpace(value))
                _company = value;
            else
                throw new ArgumentException("Название компании не может быть пустым!");
        }
    }
    
    public decimal Salary
    {
        get { return _salary; }
        set 
        { 
            if (value >= 0)
                _salary = value;
            else
                throw new ArgumentOutOfRangeException("Зарплата не может быть отрицательной!");
        }
    }
    
    public string Position
    {
        get { return _position; }
        set { _position = value; }
    }
    
    public int Experience
    {
        get { return _experience; }
        set 
        { 
            if (value >= 0)
                _experience = value;
            else
                throw new ArgumentOutOfRangeException("Стаж не может быть отрицательным!");
        }
    }
    
    public Employee(string name, int age, string gender, string company, decimal salary, string position, int experience) 
        : base(name, age, gender)
    {
        Company = company;
        Salary = salary;
        Position = position;
        Experience = experience;
    }
    
    // ========== ПЕРЕГРУЗКА МЕТОДОВ ==========
    public void AddProject(string project)
    {
        _projects.Add(project);
        Console.WriteLine($"{Name} добавлен в проект: {project}");
    }
    
    public void AddProject(params string[] projects)
    {
        foreach (var project in projects)
        {
            AddProject(project);
        }
    }
    
    public void RaiseSalary(decimal amount)
    {
        if (amount > 0)
        {
            Salary += amount;
            Console.WriteLine($"Зарплата {Name} повышена до {Salary:C}");
        }
    }
    
    public void RaiseSalary(decimal percent, bool isPercent)
    {
        if (isPercent && percent > 0)
        {
            decimal increase = Salary * percent / 100;
            Salary += increase;
            Console.WriteLine($"Зарплата {Name} повышена на {percent}% до {Salary:C}");
        }
    }
    
    public decimal CalculateBonus()
    {
        return Salary * (Experience / 100m);
    }
    
    // ========== ПЕРЕКРЫТИЕ (OVERRIDE) ==========
    public override void SayHello()
    {
        Console.WriteLine($"Здравствуйте! Я {Name}, работаю в {Company} в должности {Position}.");
    }
    
    public override string GetInfo(bool detailed)
    {
        string projectsStr = _projects.Count > 0 ? string.Join(", ", _projects) : "нет проектов";
        if (detailed)
        {
            return base.GetInfo(detailed) + $", Компания: {Company}, Должность: {Position}, " +
                   $"Стаж: {Experience} лет, Зарплата: {Salary:C}, Бонус: {CalculateBonus():C}, " +
                   $"Проекты: {projectsStr}";
        }
        return base.GetInfo(detailed);
    }
    
    public override void Work()
    {
        Console.WriteLine($"{Name} работает в {Company} над проектами: {(_projects.Count > 0 ? string.Join(", ", _projects) : "новые проекты")}");
    }
    
    public override void Relax()
    {
        Console.WriteLine($"{Name} отдыхает после рабочего дня в {Company}.");
    }
}

// ========== СЛОЖНОЕ НАСЛЕДОВАНИЕ: TEACHER (наследует от Employee) ==========
public class Teacher : Employee
{
    private string _subject;
    private string _academicDegree;
    private List<Student> _students = new List<Student>();
    
    public string Subject
    {
        get { return _subject; }
        set 
        { 
            if (!string.IsNullOrWhiteSpace(value))
                _subject = value;
            else
                throw new ArgumentException("Название предмета не может быть пустым!");
        }
    }
    
    public string AcademicDegree
    {
        get { return _academicDegree; }
        set { _academicDegree = value; }
    }
    
    public int StudentsCount => _students.Count;
    
    public Teacher(string name, int age, string gender, string company, decimal salary, 
                   string position, int experience, string subject, string academicDegree) 
        : base(name, age, gender, company, salary, position, experience)
    {
        Subject = subject;
        AcademicDegree = academicDegree;
    }
    
    // ========== ПЕРЕГРУЗКА МЕТОДОВ ==========
    public void AddStudent(Student student)
    {
        if (!_students.Contains(student))
        {
            _students.Add(student);
            Console.WriteLine($"Студент {student.Name} добавлен к преподавателю {Name}");
        }
    }
    
    public void AddStudent(params Student[] students)
    {
        foreach (var student in students)
        {
            AddStudent(student);
        }
    }
    
    public void GradeStudent(Student student, int grade)
    {
        if (_students.Contains(student))
        {
            student.AddGrade(grade, Subject);
            Console.WriteLine($"Преподаватель {Name} выставил оценку {grade} студенту {student.Name} по предмету {Subject}");
        }
        else
        {
            Console.WriteLine($"Студент {student.Name} не учится у преподавателя {Name}");
        }
    }
    
    public void ConductExam()
    {
        Console.WriteLine($"\n{Name} проводит экзамен по предмету {Subject}");
        Random rand = new Random();
        foreach (var student in _students)
        {
            int grade = rand.Next(3, 6);
            GradeStudent(student, grade);
        }
    }
    
    // ========== ПЕРЕКРЫТИЕ (OVERRIDE) ==========
    public override void SayHello()
    {
        Console.WriteLine($"Здравствуйте! Я преподаватель {Name}, {AcademicDegree}, веду предмет '{Subject}'.");
    }
    
    public override void SayHello(string language)
    {
        if (language.ToLower() == "english")
            Console.WriteLine($"Hello! I'm Professor {Name}, teaching {Subject}.");
        else
            base.SayHello(language);
    }
    
    public override string GetInfo(bool detailed)
    {
        if (detailed)
        {
            return base.GetInfo(detailed) + $", Предмет: {Subject}, Ученая степень: {AcademicDegree}, " +
                   $"Студентов: {StudentsCount}";
        }
        return base.GetInfo(detailed);
    }
    
    public override void Work()
    {
        Console.WriteLine($"{Name} преподает {Subject} студентам в {Company}.");
        if (StudentsCount > 0)
            Console.WriteLine($"  Количество студентов: {StudentsCount}");
    }
    
    public override void Study()
    {
        Console.WriteLine($"{Name} повышает квалификацию по предмету {Subject}.");
    }
}

// ========== ИНТЕРФЕЙСЫ ДЛЯ МНОЖЕСТВЕННОГО НАСЛЕДОВАНИЯ ==========
public interface IResearcher
{
    int PublicationsCount { get; set; }
    void DoResearch();
    void PublishArticle(string title);
}

public interface IMentor
{
    int StudentsMentored { get; set; }
    void MentorStudent(Student student);
    void GiveAdvice();
}

public interface IConferenceSpeaker
{
    int ConferencesAttended { get; set; }
    void SpeakAtConference(string conferenceName);
}

// ========== МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ: PROFESSOR ==========
public class Professor : Teacher, IResearcher, IMentor, IConferenceSpeaker
{
    private List<string> _publications = new List<string>();
    private List<Student> _mentoredStudents = new List<Student>();
    private List<string> _conferences = new List<string>();
    
    public string ResearchField { get; set; }
    public string LaboratoryName { get; set; }
    
    // IResearcher
    public int PublicationsCount { get; set; }
    
    // IMentor
    public int StudentsMentored { get; set; }
    
    // IConferenceSpeaker
    public int ConferencesAttended { get; set; }
    
    public Professor(string name, int age, string gender, string company, decimal salary,
                     string position, int experience, string subject, string academicDegree,
                     string researchField, string laboratoryName) 
        : base(name, age, gender, company, salary, position, experience, subject, academicDegree)
    {
        ResearchField = researchField;
        LaboratoryName = laboratoryName;
        PublicationsCount = 0;
        StudentsMentored = 0;
        ConferencesAttended = 0;
    }
    
    // Реализация IResearcher
    public void DoResearch()
    {
        Console.WriteLine($"Профессор {Name} проводит исследования в области {ResearchField} в лаборатории {LaboratoryName}");
    }
    
    public void PublishArticle(string title)
    {
        _publications.Add(title);
        PublicationsCount++;
        Console.WriteLine($"Профессор {Name} опубликовал статью: '{title}' (Всего: {PublicationsCount})");
    }
    
    // Реализация IMentor
    public void MentorStudent(Student student)
    {
        if (!_mentoredStudents.Contains(student))
        {
            _mentoredStudents.Add(student);
            StudentsMentored++;
            Console.WriteLine($"Профессор {Name} взял шефство над студентом {student.Name}");
        }
    }
    
    public void GiveAdvice()
    {
        Console.WriteLine($"Профессор {Name} дает ценный совет по научной карьере студентам");
    }
    
    // Реализация IConferenceSpeaker
    public void SpeakAtConference(string conferenceName)
    {
        _conferences.Add(conferenceName);
        ConferencesAttended++;
        Console.WriteLine($"Профессор {Name} выступает на конференции: {conferenceName}");
    }
    
    // ========== ПЕРЕКРЫТИЕ ==========
    public override void SayHello()
    {
        Console.WriteLine($"Здравствуйте! Я профессор {Name}, {AcademicDegree}, заведующий лабораторией {LaboratoryName}.");
    }
    
    public override string GetInfo(bool detailed)
    {
        if (detailed)
        {
            return base.GetInfo(detailed) + $", Лаборатория: {LaboratoryName}, Область исследований: {ResearchField}, " +
                   $"Публикаций: {PublicationsCount}, Студентов под наставничеством: {StudentsMentored}, " +
                   $"Конференций: {ConferencesAttended}";
        }
        return base.GetInfo(detailed);
    }
    
    public override void Work()
    {
        base.Work();
        DoResearch();
    }
}

// ========== GENERIC КЛАССЫ ==========

// Generic класс для коллекции людей
public class PersonCollection<T> where T : Person
{
    private List<T> _people = new List<T>();
    private static int _totalCreated = 0;
    
    public PersonCollection()
    {
        Console.WriteLine($"Создана коллекция для типа {typeof(T).Name}");
    }
    
    public void Add(T person)
    {
        _people.Add(person);
        _totalCreated++;
        Console.WriteLine($"Добавлен: {person.Name} ({person.GetType().Name})");
    }
    
    public void Add(params T[] people)
    {
        foreach (var person in people)
        {
            Add(person);
        }
    }
    
    public bool Remove(T person)
    {
        bool removed = _people.Remove(person);
        if (removed)
        {
            Console.WriteLine($"Удален: {person.Name}");
            _totalCreated--;
        }
        return removed;
    }
    
    public int Count => _people.Count;
    public static int TotalCreated => _totalCreated;
    
    public void DisplayAll()
    {
        Console.WriteLine($"\nВсего людей в коллекции: {Count}");
        foreach (var person in _people)
        {
            Console.WriteLine(person.GetInfo(true));
            person.SayHello();
            Console.WriteLine();
        }
    }
    
    public List<T> FindByName(string name)
    {
        return _people.FindAll(p => p.Name.Equals(name, StringComparison.OrdinalIgnoreCase));
    }
    
    public List<T> FindByAge(int minAge, int maxAge)
    {
        return _people.FindAll(p => p.Age >= minAge && p.Age <= maxAge);
    }
    
    public T GetAt(int index)
    {
        if (index >= 0 && index < _people.Count)
            return _people[index];
        throw new IndexOutOfRangeException("Индекс вне диапазона");
    }
    
    public void ForEach(Action<T> action)
    {
        foreach (var person in _people)
        {
            action(person);
        }
    }
}

// Generic класс для университета
public class University<T> where T : Person
{
    public string Name { get; set; }
    private PersonCollection<T> _members;
    
    public University(string name)
    {
        Name = name;
        _members = new PersonCollection<T>();
        Console.WriteLine($"Создан университет '{Name}' для типа {typeof(T).Name}");
    }
    
    public void AddMember(T member)
    {
        _members.Add(member);
    }
    
    public void DisplayInfo()
    {
        Console.WriteLine($"\n=== Университет '{Name}' ===");
        Console.WriteLine($"Тип членов: {typeof(T).Name}");
        _members.DisplayAll();
    }
    
    public PersonCollection<T> GetMembers() => _members;
}

// Generic класс для пары людей
public class PersonPair<T1, T2> where T1 : Person where T2 : Person
{
    public T1 First { get; set; }
    public T2 Second { get; set; }
    public string Relationship { get; set; }
    
    public PersonPair(T1 first, T2 second, string relationship)
    {
        First = first;
        Second = second;
        Relationship = relationship;
    }
    
    public void DisplayPair()
    {
        Console.WriteLine($"\nПара ({Relationship}):");
        Console.WriteLine($"Первый: {First.GetInfo("short")}");
        Console.WriteLine($"Второй: {Second.GetInfo("short")}");
    }
    
    public void Introduce()
    {
        Console.WriteLine($"\nПредставление пары ({Relationship}):");
        First.SayHello();
        Second.SayHello();
    }
}

// Generic класс для статистики
public class PersonStatistics<T> where T : Person
{
    private List<T> _people;
    
    public PersonStatistics(List<T> people)
    {
        _people = people;
    }
    
    public double GetAverageAge()
    {
        if (_people.Count == 0) return 0;
        return _people.Average(p => p.Age);
    }
    
    public T GetOldest()
    {
        if (_people.Count == 0) return null;
        return _people.OrderByDescending(p => p.Age).First();
    }
    
    public T GetYoungest()
    {
        if (_people.Count == 0) return null;
        return _people.OrderBy(p => p.Age).First();
    }
    
    public Dictionary<string, int> GetGenderStats()
    {
        return _people.GroupBy(p => p.Gender)
                      .ToDictionary(g => g.Key, g => g.Count());
    }
    
    public void DisplayStatistics()
    {
        Console.WriteLine($"\n=== Статистика для {typeof(T).Name} ===");
        Console.WriteLine($"Количество: {_people.Count}");
        Console.WriteLine($"Средний возраст: {GetAverageAge():F1} лет");
        
        var oldest = GetOldest();
        if (oldest != null)
            Console.WriteLine($"Самый старший: {oldest.Name} ({oldest.Age} лет)");
        
        var youngest = GetYoungest();
        if (youngest != null)
            Console.WriteLine($"Самый младший: {youngest.Name} ({youngest.Age} лет)");
        
        Console.WriteLine("\nСтатистика по полу:");
        foreach (var stat in GetGenderStats())
        {
            Console.WriteLine($"  {stat.Key}: {stat.Value} чел.");
        }
    }
}

// ========== ДЕМОНСТРАЦИЯ ==========
Console.WriteLine("========== ИНДИВИДУАЛЬНЫЙ ПРОЕКТ: СИСТЕМА УПРАВЛЕНИЯ ПЕРСОНАЛОМ ==========\n");

// ========== 1. СОЗДАНИЕ ОБЪЕКТОВ ==========
Console.WriteLine("=== 1. СОЗДАНИЕ ОБЪЕКТОВ ===\n");

Student student1 = new Student("Анна Смирнова", 20, "Женский", "МГУ", 3, "Программирование", "anna@mail.ru");
Student student2 = new Student("Олег Новиков", 22, "Мужской", "МГУ", 4, "Искусственный интеллект");
Student student3 = new Student("Елена Попова", 19, "Женский", "МГУ", 2, "Биоинформатика");

Employee employee = new Employee("Пётр Сидоров", 35, "Мужской", "ООО Ромашка", 75000, "Разработчик", 5);

Teacher teacher = new Teacher("Елена Иванова", 45, "Женский", "МГУ", 120000, "Доцент", 15, "Математика", "Кандидат наук");

Professor professor = new Professor("Сергей Николаев", 50, "Мужской", "МГУ", 200000, "Профессор", 25, 
                                    "Программирование", "Доктор наук", "Искусственный интеллект", "Лаборатория ИИ");

// ========== 2. ДЕМОНСТРАЦИЯ ПЕРЕГРУЗКИ МЕТОДОВ ==========
Console.WriteLine("=== 2. ДЕМОНСТРАЦИЯ ПЕРЕГРУЗКИ МЕТОДОВ (OVERLOAD) ===\n");

Console.WriteLine("--- Метод GetInfo() ---");
Console.WriteLine(student1.GetInfo());
Console.WriteLine(student1.GetInfo(true));
Console.WriteLine(student1.GetInfo("short"));

Console.WriteLine("\n--- Метод SayHello() ---");
student1.SayHello();
student1.SayHello("english");
student1.SayHello(true);

Console.WriteLine("\n--- Метод Communicate() ---");
student1.Communicate("Привет!");
student1.Communicate(student2, "Как дела?");
student1.Communicate("Учусь!", 2);

Console.WriteLine("\n--- Метод AddGrade() ---");
student1.AddGrade(5);
student1.AddGrade(4, "Программирование");
student1.AddGrade(5, 4, 5);

Console.WriteLine("\n--- Метод AddProject() ---");
employee.AddProject("Веб-разработка");
employee.AddProject("Мобильное приложение", "База данных");

Console.WriteLine("\n--- Метод RaiseSalary() ---");
employee.RaiseSalary(15000);
employee.RaiseSalary(10, true);

// ========== 3. ДЕМОНСТРАЦИЯ ПЕРЕКРЫТИЯ МЕТОДОВ (OVERRIDE) ==========
Console.WriteLine("\n=== 3. ДЕМОНСТРАЦИЯ ПЕРЕКРЫТИЯ МЕТОДОВ (OVERRIDE) ===\n");

Person[] people = new Person[] { student1, student2, employee, teacher, professor };

foreach (var person in people)
{
    Console.WriteLine($"--- {person.GetType().Name} ---");
    person.SayHello();
    person.Work();
    person.Study();
    person.Relax();
    Console.WriteLine();
}

// ========== 4. ПОЛИМОРФИЗМ ==========
Console.WriteLine("=== 4. ПОЛИМОРФИЗМ ===\n");

Console.WriteLine("Вызов переопределенных методов через массив Person:");
foreach (var person in people)
{
    Console.WriteLine(person.GetInfo(true));
    person.SayHello();
    Console.WriteLine();
}

// ========== 5. GENERIC КЛАССЫ ==========
Console.WriteLine("=== 5. GENERIC КЛАССЫ ===\n");

Console.WriteLine("--- PersonCollection<Student> ---");
PersonCollection<Student> students = new PersonCollection<Student>();
students.Add(student1, student2, student3);
students.DisplayAll();

Console.WriteLine("\n--- PersonCollection<Teacher> ---");
PersonCollection<Teacher> teachers = new PersonCollection<Teacher>();
teachers.Add(teacher);
teachers.Add(professor);
teachers.DisplayAll();

Console.WriteLine("\n--- University<Student> ---");
University<Student> msu = new University<Student>("МГУ");
msu.AddMember(student1);
msu.AddMember(student2);
msu.AddMember(student3);
msu.DisplayInfo();

Console.WriteLine("\n--- PersonPair<Student, Teacher> ---");
PersonPair<Student, Teacher> pair = new PersonPair<Student, Teacher>(student1, teacher, "Учитель-ученик");
pair.DisplayPair();
pair.Introduce();

// ========== 6. СТАТИСТИКА ==========
Console.WriteLine("\n=== 6. СТАТИСТИКА ===\n");

List<Student> allStudents = new List<Student> { student1, student2, student3 };
PersonStatistics<Student> studentStats = new PersonStatistics<Student>(allStudents);
studentStats.DisplayStatistics();

List<Person> allPeople = new List<Person> { student1, student2, student3, employee, teacher, professor };
PersonStatistics<Person> personStats = new PersonStatistics<Person>(allPeople);
personStats.DisplayStatistics();

// ========== 7. ВЗАИМОДЕЙСТВИЕ ОБЪЕКТОВ ==========
Console.WriteLine("\n=== 7. ВЗАИМОДЕЙСТВИЕ ОБЪЕКТОВ ===\n");

teacher.AddStudent(student1);
teacher.AddStudent(student2);
teacher.ConductExam();

professor.MentorStudent(student3);
professor.DoResearch();
professor.PublishArticle("Новые методы ИИ");
professor.SpeakAtConference("AI Conference 2024");

// ========== 8. ИТОГОВАЯ СТАТИСТИКА ==========
Console.WriteLine("\n=== 8. ИТОГОВАЯ СТАТИСТИКА ===\n");
Person.DisplayTotalPeople();
Console.WriteLine($"Всего создано через Generic коллекции: {PersonCollection<Person>.TotalCreated}");

Console.WriteLine("\n========== ПРОЕКТ ЗАВЕРШЕН ==========");
```
