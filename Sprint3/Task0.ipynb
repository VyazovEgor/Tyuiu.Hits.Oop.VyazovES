### Понятие полиморфизма в C# с параметрами

Полиморфизм в объектно-ориентированном программировании позволяет объектам разных классов обрабатывать одни и те же сообщения по-разному. В C#, полиморфизм часто реализуется через наследование и переопределение методов. Давайте рассмотрим пример с классами `Vehicle`, `Car` и `Motorcycle`, чтобы продемонстрировать, как полиморфизм работает с параметрами.

#### Определение классов

Сначала определим базовый класс `Vehicle` с виртуальным методом `Drive()`, который принимает параметры. Затем создадим два производных класса: `Car` и `Motorcycle`, которые переопределят этот метод.

```csharp
// Базовый класс Vehicle с виртуальным методом Drive(), принимающим параметры
public class Vehicle
{
    public virtual void Drive(int speed)
    {
        Console.WriteLine($"Транспорт движется {speed} km/h.");
    }
}

// Производный класс Car, переопределяющий метод Drive()
public class Car : Vehicle
{
    public override void Drive(int speed)
    {
        Console.WriteLine($"Автомобиль движется {speed} km/h.");
    }
}

// Производный класс Motorcycle, также переопределяющий метод Drive()
public class Motorcycle : Vehicle
{
    public override void Drive(int speed)
    {
        Console.WriteLine($"Мотоцикл движется {speed} km/h.");
    }
}
```

#### Использование полиморфизма

Теперь создадим список объектов типа `Vehicle` и добавим в него экземпляры `Car` и `Motorcycle`. Затем вызовем метод `Drive()` для каждого элемента списка, передавая разные скорости. Это демонстрирует, как полиморфизм позволяет нам работать с объектами разных типов через общий интерфейс, но при этом выполнять различные действия в зависимости от "реального" типа объекта.

```csharp
List<Vehicle> vehicles = new List<Vehicle>
{
    new Car(),
    new Motorcycle(),
    new Vehicle() // Добавляем экземпляр базового класса для демонстрации
};

foreach (var vehicle in vehicles)
{
    vehicle.Drive(100); // Вызываем метод Drive() 
}

```

В результате выполнения этого кода, каждый объект в списке вызывает свою версию метода `Drive()`, демонстрируя, что при работе с объектами через базовый класс (`Vehicle`) можно достичь различных действий в зависимости от "реального" типа объекта, даже когда используются разные скорости.

Этот пример иллюстрирует ключевое понятие полиморфизма в ООП: возможность работы с объектами разных типов через общий интерфейс или базовый класс, что позволяет писать более гибкий и расширяемый код.

````
using System;
using System.Collections.Generic;

// ========== БАЗОВЫЙ КЛАСС VEHICLE ==========
public class Vehicle
{
    // Виртуальный метод Drive(), принимающий параметр speed
    public virtual void Drive(int speed)
    {
        Console.WriteLine($"Транспорт движется {speed} км/ч.");
    }
    
    // Дополнительный виртуальный метод для демонстрации полиморфизма
    public virtual void Stop()
    {
        Console.WriteLine("Транспорт остановлен.");
    }
    
    // Метод для вывода информации
    public virtual void DisplayInfo()
    {
        Console.WriteLine("Транспортное средство");
    }
}

// ========== ПРОИЗВОДНЫЙ КЛАСС CAR ==========
public class Car : Vehicle
{
    // Свойства автомобиля
    public string Model { get; set; }
    public int Doors { get; set; }
    
    public Car(string model, int doors)
    {
        Model = model;
        Doors = doors;
    }
    
    // Переопределение метода Drive()
    public override void Drive(int speed)
    {
        Console.WriteLine($"Автомобиль {Model} движется {speed} км/ч.");
        if (speed > 120)
        {
            Console.WriteLine("  Внимание! Превышение скорости!");
        }
    }
    
    // Переопределение метода Stop()
    public override void Stop()
    {
        Console.WriteLine($"Автомобиль {Model} остановлен. Двери {Doors} закрыты.");
    }
    
    // Переопределение метода DisplayInfo()
    public override void DisplayInfo()
    {
        Console.WriteLine($"Автомобиль: {Model}, Количество дверей: {Doors}");
    }
}

// ========== ПРОИЗВОДНЫЙ КЛАСС MOTORCYCLE ==========
public class Motorcycle : Vehicle
{
    // Свойства мотоцикла
    public string Model { get; set; }
    public string EngineType { get; set; }
    
    public Motorcycle(string model, string engineType)
    {
        Model = model;
        EngineType = engineType;
    }
    
    // Переопределение метода Drive()
    public override void Drive(int speed)
    {
        Console.WriteLine($"Мотоцикл {Model} движется {speed} км/ч.");
        if (speed > 100)
        {
            Console.WriteLine("  Внимание! Мотоцикл разгоняется слишком быстро!");
        }
    }
    
    // Переопределение метода Stop()
    public override void Stop()
    {
        Console.WriteLine($"Мотоцикл {Model} остановлен. Двигатель {EngineType} заглушен.");
    }
    
    // Переопределение метода DisplayInfo()
    public override void DisplayInfo()
    {
        Console.WriteLine($"Мотоцикл: {Model}, Тип двигателя: {EngineType}");
    }
}

// ========== ПРОИЗВОДНЫЙ КЛАСС TRUCK (для дополнительной демонстрации) ==========
public class Truck : Vehicle
{
    // Свойства грузовика
    public string Model { get; set; }
    public double LoadCapacity { get; set; }
    
    public Truck(string model, double loadCapacity)
    {
        Model = model;
        LoadCapacity = loadCapacity;
    }
    
    // Переопределение метода Drive()
    public override void Drive(int speed)
    {
        Console.WriteLine($"Грузовик {Model} движется {speed} км/ч с грузом {LoadCapacity} тонн.");
        if (speed > 80)
        {
            Console.WriteLine("  Внимание! Грузовик превышает безопасную скорость!");
        }
    }
    
    // Переопределение метода Stop()
    public override void Stop()
    {
        Console.WriteLine($"Грузовик {Model} остановлен. Груз {LoadCapacity} тонн на месте.");
    }
    
    // Переопределение метода DisplayInfo()
    public override void DisplayInfo()
    {
        Console.WriteLine($"Грузовик: {Model}, Грузоподъемность: {LoadCapacity} тонн");
    }
}

// ========== ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА ==========
Console.WriteLine("========== ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА В КЛАССАХ VEHICLE ==========\n");

// ========== 1. СОЗДАНИЕ ОБЪЕКТОВ РАЗНЫХ КЛАССОВ ==========
Console.WriteLine("=== 1. СОЗДАНИЕ ОБЪЕКТОВ ===\n");

Car car = new Car("Toyota Camry", 4);
Motorcycle motorcycle = new Motorcycle("Honda CBR600RR", "Inline Four");
Truck truck = new Truck("Volvo FH16", 25.5);

Console.WriteLine("Созданы объекты:");
car.DisplayInfo();
motorcycle.DisplayInfo();
truck.DisplayInfo();

// ========== 2. ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА С РАЗНЫМИ ПАРАМЕТРАМИ ==========
Console.WriteLine("\n=== 2. ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА (РАЗНЫЕ СКОРОСТИ) ===\n");

Console.WriteLine("--- Метод Drive() с разными скоростями ---");
car.Drive(60);
car.Drive(130);
Console.WriteLine();

motorcycle.Drive(80);
motorcycle.Drive(150);
Console.WriteLine();

truck.Drive(50);
truck.Drive(95);
Console.WriteLine();

// ========== 3. ПОЛИМОРФИЗМ ЧЕРЕЗ МАССИВ БАЗОВОГО КЛАССА ==========
Console.WriteLine("=== 3. ПОЛИМОРФИЗМ ЧЕРЕЗ МАССИВ Vehicle ===\n");

// Создаем массив объектов базового класса Vehicle
Vehicle[] vehicles = new Vehicle[]
{
    new Car("BMW X5", 5),
    new Motorcycle("Yamaha R1", "V4"),
    new Truck("MAN TGX", 30),
    new Car("Ford Focus", 4),
    new Motorcycle("Ducati Panigale", "V2"),
    new Vehicle() // Экземпляр базового класса
};

Console.WriteLine("Вызов метода Drive() для всех транспортных средств со скоростью 100 км/ч:\n");

// Полиморфный вызов метода Drive()
foreach (var vehicle in vehicles)
{
    vehicle.DisplayInfo();
    vehicle.Drive(100);
    Console.WriteLine();
}

// ========== 4. ДЕМОНСТРАЦИЯ МЕТОДА STOP ЧЕРЕЗ МАССИВ ==========
Console.WriteLine("=== 4. ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА (МЕТОД Stop) ===\n");

foreach (var vehicle in vehicles)
{
    vehicle.Stop();
}
Console.WriteLine();

// ========== 5. ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА С РАЗНЫМИ СКОРОСТЯМИ ==========
Console.WriteLine("=== 5. ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА (РАЗНЫЕ СКОРОСТИ ДВИЖЕНИЯ) ===\n");

List<Vehicle> vehicleList = new List<Vehicle>
{
    new Car("Tesla Model S", 4),
    new Motorcycle("Kawasaki Ninja", "Inline-4"),
    new Truck("Scania R500", 28)
};

int[] testSpeeds = { 30, 60, 90, 120, 150 };

Console.WriteLine("Тестирование движения транспортных средств с разными скоростями:\n");
foreach (var vehicle in vehicleList)
{
    Console.WriteLine($"{vehicle.GetType().Name}:");
    foreach (var speed in testSpeeds)
    {
        vehicle.Drive(speed);
    }
    Console.WriteLine();
}

// ========== 6. ДОПОЛНИТЕЛЬНАЯ ДЕМОНСТРАЦИЯ: ПОЛИМОРФИЗМ В ДЕЙСТВИИ ==========
Console.WriteLine("=== 6. ПОЛИМОРФИЗМ В ДЕЙСТВИИ (ОДИНАКОВЫЕ ВЫЗОВЫ - РАЗНОЕ ПОВЕДЕНИЕ) ===\n");

Vehicle[] demoVehicles = new Vehicle[]
{
    new Car("Audi A6", 4),
    new Motorcycle("Harley Davidson", "V-Twin"),
    new Truck("Mercedes Actros", 35)
};

int demoSpeed = 85;

Console.WriteLine($"Все транспортные средства получают одинаковую скорость: {demoSpeed} км/ч\n");

foreach (var vehicle in demoVehicles)
{
    vehicle.DisplayInfo();
    vehicle.Drive(demoSpeed);
    vehicle.Stop();
    Console.WriteLine();
}

// ========== 7. ДЕМОНСТРАЦИЯ В СООТВЕТСТВИИ С ПЛЕЙБУКОМ ==========
Console.WriteLine("=== 7. ДЕМОНСТРАЦИЯ В СООТВЕТСТВИИ С ПРИМЕРОМ ИЗ ПЛЕЙБУКА ===\n");

// Создаем список объектов типа Vehicle (как в примере)
List<Vehicle> vehiclesExample = new List<Vehicle>
{
    new Car("Toyota Corolla", 4),
    new Motorcycle("Honda CBR", "Inline Four"),
    new Vehicle() // Добавляем экземпляр базового класса для демонстрации
};

Console.WriteLine("Вызов метода Drive(100) для всех транспортных средств (как в примере плейбука):\n");

foreach (var vehicle in vehiclesExample)
{
    vehicle.DisplayInfo();
    vehicle.Drive(100); // Вызываем метод Drive() с параметром 100
}

// ========== 8. ВЫВОД СТАТИСТИКИ ==========
Console.WriteLine("\n=== 8. СТАТИСТИКА ДЕМОНСТРАЦИИ ===\n");
Console.WriteLine($"Всего продемонстрировано транспортных средств: {vehicles.Length + vehicleList.Count + demoVehicles.Length + vehiclesExample.Count}");
Console.WriteLine($"Типы транспортных средств: Car, Motorcycle, Truck, Vehicle");
Console.WriteLine("\n========== ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА ЗАВЕРШЕНА ==========");
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также механизм полиморфизм.

````

using System;
using System.Collections.Generic;

// ========== БАЗОВЫЙ КЛАСС ANIMAL ==========
public class Animal
{
    // Свойства с геттерами и сеттерами
    public string Name { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }
    
    // Конструктор
    public Animal(string name, int age, double weight)
    {
        Name = name;
        Age = age;
        Weight = weight;
    }
    
    // Виртуальный метод MakeSound() с параметром (аналогия с Drive(int speed))
    public virtual void MakeSound(int volume)
    {
        string volumeLevel = volume switch
        {
            <= 30 => "тихо",
            <= 70 => "средне",
            _ => "громко"
        };
        Console.WriteLine($"{Name} издает звук {volumeLevel} (громкость: {volume} дБ).");
    }
    
    // Виртуальный метод Move() с параметром (дополнительная демонстрация полиморфизма)
    public virtual void Move(int speed)
    {
        Console.WriteLine($"{Name} двигается со скоростью {speed} км/ч.");
    }
    
    // Виртуальный метод Eat() с параметром
    public virtual void Eat(string food, double amount)
    {
        Console.WriteLine($"{Name} ест {amount} кг {food}.");
    }
    
    // Метод для вывода информации
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Имя: {Name}, Возраст: {Age} лет, Вес: {Weight} кг");
    }
}

// ========== ПРОИЗВОДНЫЙ КЛАСС DOG (СОБАКА) ==========
public class Dog : Animal
{
    // Специфичные свойства
    public string Breed { get; set; }
    public bool IsTrained { get; set; }
    
    // Конструктор
    public Dog(string name, int age, double weight, string breed, bool isTrained) 
        : base(name, age, weight)
    {
        Breed = breed;
        IsTrained = isTrained;
    }
    
    // Переопределение метода MakeSound()
    public override void MakeSound(int volume)
    {
        string volumeLevel = volume switch
        {
            <= 30 => "тихо",
            <= 70 => "средне",
            _ => "громко"
        };
        
        string sound = volume > 80 ? "ГАВ-ГАВ-ГАВ!!!" : "гав-гав";
        Console.WriteLine($"Собака {Name} лает {volumeLevel}: {sound} (громкость: {volume} дБ).");
    }
    
    // Переопределение метода Move()
    public override void Move(int speed)
    {
        if (speed > 60)
            Console.WriteLine($"Собака {Name} бежит очень быстро! Скорость: {speed} км/ч");
        else if (speed > 20)
            Console.WriteLine($"Собака {Name} бежит трусцой со скоростью {speed} км/ч");
        else
            Console.WriteLine($"Собака {Name} медленно идет со скоростью {speed} км/ч");
    }
    
    // Переопределение метода Eat()
    public override void Eat(string food, double amount)
    {
        Console.WriteLine($"Собака {Name} с аппетитом съедает {amount} кг {food}.");
        if (IsTrained)
            Console.WriteLine($"  {Name} ждет команду после еды!");
    }
    
    // Переопределение DisplayInfo
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}, Дрессирована: {(IsTrained ? "Да" : "Нет")}");
    }
}

// ========== ПРОИЗВОДНЫЙ КЛАСС CAT (КОШКА) ==========
public class Cat : Animal
{
    // Специфичные свойства
    public string Color { get; set; }
    public bool IsIndoor { get; set; }
    
    // Конструктор
    public Cat(string name, int age, double weight, string color, bool isIndoor) 
        : base(name, age, weight)
    {
        Color = color;
        IsIndoor = isIndoor;
    }
    
    // Переопределение метода MakeSound()
    public override void MakeSound(int volume)
    {
        string volumeLevel = volume switch
        {
            <= 30 => "тихо",
            <= 70 => "средне",
            _ => "громко"
        };
        
        string sound = volume > 70 ? "МЯУ-МЯУ-МЯУ!!!" : "мяу-мяу";
        Console.WriteLine($"Кошка {Name} мяукает {volumeLevel}: {sound} (громкость: {volume} дБ).");
    }
    
    // Переопределение метода Move()
    public override void Move(int speed)
    {
        if (speed > 40)
            Console.WriteLine($"Кошка {Name} быстро бежит! Скорость: {speed} км/ч");
        else if (speed > 10)
            Console.WriteLine($"Кошка {Name} грациозно крадется со скоростью {speed} км/ч");
        else
            Console.WriteLine($"Кошка {Name} медленно крадется, подкрадываясь к добыче");
    }
    
    // Переопределение метода Eat()
    public override void Eat(string food, double amount)
    {
        Console.WriteLine($"Кошка {Name} изящно пробует {amount} кг {food}.");
        if (!IsIndoor)
            Console.WriteLine($"  {Name} может принести мышь в благодарность!");
    }
    
    // Переопределение DisplayInfo
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Окрас: {Color}, Домашняя: {(IsIndoor ? "Да" : "Нет")}");
    }
}

// ========== ПРОИЗВОДНЫЙ КЛАСС BIRD (ПТИЦА) ==========
public class Bird : Animal
{
    // Специфичные свойства
    public double WingSpan { get; set; }
    public bool CanFly { get; set; }
    
    // Конструктор
    public Bird(string name, int age, double weight, double wingSpan, bool canFly) 
        : base(name, age, weight)
    {
        WingSpan = wingSpan;
        CanFly = canFly;
    }
    
    // Переопределение метода MakeSound()
    public override void MakeSound(int volume)
    {
        string volumeLevel = volume switch
        {
            <= 30 => "тихо",
            <= 70 => "средне",
            _ => "громко"
        };
        
        Console.WriteLine($"Птица {Name} чирикает {volumeLevel}: Чик-чирик! (громкость: {volume} дБ).");
    }
    
    // Переопределение метода Move()
    public override void Move(int speed)
    {
        if (CanFly)
        {
            Console.WriteLine($"Птица {Name} летит со скоростью {speed} км/ч, размах крыльев {WingSpan} м");
        }
        else
        {
            Console.WriteLine($"Птица {Name} ходит по земле со скоростью {speed} км/ч");
        }
    }
    
    // Переопределение метода Eat()
    public override void Eat(string food, double amount)
    {
        Console.WriteLine($"Птица {Name} клюет {amount} кг {food}.");
    }
    
    // Переопределение DisplayInfo
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Размах крыльев: {WingSpan} м, Умеет летать: {(CanFly ? "Да" : "Нет")}");
    }
}

// ========== ПРОИЗВОДНЫЙ КЛАСС FISH (РЫБА) ==========
public class Fish : Animal
{
    // Специфичные свойства
    public string WaterType { get; set; }
    public double MaxDepth { get; set; }
    
    // Конструктор
    public Fish(string name, int age, double weight, string waterType, double maxDepth) 
        : base(name, age, weight)
    {
        WaterType = waterType;
        MaxDepth = maxDepth;
    }
    
    // Переопределение метода MakeSound()
    public override void MakeSound(int volume)
    {
        Console.WriteLine($"Рыба {Name} издает булькающие звуки: Буль-буль! (громкость: {volume} дБ)");
    }
    
    // Переопределение метода Move()
    public override void Move(int speed)
    {
        Console.WriteLine($"Рыба {Name} плавает в {WaterType} воде со скоростью {speed} км/ч на глубине {MaxDepth} м");
    }
    
    // Переопределение метода Eat()
    public override void Eat(string food, double amount)
    {
        Console.WriteLine($"Рыба {Name} проглатывает {amount} кг {food}.");
    }
    
    // Переопределение DisplayInfo
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Тип воды: {WaterType}, Макс. глубина: {MaxDepth} м");
    }
}

// ========== ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА ==========

Console.WriteLine("========== ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА В КЛАССАХ ЖИВОТНЫХ ==========\n");

// ========== 1. СОЗДАНИЕ ОБЪЕКТОВ РАЗНЫХ ТИПОВ ==========
Console.WriteLine("=== 1. СОЗДАНИЕ ОБЪЕКТОВ ===\n");

Dog dog = new Dog("Рекс", 3, 25.5, "Овчарка", true);
Cat cat = new Cat("Муся", 2, 4.2, "Серый", true);
Bird bird = new Bird("Кеша", 1, 0.35, 0.5, true);
Fish fish = new Fish("Немо", 2, 0.15, "Соленая", 50);

Console.WriteLine("Созданы объекты разных классов:");
dog.DisplayInfo();
cat.DisplayInfo();
bird.DisplayInfo();
fish.DisplayInfo();

// ========== 2. ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА С РАЗНЫМИ ПАРАМЕТРАМИ ==========
Console.WriteLine("\n=== 2. ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА (РАЗНЫЕ ПАРАМЕТРЫ) ===\n");

Console.WriteLine("--- Метод MakeSound() с разными уровнями громкости ---");
dog.MakeSound(20);   // тихо
dog.MakeSound(50);   // средне
dog.MakeSound(90);   // громко
Console.WriteLine();

cat.MakeSound(15);
cat.MakeSound(45);
cat.MakeSound(85);
Console.WriteLine();

bird.MakeSound(25);
bird.MakeSound(60);
Console.WriteLine();

// ========== 3. ПОЛИМОРФИЗМ ЧЕРЕЗ МАССИВ БАЗОВОГО КЛАССА ==========
Console.WriteLine("=== 3. ПОЛИМОРФИЗМ ЧЕРЕЗ МАССИВ Animal (аналогия с Vehicle) ===\n");

// Создаем массив объектов базового класса Animal
Animal[] animals = new Animal[]
{
    new Dog("Шарик", 4, 20.0, "Дворняга", false),
    new Cat("Васька", 3, 3.8, "Черный", false),
    new Bird("Пингвин", 5, 30.0, 0.8, false),
    new Fish("Золотая рыбка", 1, 0.1, "Пресная", 2),
    new Dog("Лайка", 2, 18.5, "Хаски", true),
    new Cat("Снежок", 1, 3.2, "Белый", true),
    new Animal("Безымянный", 1, 10.0) // Экземпляр базового класса
};

Console.WriteLine("Вызов метода Move() для всех животных с разной скоростью:\n");

// Демонстрация полиморфизма - вызов метода Move() с разными параметрами
int[] speeds = { 10, 25, 50, 75, 100 };

for (int i = 0; i < animals.Length; i++)
{
    Console.WriteLine($"Животное {i + 1}:");
    animals[i].DisplayInfo();
    animals[i].Move(speeds[i % speeds.Length]);
    Console.WriteLine();
}

// ========== 4. ДЕМОНСТРАЦИЯ МЕТОДА MAKE SOUND ЧЕРЕЗ МАССИВ ==========
Console.WriteLine("=== 4. ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА (МЕТОД MakeSound) ===\n");

int[] volumes = { 20, 50, 80 };

foreach (var animal in animals)
{
    foreach (var volume in volumes)
    {
        animal.MakeSound(volume);
    }
    Console.WriteLine();
}

// ========== 5. ДЕМОНСТРАЦИЯ МЕТОДА EAT С РАЗНЫМИ ПАРАМЕТРАМИ ==========
Console.WriteLine("=== 5. ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА (МЕТОД Eat) ===\n");

var foods = new[] { "корм", "рыбу", "мясо", "зерно" };
var amounts = new[] { 0.2, 0.5, 1.0, 1.5 };

foreach (var animal in animals)
{
    animal.Eat(foods[new Random().Next(foods.Length)], amounts[new Random().Next(amounts.Length)]);
}
Console.WriteLine();

// ========== 6. ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА С РАЗНЫМИ СКОРОСТЯМИ ==========
Console.WriteLine("=== 6. ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА (РАЗНЫЕ СКОРОСТИ ДВИЖЕНИЯ) ===\n");

List<Animal> animalList = new List<Animal>
{
    new Dog("Барбос", 5, 28.0, "Бульдог", true),
    new Cat("Мурзик", 4, 5.0, "Серый", false),
    new Bird("Орел", 3, 3.5, 2.0, true),
    new Fish("Карп", 2, 1.2, "Пресная", 10)
};

int[] testSpeeds = { 5, 15, 30, 45, 60 };

Console.WriteLine("Тестирование движения животных с разными скоростями:\n");
foreach (var animal in animalList)
{
    Console.WriteLine($"{animal.GetType().Name} {animal.Name}:");
    foreach (var speed in testSpeeds)
    {
        animal.Move(speed);
    }
    Console.WriteLine();
}

// ========== 7. ДОПОЛНИТЕЛЬНАЯ ДЕМОНСТРАЦИЯ: ПОЛИМОРФИЗМ В ДЕЙСТВИИ ==========
Console.WriteLine("=== 7. ПОЛИМОРФИЗМ В ДЕЙСТВИИ (ОДИНАКОВЫЕ ВЫЗОВЫ - РАЗНОЕ ПОВЕДЕНИЕ) ===\n");

Animal[] demoAnimals = new Animal[]
{
    new Dog("Дружок", 2, 15.0, "Такса", true),
    new Cat("Барсик", 3, 4.5, "Рыжий", true),
    new Bird("Воробей", 1, 0.05, 0.2, true),
    new Fish("Лосось", 4, 2.5, "Соленая", 100)
};

int demoVolume = 65;
int demoSpeed = 40;
string demoFood = "корм";
double demoAmount = 0.5;

Console.WriteLine($"Все животные получают одинаковые параметры:");
Console.WriteLine($"  - Громкость звука: {demoVolume} дБ");
Console.WriteLine($"  - Скорость движения: {demoSpeed} км/ч");
Console.WriteLine($"  - Еда: {demoFood} ({demoAmount} кг)\n");

foreach (var animal in demoAnimals)
{
    Console.WriteLine($"{animal.GetType().Name} {animal.Name}:");
    animal.MakeSound(demoVolume);
    animal.Move(demoSpeed);
    animal.Eat(demoFood, demoAmount);
    Console.WriteLine();
}

// ========== 8. ВЫВОД СТАТИСТИКИ ==========
Console.WriteLine("=== 8. СТАТИСТИКА ДЕМОНСТРАЦИИ ===\n");
Console.WriteLine($"Всего продемонстрировано животных: {animals.Length + animalList.Count + demoAnimals.Length}");
Console.WriteLine($"Типы животных: Dog, Cat, Bird, Fish");
Console.WriteLine("\n========== ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА ЗАВЕРШЕНА ==========");
````
