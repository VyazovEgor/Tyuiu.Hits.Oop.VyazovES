<h1 style="color:DodgerBlue">Generic классы в C#</h1>


Генерические классы в C# представляют собой шаблон класса, который может работать с различными типами данных без потери производительности. Они позволяют создавать компоненты, которые могут быть безопасно повторно использованы в различных контекстах, обеспечивая при этом строгую типизацию и избегая необходимости приведения типов.

#### Основные преимущества генерических классов:

1. **Безопасность типов**: Генерические классы обеспечивают проверку типов на этапе компиляции, что помогает предотвратить ошибки во время выполнения.
2. **Универсальность**: Генерические классы могут работать с различными типами данных, делая их очень гибкими и универсальными.
3. **Производительность**: Генерические классы не приводят к значительной потере производительности, поскольку информация о типе хранится вместе с объектом.

#### Пример с классами `Vehicle`, `Car` и `Motorcycle`

Давайте рассмотрим пример генерического класса `VehicleCollection<T>`, который может хранить коллекцию объектов типа `T`, где `T` может быть любой строкой, например, `Vehicle`, `Car` или `Motorcycle`.

```csharp
using System;
using System.Collections.Generic;

// Определение базового класса Vehicle
public class Vehicle
{
    public virtual void Drive()
    {
        Console.WriteLine("Транспортное средство находится в движении.");
    }
}

// Определение производного класса Car
public class Car : Vehicle
{
    public override void Drive()
    {
        Console.WriteLine("Машина едет.");
    }
}

// Определение производного класса Motorcycle
public class Motorcycle : Vehicle
{
    public override void Drive()
    {
        Console.WriteLine("Мотоцикл едет.");
    }
}

// Определение генерического класса VehicleCollection<T>
public class VehicleCollection<T> where T : Vehicle
{
    private List<T> _vehicles = new List<T>();

    public void Add(T vehicle)
    {
        _vehicles.Add(vehicle);
    }

    public void Remove(T vehicle)
    {
        _vehicles.Remove(vehicle);
    }

    public void DisplayVehicles()
    {
        foreach (var vehicle in _vehicles)
        {
            vehicle.Drive(); // Вызов метода Drive() для каждого транспортного средства
        }
    }
}


VehicleCollection<Car> cars = new VehicleCollection<Car>();
cars.Add(new Car());
cars.Add(new Motorcycle()); // Это возможно благодаря ограничению "where T : Vehicle"

cars.DisplayVehicles(); // Выводит: "Машина едет." и "Мотоцикл едет."


```

В этом примере `VehicleCollection<T>` является генерическим классом, который может хранить коллекцию объектов типа `T`, где `T` должен быть подклассом `Vehicle`. Это позволяет создавать коллекции автомобилей, мотоциклов и других видов транспортных средств, обеспечивая при этом строгую типизацию и безопасность типов.

### Заключение

Генерические классы в C# предоставляют мощный и гибкий способ создания компонентов, которые могут работать с различными типами данных, обеспечивая при этом строгую типизацию и безопасность типов. Это делает их идеальным выбором для многих задач, требующих повторного использования кода с высокой степенью абстракции.

````
using System;
using System.Collections.Generic;

// ========== БАЗОВЫЙ КЛАСС VEHICLE ==========
public class Vehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    public int Year { get; set; }
    
    public Vehicle(string make, string model, int year)
    {
        Make = make;
        Model = model;
        Year = year;
    }
    
    public virtual void Drive()
    {
        Console.WriteLine($"Транспортное средство {Make} {Model} находится в движении.");
    }
    
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"{Make} {Model}, {Year} год");
    }
}

// ========== ПРОИЗВОДНЫЙ КЛАСС CAR ==========
public class Car : Vehicle
{
    public int Doors { get; set; }
    public string BodyType { get; set; }
    
    public Car(string make, string model, int year, int doors, string bodyType) 
        : base(make, model, year)
    {
        Doors = doors;
        BodyType = bodyType;
    }
    
    public override void Drive()
    {
        Console.WriteLine($"Автомобиль {Make} {Model} едет по дороге.");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип: Автомобиль, Дверей: {Doors}, Кузов: {BodyType}");
    }
}

// ========== ПРОИЗВОДНЫЙ КЛАСС MOTORCYCLE ==========
public class Motorcycle : Vehicle
{
    public string EngineType { get; set; }
    public bool HasSidecar { get; set; }
    
    public Motorcycle(string make, string model, int year, string engineType, bool hasSidecar) 
        : base(make, model, year)
    {
        EngineType = engineType;
        HasSidecar = hasSidecar;
    }
    
    public override void Drive()
    {
        Console.WriteLine($"Мотоцикл {Make} {Model} мчится по трассе.");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип: Мотоцикл, Двигатель: {EngineType}, Коляска: {(HasSidecar ? "Есть" : "Нет")}");
    }
}

// ========== ПРОИЗВОДНЫЙ КЛАСС TRUCK ==========
public class Truck : Vehicle
{
    public double LoadCapacity { get; set; }
    public int Axles { get; set; }
    
    public Truck(string make, string model, int year, double loadCapacity, int axles) 
        : base(make, model, year)
    {
        LoadCapacity = loadCapacity;
        Axles = axles;
    }
    
    public override void Drive()
    {
        Console.WriteLine($"Грузовик {Make} {Model} перевозит {LoadCapacity} тонн груза.");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип: Грузовик, Грузоподъемность: {LoadCapacity} т, Оси: {Axles}");
    }
}

// ========== GENERIC КЛАСС VEHICLE COLLECTION ==========
public class VehicleCollection<T> where T : Vehicle
{
    private List<T> _vehicles = new List<T>();
    
    // Добавление транспортного средства
    public void Add(T vehicle)
    {
        _vehicles.Add(vehicle);
        Console.WriteLine($"Добавлен: {vehicle.Make} {vehicle.Model}");
    }
    
    // Удаление транспортного средства
    public bool Remove(T vehicle)
    {
        bool removed = _vehicles.Remove(vehicle);
        if (removed)
            Console.WriteLine($"Удален: {vehicle.Make} {vehicle.Model}");
        return removed;
    }
    
    // Получение количества транспортных средств
    public int Count => _vehicles.Count;
    
    // Отображение всех транспортных средств
    public void DisplayAll()
    {
        Console.WriteLine($"\nВсего транспортных средств: {Count}");
        foreach (var vehicle in _vehicles)
        {
            vehicle.DisplayInfo();
            vehicle.Drive();
            Console.WriteLine();
        }
    }
    
    // Поиск по производителю
    public List<T> FindByMake(string make)
    {
        return _vehicles.FindAll(v => v.Make.Equals(make, StringComparison.OrdinalIgnoreCase));
    }
    
    // Получение транспортного средства по индексу
    public T GetAt(int index)
    {
        if (index >= 0 && index < _vehicles.Count)
            return _vehicles[index];
        throw new IndexOutOfRangeException("Индекс вне диапазона");
    }
}

// ========== GENERIC КЛАСС ДЛЯ ПАРЫ ЗНАЧЕНИЙ ==========
public class Pair<T1, T2> where T1 : Vehicle where T2 : Vehicle
{
    public T1 First { get; set; }
    public T2 Second { get; set; }
    
    public Pair(T1 first, T2 second)
    {
        First = first;
        Second = second;
    }
    
    public void DisplayComparison()
    {
        Console.WriteLine($"Сравнение:");
        First.DisplayInfo();
        Second.DisplayInfo();
    }
}

// ========== ДЕМОНСТРАЦИЯ ДЛЯ VEHICLE ==========
Console.WriteLine("========== GENERIC КЛАССЫ: ТРАНСПОРТНЫЕ СРЕДСТВА ==========\n");

// Создание коллекции для Car
Console.WriteLine("=== 1. VehicleCollection<Car> ===");
VehicleCollection<Car> carCollection = new VehicleCollection<Car>();
carCollection.Add(new Car("Toyota", "Camry", 2022, 4, "Седан"));
carCollection.Add(new Car("Honda", "Accord", 2023, 4, "Седан"));
carCollection.Add(new Car("BMW", "X5", 2024, 5, "Внедорожник"));
carCollection.DisplayAll();

// Создание коллекции для Motorcycle
Console.WriteLine("\n=== 2. VehicleCollection<Motorcycle> ===");
VehicleCollection<Motorcycle> motorcycleCollection = new VehicleCollection<Motorcycle>();
motorcycleCollection.Add(new Motorcycle("Honda", "CBR600RR", 2023, "Inline-4", false));
motorcycleCollection.Add(new Motorcycle("Harley-Davidson", "Street Glide", 2022, "V-Twin", false));
motorcycleCollection.Add(new Motorcycle("Yamaha", "R1", 2024, "V4", false));
motorcycleCollection.DisplayAll();

// Создание коллекции для Truck
Console.WriteLine("\n=== 3. VehicleCollection<Truck> ===");
VehicleCollection<Truck> truckCollection = new VehicleCollection<Truck>();
truckCollection.Add(new Truck("Volvo", "FH16", 2023, 25.5, 3));
truckCollection.Add(new Truck("Scania", "R500", 2022, 28.0, 3));
truckCollection.DisplayAll();

// Создание универсальной коллекции VehicleCollection<Vehicle>
Console.WriteLine("\n=== 4. VehicleCollection<Vehicle> (универсальная) ===");
VehicleCollection<Vehicle> allVehicles = new VehicleCollection<Vehicle>();
allVehicles.Add(new Car("Tesla", "Model S", 2024, 4, "Лифтбек"));
allVehicles.Add(new Motorcycle("Ducati", "Panigale V4", 2023, "V4", false));
allVehicles.Add(new Truck("MAN", "TGX", 2022, 30.0, 4));
allVehicles.Add(new Car("Ford", "Mustang", 2023, 2, "Купе"));
allVehicles.DisplayAll();

// Демонстрация методов Generic коллекции
Console.WriteLine("\n=== 5. Дополнительные методы ===");
Console.WriteLine($"Количество автомобилей: {carCollection.Count}");

var bmwCars = carCollection.FindByMake("BMW");
Console.WriteLine($"\nНайдено BMW: {bmwCars.Count}");
foreach (var car in bmwCars)
{
    car.DisplayInfo();
}

// Демонстрация Generic класса Pair
Console.WriteLine("\n=== 6. Generic класс Pair<T1, T2> ===");
Pair<Car, Motorcycle> pair = new Pair<Car, Motorcycle>(
    new Car("Porsche", "911", 2024, 2, "Купе"),
    new Motorcycle("Kawasaki", "Ninja H2", 2023, "Supercharged", false)
);
pair.DisplayComparison();

Console.WriteLine("\n========== ДЕМОНСТРАЦИЯ VEHICLE ЗАВЕРШЕНА ==========\n");
````

````
public class Animal
{
    public string Name { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }
    
    public Animal(string name, int age, double weight)
    {
        Name = name;
        Age = age;
        Weight = weight;
    }
    
    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} издает звук.");
    }
    
    public virtual void Move()
    {
        Console.WriteLine($"{Name} двигается.");
    }
    
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Имя: {Name}, Возраст: {Age} лет, Вес: {Weight} кг");
    }
}

// ========== ПРОИЗВОДНЫЙ КЛАСС DOG ==========
public class Dog : Animal
{
    public string Breed { get; set; }
    public bool IsTrained { get; set; }
    
    public Dog(string name, int age, double weight, string breed, bool isTrained) 
        : base(name, age, weight)
    {
        Breed = breed;
        IsTrained = isTrained;
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} лает: Гав-гав!");
    }
    
    public override void Move()
    {
        Console.WriteLine($"{Name} бегает на четырех лапах.");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Порода: {Breed}, Дрессирована: {(IsTrained ? "Да" : "Нет")}");
    }
    
    public void Fetch()
    {
        Console.WriteLine($"{Name} приносит мяч!");
    }
}

// ========== ПРОИЗВОДНЫЙ КЛАСС CAT ==========
public class Cat : Animal
{
    public string Color { get; set; }
    public bool IsIndoor { get; set; }
    
    public Cat(string name, int age, double weight, string color, bool isIndoor) 
        : base(name, age, weight)
    {
        Color = color;
        IsIndoor = isIndoor;
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} мяукает: Мяу-мяу!");
    }
    
    public override void Move()
    {
        Console.WriteLine($"{Name} грациозно крадется.");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Окрас: {Color}, Домашняя: {(IsIndoor ? "Да" : "Нет")}");
    }
    
    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет: Мур-мур-мур...");
    }
}

// ========== ПРОИЗВОДНЫЙ КЛАСС BIRD ==========
public class Bird : Animal
{
    public double WingSpan { get; set; }
    public bool CanFly { get; set; }
    
    public Bird(string name, int age, double weight, double wingSpan, bool canFly) 
        : base(name, age, weight)
    {
        WingSpan = wingSpan;
        CanFly = canFly;
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} чирикает: Чик-чирик!");
    }
    
    public override void Move()
    {
        if (CanFly)
            Console.WriteLine($"{Name} летит с размахом крыльев {WingSpan} м.");
        else
            Console.WriteLine($"{Name} ходит по земле.");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Размах крыльев: {WingSpan} м, Умеет летать: {(CanFly ? "Да" : "Нет")}");
    }
    
    public void BuildNest()
    {
        Console.WriteLine($"{Name} строит гнездо.");
    }
}

// ========== ПРОИЗВОДНЫЙ КЛАСС FISH ==========
public class Fish : Animal
{
    public string WaterType { get; set; }
    public double MaxDepth { get; set; }
    
    public Fish(string name, int age, double weight, string waterType, double maxDepth) 
        : base(name, age, weight)
    {
        WaterType = waterType;
        MaxDepth = maxDepth;
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} издает булькающие звуки: Буль-буль!");
    }
    
    public override void Move()
    {
        Console.WriteLine($"{Name} плавает в {WaterType} воде на глубине до {MaxDepth} м.");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип воды: {WaterType}, Макс. глубина: {MaxDepth} м");
    }
    
    public void Swim()
    {
        Console.WriteLine($"{Name} плавает.");
    }
}

// ========== GENERIC КЛАСС ANIMAL COLLECTION ==========
public class AnimalCollection<T> where T : Animal
{
    private List<T> _animals = new List<T>();
    private static int _totalAnimalsCreated = 0;
    
    public AnimalCollection()
    {
        Console.WriteLine($"Создана новая коллекция для типа {typeof(T).Name}");
    }
    
    // Добавление животного
    public void Add(T animal)
    {
        _animals.Add(animal);
        _totalAnimalsCreated++;
        Console.WriteLine($"Добавлено животное: {animal.Name} ({animal.GetType().Name})");
    }
    
    // Удаление животного
    public bool Remove(T animal)
    {
        bool removed = _animals.Remove(animal);
        if (removed)
        {
            Console.WriteLine($"Удалено животное: {animal.Name}");
            _totalAnimalsCreated--;
        }
        return removed;
    }
    
    // Получение количества животных
    public int Count => _animals.Count;
    
    // Статическое свойство для общего количества созданных животных
    public static int TotalAnimalsCreated => _totalAnimalsCreated;
    
    // Отображение всех животных
    public void DisplayAll()
    {
        Console.WriteLine($"\nВсего животных в коллекции: {Count}");
        foreach (var animal in _animals)
        {
            animal.DisplayInfo();
            animal.MakeSound();
            animal.Move();
            Console.WriteLine();
        }
    }
    
    // Поиск по имени
    public List<T> FindByName(string name)
    {
        return _animals.FindAll(a => a.Name.Equals(name, StringComparison.OrdinalIgnoreCase));
    }
    
    // Поиск по возрасту
    public List<T> FindByAge(int minAge, int maxAge)
    {
        return _animals.FindAll(a => a.Age >= minAge && a.Age <= maxAge);
    }
    
    // Получение животного по индексу
    public T GetAt(int index)
    {
        if (index >= 0 && index < _animals.Count)
            return _animals[index];
        throw new IndexOutOfRangeException("Индекс вне диапазона");
    }
    
    // Сортировка по имени
    public void SortByName()
    {
        _animals.Sort((a, b) => string.Compare(a.Name, b.Name, StringComparison.Ordinal));
        Console.WriteLine("Коллекция отсортирована по имени");
    }
    
    // Сортировка по возрасту
    public void SortByAge()
    {
        _animals.Sort((a, b) => a.Age.CompareTo(b.Age));
        Console.WriteLine("Коллекция отсортирована по возрасту");
    }
    
    // Применить действие ко всем животным
    public void ForEach(Action<T> action)
    {
        foreach (var animal in _animals)
        {
            action(animal);
        }
    }
}

// ========== GENERIC КЛАСС ДЛЯ ЗООПАРКА ==========
public class Zoo<T> where T : Animal
{
    private string _name;
    private AnimalCollection<T> _animals;
    
    public Zoo(string name)
    {
        _name = name;
        _animals = new AnimalCollection<T>();
        Console.WriteLine($"Создан зоопарк '{_name}' для животных типа {typeof(T).Name}");
    }
    
    public void AddAnimal(T animal)
    {
        _animals.Add(animal);
    }
    
    public void DisplayZooInfo()
    {
        Console.WriteLine($"\n=== Зоопарк '{_name}' ===");
        Console.WriteLine($"Тип животных: {typeof(T).Name}");
        _animals.DisplayAll();
    }
    
    public AnimalCollection<T> GetCollection() => _animals;
}

// ========== GENERIC КЛАСС ДЛЯ ПАРЫ ЖИВОТНЫХ ==========
public class AnimalPair<T1, T2> where T1 : Animal where T2 : Animal
{
    public T1 First { get; set; }
    public T2 Second { get; set; }
    public string Relationship { get; set; }
    
    public AnimalPair(T1 first, T2 second, string relationship)
    {
        First = first;
        Second = second;
        Relationship = relationship;
    }
    
    public void DisplayPair()
    {
        Console.WriteLine($"\nПара животных ({Relationship}):");
        Console.Write("Первое: ");
        First.DisplayInfo();
        Console.Write("Второе: ");
        Second.DisplayInfo();
    }
    
    public void MakeBothSounds()
    {
        Console.WriteLine($"\nЗвуки пары ({Relationship}):");
        First.MakeSound();
        Second.MakeSound();
    }
}

// ========== GENERIC КЛАСС ДЛЯ СТАТИСТИКИ ==========
public class AnimalStatistics<T> where T : Animal
{
    private List<T> _animals;
    
    public AnimalStatistics(List<T> animals)
    {
        _animals = animals;
    }
    
    public double GetAverageAge()
    {
        if (_animals.Count == 0) return 0;
        double sum = 0;
        foreach (var animal in _animals)
            sum += animal.Age;
        return sum / _animals.Count;
    }
    
    public double GetAverageWeight()
    {
        if (_animals.Count == 0) return 0;
        double sum = 0;
        foreach (var animal in _animals)
            sum += animal.Weight;
        return sum / _animals.Count;
    }
    
    public T GetOldest()
    {
        if (_animals.Count == 0) return null;
        T oldest = _animals[0];
        foreach (var animal in _animals)
        {
            if (animal.Age > oldest.Age)
                oldest = animal;
        }
        return oldest;
    }
    
    public T GetHeaviest()
    {
        if (_animals.Count == 0) return null;
        T heaviest = _animals[0];
        foreach (var animal in _animals)
        {
            if (animal.Weight > heaviest.Weight)
                heaviest = animal;
        }
        return heaviest;
    }
    
    public void DisplayStatistics()
    {
        Console.WriteLine($"\n=== Статистика для {typeof(T).Name} ===");
        Console.WriteLine($"Количество: {_animals.Count}");
        Console.WriteLine($"Средний возраст: {GetAverageAge():F1} лет");
        Console.WriteLine($"Средний вес: {GetAverageWeight():F1} кг");
        
        var oldest = GetOldest();
        if (oldest != null)
            Console.WriteLine($"Самое старое: {oldest.Name} ({oldest.Age} лет)");
        
        var heaviest = GetHeaviest();
        if (heaviest != null)
            Console.WriteLine($"Самое тяжелое: {heaviest.Name} ({heaviest.Weight} кг)");
    }
}

// ========== ДЕМОНСТРАЦИЯ ДЛЯ ANIMAL ==========
class AnimalGenericDemo
{
    public static void Run()
    {
        Console.WriteLine("========== GENERIC КЛАССЫ: ЖИВОТНЫЕ ==========\n");
        
        // 1. Создание коллекции для Dog
        Console.WriteLine("=== 1. AnimalCollection<Dog> ===");
        AnimalCollection<Dog> dogCollection = new AnimalCollection<Dog>();
        dogCollection.Add(new Dog("Рекс", 3, 25.5, "Овчарка", true));
        dogCollection.Add(new Dog("Шарик", 2, 18.0, "Дворняга", false));
        dogCollection.Add(new Dog("Лайка", 4, 22.0, "Хаски", true));
        dogCollection.DisplayAll();
        
        // 2. Создание коллекции для Cat
        Console.WriteLine("\n=== 2. AnimalCollection<Cat> ===");
        AnimalCollection<Cat> catCollection = new AnimalCollection<Cat>();
        catCollection.Add(new Cat("Муся", 2, 4.2, "Серый", true));
        catCollection.Add(new Cat("Васька", 3, 3.8, "Черный", false));
        catCollection.Add(new Cat("Снежок", 1, 3.5, "Белый", true));
        catCollection.DisplayAll();
        
        // 3. Создание коллекции для Bird
        Console.WriteLine("\n=== 3. AnimalCollection<Bird> ===");
        AnimalCollection<Bird> birdCollection = new AnimalCollection<Bird>();
        birdCollection.Add(new Bird("Кеша", 1, 0.35, 0.5, true));
        birdCollection.Add(new Bird("Пингвин", 5, 30.0, 0.8, false));
        birdCollection.Add(new Bird("Орел", 4, 6.0, 2.2, true));
        birdCollection.DisplayAll();
        
        // 4. Создание коллекции для Fish
        Console.WriteLine("\n=== 4. AnimalCollection<Fish> ===");
        AnimalCollection<Fish> fishCollection = new AnimalCollection<Fish>();
        fishCollection.Add(new Fish("Немо", 2, 0.15, "Соленая", 50));
        fishCollection.Add(new Fish("Золотая", 1, 0.1, "Пресная", 2));
        fishCollection.Add(new Fish("Акула", 8, 500.0, "Соленая", 1000));
        fishCollection.DisplayAll();
        
        // 5. Универсальная коллекция AnimalCollection<Animal>
        Console.WriteLine("\n=== 5. AnimalCollection<Animal> (универсальная) ===");
        AnimalCollection<Animal> allAnimals = new AnimalCollection<Animal>();
        allAnimals.Add(new Dog("Бобик", 3, 15.0, "Такса", false));
        allAnimals.Add(new Cat("Барсик", 2, 4.0, "Серый", true));
        allAnimals.Add(new Bird("Воробей", 1, 0.05, 0.2, true));
        allAnimals.Add(new Fish("Карп", 3, 2.5, "Пресная", 10));
        allAnimals.DisplayAll();
        
        // 6. Демонстрация методов Generic коллекции
        Console.WriteLine("\n=== 6. Дополнительные методы ===");
        Console.WriteLine($"Количество собак: {dogCollection.Count}");
        Console.WriteLine($"Общее количество созданных животных: {AnimalCollection<Animal>.TotalAnimalsCreated}");
        
        var foundDogs = dogCollection.FindByName("Рекс");
        Console.WriteLine($"\nНайдено собак по имени 'Рекс': {foundDogs.Count}");
        
        var youngCats = catCollection.FindByAge(1, 2);
        Console.WriteLine($"\nМолодые кошки (1-2 года): {youngCats.Count}");
        
        // 7. Сортировка
        Console.WriteLine("\n=== 7. Сортировка коллекции ===");
        dogCollection.SortByName();
        Console.WriteLine("После сортировки по имени:");
        dogCollection.DisplayAll();
        
        dogCollection.SortByAge();
        Console.WriteLine("После сортировки по возрасту:");
        dogCollection.DisplayAll();
        
        // 8. Использование ForEach
        Console.WriteLine("\n=== 8. Использование ForEach ===");
        Console.WriteLine("Все собаки выполняют команду 'Fetch':");
        dogCollection.ForEach(dog => dog.Fetch());
        
        // 9. Generic класс Zoo
        Console.WriteLine("\n=== 9. Generic класс Zoo<T> ===");
        Zoo<Dog> dogZoo = new Zoo<Dog>("Собачий рай");
        dogZoo.AddAnimal(new Dog("Тузик", 2, 20.0, "Бульдог", false));
        dogZoo.AddAnimal(new Dog("Жучка", 3, 18.0, "Дворняга", true));
        dogZoo.DisplayZooInfo();
        
        Zoo<Cat> catZoo = new Zoo<Cat>("Кошачье царство");
        catZoo.AddAnimal(new Cat("Мурзик", 4, 5.0, "Серый", true));
        catZoo.AddAnimal(new Cat("Рыжик", 1, 3.2, "Рыжий", true));
        catZoo.DisplayZooInfo();
        
        // 10. Generic класс AnimalPair
        Console.WriteLine("\n=== 10. Generic класс AnimalPair<T1, T2> ===");
        AnimalPair<Dog, Cat> bestFriends = new AnimalPair<Dog, Cat>(
            new Dog("Рекс", 3, 25.5, "Овчарка", true),
            new Cat("Муся", 2, 4.2, "Серый", true),
            "Лучшие друзья"
        );
        bestFriends.DisplayPair();
        bestFriends.MakeBothSounds();
        
        AnimalPair<Bird, Fish> skyAndSea = new AnimalPair<Bird, Fish>(
            new Bird("Орел", 4, 6.0, 2.2, true),
            new Fish("Немо", 2, 0.15, "Соленая", 50),
            "Небо и море"
        );
        skyAndSea.DisplayPair();
        skyAndSea.MakeBothSounds();
        
        // 11. Generic класс AnimalStatistics
        Console.WriteLine("\n=== 11. Generic класс AnimalStatistics<T> ===");
        
        List<Dog> dogsList = new List<Dog>
        {
            new Dog("Рекс", 3, 25.5, "Овчарка", true),
            new Dog("Шарик", 2, 18.0, "Дворняга", false),
            new Dog("Лайка", 4, 22.0, "Хаски", true)
        };
        AnimalStatistics<Dog> dogStats = new AnimalStatistics<Dog>(dogsList);
        dogStats.DisplayStatistics();
        
        List<Cat> catsList = new List<Cat>
        {
            new Cat("Муся", 2, 4.2, "Серый", true),
            new Cat("Васька", 3, 3.8, "Черный", false),
            new Cat("Снежок", 1, 3.5, "Белый", true)
        };
        AnimalStatistics<Cat> catStats = new AnimalStatistics<Cat>(catsList);
        catStats.DisplayStatistics();
        
        Console.WriteLine("\n========== ДЕМОНСТРАЦИЯ ANIMAL ЗАВЕРШЕНА ==========");
    }
}

// ========== ЗАПУСК ВСЕХ ДЕМОНСТРАЦИЙ ==========
Console.WriteLine("\n" + new string('=', 100) + "\n");
AnimalGenericDemo.Run();
````
