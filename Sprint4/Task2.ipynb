### Интерфейсы в C#

**Интерфейс** — это контракт, который определяет набор методов и свойств, которые класс должен реализовать. Интерфейсы позволяют создать гибкую и расширяемую архитектуру, поскольку они отделяют определение поведения от его реализации.

Давайте подробно разберем понятия интерфейсов и управления зависимостями в C# на примере классов `Vehicle`, `Car` и `Motorcycle`.

#### Пример интерфейса

Предположим, мы имеем интерфейс `IVehicle`, который определяет поведение транспортных средств. Ниже приведен пример его определения:

```csharp
public interface IVehicle
{
    void Start();            // Метод для запуска транспортного средства
    void Stop();             // Метод для остановки транспортного средства
    int GetNumberOfWheels(); // Метод для получения количества колес
}
```

Интерфейс `IVehicle` требует, чтобы любой класс, реализующий его, обеспечил реализацию этих методов.

### Реализация интерфейса

Теперь создадим классы `Car` и `Motorcycle`, которые будут реализовывать этот интерфейс:

```csharp
public class Car : IVehicle
{
    public void Start()
    {
        Console.WriteLine("Машина запущена.");
    }

    public void Stop()
    {
        Console.WriteLine("Машина остановлена.");
    }

    public int GetNumberOfWheels()
    {
        return 4; // У машины 4 колеса
    }
}

public class Motorcycle : IVehicle
{
    public void Start()
    {
        Console.WriteLine("Мотоцикл запущен.");
    }

    public void Stop()
    {
        Console.WriteLine("Мотоцикл остановлен.");
    }

    public int GetNumberOfWheels()
    {
        return 2; // У мотоцикла 2 колеса
    }
}
```

Каждый класс (`Car` и `Motorcycle`) реализует все методы, определенные в интерфейсе `IVehicle`, что позволяет создать единый интерфейс для работы с разными типами транспортных средств.

### Управление зависимостями (Dependency Injection)

**Управление зависимостями** — это паттерн проектирования, который помогает управлять созданием зависимостей в классе. Этот подход позволяет внедрять (или "вводить") зависимости через конструктор, методы или свойства, вместо того чтобы создавать их внутри класса. Это делает код более чистым и облегчает его тестирование.

#### Пример управления зависимостями

Создадим класс `VehicleService`, который будет зависеть от интерфейса `IVehicle`:

```csharp
public class VehicleService
{
    private readonly IVehicle _vehicle;

    // Внедрение зависимости через конструктор
    public VehicleService(IVehicle vehicle)
    {
        _vehicle = vehicle; 
    }

    public void UseVehicle()
    {
        _vehicle.Start(); // Запускаем транспортное средство
        Console.WriteLine($"У этого транспортного средства {_vehicle.GetNumberOfWheels()} колеса.");
        _vehicle.Stop(); // Останавливаем транспортное средство
    }
}
```

В этом примере класс `VehicleService` принимает в своем конструкторе объект, реализующий интерфейс `IVehicle`. Это позволяет `VehicleService` работать с любым объектом, который реализует `IVehicle`, обеспечивая гибкость и уменьшая связывание между классами.

### Использование классов

Теперь давайте создадим объекты `Car` и `Motorcycle` и используем наш сервис:

```csharp
IVehicle myCar = new Car(); // Создаем объект машины
VehicleService carService = new VehicleService(myCar); // Передаем машину в сервис
carService.UseVehicle(); // Используем сервис с машиной

IVehicle myMotorcycle = new Motorcycle(); // Создаем объект мотоцикла
VehicleService motorcycleService = new VehicleService(myMotorcycle); // Передаем мотоцикл в сервис
motorcycleService.UseVehicle(); // Используем сервис с мотоциклом

```

### Результат выполнения

При запуске данной программы будет следующий вывод:

```

У этого транспортного средства 4 колеса.
Машина остановлена.
Мотоцикл запущен.
У этого транспортного средства 2 колеса.
Мотоцикл остановлен.
```

### Заключение

Использование интерфейсов и управления зависимостями позволяет создать более организованный и тестируемый код. Интерфейсы определяют стандартные поведения, а управление зависимостями снижает связность между компонентами. Это делает систему более гибкой и упрощает процесс её поддержки и расширения в будущем.

````
using System;
using System.Collections.Generic;
using System.Linq;

// ========== ИНТЕРФЕЙСЫ ==========
public interface IVehicle
{
    void Start();
    void Stop();
    int GetCurrentSpeed();
    int GetNumberOfWheels();
    string GetInfo();
    void Accelerate(int amount);
    void Brake(int amount);
    void Honk(); // Добавлен метод Honk в интерфейс
}

public interface IElectricVehicle
{
    void Charge();
    int GetBatteryLevel();
    string GetChargingStatus();
}

public interface IRaceVehicle
{
    void ActivateSportMode();
    int GetMaxSpeed();
    double GetAccelerationTo100();
}

public interface IAutopilot
{
    void EnableAutopilot();
    void DisableAutopilot();
    bool IsAutopilotEnabled { get; }
}

// ========== БАЗОВЫЙ АБСТРАКТНЫЙ КЛАСС ==========
public abstract class Vehicle : IVehicle
{
    // Свойства
    public string Make { get; set; }
    public string Model { get; set; }
    public int Year { get; set; }
    public string Color { get; set; }
    protected int currentSpeed;
    protected bool isRunning;
    
    // Конструктор
    public Vehicle(string make, string model, int year, string color)
    {
        Make = make;
        Model = model;
        Year = year;
        Color = color;
        currentSpeed = 0;
        isRunning = false;
    }
    
    // Реализация IVehicle
    public virtual void Start()
    {
        isRunning = true;
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} ({Year}) запущен. Цвет: {Color}");
    }
    
    public virtual void Stop()
    {
        currentSpeed = 0;
        isRunning = false;
        Console.WriteLine($"{Make} {Model} остановлен.");
    }
    
    public int GetCurrentSpeed()
    {
        return currentSpeed;
    }
    
    public abstract int GetNumberOfWheels();
    
    public virtual string GetInfo()
    {
        return $"{Make} {Model} ({Year}), цвет: {Color}";
    }
    
    public virtual void Accelerate(int amount)
    {
        if (isRunning)
        {
            currentSpeed += amount;
            Console.WriteLine($"{Make} {Model} разгоняется до {currentSpeed} км/ч.");
        }
        else
        {
            Console.WriteLine($"Сначала нужно запустить двигатель {Make} {Model}!");
        }
    }
    
    public virtual void Brake(int amount)
    {
        if (currentSpeed > 0)
        {
            currentSpeed = Math.Max(0, currentSpeed - amount);
            Console.WriteLine($"{Make} {Model} тормозит до {currentSpeed} км/ч.");
        }
    }
    
    public abstract void Honk(); // Абстрактный метод Honk
    
    public virtual void ShowStatus()
    {
        string status = isRunning ? "запущен" : "остановлен";
        Console.WriteLine($"{Make} {Model}: {status}, скорость: {currentSpeed} км/ч");
    }
}

// ========== КЛАСС CAR ==========
public class Car : Vehicle, IRaceVehicle
{
    public int Doors { get; set; }
    public string BodyType { get; set; }
    public int Horsepower { get; set; }
    private int maxSpeed;
    private double accelerationTo100;
    
    public Car(string make, string model, int year, string color, int doors, string bodyType, int horsepower, int maxSpeed, double accelerationTo100) 
        : base(make, model, year, color)
    {
        Doors = doors;
        BodyType = bodyType;
        Horsepower = horsepower;
        this.maxSpeed = maxSpeed;
        this.accelerationTo100 = accelerationTo100;
    }
    
    public override int GetNumberOfWheels()
    {
        return 4;
    }
    
    public override void Honk()
    {
        Console.WriteLine($"{Make} {Model} сигналит: Бип-бип!");
    }
    
    public override void Start()
    {
        base.Start();
        Console.WriteLine($"  Мощность: {Horsepower} л.с., Дверей: {Doors}, Кузов: {BodyType}");
    }
    
    // Реализация IRaceVehicle
    public void ActivateSportMode()
    {
        Console.WriteLine($"Спортивный режим активирован! {Make} {Model}");
        Console.WriteLine($"  Разгон 0-100 км/ч: {accelerationTo100} сек");
        Console.WriteLine($"  Максимальная скорость: {maxSpeed} км/ч");
    }
    
    public int GetMaxSpeed()
    {
        return maxSpeed;
    }
    
    public double GetAccelerationTo100()
    {
        return accelerationTo100;
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", {Doors} двери, {BodyType}, {Horsepower} л.с., макс. скорость: {maxSpeed} км/ч";
    }
}

// ========== КЛАСС MOTORCYCLE ==========
public class Motorcycle : Vehicle, IRaceVehicle
{
    public string EngineType { get; set; }
    public bool HasSidecar { get; set; }
    public int Horsepower { get; set; }
    private int maxSpeed;
    private double accelerationTo100;
    
    public Motorcycle(string make, string model, int year, string color, string engineType, bool hasSidecar, int horsepower, int maxSpeed, double accelerationTo100) 
        : base(make, model, year, color)
    {
        EngineType = engineType;
        HasSidecar = hasSidecar;
        Horsepower = horsepower;
        this.maxSpeed = maxSpeed;
        this.accelerationTo100 = accelerationTo100;
    }
    
    public override int GetNumberOfWheels()
    {
        return 2;
    }
    
    public override void Honk()
    {
        Console.WriteLine($"{Make} {Model} сигналит: Бип!");
    }
    
    public override void Start()
    {
        base.Start();
        Console.WriteLine($"  Двигатель: {EngineType}, Мощность: {Horsepower} л.с., Коляска: {(HasSidecar ? "Есть" : "Нет")}");
    }
    
    public void Wheelie()
    {
        if (isRunning && currentSpeed > 20)
        {
            Console.WriteLine($"Мотоцикл {Make} {Model} едет на заднем колесе!");
        }
        else
        {
            Console.WriteLine($"Невозможно сделать вили на мотоцикле {Make} {Model} (нужна скорость > 20 км/ч)");
        }
    }
    
    // Реализация IRaceVehicle
    public void ActivateSportMode()
    {
        Console.WriteLine($"Спортивный режим активирован! {Make} {Model}");
        Console.WriteLine($"  Разгон 0-100 км/ч: {accelerationTo100} сек");
        Console.WriteLine($"  Максимальная скорость: {maxSpeed} км/ч");
    }
    
    public int GetMaxSpeed()
    {
        return maxSpeed;
    }
    
    public double GetAccelerationTo100()
    {
        return accelerationTo100;
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", {EngineType}, {Horsepower} л.с., макс. скорость: {maxSpeed} км/ч, коляска: {(HasSidecar ? "Да" : "Нет")}";
    }
}

// ========== КЛАСС ELECTRIC_CAR ==========
public class ElectricCar : Car, IElectricVehicle, IAutopilot
{
    private int batteryLevel;
    private bool isAutopilotEnabled;
    
    public int BatteryCapacity { get; set; }
    
    public ElectricCar(string make, string model, int year, string color, int doors, string bodyType, int horsepower, int maxSpeed, double accelerationTo100, int batteryCapacity) 
        : base(make, model, year, color, doors, bodyType, horsepower, maxSpeed, accelerationTo100)
    {
        BatteryCapacity = batteryCapacity;
        batteryLevel = 100;
        isAutopilotEnabled = false;
    }
    
    // Реализация IElectricVehicle
    public void Charge()
    {
        batteryLevel = 100;
        Console.WriteLine($"Электромобиль {Make} {Model} заряжен до {batteryLevel}%");
    }
    
    public int GetBatteryLevel()
    {
        return batteryLevel;
    }
    
    public string GetChargingStatus()
    {
        return batteryLevel >= 80 ? "Готов к поездке" : "Требуется зарядка";
    }
    
    // Реализация IAutopilot
    public bool IsAutopilotEnabled => isAutopilotEnabled;
    
    public void EnableAutopilot()
    {
        if (batteryLevel >= 20)
        {
            isAutopilotEnabled = true;
            Console.WriteLine($"Автопилот {Make} {Model} активирован! Машина движется сама.");
        }
        else
        {
            Console.WriteLine($"Невозможно активировать автопилот: низкий заряд батареи ({batteryLevel}%)");
        }
    }
    
    public void DisableAutopilot()
    {
        isAutopilotEnabled = false;
        Console.WriteLine($"Автопилот {Make} {Model} деактивирован.");
    }
    
    public override void Start()
    {
        isRunning = true;
        currentSpeed = 0;
        Console.WriteLine($"Электромобиль {Make} {Model} запущен (бесшумно). Заряд: {batteryLevel}%");
        if (batteryLevel < 20)
        {
            Console.WriteLine($"  ВНИМАНИЕ: Низкий заряд батареи! {batteryLevel}%");
        }
    }
    
    public override void Accelerate(int amount)
    {
        if (isRunning && batteryLevel > 0)
        {
            base.Accelerate(amount);
            batteryLevel = Math.Max(0, batteryLevel - amount / 10);
            Console.WriteLine($"  Заряд батареи: {batteryLevel}%");
        }
        else if (batteryLevel <= 0)
        {
            Console.WriteLine($"Электромобиль {Make} {Model} разряжен! Требуется зарядка.");
        }
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", батарея: {BatteryCapacity} кВт/ч, заряд: {batteryLevel}%, автопилот: {(isAutopilotEnabled ? "вкл" : "выкл")}";
    }
}

// ========== КЛАСС TRUCK ==========
public class Truck : Vehicle
{
    public double LoadCapacity { get; set; }
    public int Axles { get; set; }
    
    public Truck(string make, string model, int year, string color, double loadCapacity, int axles) 
        : base(make, model, year, color)
    {
        LoadCapacity = loadCapacity;
        Axles = axles;
    }
    
    public override int GetNumberOfWheels()
    {
        return Axles * 2;
    }
    
    public override void Honk()
    {
        Console.WriteLine($"{Make} {Model} гудит: БИИИП-БИИИП!");
    }
    
    public override void Start()
    {
        base.Start();
        Console.WriteLine($"  Грузоподъемность: {LoadCapacity} т, Оси: {Axles}, Колес: {GetNumberOfWheels()}");
    }
    
    public void LoadCargo(double weight)
    {
        if (weight <= LoadCapacity)
        {
            Console.WriteLine($"Грузовик {Make} {Model} загружен на {weight} тонн.");
        }
        else
        {
            Console.WriteLine($"Грузовик {Make} {Model} перегружен! Максимум {LoadCapacity} тонн.");
        }
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", грузоподъемность: {LoadCapacity} т, оси: {Axles}";
    }
}

// ========== СЕРВИСНЫЕ КЛАССЫ ДЛЯ УПРАВЛЕНИЯ ЗАВИСИМОСТЯМИ ==========

// Сервис для работы с транспортными средствами (Dependency Injection)
public class VehicleService
{
    private readonly IVehicle _vehicle;
    
    // Внедрение зависимости через конструктор
    public VehicleService(IVehicle vehicle)
    {
        _vehicle = vehicle;
    }
    
    public void DemonstrateVehicle()
    {
        Console.WriteLine($"\n--- Демонстрация транспортного средства: {_vehicle.GetInfo()} ---");
        Console.WriteLine($"Количество колес: {_vehicle.GetNumberOfWheels()}");
        _vehicle.Start();
        _vehicle.Accelerate(50);
        Console.WriteLine($"Текущая скорость: {_vehicle.GetCurrentSpeed()} км/ч");
        _vehicle.Brake(20);
        _vehicle.Honk(); // Теперь Honk доступен через интерфейс
        _vehicle.Stop();
    }
    
    public void StartVehicle()
    {
        _vehicle.Start();
    }
    
    public void StopVehicle()
    {
        _vehicle.Stop();
    }
    
    public void AccelerateVehicle(int amount)
    {
        _vehicle.Accelerate(amount);
    }
}

// Сервис для работы с гоночными транспортными средствами
public class RaceService
{
    private readonly IRaceVehicle _raceVehicle;
    private readonly string _vehicleName;
    
    public RaceService(IRaceVehicle raceVehicle, string vehicleName)
    {
        _raceVehicle = raceVehicle;
        _vehicleName = vehicleName;
    }
    
    public void ShowRaceCapabilities()
    {
        Console.WriteLine($"\n--- Гоночные характеристики {_vehicleName} ---");
        _raceVehicle.ActivateSportMode();
        Console.WriteLine($"Максимальная скорость: {_raceVehicle.GetMaxSpeed()} км/ч");
        Console.WriteLine($"Разгон 0-100 км/ч: {_raceVehicle.GetAccelerationTo100()} сек");
    }
    
    public void StartRace()
    {
        _raceVehicle.ActivateSportMode();
    }
}

// Сервис для работы с электромобилями
public class ElectricVehicleService
{
    private readonly IElectricVehicle _electricVehicle;
    private readonly string _vehicleName;
    
    public ElectricVehicleService(IElectricVehicle electricVehicle, string vehicleName)
    {
        _electricVehicle = electricVehicle;
        _vehicleName = vehicleName;
    }
    
    public void ShowBatteryStatus()
    {
        Console.WriteLine($"\n--- Состояние батареи {_vehicleName} ---");
        Console.WriteLine($"Уровень заряда: {_electricVehicle.GetBatteryLevel()}%");
        Console.WriteLine($"Статус: {_electricVehicle.GetChargingStatus()}");
    }
    
    public void ChargeVehicle()
    {
        Console.WriteLine($"\n--- Зарядка {_vehicleName} ---");
        _electricVehicle.Charge();
    }
}

// Сервис для работы с автопилотом
public class AutopilotService
{
    private readonly IAutopilot _autopilot;
    private readonly string _vehicleName;
    
    public AutopilotService(IAutopilot autopilot, string vehicleName)
    {
        _autopilot = autopilot;
        _vehicleName = vehicleName;
    }
    
    public void TestAutopilot()
    {
        Console.WriteLine($"\n--- Тестирование автопилота {_vehicleName} ---");
        _autopilot.EnableAutopilot();
        Console.WriteLine($"Автопилот включен: {_autopilot.IsAutopilotEnabled}");
        _autopilot.DisableAutopilot();
        Console.WriteLine($"Автопилот включен: {_autopilot.IsAutopilotEnabled}");
    }
}

// Автосалон с управлением зависимостями
public class Dealership
{
    private readonly List<IVehicle> _vehicles;
    private readonly string _name;
    
    public Dealership(string name, List<IVehicle> vehicles)
    {
        _name = name;
        _vehicles = vehicles;
    }
    
    public void ShowAllVehicles()
    {
        Console.WriteLine($"\n=== Автосалон '{_name}' ===");
        Console.WriteLine($"Всего транспортных средств: {_vehicles.Count}");
        
        foreach (var vehicle in _vehicles)
        {
            Console.WriteLine($"\n{vehicle.GetInfo()}");
            Console.WriteLine($"Колес: {vehicle.GetNumberOfWheels()}");
            vehicle.Start();
            vehicle.Honk(); // Теперь Honk доступен через интерфейс
            vehicle.Stop();
        }
    }
    
    public void TestDriveAll()
    {
        Console.WriteLine($"\n=== Тест-драйв в автосалоне '{_name}' ===");
        foreach (var vehicle in _vehicles)
        {
            Console.WriteLine($"\n--- {vehicle.GetInfo()} ---");
            vehicle.Start();
            vehicle.Accelerate(60);
            vehicle.Brake(30);
            vehicle.Stop();
        }
    }
}

// ========== ДЕМОНСТРАЦИЯ УПРАВЛЕНИЯ ЗАВИСИМОСТЯМИ ==========
class VehicleDemo
{
    public static void Run()
    {
        Console.WriteLine("========== УПРАВЛЕНИЕ ЗАВИСИМОСТЯМИ (Dependency Injection): ТРАНСПОРТНЫЕ СРЕДСТВА ==========\n");
        
        // Создание объектов
        Car car = new Car("Toyota", "Camry", 2022, "Красный", 4, "Седан", 203, 220, 7.5);
        Motorcycle motorcycle = new Motorcycle("Harley-Davidson", "Street Glide", 2023, "Черный", "V-Twin", false, 95, 180, 4.2);
        ElectricCar electricCar = new ElectricCar("Tesla", "Model S", 2024, "Белый", 4, "Лифтбек", 1020, 320, 2.1, 100);
        Truck truck = new Truck("Volvo", "FH16", 2023, "Синий", 25.5, 3);
        
        // ========== 1. ДЕМОНСТРАЦИЯ ВНЕДРЕНИЯ ЗАВИСИМОСТЕЙ (DI) ==========
        Console.WriteLine("=== 1. ВНЕДРЕНИЕ ЗАВИСИМОСТЕЙ ЧЕРЕЗ КОНСТРУКТОР ===\n");
        
        // Внедрение зависимости IVehicle в VehicleService
        VehicleService carService = new VehicleService(car);
        VehicleService motorcycleService = new VehicleService(motorcycle);
        VehicleService electricCarService = new VehicleService(electricCar);
        
        Console.WriteLine("--- Автомобиль через VehicleService ---");
        carService.DemonstrateVehicle();
        
        Console.WriteLine("\n--- Мотоцикл через VehicleService ---");
        motorcycleService.DemonstrateVehicle();
        
        Console.WriteLine("\n--- Электромобиль через VehicleService ---");
        electricCarService.DemonstrateVehicle();
        
        // ========== 2. РАБОТА С ГОНОЧНЫМИ ТРАНСПОРТНЫМИ СРЕДСТВАМИ ==========
        Console.WriteLine("\n=== 2. СЕРВИС ДЛЯ ГОНОЧНЫХ ТРАНСПОРТНЫХ СРЕДСТВ (IRaceVehicle) ===\n");
        
        RaceService carRaceService = new RaceService(car, "Toyota Camry");
        RaceService bikeRaceService = new RaceService(motorcycle, "Harley-Davidson Street Glide");
        RaceService evRaceService = new RaceService(electricCar, "Tesla Model S");
        
        carRaceService.ShowRaceCapabilities();
        bikeRaceService.ShowRaceCapabilities();
        evRaceService.ShowRaceCapabilities();
        
        // ========== 3. РАБОТА С ЭЛЕКТРОМОБИЛЯМИ ==========
        Console.WriteLine("\n=== 3. СЕРВИС ДЛЯ ЭЛЕКТРОМОБИЛЕЙ (IElectricVehicle) ===\n");
        
        ElectricVehicleService evService = new ElectricVehicleService(electricCar, "Tesla Model S");
        evService.ShowBatteryStatus();
        evService.ChargeVehicle();
        evService.ShowBatteryStatus();
        
        // ========== 4. РАБОТА С АВТОПИЛОТОМ ==========
        Console.WriteLine("\n=== 4. СЕРВИС ДЛЯ АВТОПИЛОТА (IAutopilot) ===\n");
        
        AutopilotService autopilotService = new AutopilotService(electricCar, "Tesla Model S");
        autopilotService.TestAutopilot();
        
        // ========== 5. АВТОСАЛОН С ВНЕДРЕНИЕМ ЗАВИСИМОСТЕЙ ==========
        Console.WriteLine("\n=== 5. АВТОСАЛОН (коллекция IVehicle) ===\n");
        
        List<IVehicle> vehicles = new List<IVehicle>
        {
            new Car("BMW", "X5", 2023, "Черный", 5, "Внедорожник", 340, 250, 5.5),
            new Motorcycle("Yamaha", "R1", 2024, "Синий", "Inline-4", false, 200, 299, 3.0),
            new ElectricCar("Rivian", "R1T", 2023, "Зеленый", 4, "Пикап", 835, 200, 3.5, 135),
            new Truck("MAN", "TGX", 2022, "Белый", 30, 4),
            car,
            motorcycle,
            electricCar,
            truck
        };
        
        Dealership dealership = new Dealership("Автомир", vehicles);
        dealership.ShowAllVehicles();
        dealership.TestDriveAll();
        
        // ========== 6. ДОПОЛНИТЕЛЬНАЯ ДЕМОНСТРАЦИЯ ==========
        Console.WriteLine("\n=== 6. ДОПОЛНИТЕЛЬНАЯ ДЕМОНСТРАЦИЯ ===\n");
        
        // Демонстрация специфичных методов
        Console.WriteLine("Специфичные методы:");
        motorcycle.Wheelie();
        truck.LoadCargo(20);
        
        // Демонстрация статуса
        Console.WriteLine("\n--- Статус транспортных средств ---");
        car.ShowStatus();
        electricCar.ShowStatus();
        truck.ShowStatus();
        
        // Демонстрация работы с электромобилем после разрядки
        Console.WriteLine("\n--- Демонстрация разрядки электромобиля ---");
        for (int i = 0; i < 5; i++)
        {
            electricCar.Accelerate(30);
        }
        
        Console.WriteLine("\n========== ДЕМОНСТРАЦИЯ УПРАВЛЕНИЯ ЗАВИСИМОСТЯМИ ЗАВЕРШЕНА ==========");
    }
}

// ========== ЗАПУСК ДЕМОНСТРАЦИИ ==========
VehicleDemo.Run();
````

````
using System;
using System.Collections.Generic;
using System.Linq;

// ========== ИНТЕРФЕЙСЫ ==========
public interface IAnimal
{
    void MakeSound();
    void Move();
    string GetInfo();
    void Eat();
    void Sleep();
    int GetEnergyLevel();
}

public interface IPet
{
    string OwnerName { get; }
    void Play();
    void Feed();
    void ShowAffection();
    void Walk();
}

public interface IWild
{
    string Habitat { get; }
    void Hunt();
    void DefendTerritory();
    void Hide();
}

public interface ITrainable
{
    bool IsTrained { get; }
    void Train();
    void PerformTrick(string trick);
    string[] GetKnownTricks();
}

// ========== БАЗОВЫЙ АБСТРАКТНЫЙ КЛАСС ==========
public abstract class Animal : IAnimal
{
    // Свойства
    public string Name { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }
    protected int energyLevel;
    protected bool isAlive;
    
    // Конструктор
    public Animal(string name, int age, double weight)
    {
        Name = name;
        Age = age;
        Weight = weight;
        energyLevel = 100;
        isAlive = true;
    }
    
    // Реализация IAnimal
    public abstract void MakeSound();
    public abstract void Move();
    public abstract void Eat();
    public abstract void Sleep();
    
    public virtual string GetInfo()
    {
        return $"{Name}, {Age} лет, {Weight} кг, энергия: {energyLevel}%";
    }
    
    public int GetEnergyLevel()
    {
        return energyLevel;
    }
    
    public virtual void ShowStatus()
    {
        string status = isAlive ? "жив" : "мертв";
        Console.WriteLine($"{Name}: {status}, энергия: {energyLevel}%");
    }
    
    public abstract void Breathe();
}

// ========== КЛАСС DOG ==========
public class Dog : Animal, IPet, ITrainable
{
    public string Breed { get; set; }
    private string ownerName;
    private bool isTrained;
    private List<string> tricks = new List<string>();
    
    public Dog(string name, int age, double weight, string breed, string ownerName) 
        : base(name, age, weight)
    {
        Breed = breed;
        this.ownerName = ownerName;
        isTrained = false;
    }
    
    // Реализация IPet
    public string OwnerName => ownerName;
    
    public void Play()
    {
        if (energyLevel >= 15)
        {
            energyLevel -= 15;
            Console.WriteLine($"{Name} играет с мячом! Энергия: {energyLevel}%");
        }
        else
        {
            Console.WriteLine($"{Name} слишком устал для игр!");
        }
    }
    
    public void Feed()
    {
        Eat();
    }
    
    public void ShowAffection()
    {
        Console.WriteLine($"{Name} виляет хвостом и лижет руку хозяину {ownerName}!");
    }
    
    public void Walk()
    {
        Console.WriteLine($"{Name} гуляет на поводке с {ownerName}.");
        Move();
    }
    
    // Реализация ITrainable
    public bool IsTrained => isTrained;
    
    public void Train()
    {
        if (!isTrained)
        {
            isTrained = true;
            Console.WriteLine($"{Name} прошел обучение! Теперь он знает команды.");
        }
        else
        {
            Console.WriteLine($"{Name} уже обучен.");
        }
    }
    
    public void PerformTrick(string trick)
    {
        if (isTrained)
        {
            if (!tricks.Contains(trick))
                tricks.Add(trick);
            Console.WriteLine($"{Name} выполняет трюк: {trick}!");
        }
        else
        {
            Console.WriteLine($"{Name} еще не обучен. Нужно сначала обучить собаку!");
        }
    }
    
    public string[] GetKnownTricks()
    {
        return tricks.ToArray();
    }
    
    // Реализация IAnimal
    public override void MakeSound()
    {
        Console.WriteLine($"Собака {Name} лает: Гав-гав!");
    }
    
    public override void Move()
    {
        if (energyLevel >= 10)
        {
            energyLevel -= 10;
            Console.WriteLine($"Собака {Name} бегает на четырех лапах. Энергия: {energyLevel}%");
        }
        else
        {
            Console.WriteLine($"Собака {Name} слишком устала, чтобы двигаться!");
        }
    }
    
    public override void Eat()
    {
        energyLevel = Math.Min(100, energyLevel + 20);
        Console.WriteLine($"Собака {Name} ест корм. Энергия: {energyLevel}%");
    }
    
    public override void Sleep()
    {
        energyLevel = 100;
        Console.WriteLine($"Собака {Name} выспалась. Энергия: {energyLevel}%");
    }
    
    public override void Breathe()
    {
        Console.WriteLine($"Собака {Name} дышит, высунув язык.");
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", порода: {Breed}, хозяин: {ownerName}";
    }
}

// ========== КЛАСС CAT ==========
public class Cat : Animal, IPet
{
    public string Color { get; set; }
    public bool IsIndoor { get; set; }
    private string ownerName;
    
    public Cat(string name, int age, double weight, string color, bool isIndoor, string ownerName) 
        : base(name, age, weight)
    {
        Color = color;      // цвет "серый"
        IsIndoor = isIndoor;
        this.ownerName = ownerName;
    }
    
    // Реализация IPet
    public string OwnerName => ownerName;
    
    public void Play()
    {
        if (energyLevel >= 10)
        {
            energyLevel -= 10;
            Console.WriteLine($"{Name} играет с клубком ниток! Энергия: {energyLevel}%");
        }
        else
        {
            Console.WriteLine($"{Name} слишком устала для игр!");
        }
    }
    
    public void Feed()
    {
        Eat();
    }
    
    public void ShowAffection()
    {
        Console.WriteLine($"{Name} трется о ноги хозяина {ownerName} и мурлычет: Мур-мур-мур...");
    }
    
    public void Walk()
    {
        if (IsIndoor)
            Console.WriteLine($"{Name} гуляет по квартире с {ownerName}.");
        else
            Console.WriteLine($"{Name} гуляет на улице, исследуя окрестности.");
        Move();
    }
    
    // Реализация IAnimal
    public override void MakeSound()
    {
        Console.WriteLine($"Кошка {Name} мяукает: Мяу-мяу!");
    }
    
    public override void Move()
    {
        if (energyLevel >= 5)
        {
            energyLevel -= 5;
            Console.WriteLine($"Кошка {Name} грациозно крадется. Энергия: {energyLevel}%");
        }
        else
        {
            Console.WriteLine($"Кошка {Name} слишком устала, чтобы двигаться!");
        }
    }
    
    public override void Eat()
    {
        energyLevel = Math.Min(100, energyLevel + 20);
        Console.WriteLine($"Кошка {Name} ест рыбу. Энергия: {energyLevel}%");
    }
    
    public override void Sleep()
    {
        energyLevel = 100;
        Console.WriteLine($"Кошка {Name} выспалась. Энергия: {energyLevel}%");
    }
    
    public override void Breathe()
    {
        Console.WriteLine($"Кошка {Name} тихо дышит, свернувшись клубочком.");
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", окрас: {Color}, хозяин: {ownerName}";
    }
    
    // Специфичный метод
    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет от удовольствия!");
    }
}

// ========== КЛАСС BIRD ==========
public class Bird : Animal, IWild
{
    public double WingSpan { get; set; }
    public bool CanFly { get; set; }
    private string habitat;
    
    public Bird(string name, int age, double weight, double wingSpan, bool canFly, string habitat) 
        : base(name, age, weight)
    {
        WingSpan = wingSpan;
        CanFly = canFly;
        this.habitat = habitat;
    }
    
    // Реализация IWild
    public string Habitat => habitat;
    
    public void Hunt()
    {
        if (energyLevel >= 20)
        {
            energyLevel -= 20;
            Console.WriteLine($"Птица {Name} охотится в {habitat}!");
        }
        else
        {
            Console.WriteLine($"{Name} слишком устала для охоты!");
        }
    }
    
    public void DefendTerritory()
    {
        Console.WriteLine($"Птица {Name} защищает свою территорию в {habitat}!");
    }
    
    public void Hide()
    {
        Console.WriteLine($"Птица {Name} прячется в {habitat} от хищников.");
    }
    
    // Реализация IAnimal
    public override void MakeSound()
    {
        Console.WriteLine($"Птица {Name} чирикает: Чик-чирик!");
    }
    
    public override void Move()
    {
        if (CanFly && energyLevel >= 15)
        {
            energyLevel -= 15;
            Console.WriteLine($"Птица {Name} летит с размахом крыльев {WingSpan} м. Энергия: {energyLevel}%");
        }
        else if (!CanFly && energyLevel >= 5)
        {
            energyLevel -= 5;
            Console.WriteLine($"Птица {Name} ходит по земле. Энергия: {energyLevel}%");
        }
        else
        {
            Console.WriteLine($"Птица {Name} слишком устала, чтобы двигаться!");
        }
    }
    
    public override void Eat()
    {
        energyLevel = Math.Min(100, energyLevel + 15);
        Console.WriteLine($"Птица {Name} клюет зерна. Энергия: {energyLevel}%");
    }
    
    public override void Sleep()
    {
        energyLevel = 100;
        Console.WriteLine($"Птица {Name} выспалась. Энергия: {energyLevel}%");
    }
    
    public override void Breathe()
    {
        Console.WriteLine($"Птица {Name} дышит легко и быстро.");
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", размах крыльев: {WingSpan} м, среда обитания: {habitat}";
    }
}

// ========== КЛАСС FISH ==========
public class Fish : Animal, IWild
{
    public string WaterType { get; set; }
    public double MaxDepth { get; set; }
    private string habitat;
    
    public Fish(string name, int age, double weight, string waterType, double maxDepth, string habitat) 
        : base(name, age, weight)
    {
        WaterType = waterType;
        MaxDepth = maxDepth;
        this.habitat = habitat;
    }
    
    // Реализация IWild
    public string Habitat => habitat;
    
    public void Hunt()
    {
        if (energyLevel >= 15)
        {
            energyLevel -= 15;
            Console.WriteLine($"Рыба {Name} охотится на мелкую рыбу в {habitat}!");
        }
        else
        {
            Console.WriteLine($"{Name} слишком устала для охоты!");
        }
    }
    
    public void DefendTerritory()
    {
        Console.WriteLine($"Рыба {Name} защищает свою территорию в {habitat} на глубине {MaxDepth} м!");
    }
    
    public void Hide()
    {
        Console.WriteLine($"Рыба {Name} прячется среди кораллов в {habitat}.");
    }
    
    // Реализация IAnimal
    public override void MakeSound()
    {
        Console.WriteLine($"Рыба {Name} издает булькающие звуки: Буль-буль!");
    }
    
    public override void Move()
    {
        if (energyLevel >= 5)
        {
            energyLevel -= 5;
            Console.WriteLine($"Рыба {Name} плавает в {WaterType} воде. Энергия: {energyLevel}%");
        }
        else
        {
            Console.WriteLine($"Рыба {Name} слишком устала, чтобы плавать!");
        }
    }
    
    public override void Eat()
    {
        energyLevel = Math.Min(100, energyLevel + 15);
        Console.WriteLine($"Рыба {Name} проглатывает планктон. Энергия: {energyLevel}%");
    }
    
    public override void Sleep()
    {
        energyLevel = 100;
        Console.WriteLine($"Рыба {Name} выспалась. Энергия: {energyLevel}%");
    }
    
    public override void Breathe()
    {
        Console.WriteLine($"Рыба {Name} дышит жабрами под водой.");
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", вода: {WaterType}, макс. глубина: {MaxDepth} м, среда: {habitat}";
    }
}

// ========== СЕРВИСНЫЕ КЛАССЫ ДЛЯ УПРАВЛЕНИЯ ЗАВИСИМОСТЯМИ ==========

// Сервис для работы с животными (Dependency Injection)
public class AnimalService
{
    private readonly IAnimal _animal;
    
    // Внедрение зависимости через конструктор
    public AnimalService(IAnimal animal)
    {
        _animal = animal;
    }
    
    public void DemonstrateAnimal()
    {
        Console.WriteLine($"\n--- Демонстрация животного: {_animal.GetInfo()} ---");
        _animal.MakeSound();
        _animal.Move();
        _animal.Eat();
        _animal.Sleep();
        Console.WriteLine($"Энергия после всех действий: {_animal.GetEnergyLevel()}%");
    }
    
    public void MakeAnimalSound()
    {
        _animal.MakeSound();
    }
    
    public void FeedAnimal()
    {
        _animal.Eat();
    }
}

// Сервис для работы с домашними животными
public class PetService
{
    private readonly IPet _pet;
    
    public PetService(IPet pet)
    {
        _pet = pet;
    }
    
    public void InteractWithPet()
    {
        Console.WriteLine($"\n--- Взаимодействие с домашним питомцем {_pet.OwnerName} ---");
        _pet.Play();
        _pet.Feed();
        _pet.ShowAffection();
        _pet.Walk();
    }
    
    public void PlayWithPet()
    {
        _pet.Play();
    }
    
    public void FeedPet()
    {
        _pet.Feed();
    }
}

// Сервис для работы с дикими животными
public class WildAnimalService
{
    private readonly IWild _wildAnimal;
    
    public WildAnimalService(IWild wildAnimal)
    {
        _wildAnimal = wildAnimal;
    }
    
    public void ObserveWildAnimal()
    {
        Console.WriteLine($"\n--- Наблюдение за диким животным в среде {_wildAnimal.Habitat} ---");
        _wildAnimal.Hunt();
        _wildAnimal.DefendTerritory();
        _wildAnimal.Hide();
    }
}

// Сервис для работы с обучаемыми животными
public class TrainingService
{
    private readonly ITrainable _trainableAnimal;
    
    public TrainingService(ITrainable trainableAnimal)
    {
        _trainableAnimal = trainableAnimal;
    }
    
    public void TrainAnimal()
    {
        Console.WriteLine("\n--- Процесс обучения ---");
        _trainableAnimal.Train();
        
        if (_trainableAnimal.IsTrained)
        {
            _trainableAnimal.PerformTrick("Сидеть");
            _trainableAnimal.PerformTrick("Лежать");
            _trainableAnimal.PerformTrick("Дай лапу");
            Console.WriteLine($"Известные трюки: {string.Join(", ", _trainableAnimal.GetKnownTricks())}");
        }
    }
}

// Зоопарк с управлением зависимостями
public class Zoo
{
    private readonly List<IAnimal> _animals;
    private readonly string _name;
    
    public Zoo(string name, List<IAnimal> animals)
    {
        _name = name;
        _animals = animals;
    }
    
    public void ShowAllAnimals()
    {
        Console.WriteLine($"\n=== Зоопарк '{_name}' ===");
        Console.WriteLine($"Всего животных: {_animals.Count}");
        
        foreach (var animal in _animals)
        {
            Console.WriteLine($"\n{animal.GetInfo()}");
            animal.MakeSound();
            animal.Move();
        }
    }
    
    public void FeedAllAnimals()
    {
        Console.WriteLine($"\n=== Кормление всех животных в зоопарке '{_name}' ===");
        foreach (var animal in _animals)
        {
            animal.Eat();
        }
    }
}

// ========== ДЕМОНСТРАЦИЯ УПРАВЛЕНИЯ ЗАВИСИМОСТЯМИ ==========
class AnimalDemo
{
    public static void Run()
    {
        Console.WriteLine("========== УПРАВЛЕНИЕ ЗАВИСИМОСТЯМИ (Dependency Injection) ==========\n");
        
        // Создание объектов с указанными параметрами
        Dog dog = new Dog("Рекс", 3, 25.5, "Овчарка", "Иван");
        Cat cat = new Cat("Муся", 2, 4.2, "серый", true, "Элина");
        Bird bird = new Bird("Кеша", 1, 0.35, 0.5, true, "лес");
        Fish fish = new Fish("Немо", 2, 0.15, "Соленая", 50, "океан");
        
        // ========== 1. ДЕМОНСТРАЦИЯ ВНЕДРЕНИЯ ЗАВИСИМОСТЕЙ (DI) ==========
        Console.WriteLine("=== 1. ВНЕДРЕНИЕ ЗАВИСИМОСТЕЙ ЧЕРЕЗ КОНСТРУКТОР ===\n");
        
        // Внедрение зависимости IAnimal в AnimalService
        AnimalService dogService = new AnimalService(dog);
        AnimalService catService = new AnimalService(cat);
        AnimalService birdService = new AnimalService(bird);
        
        Console.WriteLine("--- Собака через AnimalService ---");
        dogService.DemonstrateAnimal();
        
        Console.WriteLine("\n--- Кошка Муся через AnimalService ---");
        catService.DemonstrateAnimal();
        
        // ========== 2. РАБОТА С ДОМАШНИМИ ЖИВОТНЫМИ ==========
        Console.WriteLine("\n=== 2. СЕРВИС ДЛЯ ДОМАШНИХ ЖИВОТНЫХ (IPet) ===\n");
        
        PetService petDogService = new PetService(dog);
        PetService petCatService = new PetService(cat);
        
        Console.WriteLine("--- Собака как домашний питомец ---");
        petDogService.InteractWithPet();
        
        Console.WriteLine("\n--- Кошка Муся как домашний питомец ---");
        petCatService.InteractWithPet();
        
        // ========== 3. РАБОТА С ДИКИМИ ЖИВОТНЫМИ ==========
        Console.WriteLine("\n=== 3. СЕРВИС ДЛЯ ДИКИХ ЖИВОТНЫХ (IWild) ===\n");
        
        WildAnimalService wildBirdService = new WildAnimalService(bird);
        WildAnimalService wildFishService = new WildAnimalService(fish);
        
        wildBirdService.ObserveWildAnimal();
        wildFishService.ObserveWildAnimal();
        
        // ========== 4. РАБОТА С ОБУЧАЕМЫМИ ЖИВОТНЫМИ ==========
        Console.WriteLine("\n=== 4. СЕРВИС ДЛЯ ОБУЧАЕМЫХ ЖИВОТНЫХ (ITrainable) ===\n");
        
        TrainingService trainingService = new TrainingService(dog);
        trainingService.TrainAnimal();
        
        // ========== 5. ЗООПАРК С ВНЕДРЕНИЕМ ЗАВИСИМОСТЕЙ ==========
        Console.WriteLine("\n=== 5. ЗООПАРК (коллекция IAnimal) ===\n");
        
        List<IAnimal> zooAnimals = new List<IAnimal>
        {
            new Dog("Шарик", 4, 20, "Дворняга", "Петр"),
            new Cat("Васька", 3, 3.8, "черный", false, "Анна"),
            new Bird("Воробей", 1, 0.05, 0.2, true, "город"),
            new Fish("Золотая", 1, 0.1, "Пресная", 2, "пруд"),
            dog,
            cat,
            bird,
            fish
        };
        
        Zoo cityZoo = new Zoo("Городской зоопарк", zooAnimals);
        cityZoo.ShowAllAnimals();
        cityZoo.FeedAllAnimals();
        
        // ========== 6. ДОПОЛНИТЕЛЬНАЯ ДЕМОНСТРАЦИЯ ==========
        Console.WriteLine("\n=== 6. ДОПОЛНИТЕЛЬНАЯ ДЕМОНСТРАЦИЯ ===\n");
        
        // Демонстрация специфичных методов
        Console.WriteLine("Специфичные методы:");
        dog.PerformTrick("Дай лапу");
        cat.Purr();
        
        // Демонстрация информации о кошке Мусе с новым хозяином
        Console.WriteLine("\n--- Информация о кошке Мусе ---");
        Console.WriteLine($"Имя: {cat.Name}");
        Console.WriteLine($"Возраст: {cat.Age} года");
        Console.WriteLine($"Цвет: {cat.Color}");
        Console.WriteLine($"Хозяин: {(cat as IPet).OwnerName}");
        Console.WriteLine($"Домашняя: {cat.IsIndoor}");
        
        // Демонстрация работы с энергетической системой
        Console.WriteLine("\n--- Демонстрация энергетической системы ---");
        Console.WriteLine($"Начальная энергия Муси: {cat.GetEnergyLevel()}%");
        cat.Move();
        cat.Move();
        cat.Move();
        Console.WriteLine($"Энергия после движений: {cat.GetEnergyLevel()}%");
        cat.Sleep();
        Console.WriteLine($"Энергия после сна: {cat.GetEnergyLevel()}%");
        
        Console.WriteLine("\n========== ДЕМОНСТРАЦИЯ УПРАВЛЕНИЯ ЗАВИСИМОСТЯМИ ЗАВЕРШЕНА ==========");
    }
}

// ========== ЗАПУСК ДЕМОНСТРАЦИИ ==========
AnimalDemo.Run();
````
