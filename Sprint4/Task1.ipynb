<h1 style="color:DodgerBlue">Явная реализация интерфейса</h1>

В C# интерфейсы могут быть реализованы явным и неявным образом. Давайте подробно рассмотрим, что такое явная реализация интерфейсов, реализация интерфейсов в базовых и производных классах, а также наследование интерфейсов, используя примеры классов `Vehicle`, `Car` и `Motorcycle`.

### Явная реализация интерфейсов

Явная реализация интерфейса используется для того, чтобы скрыть методы интерфейса от стандартного контекста класса. Это означает, что методы интерфейса могут быть вызваны только через ссылку на интерфейс, а не через экземпляр класса.

#### Пример явной реализации

Рассмотрим интерфейс `IVehicle` и его явную реализацию в классе `Car`.

```csharp
public interface IVehicle
{
    void Start();
    void Stop();
    int GetCurrentSpeed();
}

public class Car : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    private int currentSpeed;

    // Явная реализация интерфейса
    void IVehicle.Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} запущен.");
    }

    void IVehicle.Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} остановлен.");
    }

    // Неявная реализация метода
    public int GetCurrentSpeed()
    {
        return currentSpeed;
    }
}
```

В этом примере методы `Start` и `Stop` реализованы явно, и для их вызова необходимо использовать объект типа `IVehicle`:

```csharp

IVehicle myCar = new Car { Make = "Toyota", Model = "Camry" };
myCar.Start(); // Корректно, вызывает явную реализацию
// myCar.Stop(); // Корректно, вызывает явную реализацию
```

Если вы попытаетесь вызвать `Start` или `Stop` непосредственно через экземпляр `Car`, это приведёт к ошибке компиляции:

```csharp
Car myCar = new Car();
myCar.Start(); // Ошибка компиляции
```

### Реализация интерфейсов в базовых и производных классах

Интерфейсы могут быть реализованы в базовых классах и наследоваться в производных классах. Давайте рассмотрим эту концепцию на примере, включающем классы `Vehicle`, `Car` и `Motorcycle`.

#### Базовый класс Vehicle

```csharp
public abstract class Vehicle : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    protected int currentSpeed;

    public abstract void Start();  // Оставляем абстрактным
    public abstract void Stop();   // Оставляем абстрактным

    public virtual int GetCurrentSpeed()
    {
        return currentSpeed;
    }
}
```

#### Реализация в производных классах

Теперь реализуем `Car` и `Motorcycle`.

```csharp
public class Car : Vehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} остановлен.");
    }
}

public class Motorcycle : Vehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл остановлен.");
    }
}
```

Теперь оба класса `Car` и `Motorcycle` обязаны реализовать методы `Start` и `Stop`, которые определены в абстрактном классе `Vehicle`.

### Наследование интерфейсов

Интерфейсы могут наследовать друг от друга. Это позволяет создавать сложные иерархии интерфейсов. Например, мы можем создать интерфейс `IMotorizedVehicle`, который наследует `IVehicle`.

```csharp
public interface IMotorizedVehicle : IVehicle
{
    void RevEngine();
}

public class Motorcycle : Vehicle, IMotorizedVehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл остановлен.");
    }

    public void RevEngine()
    {
        Console.WriteLine("Мотоцикл рычит!");
    }
}
```

Таким образом, `Motorcycle` теперь может реализовать все методы из обоих интерфейсов `IVehicle` и `IMotorizedVehicle`.

### Заключение

1. **Явная реализация** интерфейсов скрывает функциональность классов от общего доступа, что помогает избежать путаницы с методами.
   
2. **Реализация интерфейсов в базовых и производных классах** позволяет создавать иерархию классов и обеспечивает обязательность реализации методов в производных классах.

3. **Наследование интерфейсов** позволяет комбинировать несколько интерфейсов, создавая более сложные структуры, что делает код более гибким и расширяемым.

Эти механизмы являются важными инструментами в инструменте разработчика, позволяя создавать более структурированные и управляемые программы.

<h4 style="color:DodgerBlue">Для проверки напишите пример кода на основе классов Vehicle, Car и Motorcycle ниже в блоке с применением интерфейса и абстактного класса:</h4>

````
using System;
using System.Collections.Generic;

// ========== ИНТЕРФЕЙСЫ ==========
public interface IVehicle
{
    void Start();
    void Stop();
    int GetCurrentSpeed();
    string GetInfo();
    void Honk();
}

public interface IMotorizedVehicle : IVehicle
{
    void RevEngine();
    int GetHorsepower();
}

public interface IElectricVehicle
{
    void Charge();
    int GetBatteryLevel();
    string GetChargingStatus();
}

// ========== БАЗОВЫЙ АБСТРАКТНЫЙ КЛАСС ==========
public abstract class Vehicle
{
    // Свойства
    public string Make { get; set; }
    public string Model { get; set; }
    public int Year { get; set; }
    protected int currentSpeed;
    protected bool isRunning;
    
    // Конструктор
    public Vehicle(string make, string model, int year)
    {
        Make = make;
        Model = model;
        Year = year;
        currentSpeed = 0;
        isRunning = false;
    }
    
    // Абстрактные методы (должны быть реализованы в производных классах)
    public abstract void Start();
    public abstract void Stop();
    
    // Виртуальный метод
    public virtual int GetCurrentSpeed()
    {
        return currentSpeed;
    }
    
    public virtual string GetInfo()
    {
        return $"{Make} {Model} ({Year})";
    }
    
    public virtual void ShowStatus()
    {
        string status = isRunning ? "запущен" : "остановлен";
        Console.WriteLine($"{Make} {Model}: {status}, скорость: {currentSpeed} км/ч");
    }
}

// ========== КЛАСС CAR ==========
public class Car : Vehicle, IMotorizedVehicle
{
    public int Doors { get; set; }
    public string BodyType { get; set; }
    public int Horsepower { get; set; }
    
    public Car(string make, string model, int year, int doors, string bodyType, int horsepower) 
        : base(make, model, year)
    {
        Doors = doors;
        BodyType = bodyType;
        Horsepower = horsepower;
    }
    
    // Явная реализация IMotorizedVehicle.RevEngine()
    void IMotorizedVehicle.RevEngine()
    {
        Console.WriteLine($"Автомобиль {Make} {Model} рычит: VRRROOOOOM! ({(this as IMotorizedVehicle).GetHorsepower()} л.с.)");
    }
    
    // Явная реализация IMotorizedVehicle.GetHorsepower()
    int IMotorizedVehicle.GetHorsepower()
    {
        return Horsepower;
    }
    
    // Явная реализация IVehicle.Start()
    void IVehicle.Start()
    {
        isRunning = true;
        currentSpeed = 0;
        Console.WriteLine($"Автомобиль {Make} {Model} запущен (через интерфейс).");
        Console.WriteLine($"  Мощность: {Horsepower} л.с., Дверей: {Doors}");
    }
    
    // Явная реализация IVehicle.Stop()
    void IVehicle.Stop()
    {
        currentSpeed = 0;
        isRunning = false;
        Console.WriteLine($"Автомобиль {Make} {Model} остановлен (через интерфейс).");
    }
    
    // Явная реализация IVehicle.GetCurrentSpeed()
    int IVehicle.GetCurrentSpeed()
    {
        return currentSpeed;
    }
    
    // Явная реализация IVehicle.GetInfo()
    string IVehicle.GetInfo()
    {
        return $"{Make} {Model} ({Year}), {Doors} двери, {BodyType}, {Horsepower} л.с.";
    }
    
    // Явная реализация IVehicle.Honk()
    void IVehicle.Honk()
    {
        Console.WriteLine($"{Make} {Model} сигналит: Бип-бип!");
    }
    
    // Неявная реализация (публичные методы для прямого доступа)
    public void Accelerate(int amount)
    {
        if (isRunning)
        {
            currentSpeed += amount;
            Console.WriteLine($"Автомобиль {Make} {Model} разгоняется до {currentSpeed} км/ч.");
        }
        else
        {
            Console.WriteLine($"Сначала нужно запустить двигатель {Make} {Model}!");
        }
    }
    
    public void Brake(int amount)
    {
        if (currentSpeed > 0)
        {
            currentSpeed = Math.Max(0, currentSpeed - amount);
            Console.WriteLine($"Автомобиль {Make} {Model} тормозит до {currentSpeed} км/ч.");
        }
    }
    
    // Переопределение абстрактных методов
    public override void Start()
    {
        isRunning = true;
        currentSpeed = 0;
        Console.WriteLine($"Автомобиль {Make} {Model} запущен.");
    }
    
    public override void Stop()
    {
        currentSpeed = 0;
        isRunning = false;
        Console.WriteLine($"Автомобиль {Make} {Model} остановлен.");
    }
}

// ========== КЛАСС MOTORCYCLE ==========
public class Motorcycle : Vehicle, IMotorizedVehicle
{
    public string EngineType { get; set; }
    public bool HasSidecar { get; set; }
    public int Horsepower { get; set; }
    
    public Motorcycle(string make, string model, int year, string engineType, bool hasSidecar, int horsepower) 
        : base(make, model, year)
    {
        EngineType = engineType;
        HasSidecar = hasSidecar;
        Horsepower = horsepower;
    }
    
    // Явная реализация IMotorizedVehicle.RevEngine()
    void IMotorizedVehicle.RevEngine()
    {
        Console.WriteLine($"Мотоцикл {Make} {Model} рычит: BRRRAAAP! ({(this as IMotorizedVehicle).GetHorsepower()} л.с.)");
    }
    
    // Явная реализация IMotorizedVehicle.GetHorsepower()
    int IMotorizedVehicle.GetHorsepower()
    {
        return Horsepower;
    }
    
    // Явная реализация IVehicle.Start()
    void IVehicle.Start()
    {
        isRunning = true;
        currentSpeed = 0;
        Console.WriteLine($"Мотоцикл {Make} {Model} запущен (через интерфейс).");
        Console.WriteLine($"  Двигатель: {EngineType}, Мощность: {Horsepower} л.с.");
    }
    
    // Явная реализация IVehicle.Stop()
    void IVehicle.Stop()
    {
        currentSpeed = 0;
        isRunning = false;
        Console.WriteLine($"Мотоцикл {Make} {Model} остановлен (через интерфейс).");
    }
    
    // Явная реализация IVehicle.GetCurrentSpeed()
    int IVehicle.GetCurrentSpeed()
    {
        return currentSpeed;
    }
    
    // Явная реализация IVehicle.GetInfo()
    string IVehicle.GetInfo()
    {
        return $"{Make} {Model} ({Year}), {EngineType}, {Horsepower} л.с., Коляска: {(HasSidecar ? "Да" : "Нет")}";
    }
    
    // Явная реализация IVehicle.Honk()
    void IVehicle.Honk()
    {
        Console.WriteLine($"{Make} {Model} сигналит: Бип!");
    }
    
    // Неявная реализация (публичные методы)
    public void Wheelie()
    {
        if (isRunning && currentSpeed > 20)
        {
            Console.WriteLine($"Мотоцикл {Make} {Model} едет на заднем колесе!");
        }
        else
        {
            Console.WriteLine($"Невозможно сделать вили на мотоцикле {Make} {Model} (нужна скорость > 20 км/ч)");
        }
    }
    
    // Переопределение абстрактных методов
    public override void Start()
    {
        isRunning = true;
        currentSpeed = 0;
        Console.WriteLine($"Мотоцикл {Make} {Model} запущен.");
    }
    
    public override void Stop()
    {
        currentSpeed = 0;
        isRunning = false;
        Console.WriteLine($"Мотоцикл {Make} {Model} остановлен.");
    }
}

// ========== КЛАСС ELECTRIC_CAR ==========
public class ElectricCar : Car, IElectricVehicle
{
    private int batteryLevel;
    
    public int BatteryCapacity { get; set; }
    
    public ElectricCar(string make, string model, int year, int doors, string bodyType, int horsepower, int batteryCapacity) 
        : base(make, model, year, doors, bodyType, horsepower)
    {
        BatteryCapacity = batteryCapacity;
        batteryLevel = 100;
    }
    
    // Явная реализация IElectricVehicle.Charge()
    void IElectricVehicle.Charge()
    {
        batteryLevel = 100;
        Console.WriteLine($"Электромобиль {Make} {Model} заряжен до {batteryLevel}% ({(this as IElectricVehicle).GetChargingStatus()})");
    }
    
    // Явная реализация IElectricVehicle.GetBatteryLevel()
    int IElectricVehicle.GetBatteryLevel()
    {
        return batteryLevel;
    }
    
    // Явная реализация IElectricVehicle.GetChargingStatus()
    string IElectricVehicle.GetChargingStatus()
    {
        return batteryLevel >= 80 ? "Готов к поездке" : "Требуется зарядка";
    }
    
    // Переопределение методов
    public override void Start()
    {
        isRunning = true;
        currentSpeed = 0;
        Console.WriteLine($"Электромобиль {Make} {Model} запущен (бесшумно). Заряд: {batteryLevel}%");
    }
}

// ========== ДЕМОНСТРАЦИЯ ДЛЯ VEHICLE ==========
Console.WriteLine("========== ЯВНАЯ РЕАЛИЗАЦИЯ ИНТЕРФЕЙСОВ: ТРАНСПОРТНЫЕ СРЕДСТВА ==========\n");

// Создание объектов
Car car = new Car("Toyota", "Camry", 2022, 4, "Седан", 203);
Motorcycle motorcycle = new Motorcycle("Harley-Davidson", "Street Glide", 2023, "V-Twin", false, 95);
ElectricCar electricCar = new ElectricCar("Tesla", "Model S", 2024, 4, "Лифтбек", 1020, 100);

Console.WriteLine("=== 1. НЕЯВНАЯ РЕАЛИЗАЦИЯ (прямой вызов) ===\n");

car.Start();
car.Accelerate(50);
car.Stop();

Console.WriteLine();
motorcycle.Start();
motorcycle.Wheelie();
motorcycle.Stop();

Console.WriteLine();
electricCar.Start();
(electricCar as IElectricVehicle).Charge();

Console.WriteLine("\n=== 2. ЯВНАЯ РЕАЛИЗАЦИЯ (вызов через интерфейс) ===\n");

// Вызов явно реализованных методов через интерфейс
IVehicle vehicleCar = car;
Console.WriteLine("Через интерфейс IVehicle для Car:");
vehicleCar.Start();
Console.WriteLine($"Информация: {vehicleCar.GetInfo()}");
vehicleCar.Honk();
vehicleCar.Stop();

Console.WriteLine();

IVehicle vehicleBike = motorcycle;
Console.WriteLine("Через интерфейс IVehicle для Motorcycle:");
vehicleBike.Start();
Console.WriteLine($"Информация: {vehicleBike.GetInfo()}");
vehicleBike.Honk();
vehicleBike.Stop();

Console.WriteLine("\n=== 3. ВЫЗОВ МЕТОДОВ ИЗ НАСЛЕДОВАННЫХ ИНТЕРФЕЙСОВ ===\n");

IMotorizedVehicle motorCar = car;
Console.WriteLine("Через интерфейс IMotorizedVehicle для Car:");
motorCar.RevEngine();
Console.WriteLine($"Мощность: {motorCar.GetHorsepower()} л.с.");

Console.WriteLine();

IMotorizedVehicle motorBike = motorcycle;
Console.WriteLine("Через интерфейс IMotorizedVehicle для Motorcycle:");
motorBike.RevEngine();
Console.WriteLine($"Мощность: {motorBike.GetHorsepower()} л.с.");

Console.WriteLine("\n=== 4. НАСЛЕДОВАНИЕ ИНТЕРФЕЙСОВ ===\n");

IElectricVehicle ev = electricCar;
Console.WriteLine("Через интерфейс IElectricVehicle:");
ev.Charge();
Console.WriteLine($"Уровень заряда: {ev.GetBatteryLevel()}%");
Console.WriteLine($"Статус: {ev.GetChargingStatus()}");

Console.WriteLine("\n=== 5. ПОЛИМОРФИЗМ ЧЕРЕЗ ИНТЕРФЕЙСЫ ===\n");

List<IVehicle> vehicles = new List<IVehicle>
{
    new Car("BMW", "X5", 2023, 5, "Внедорожник", 340),
    new Motorcycle("Yamaha", "R1", 2024, "Inline-4", false, 200),
    new ElectricCar("Rivian", "R1T", 2023, 4, "Пикап", 835, 135)
};

foreach (var vehicle in vehicles)
{
    Console.WriteLine($"Транспорт: {vehicle.GetInfo()}");
    vehicle.Start();
    vehicle.Honk();
    vehicle.Stop();
    Console.WriteLine();
}

Console.WriteLine("========== ДЕМОНСТРАЦИЯ VEHICLE ЗАВЕРШЕНА ==========\n");
````

````
using System;
using System.Collections.Generic;
using System.Linq;

// ========== ИНТЕРФЕЙСЫ ==========
public interface IAnimal
{
    void MakeSound();
    void Move();
    string GetInfo();
    void Eat();
    void Sleep();
}

public interface IPet
{
    string OwnerName { get; }
    void Play();
    void Feed();
    void ShowAffection();
    void Walk();
}

public interface IWild
{
    string Habitat { get; }
    void Hunt();
    void DefendTerritory();
    void Hide();
}

public interface ITrainable
{
    bool IsTrained { get; }
    void Train();
    void PerformTrick(string trick);
    string[] GetKnownTricks();
}

// ========== БАЗОВЫЙ АБСТРАКТНЫЙ КЛАСС ==========
public abstract class Animal
{
    // Свойства
    public string Name { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }
    protected int energyLevel;
    protected bool isAlive;
    
    // Конструктор
    public Animal(string name, int age, double weight)
    {
        Name = name;
        Age = age;
        Weight = weight;
        energyLevel = 100;
        isAlive = true;
    }
    
    // Абстрактные методы
    public abstract void MakeSound();
    public abstract void Move();
    public abstract void Breathe();
    
    // Виртуальные методы
    public virtual void Eat()
    {
        energyLevel = Math.Min(100, energyLevel + 20);
        Console.WriteLine($"{Name} поел. Энергия: {energyLevel}%");
    }
    
    public virtual void Sleep()
    {
        energyLevel = 100;
        Console.WriteLine($"{Name} выспался. Энергия восстановлена до {energyLevel}%");
    }
    
    public virtual string GetInfo()
    {
        return $"{Name}, {Age} лет, {Weight} кг, энергия: {energyLevel}%";
    }
    
    public virtual void ShowStatus()
    {
        string status = isAlive ? "жив" : "мертв";
        Console.WriteLine($"{Name}: {status}, энергия: {energyLevel}%");
    }
}

// ========== КЛАСС DOG ==========
// Исправлено: добавлен IAnimal в список реализуемых интерфейсов
public class Dog : Animal, IAnimal, IPet, ITrainable
{
    public string Breed { get; set; }
    private string ownerName;
    private bool isTrained;
    private List<string> tricks = new List<string>();
    
    public Dog(string name, int age, double weight, string breed, string ownerName) 
        : base(name, age, weight)
    {
        Breed = breed;
        this.ownerName = ownerName;
        isTrained = false;
    }
    
    // Явная реализация IPet.OwnerName
    string IPet.OwnerName => ownerName;
    
    // Явная реализация ITrainable.IsTrained
    bool ITrainable.IsTrained => isTrained;
    
    // Явная реализация IAnimal.MakeSound()
    void IAnimal.MakeSound()
    {
        Console.WriteLine($"Собака {Name} лает: Гав-гав! (через интерфейс)");
    }
    
    // Явная реализация IAnimal.Move()
    void IAnimal.Move()
    {
        if (energyLevel >= 10)
        {
            energyLevel -= 10;
            Console.WriteLine($"Собака {Name} бегает на четырех лапах. Энергия: {energyLevel}% (через интерфейс)");
        }
        else
        {
            Console.WriteLine($"Собака {Name} слишком устала, чтобы двигаться!");
        }
    }
    
    // Явная реализация IAnimal.GetInfo()
    string IAnimal.GetInfo()
    {
        return $"{Name}, {Age} лет, {Weight} кг, порода: {Breed}, хозяин: {ownerName}";
    }
    
    // Явная реализация IAnimal.Eat()
    void IAnimal.Eat()
    {
        energyLevel = Math.Min(100, energyLevel + 20);
        Console.WriteLine($"Собака {Name} ест корм. Энергия: {energyLevel}% (через интерфейс)");
    }
    
    // Явная реализация IAnimal.Sleep()
    void IAnimal.Sleep()
    {
        energyLevel = 100;
        Console.WriteLine($"Собака {Name} спит. Энергия восстановлена (через интерфейс)");
    }
    
    // Явная реализация IPet.Play()
    void IPet.Play()
    {
        if (energyLevel >= 15)
        {
            energyLevel -= 15;
            Console.WriteLine($"{Name} играет с мячом! Энергия: {energyLevel}%");
        }
        else
        {
            Console.WriteLine($"{Name} слишком устал для игр!");
        }
    }
    
    // Явная реализация IPet.Feed()
    void IPet.Feed()
    {
        (this as IAnimal).Eat();
    }
    
    // Явная реализация IPet.ShowAffection()
    void IPet.ShowAffection()
    {
        Console.WriteLine($"{Name} виляет хвостом и лижет руку хозяину {ownerName}!");
    }
    
    // Явная реализация IPet.Walk()
    void IPet.Walk()
    {
        Console.WriteLine($"{Name} гуляет на поводке с {ownerName}.");
        (this as IAnimal).Move();
    }
    
    // Явная реализация ITrainable.Train()
    void ITrainable.Train()
    {
        if (!isTrained)
        {
            isTrained = true;
            Console.WriteLine($"{Name} прошел обучение! Теперь он знает команды.");
        }
        else
        {
            Console.WriteLine($"{Name} уже обучен.");
        }
    }
    
    // Явная реализация ITrainable.PerformTrick()
    void ITrainable.PerformTrick(string trick)
    {
        if (isTrained)
        {
            if (!tricks.Contains(trick))
                tricks.Add(trick);
            Console.WriteLine($"{Name} выполняет трюк: {trick}!");
        }
        else
        {
            Console.WriteLine($"{Name} еще не обучен. Нужно сначала обучить собаку!");
        }
    }
    
    // Явная реализация ITrainable.GetKnownTricks()
    string[] ITrainable.GetKnownTricks()
    {
        return tricks.ToArray();
    }
    
    // Переопределение абстрактных методов
    public override void MakeSound()
    {
        Console.WriteLine($"Собака {Name} лает: Гав-гав!");
    }
    
    public override void Move()
    {
        if (energyLevel >= 10)
        {
            energyLevel -= 10;
            Console.WriteLine($"Собака {Name} бегает на четырех лапах. Энергия: {energyLevel}%");
        }
        else
        {
            Console.WriteLine($"Собака {Name} слишком устала, чтобы двигаться!");
        }
    }
    
    public override void Breathe()
    {
        Console.WriteLine($"Собака {Name} дышит, высунув язык.");
    }
}

// ========== КЛАСС CAT (с именем Муся и цветом серый) ==========
// Исправлено: добавлен IAnimal в список реализуемых интерфейсов
public class Cat : Animal, IAnimal, IPet
{
    public string Color { get; set; }
    public bool IsIndoor { get; set; }
    private string ownerName;
    
    public Cat(string name, int age, double weight, string color, bool isIndoor, string ownerName) 
        : base(name, age, weight)
    {
        Color = color;      // цвет "серый"
        IsIndoor = isIndoor;
        this.ownerName = ownerName;
    }
    
    // Явная реализация IPet.OwnerName
    string IPet.OwnerName => ownerName;
    
    // Явная реализация IAnimal.MakeSound()
    void IAnimal.MakeSound()
    {
        Console.WriteLine($"Кошка {Name} мяукает: Мяу-мяу! (через интерфейс)");
    }
    
    // Явная реализация IAnimal.Move()
    void IAnimal.Move()
    {
        if (energyLevel >= 5)
        {
            energyLevel -= 5;
            Console.WriteLine($"Кошка {Name} грациозно крадется. Энергия: {energyLevel}% (через интерфейс)");
        }
        else
        {
            Console.WriteLine($"Кошка {Name} слишком устала, чтобы двигаться!");
        }
    }
    
    // Явная реализация IAnimal.GetInfo()
    string IAnimal.GetInfo()
    {
        return $"{Name}, {Age} лет, {Weight} кг, окрас: {Color}, хозяин: {ownerName}";
    }
    
    // Явная реализация IAnimal.Eat()
    void IAnimal.Eat()
    {
        energyLevel = Math.Min(100, energyLevel + 20);
        Console.WriteLine($"Кошка {Name} ест рыбу. Энергия: {energyLevel}% (через интерфейс)");
    }
    
    // Явная реализация IAnimal.Sleep()
    void IAnimal.Sleep()
    {
        energyLevel = 100;
        Console.WriteLine($"Кошка {Name} спит, свернувшись клубочком. Энергия восстановлена (через интерфейс)");
    }
    
    // Явная реализация IPet.Play()
    void IPet.Play()
    {
        if (energyLevel >= 10)
        {
            energyLevel -= 10;
            Console.WriteLine($"{Name} играет с клубком ниток! Энергия: {energyLevel}%");
        }
        else
        {
            Console.WriteLine($"{Name} слишком устала для игр!");
        }
    }
    
    // Явная реализация IPet.Feed()
    void IPet.Feed()
    {
        (this as IAnimal).Eat();
    }
    
    // Явная реализация IPet.ShowAffection()
    void IPet.ShowAffection()
    {
        Console.WriteLine($"{Name} трется о ноги хозяина {ownerName} и мурлычет: Мур-мур-мур...");
    }
    
    // Явная реализация IPet.Walk()
    void IPet.Walk()
    {
        if (IsIndoor)
            Console.WriteLine($"{Name} гуляет по квартире с {ownerName}.");
        else
            Console.WriteLine($"{Name} гуляет на улице, исследуя окрестности.");
        (this as IAnimal).Move();
    }
    
    // Переопределение абстрактных методов
    public override void MakeSound()
    {
        Console.WriteLine($"Кошка {Name} мяукает: Мяу-мяу!");
    }
    
    public override void Move()
    {
        if (energyLevel >= 5)
        {
            energyLevel -= 5;
            Console.WriteLine($"Кошка {Name} грациозно крадется. Энергия: {energyLevel}%");
        }
        else
        {
            Console.WriteLine($"Кошка {Name} слишком устала, чтобы двигаться!");
        }
    }
    
    public override void Breathe()
    {
        Console.WriteLine($"Кошка {Name} тихо дышит, свернувшись клубочком.");
    }
    
    // Специфичный метод
    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет от удовольствия!");
    }
}

// ========== КЛАСС BIRD ==========
// Исправлено: добавлен IAnimal в список реализуемых интерфейсов
public class Bird : Animal, IAnimal, IWild
{
    public double WingSpan { get; set; }
    public bool CanFly { get; set; }
    private string habitat;
    
    public Bird(string name, int age, double weight, double wingSpan, bool canFly, string habitat) 
        : base(name, age, weight)
    {
        WingSpan = wingSpan;
        CanFly = canFly;
        this.habitat = habitat;
    }
    
    // Явная реализация IWild.Habitat
    string IWild.Habitat => habitat;
    
    // Явная реализация IAnimal.MakeSound()
    void IAnimal.MakeSound()
    {
        Console.WriteLine($"Птица {Name} чирикает: Чик-чирик! (через интерфейс)");
    }
    
    // Явная реализация IAnimal.Move()
    void IAnimal.Move()
    {
        if (CanFly && energyLevel >= 15)
        {
            energyLevel -= 15;
            Console.WriteLine($"Птица {Name} летит с размахом крыльев {WingSpan} м. Энергия: {energyLevel}% (через интерфейс)");
        }
        else if (!CanFly && energyLevel >= 5)
        {
            energyLevel -= 5;
            Console.WriteLine($"Птица {Name} ходит по земле. Энергия: {energyLevel}% (через интерфейс)");
        }
        else
        {
            Console.WriteLine($"Птица {Name} слишком устала, чтобы двигаться!");
        }
    }
    
    // Явная реализация IAnimal.GetInfo()
    string IAnimal.GetInfo()
    {
        return $"{Name}, {Age} лет, {Weight} кг, размах крыльев: {WingSpan} м, среда обитания: {habitat}";
    }
    
    // Явная реализация IAnimal.Eat()
    void IAnimal.Eat()
    {
        energyLevel = Math.Min(100, energyLevel + 15);
        Console.WriteLine($"Птица {Name} клюет зерна. Энергия: {energyLevel}% (через интерфейс)");
    }
    
    // Явная реализация IAnimal.Sleep()
    void IAnimal.Sleep()
    {
        energyLevel = 100;
        Console.WriteLine($"Птица {Name} спит, спрятав голову под крыло. Энергия восстановлена (через интерфейс)");
    }
    
    // Явная реализация IWild.Hunt()
    void IWild.Hunt()
    {
        if (energyLevel >= 20)
        {
            energyLevel -= 20;
            Console.WriteLine($"Птица {Name} охотится в {habitat}!");
        }
        else
        {
            Console.WriteLine($"{Name} слишком устала для охоты!");
        }
    }
    
    // Явная реализация IWild.DefendTerritory()
    void IWild.DefendTerritory()
    {
        Console.WriteLine($"Птица {Name} защищает свою территорию в {habitat}!");
    }
    
    // Явная реализация IWild.Hide()
    void IWild.Hide()
    {
        Console.WriteLine($"Птица {Name} прячется в {habitat} от хищников.");
    }
    
    // Переопределение абстрактных методов
    public override void MakeSound()
    {
        Console.WriteLine($"Птица {Name} чирикает: Чик-чирик!");
    }
    
    public override void Move()
    {
        if (CanFly && energyLevel >= 15)
        {
            energyLevel -= 15;
            Console.WriteLine($"Птица {Name} летит с размахом крыльев {WingSpan} м. Энергия: {energyLevel}%");
        }
        else if (!CanFly && energyLevel >= 5)
        {
            energyLevel -= 5;
            Console.WriteLine($"Птица {Name} ходит по земле. Энергия: {energyLevel}%");
        }
        else
        {
            Console.WriteLine($"Птица {Name} слишком устала, чтобы двигаться!");
        }
    }
    
    public override void Breathe()
    {
        Console.WriteLine($"Птица {Name} дышит легко и быстро.");
    }
}

// ========== КЛАСС FISH ==========
// Исправлено: добавлен IAnimal в список реализуемых интерфейсов
public class Fish : Animal, IAnimal, IWild
{
    public string WaterType { get; set; }
    public double MaxDepth { get; set; }
    private string habitat;
    
    public Fish(string name, int age, double weight, string waterType, double maxDepth, string habitat) 
        : base(name, age, weight)
    {
        WaterType = waterType;
        MaxDepth = maxDepth;
        this.habitat = habitat;
    }
    
    // Явная реализация IWild.Habitat
    string IWild.Habitat => habitat;
    
    // Явная реализация IAnimal.MakeSound()
    void IAnimal.MakeSound()
    {
        Console.WriteLine($"Рыба {Name} издает булькающие звуки: Буль-буль! (через интерфейс)");
    }
    
    // Явная реализация IAnimal.Move()
    void IAnimal.Move()
    {
        if (energyLevel >= 5)
        {
            energyLevel -= 5;
            Console.WriteLine($"Рыба {Name} плавает в {WaterType} воде. Энергия: {energyLevel}% (через интерфейс)");
        }
        else
        {
            Console.WriteLine($"Рыба {Name} слишком устала, чтобы плавать!");
        }
    }
    
    // Явная реализация IAnimal.GetInfo()
    string IAnimal.GetInfo()
    {
        return $"{Name}, {Age} лет, {Weight} кг, вода: {WaterType}, макс. глубина: {MaxDepth} м, среда: {habitat}";
    }
    
    // Явная реализация IAnimal.Eat()
    void IAnimal.Eat()
    {
        energyLevel = Math.Min(100, energyLevel + 15);
        Console.WriteLine($"Рыба {Name} проглатывает планктон. Энергия: {energyLevel}% (через интерфейс)");
    }
    
    // Явная реализация IAnimal.Sleep()
    void IAnimal.Sleep()
    {
        energyLevel = 100;
        Console.WriteLine($"Рыба {Name} спит, неподвижно зависнув в воде. Энергия восстановлена (через интерфейс)");
    }
    
    // Явная реализация IWild.Hunt()
    void IWild.Hunt()
    {
        if (energyLevel >= 15)
        {
            energyLevel -= 15;
            Console.WriteLine($"Рыба {Name} охотится на мелкую рыбу в {habitat}!");
        }
        else
        {
            Console.WriteLine($"{Name} слишком устала для охоты!");
        }
    }
    
    // Явная реализация IWild.DefendTerritory()
    void IWild.DefendTerritory()
    {
        Console.WriteLine($"Рыба {Name} защищает свою территорию в {habitat} на глубине {MaxDepth} м!");
    }
    
    // Явная реализация IWild.Hide()
    void IWild.Hide()
    {
        Console.WriteLine($"Рыба {Name} прячется среди кораллов в {habitat}.");
    }
    
    // Переопределение абстрактных методов
    public override void MakeSound()
    {
        Console.WriteLine($"Рыба {Name} издает булькающие звуки: Буль-буль!");
    }
    
    public override void Move()
    {
        if (energyLevel >= 5)
        {
            energyLevel -= 5;
            Console.WriteLine($"Рыба {Name} плавает в {WaterType} воде. Энергия: {energyLevel}%");
        }
        else
        {
            Console.WriteLine($"Рыба {Name} слишком устала, чтобы плавать!");
        }
    }
    
    public override void Breathe()
    {
        Console.WriteLine($"Рыба {Name} дышит жабрами под водой.");
    }
}

// ========== ДЕМОНСТРАЦИЯ ДЛЯ ANIMAL ==========
class AnimalDemo
{
    public static void Run()
    {
        Console.WriteLine("========== ЯВНАЯ РЕАЛИЗАЦИЯ ИНТЕРФЕЙСОВ: ЖИВОТНЫЕ ==========\n");
        
        // Создание объектов с указанными параметрами
        Dog dog = new Dog("Рекс", 3, 25.5, "Овчарка", "Иван");
        Cat cat = new Cat("Муся", 2, 4.2, "серый", true, "Элина");
        Bird bird = new Bird("Кеша", 1, 0.35, 0.5, true, "лес");
        Fish fish = new Fish("Немо", 2, 0.15, "Соленая", 50, "океан");
        
        Console.WriteLine("=== 1. НЕЯВНАЯ РЕАЛИЗАЦИЯ (прямой вызов) ===\n");
        
        dog.MakeSound();
        dog.Move();
        dog.Eat();
        
        Console.WriteLine();
        cat.MakeSound();
        cat.Move();
        cat.Purr(); // Специфичный метод
        
        Console.WriteLine();
        bird.MakeSound();
        bird.Move();
        
        Console.WriteLine("\n=== 2. ЯВНАЯ РЕАЛИЗАЦИЯ (вызов через интерфейс IAnimal) ===\n");
        
        IAnimal animalDog = dog;
        Console.WriteLine("Через интерфейс IAnimal для Dog:");
        animalDog.MakeSound();
        animalDog.Move();
        animalDog.Eat();
        animalDog.Sleep();
        Console.WriteLine($"Информация: {animalDog.GetInfo()}");
        
        Console.WriteLine();
        
        IAnimal animalCat = cat;
        Console.WriteLine("Через интерфейс IAnimal для Cat (Муся):");
        animalCat.MakeSound();
        animalCat.Move();
        animalCat.Eat();
        animalCat.Sleep();
        Console.WriteLine($"Информация: {animalCat.GetInfo()}");
        
        Console.WriteLine("\n=== 3. ЯВНАЯ РЕАЛИЗАЦИЯ (вызов через интерфейс IPet) ===\n");
        
        IPet petDog = dog;
        Console.WriteLine("Через интерфейс IPet для Dog:");
        petDog.Play();
        petDog.Feed();
        petDog.ShowAffection();
        petDog.Walk();
        Console.WriteLine($"Хозяин: {petDog.OwnerName}");
        
        Console.WriteLine();
        
        IPet petCat = cat;
        Console.WriteLine("Через интерфейс IPet для Cat (Муся):");
        petCat.Play();
        petCat.Feed();
        petCat.ShowAffection();
        petCat.Walk();
        Console.WriteLine($"Хозяин: {petCat.OwnerName}");
        
        Console.WriteLine("\n=== 4. ЯВНАЯ РЕАЛИЗАЦИЯ (вызов через интерфейс ITrainable) ===\n");
        
        ITrainable trainableDog = dog;
        Console.WriteLine("Через интерфейс ITrainable для Dog:");
        trainableDog.Train();
        trainableDog.PerformTrick("Сидеть");
        trainableDog.PerformTrick("Лежать");
        trainableDog.PerformTrick("Дай лапу");
        Console.WriteLine($"Обучен: {trainableDog.IsTrained}");
        Console.WriteLine($"Известные трюки: {string.Join(", ", trainableDog.GetKnownTricks())}");
        
        Console.WriteLine("\n=== 5. ЯВНАЯ РЕАЛИЗАЦИЯ (вызов через интерфейс IWild) ===\n");
        
        IWild wildBird = bird;
        Console.WriteLine("Через интерфейс IWild для Bird:");
        wildBird.Hunt();
        wildBird.DefendTerritory();
        wildBird.Hide();
        Console.WriteLine($"Среда обитания: {wildBird.Habitat}");
        
        Console.WriteLine();
        
        IWild wildFish = fish;
        Console.WriteLine("Через интерфейс IWild для Fish:");
        wildFish.Hunt();
        wildFish.DefendTerritory();
        wildFish.Hide();
        Console.WriteLine($"Среда обитания: {wildFish.Habitat}");
        
        Console.WriteLine("\n=== 6. ПОЛИМОРФИЗМ ЧЕРЕЗ ИНТЕРФЕЙСЫ ===\n");
        
        List<IAnimal> animals = new List<IAnimal>
        {
            new Dog("Шарик", 4, 20, "Дворняга", "Петр"),
            new Cat("Васька", 3, 3.8, "черный", false, "Анна"),
            new Bird("Воробей", 1, 0.05, 0.2, true, "город"),
            new Fish("Золотая", 1, 0.1, "Пресная", 2, "пруд")
        };
        
        foreach (var animal in animals)
        {
            Console.WriteLine($"Животное: {animal.GetInfo()}");
            animal.MakeSound();
            animal.Move();
            Console.WriteLine();
        }
        
        Console.WriteLine("\n=== 7. ДЕМОНСТРАЦИЯ АБСТРАКТНЫХ МЕТОДОВ ===\n");
        
        Animal[] animalsArray = new Animal[] { dog, cat, bird, fish };
        
        foreach (var animal in animalsArray)
        {
            animal.ShowStatus();
            animal.Breathe();
            Console.WriteLine();
        }
        
        Console.WriteLine("========== ДЕМОНСТРАЦИЯ ANIMAL ЗАВЕРШЕНА ==========");
    }
}

// ========== ЗАПУСК ДЕМОНСТРАЦИИ ==========
AnimalDemo.Run();
````
