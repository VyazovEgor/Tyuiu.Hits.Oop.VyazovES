<h1 style="color:DodgerBlue">Определение и применение интерфейса</h1>



Интерфейсы в C# — это контракты, которые определяют набор методов и свойств, которые класс должен реализовать, если он "подписывается" на этот интерфейс. Они не содержат реализации, а лишь описывают, какие методы и свойства должны быть у класса. Интерфейсы позволяют создавать гибкую и расширяемую архитектуру, так как разные классы могут реализовывать один и тот же интерфейс, предоставляя свою индивидуальную логику.

### Пример интерфейса

Рассмотрим интерфейс `IVehicle`, который будет использоваться для классов `Vehicle`, `Car`, и `Motorcycle`.

```csharp
public interface IVehicle
{
    void Start();
    void Stop();
    int GetCurrentSpeed();
}
```

### Реализация интерфейса в классах

Теперь создадим абстрактный класс `Vehicle`, который будет предоставлять базовые свойства, а потом создадим классы `Car` и `Motorcycle`, которые будут реализовывать интерфейс `IVehicle`.

```csharp
public abstract class Vehicle : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    protected int currentSpeed;

    public virtual void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} запущен.");
    }

    public virtual void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} остановлен.");
    }

    public virtual int GetCurrentSpeed()
    {
        return currentSpeed;
    }
}
```

### Реализация классов Car и Motorcycle

Теперь реализуем классы `Car` и `Motorcycle`, которые наследуются от `Vehicle` и могут добавлять свою специфическую логику.

```csharp
public class Car : Vehicle
{
    public int NumberOfDoors { get; set; }

    public override void Start()
    {
        base.Start();
        Console.WriteLine("Автомобиль готов к поездке.");
    }

    public override void Stop()
    {
        base.Stop();
        Console.WriteLine("Автомобиль полностью остановлен.");
    }

    public void Accelerate(int amount)
    {
        currentSpeed += amount;
        Console.WriteLine($"Автомобиль разгоняется до {currentSpeed} км/ч.");
    }
}

public class Motorcycle : Vehicle
{
    public bool HasSidecar { get; set; }

    public override void Start()
    {
        base.Start();
        Console.WriteLine("Мотоцикл готов к поездке.");
    }

    public override void Stop()
    {
        base.Stop();
        Console.WriteLine("Мотоцикл полностью остановлен.");
    }

    public void RevEngine()
    {
        Console.WriteLine("Мотоцикл рычит!");
    }
}
```

### Использование классов

Теперь мы можем создать экземпляры классов `Car` и `Motorcycle` и использовать методы, определенные в интерфейсе:

```csharp

IVehicle myCar = new Car { Make = "Toyota", Model = "Camry", NumberOfDoors = 4 };
myCar.Start();
((Car)myCar).Accelerate(50);
Console.WriteLine($"Текущая скорость: {myCar.GetCurrentSpeed()} км/ч");
        
IVehicle myMotorcycle = new Motorcycle { Make = "Harley-Davidson", Model = "Street", HasSidecar = false };
myMotorcycle.Start();
((Motorcycle)myMotorcycle).RevEngine();
myMotorcycle.Stop();
```

### Пояснения

1. **Интерфейс (IVehicle)**: Определяет методы, которые должны быть реализованы в классах, использующих этот интерфейс.
2. **Абстрактный класс (Vehicle)**: Реализует часть общей логики (например, свойства `Make`, `Model` и методы `Start`, `Stop`). Это позволяет избежать дублирования кода.
3. **Наследование и Полиморфизм**: Классы `Car` и `Motorcycle` наследуют `Vehicle` и могут расширять поведение класса, переопределяя методы и добавляя новые.
4. **Гибкость**: Используя интерфейсы, можно легко добавить другие транспортные средства, такие как `Truck`, просто реализовав тот же интерфейс.

Это создаёт хорошо структурированную и масштабируемую архитектуру, позволяя легко добавлять новые типы транспортных средств без изменения существующего кода. 

### Заключение
Интерфейсы в C# предоставляют мощный механизм для определения контрактов и достижения полиморфизма. Они позволяют разработчикам создавать гибкие и расширяемые системы, где классы могут быть легко заменены или добавлены без изменения существующего кода, который использует эти интерфейсы.

````
using System;
using System.Collections.Generic;

// ========== ИНТЕРФЕЙСЫ ==========
public interface IVehicle
{
    void Start();
    void Stop();
    int GetCurrentSpeed();
    string GetInfo();
}

public interface IElectricVehicle
{
    void Charge();
    int GetBatteryLevel();
}

public interface IRaceVehicle
{
    void ActivateSportMode();
    int GetMaxSpeed();
}

// ========== АБСТРАКТНЫЙ БАЗОВЫЙ КЛАСС ==========
public abstract class Vehicle : IVehicle
{
    // Свойства
    public string Make { get; set; }
    public string Model { get; set; }
    public int Year { get; set; }
    protected int currentSpeed;
    protected bool isRunning;
    
    // Конструктор
    public Vehicle(string make, string model, int year)
    {
        Make = make;
        Model = model;
        Year = year;
        currentSpeed = 0;
        isRunning = false;
    }
    
    // Реализация интерфейса IVehicle
    public virtual void Start()
    {
        isRunning = true;
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} ({Year}) запущен.");
    }
    
    public virtual void Stop()
    {
        currentSpeed = 0;
        isRunning = false;
        Console.WriteLine($"{Make} {Model} остановлен.");
    }
    
    public virtual int GetCurrentSpeed()
    {
        return currentSpeed;
    }
    
    public virtual string GetInfo()
    {
        return $"Производитель: {Make}, Модель: {Model}, Год: {Year}";
    }
    
    // Абстрактный метод (должен быть реализован в производных классах)
    public abstract void Honk();
    
    // Виртуальный метод для отображения состояния
    public virtual void ShowStatus()
    {
        string status = isRunning ? "запущен" : "остановлен";
        Console.WriteLine($"{Make} {Model}: {status}, скорость: {currentSpeed} км/ч");
    }
}

// ========== КЛАСС CAR ==========
public class Car : Vehicle
{
    public int Doors { get; set; }
    public string BodyType { get; set; }
    
    public Car(string make, string model, int year, int doors, string bodyType) 
        : base(make, model, year)
    {
        Doors = doors;
        BodyType = bodyType;
    }
    
    public override void Start()
    {
        base.Start();
        Console.WriteLine($"  Автомобиль {Make} {Model} готов к поездке.");
    }
    
    public override void Stop()
    {
        base.Stop();
        Console.WriteLine($"  Автомобиль {Make} {Model} полностью остановлен, ручник включен.");
    }
    
    public override void Honk()
    {
        Console.WriteLine($"{Make} {Model} сигналит: Бип-бип!");
    }
    
    public void Accelerate(int amount)
    {
        if (isRunning)
        {
            currentSpeed += amount;
            Console.WriteLine($"Автомобиль {Make} {Model} разгоняется до {currentSpeed} км/ч.");
        }
        else
        {
            Console.WriteLine($"Сначала нужно запустить двигатель {Make} {Model}!");
        }
    }
    
    public void Brake(int amount)
    {
        if (currentSpeed > 0)
        {
            currentSpeed = Math.Max(0, currentSpeed - amount);
            Console.WriteLine($"Автомобиль {Make} {Model} тормозит до {currentSpeed} км/ч.");
        }
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Дверей: {Doors}, Кузов: {BodyType}";
    }
}

// ========== КЛАСС MOTORCYCLE ==========
public class Motorcycle : Vehicle
{
    public string EngineType { get; set; }
    public bool HasSidecar { get; set; }
    
    public Motorcycle(string make, string model, int year, string engineType, bool hasSidecar) 
        : base(make, model, year)
    {
        EngineType = engineType;
        HasSidecar = hasSidecar;
    }
    
    public override void Start()
    {
        base.Start();
        Console.WriteLine($"  Мотоцикл {Make} {Model} готов к поездке.");
    }
    
    public override void Stop()
    {
        base.Stop();
        Console.WriteLine($"  Мотоцикл {Make} {Model} полностью остановлен.");
    }
    
    public override void Honk()
    {
        Console.WriteLine($"{Make} {Model} сигналит: Бип!");
    }
    
    public void RevEngine()
    {
        if (isRunning)
        {
            Console.WriteLine($"Мотоцикл {Make} {Model} рычит! BRRRAAAP!");
        }
        else
        {
            Console.WriteLine($"Сначала нужно запустить двигатель {Make} {Model}!");
        }
    }
    
    public void Wheelie()
    {
        if (isRunning && currentSpeed > 20)
        {
            Console.WriteLine($"Мотоцикл {Make} {Model} едет на заднем колесе!");
        }
        else
        {
            Console.WriteLine($"Невозможно сделать вили на мотоцикле {Make} {Model} (нужна скорость > 20 км/ч)");
        }
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Тип двигателя: {EngineType}, Коляска: {(HasSidecar ? "Есть" : "Нет")}";
    }
}

// ========== КЛАСС TRUCK (ДОПОЛНИТЕЛЬНЫЙ) ==========
public class Truck : Vehicle, IElectricVehicle
{
    public double LoadCapacity { get; set; }
    public int Axles { get; set; }
    private int batteryLevel;
    
    public Truck(string make, string model, int year, double loadCapacity, int axles) 
        : base(make, model, year)
    {
        LoadCapacity = loadCapacity;
        Axles = axles;
        batteryLevel = 100;
    }
    
    public override void Start()
    {
        base.Start();
        Console.WriteLine($"  Грузовик {Make} {Model} готов к работе.");
    }
    
    public override void Stop()
    {
        base.Stop();
        Console.WriteLine($"  Грузовик {Make} {Model} остановлен, пневмотормоза активированы.");
    }
    
    public override void Honk()
    {
        Console.WriteLine($"{Make} {Model} гудит: БИИИП-БИИИП!");
    }
    
    public void LoadCargo(double weight)
    {
        if (weight <= LoadCapacity)
        {
            Console.WriteLine($"Грузовик {Make} {Model} загружен на {weight} тонн.");
        }
        else
        {
            Console.WriteLine($"Грузовик {Make} {Model} перегружен! Максимум {LoadCapacity} тонн.");
        }
    }
    
    // Реализация IElectricVehicle
    public void Charge()
    {
        batteryLevel = 100;
        Console.WriteLine($"Электрогрузовик {Make} {Model} заряжен до {batteryLevel}%");
    }
    
    public int GetBatteryLevel()
    {
        return batteryLevel;
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Грузоподъемность: {LoadCapacity} т, Оси: {Axles}";
    }
}

// ========== ДЕМОНСТРАЦИЯ ДЛЯ VEHICLE ==========
Console.WriteLine("========== ИНТЕРФЕЙСЫ: ТРАНСПОРТНЫЕ СРЕДСТВА ==========\n");

// Создание объектов через интерфейс
Console.WriteLine("=== 1. РАБОТА ЧЕРЕЗ ИНТЕРФЕЙС IVehicle ===\n");

IVehicle myCar = new Car("Toyota", "Camry", 2022, 4, "Седан");
myCar.Start();
((Car)myCar).Accelerate(50);
Console.WriteLine($"Текущая скорость: {myCar.GetCurrentSpeed()} км/ч");
((Car)myCar).Brake(20);
myCar.Stop();

Console.WriteLine();

IVehicle myMotorcycle = new Motorcycle("Harley-Davidson", "Street Glide", 2023, "V-Twin", false);
myMotorcycle.Start();
((Motorcycle)myMotorcycle).RevEngine();
((Motorcycle)myMotorcycle).Wheelie();
myMotorcycle.Stop();

Console.WriteLine();

IVehicle myTruck = new Truck("Volvo", "FH16", 2024, 25.5, 3);
myTruck.Start();
((Truck)myTruck).LoadCargo(20);
((IElectricVehicle)myTruck).Charge();
myTruck.Stop();

// Демонстрация полиморфизма через интерфейс
Console.WriteLine("\n=== 2. ПОЛИМОРФИЗМ ЧЕРЕЗ ИНТЕРФЕЙС ===\n");

List<IVehicle> vehicles = new List<IVehicle>
{
    new Car("BMW", "X5", 2023, 5, "Внедорожник"),
    new Motorcycle("Yamaha", "R1", 2024, "Inline-4", false),
    new Truck("MAN", "TGX", 2022, 30, 4)
};

foreach (var vehicle in vehicles)
{
    Console.WriteLine(vehicle.GetInfo());
    vehicle.Start();
    Console.WriteLine($"  Скорость: {vehicle.GetCurrentSpeed()} км/ч");
    vehicle.Stop();
    Console.WriteLine();
}

// Демонстрация метода Honk (абстрактный метод)
Console.WriteLine("=== 3. ДЕМОНСТРАЦИЯ АБСТРАКТНОГО МЕТОДА HONK ===\n");

Vehicle[] vehiclesArray = new Vehicle[]
{
    new Car("Ford", "Focus", 2022, 4, "Хэтчбек"),
    new Motorcycle("Ducati", "Panigale", 2023, "V4", false),
    new Truck("Scania", "R500", 2023, 28, 3)
};

foreach (var vehicle in vehiclesArray)
{
    vehicle.ShowStatus();
    vehicle.Honk();
    Console.WriteLine();
}

Console.WriteLine("========== ДЕМОНСТРАЦИЯ VEHICLE ЗАВЕРШЕНА ==========\n");
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также наследование с применением интерфейсов.

````
// ========== ИНТЕРФЕЙСЫ ==========
public interface IAnimal
{
    void MakeSound();
    void Move();
    string GetInfo();
    void Eat();
}

public interface IPet
{
    string OwnerName { get; set; }
    void Play();
    void Feed();
    void ShowAffection();
}

public interface IWild
{
    string Habitat { get; set; }
    void Hunt();
    void DefendTerritory();
}

public interface ITrainable
{
    bool IsTrained { get; set; }
    void Train();
    void PerformTrick(string trick);
}

// ========== АБСТРАКТНЫЙ БАЗОВЫЙ КЛАСС ==========
public abstract class Animal : IAnimal
{
    // Свойства
    public string Name { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }
    protected bool isAlive;
    protected int energyLevel;
    
    // Конструктор
    public Animal(string name, int age, double weight)
    {
        Name = name;
        Age = age;
        Weight = weight;
        isAlive = true;
        energyLevel = 100;
    }
    
    // Реализация интерфейса IAnimal
    public abstract void MakeSound();
    public abstract void Move();
    
    public virtual string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age} лет, Вес: {Weight} кг, Энергия: {energyLevel}%";
    }
    
    public virtual void Eat()
    {
        energyLevel = Math.Min(100, energyLevel + 20);
        Console.WriteLine($"{Name} поел. Энергия: {energyLevel}%");
    }
    
    // Виртуальные методы
    public virtual void Sleep()
    {
        energyLevel = 100;
        Console.WriteLine($"{Name} выспался. Энергия восстановлена до {energyLevel}%");
    }
    
    public virtual void ShowStatus()
    {
        string status = isAlive ? "жив" : "мертв";
        Console.WriteLine($"{Name}: {status}, энергия: {energyLevel}%");
    }
    
    // Абстрактный метод
    public abstract void Breathe();
}

// ========== КЛАСС DOG ==========
public class Dog : Animal, IPet, ITrainable
{
    public string Breed { get; set; }
    public string OwnerName { get; set; }
    public bool IsTrained { get; set; }
    private List<string> tricks = new List<string>();
    
    public Dog(string name, int age, double weight, string breed, string ownerName) 
        : base(name, age, weight)
    {
        Breed = breed;
        OwnerName = ownerName;
        IsTrained = false;
    }
    
    // Реализация IAnimal
    public override void MakeSound()
    {
        Console.WriteLine($"Собака {Name} лает: Гав-гав!");
    }
    
    public override void Move()
    {
        if (energyLevel >= 10)
        {
            energyLevel -= 10;
            Console.WriteLine($"Собака {Name} бегает на четырех лапах. Энергия: {energyLevel}%");
        }
        else
        {
            Console.WriteLine($"Собака {Name} слишком устала, чтобы двигаться!");
        }
    }
    
    public override void Eat()
    {
        base.Eat();
        Console.WriteLine($"  {Name} ест корм с большим аппетитом!");
    }
    
    public override void Breathe()
    {
        Console.WriteLine($"Собака {Name} дышит, высунув язык.");
    }
    
    // Реализация IPet
    public void Play()
    {
        if (energyLevel >= 15)
        {
            energyLevel -= 15;
            Console.WriteLine($"{Name} играет с мячом! Энергия: {energyLevel}%");
        }
        else
        {
            Console.WriteLine($"{Name} слишком устал для игр!");
        }
    }
    
    public void Feed()
    {
        Eat();
    }
    
    public void ShowAffection()
    {
        Console.WriteLine($"{Name} виляет хвостом и лижет руку хозяину {OwnerName}!");
    }
    
    // Реализация ITrainable
    public void Train()
    {
        if (!IsTrained)
        {
            IsTrained = true;
            Console.WriteLine($"{Name} прошел обучение! Теперь он знает команды.");
        }
        else
        {
            Console.WriteLine($"{Name} уже обучен.");
        }
    }
    
    public void PerformTrick(string trick)
    {
        if (IsTrained)
        {
            tricks.Add(trick);
            Console.WriteLine($"{Name} выполняет трюк: {trick}!");
        }
        else
        {
            Console.WriteLine($"{Name} еще не обучен. Нужно сначала обучить собаку!");
        }
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Порода: {Breed}, Хозяин: {OwnerName}, Обучен: {(IsTrained ? "Да" : "Нет")}";
    }
}

// ========== КЛАСС CAT ==========
public class Cat : Animal, IPet
{
    public string Color { get; set; }
    public bool IsIndoor { get; set; }
    public string OwnerName { get; set; }
    
    public Cat(string name, int age, double weight, string color, bool isIndoor, string ownerName) 
        : base(name, age, weight)
    {
        Color = color;           // Указан цвет "серый"
        IsIndoor = isIndoor;
        OwnerName = ownerName;
    }
    
    // Реализация IAnimal
    public override void MakeSound()
    {
        Console.WriteLine($"Кошка {Name} мяукает: Мяу-мяу!");
    }
    
    public override void Move()
    {
        if (energyLevel >= 5)
        {
            energyLevel -= 5;
            Console.WriteLine($"Кошка {Name} грациозно крадется. Энергия: {energyLevel}%");
        }
        else
        {
            Console.WriteLine($"Кошка {Name} слишком устала, чтобы двигаться!");
        }
    }
    
    public override void Eat()
    {
        base.Eat();
        Console.WriteLine($"  {Name} изящно пробует рыбу.");
    }
    
    public override void Breathe()
    {
        Console.WriteLine($"Кошка {Name} тихо дышит, свернувшись клубочком.");
    }
    
    // Реализация IPet
    public void Play()
    {
        if (energyLevel >= 10)
        {
            energyLevel -= 10;
            Console.WriteLine($"{Name} играет с клубком ниток! Энергия: {energyLevel}%");
        }
        else
        {
            Console.WriteLine($"{Name} слишком устала для игр!");
        }
    }
    
    public void Feed()
    {
        Eat();
    }
    
    public void ShowAffection()
    {
        Console.WriteLine($"{Name} трется о ноги хозяина {OwnerName} и мурлычет: Мур-мур-мур...");
    }
    
    // Специфичный метод
    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет от удовольствия!");
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Окрас: {Color}, Домашняя: {(IsIndoor ? "Да" : "Нет")}, Хозяин: {OwnerName}";
    }
}

// ========== КЛАСС BIRD ==========
public class Bird : Animal, IWild
{
    public double WingSpan { get; set; }
    public bool CanFly { get; set; }
    public string Habitat { get; set; }
    
    public Bird(string name, int age, double weight, double wingSpan, bool canFly, string habitat) 
        : base(name, age, weight)
    {
        WingSpan = wingSpan;
        CanFly = canFly;
        Habitat = habitat;
    }
    
    // Реализация IAnimal
    public override void MakeSound()
    {
        Console.WriteLine($"Птица {Name} чирикает: Чик-чирик!");
    }
    
    public override void Move()
    {
        if (CanFly && energyLevel >= 15)
        {
            energyLevel -= 15;
            Console.WriteLine($"Птица {Name} летит с размахом крыльев {WingSpan} м. Энергия: {energyLevel}%");
        }
        else if (!CanFly && energyLevel >= 5)
        {
            energyLevel -= 5;
            Console.WriteLine($"Птица {Name} ходит по земле. Энергия: {energyLevel}%");
        }
        else
        {
            Console.WriteLine($"Птица {Name} слишком устала, чтобы двигаться!");
        }
    }
    
    public override void Eat()
    {
        base.Eat();
        Console.WriteLine($"  {Name} клюет зерна.");
    }
    
    public override void Breathe()
    {
        Console.WriteLine($"Птица {Name} дышит легко и быстро.");
    }
    
    // Реализация IWild
    public void Hunt()
    {
        if (energyLevel >= 20)
        {
            energyLevel -= 20;
            Console.WriteLine($"Птица {Name} охотится в {Habitat}!");
        }
        else
        {
            Console.WriteLine($"{Name} слишком устала для охоты!");
        }
    }
    
    public void DefendTerritory()
    {
        Console.WriteLine($"Птица {Name} защищает свою территорию в {Habitat}!");
    }
    
    // Специфичный метод
    public void BuildNest()
    {
        Console.WriteLine($"{Name} строит гнездо в {Habitat}.");
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Размах крыльев: {WingSpan} м, Летает: {(CanFly ? "Да" : "Нет")}, Среда обитания: {Habitat}";
    }
}

// ========== КЛАСС FISH ==========
public class Fish : Animal, IWild
{
    public string WaterType { get; set; }
    public double MaxDepth { get; set; }
    public string Habitat { get; set; }
    
    public Fish(string name, int age, double weight, string waterType, double maxDepth, string habitat) 
        : base(name, age, weight)
    {
        WaterType = waterType;
        MaxDepth = maxDepth;
        Habitat = habitat;
    }
    
    // Реализация IAnimal
    public override void MakeSound()
    {
        Console.WriteLine($"Рыба {Name} издает булькающие звуки: Буль-буль!");
    }
    
    public override void Move()
    {
        if (energyLevel >= 5)
        {
            energyLevel -= 5;
            Console.WriteLine($"Рыба {Name} плавает в {WaterType} воде. Энергия: {energyLevel}%");
        }
        else
        {
            Console.WriteLine($"Рыба {Name} слишком устала, чтобы плавать!");
        }
    }
    
    public override void Eat()
    {
        base.Eat();
        Console.WriteLine($"  {Name} проглатывает планктон.");
    }
    
    public override void Breathe()
    {
        Console.WriteLine($"Рыба {Name} дышит жабрами под водой.");
    }
    
    // Реализация IWild
    public void Hunt()
    {
        if (energyLevel >= 15)
        {
            energyLevel -= 15;
            Console.WriteLine($"Рыба {Name} охотится на мелкую рыбу в {Habitat}!");
        }
        else
        {
            Console.WriteLine($"{Name} слишком устала для охоты!");
        }
    }
    
    public void DefendTerritory()
    {
        Console.WriteLine($"Рыба {Name} защищает свою территорию в {Habitat} на глубине {MaxDepth} м!");
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Тип воды: {WaterType}, Макс. глубина: {MaxDepth} м, Среда обитания: {Habitat}";
    }
}

// ========== ДЕМОНСТРАЦИЯ ДЛЯ ANIMAL ==========
class AnimalDemo
{
    public static void Run()
    {
        Console.WriteLine("========== ИНТЕРФЕЙСЫ: ЖИВОТНЫЕ ==========\n");
        
        // Создание объектов
        Console.WriteLine("=== 1. СОЗДАНИЕ ОБЪЕКТОВ ===\n");
        
        // Используем указанные параметры: кошка Муся, цвет серый
        Dog dog = new Dog("Рекс", 3, 25.5, "Овчарка", "Иван");
        Cat cat = new Cat("Муся", 2, 4.2, "серый", true, "Элина");
        Bird bird = new Bird("Кеша", 1, 0.35, 0.5, true, "лес");
        Fish fish = new Fish("Немо", 2, 0.15, "Соленая", 50, "океан");
        
        Console.WriteLine(dog.GetInfo());
        Console.WriteLine(cat.GetInfo());
        Console.WriteLine(bird.GetInfo());
        Console.WriteLine(fish.GetInfo());
        
        // Демонстрация работы через интерфейс IAnimal
        Console.WriteLine("\n=== 2. РАБОТА ЧЕРЕЗ ИНТЕРФЕЙС IAnimal ===\n");
        
        IAnimal myCat = cat;
        myCat.MakeSound();
        myCat.Move();
        myCat.Eat();
        Console.WriteLine(myCat.GetInfo());
        
        IAnimal myDog = dog;
        myDog.MakeSound();
        myDog.Move();
        myDog.Eat();
        
        // Демонстрация работы через интерфейс IPet
        Console.WriteLine("\n=== 3. РАБОТА ЧЕРЕЗ ИНТЕРФЕЙС IPet ===\n");
        
        IPet petDog = dog;
        petDog.Play();
        petDog.ShowAffection();
        petDog.Feed();
        
        IPet petCat = cat;
        petCat.Play();
        petCat.ShowAffection();
        petCat.Feed();
        
        // Демонстрация работы через интерфейс ITrainable
        Console.WriteLine("\n=== 4. РАБОТА ЧЕРЕЗ ИНТЕРФЕЙС ITrainable ===\n");
        
        ITrainable trainableDog = dog;
        trainableDog.Train();
        trainableDog.PerformTrick("Сидеть");
        trainableDog.PerformTrick("Лежать");
        
        // Демонстрация работы через интерфейс IWild
        Console.WriteLine("\n=== 5. РАБОТА ЧЕРЕЗ ИНТЕРФЕЙС IWild ===\n");
        
        IWild wildBird = bird;
        wildBird.Hunt();
        wildBird.DefendTerritory();
        
        IWild wildFish = fish;
        wildFish.Hunt();
        wildFish.DefendTerritory();
        
        // Демонстрация полиморфизма через интерфейсы
        Console.WriteLine("\n=== 6. ПОЛИМОРФИЗМ ЧЕРЕЗ ИНТЕРФЕЙСЫ ===\n");
        
        List<IAnimal> animals = new List<IAnimal>
        {
            new Dog("Шарик", 4, 20, "Дворняга", "Петр"),
            new Cat("Васька", 3, 3.8, "черный", false, "Анна"),
            new Bird("Воробей", 1, 0.05, 0.2, true, "город"),
            new Fish("Золотая", 1, 0.1, "Пресная", 2, "пруд")
        };
        
        foreach (var animal in animals)
        {
            Console.WriteLine(animal.GetInfo());
            animal.MakeSound();
            animal.Move();
            Console.WriteLine();
        }
        
        // Демонстрация абстрактного метода Breathe
        Console.WriteLine("=== 7. ДЕМОНСТРАЦИЯ АБСТРАКТНОГО МЕТОДА BREATHE ===\n");
        
        Animal[] animalsArray = new Animal[]
        {
            dog,
            cat,
            bird,
            fish
        };
        
        foreach (var animal in animalsArray)
        {
            animal.ShowStatus();
            animal.Breathe();
            Console.WriteLine();
        }
        
        // Демонстрация специфичных методов
        Console.WriteLine("=== 8. СПЕЦИФИЧНЫЕ МЕТОДЫ ===\n");
        
        // Кошка Муся мурлычет
        cat.Purr();
        
        // Собака выполняет трюки
        dog.PerformTrick("Дай лапу");
        
        // Птица строит гнездо
        bird.BuildNest();
        
        // Рыба плавает
        fish.Move();
        
        Console.WriteLine("\n========== ДЕМОНСТРАЦИЯ ANIMAL ЗАВЕРШЕНА ==========");
    }
}

// ========== ЗАПУСК ВСЕХ ДЕМОНСТРАЦИЙ ==========
Console.WriteLine("\n" + new string('=', 100) + "\n");
AnimalDemo.Run();
````
