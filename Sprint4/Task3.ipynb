```
using System;
using System.Collections.Generic;
using System.Linq;

// ========== ИНТЕРФЕЙСЫ ==========
public interface IPerson
{
    string Name { get; }
    int Age { get; }
    string Gender { get; }
    string GetInfo();
    void SayHello();
    void Work();
    void Study();
    void Relax();
}

public interface IWorker
{
    string Company { get; }
    decimal Salary { get; }
    void Work();
    void TakeBreak();
    void ReceiveBonus(decimal amount);
}

public interface IStudent
{
    string University { get; }
    int Course { get; }
    double GetAverageGrade();
    void AddGrade(int grade);
    void Study();
    void TakeExam();
}

public interface ITeacher
{
    string Subject { get; }
    string AcademicDegree { get; }
    void Teach();
    void GradeStudent(Student student, int grade);
    void ConductExam();
}

public interface IResearcher
{
    int PublicationsCount { get; }
    void DoResearch();
    void PublishArticle(string title);
}

public interface IMentor
{
    int StudentsMentored { get; }
    void MentorStudent(Student student);
    void GiveAdvice();
}

public interface IConferenceSpeaker
{
    int ConferencesAttended { get; }
    void SpeakAtConference(string conferenceName);
    void PresentResearch();
}

// ========== БАЗОВЫЙ КЛАСС PERSON ==========
public abstract class Person : IPerson
{
    // Приватные поля
    private string _name;
    private int _age;
    private string _gender;
    private string _email;
    private string _phone;
    
    // Свойства
    public string Name => _name;
    public int Age => _age;
    public string Gender => _gender;
    public string Email => _email;
    public string Phone => _phone;
    
    // Статический атрибут
    public static int TotalPeople { get; private set; } = 0;
    protected static List<Person> allPeople = new List<Person>();
    
    // Конструкторы
    public Person(string name, int age, string gender)
    {
        _name = name;
        _age = age;
        _gender = gender;
        TotalPeople++;
        allPeople.Add(this);
    }
    
    public Person(string name, int age, string gender, string email) : this(name, age, gender)
    {
        _email = email;
    }
    
    public Person(string name, int age, string gender, string email, string phone) : this(name, age, gender, email)
    {
        _phone = phone;
    }
    
    // Реализация IPerson
    public virtual string GetInfo()
    {
        string emailInfo = string.IsNullOrEmpty(Email) ? "не указан" : Email;
        return $"Имя: {Name}, Возраст: {Age}, Пол: {Gender}, Email: {emailInfo}";
    }
    
    public virtual void SayHello()
    {
        Console.WriteLine($"Привет! Меня зовут {Name}.");
    }
    
    public abstract void Work();
    public abstract void Study();
    public abstract void Relax();
    
    // Дополнительные методы
    public virtual void ShowStatus()
    {
        Console.WriteLine($"{Name}: {GetType().Name}, возраст {Age} лет");
    }
    
    public static void DisplayTotalPeople()
    {
        Console.WriteLine($"\nВсего создано людей: {TotalPeople}");
    }
    
    public static Person FindPerson(string name)
    {
        return allPeople.Find(p => p.Name.Equals(name, StringComparison.OrdinalIgnoreCase));
    }
}

// ========== КЛАСС STUDENT ==========
public class Student : Person, IStudent
{
    private string _university;
    private int _course;
    private string _major;
    private List<int> _grades = new List<int>();
    private double _scholarship;
    
    public string University => _university;
    public int Course => _course;
    public string Major => _major;
    public double Scholarship => _scholarship;
    
    public Student(string name, int age, string gender, string university, int course, string major) 
        : base(name, age, gender)
    {
        _university = university;
        _course = course;
        _major = major;
    }
    
    public Student(string name, int age, string gender, string university, int course, string major, string email) 
        : base(name, age, gender, email)
    {
        _university = university;
        _course = course;
        _major = major;
    }
    
    // Реализация IStudent
    public double GetAverageGrade()
    {
        if (_grades.Count == 0) return 0;
        return _grades.Average();
    }
    
    public void AddGrade(int grade)
    {
        if (grade >= 2 && grade <= 5)
        {
            _grades.Add(grade);
            Console.WriteLine($"Студенту {Name} добавлена оценка: {grade}");
            UpdateScholarship();
        }
        else
        {
            Console.WriteLine("Оценка должна быть от 2 до 5!");
        }
    }
    
    public void TakeExam()
    {
        Console.WriteLine($"{Name} сдает экзамен по специальности {Major}");
    }
    
    private void UpdateScholarship()
    {
        double avg = GetAverageGrade();
        if (avg >= 4.8)
            _scholarship = 5000;
        else if (avg >= 4.5)
            _scholarship = 4000;
        else if (avg >= 4.0)
            _scholarship = 3000;
        else
            _scholarship = 0;
    }
    
    public string GetGradeStatus()
    {
        double avg = GetAverageGrade();
        return avg switch
        {
            >= 4.8 => "Отлично",
            >= 4.0 => "Хорошо",
            >= 3.0 => "Удовлетворительно",
            _ => "Требуется улучшение"
        };
    }
    
    // Переопределение методов Person
    public override void SayHello()
    {
        Console.WriteLine($"Привет! Я студент {Name}, учусь на {Course}-м курсе в {University} по специальности '{Major}'.");
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Университет: {University}, Курс: {Course}, Специальность: {Major}, " +
               $"Средний балл: {GetAverageGrade():F2}, Стипендия: {Scholarship:C}";
    }
    
    public override void Work()
    {
        Console.WriteLine($"{Name} подрабатывает во время учебы.");
    }
    
    public override void Study()
    {
        Console.WriteLine($"{Name} усердно учится на {Course} курсе по специальности {Major}.");
        if (GetAverageGrade() >= 4.5)
            Console.WriteLine($"  Отличная успеваемость! Средний балл: {GetAverageGrade():F2}");
    }
    
    public override void Relax()
    {
        Console.WriteLine($"{Name} отдыхает с друзьями в студенческом кампусе.");
    }
}

// ========== КЛАСС EMPLOYEE ==========
public class Employee : Person, IWorker
{
    private string _company;
    private decimal _salary;
    private string _position;
    private int _experience;
    private List<string> _projects = new List<string>();
    
    public string Company => _company;
    public decimal Salary => _salary;
    public string Position => _position;
    public int Experience => _experience;
    public IReadOnlyList<string> Projects => _projects.AsReadOnly();
    
    public Employee(string name, int age, string gender, string company, decimal salary, string position, int experience) 
        : base(name, age, gender)
    {
        _company = company;
        _salary = salary;
        _position = position;
        _experience = experience;
    }
    
    // Реализация IWorker
    public void TakeBreak()
    {
        Console.WriteLine($"{Name} делает перерыв на обед.");
    }
    
    public void ReceiveBonus(decimal amount)
    {
        if (amount > 0)
        {
            _salary += amount;
            Console.WriteLine($"{Name} получил бонус {amount:C}. Новая зарплата: {Salary:C}");
        }
    }
    
    public void AddProject(string project)
    {
        _projects.Add(project);
        Console.WriteLine($"{Name} добавлен в проект: {project}");
    }
    
    public decimal CalculateBonus()
    {
        return Salary * (Experience / 100m);
    }
    
    // Переопределение методов Person
    public override void SayHello()
    {
        Console.WriteLine($"Здравствуйте! Я {Name}, работаю в {Company} в должности {Position}.");
    }
    
    public override string GetInfo()
    {
        string projectsStr = _projects.Count > 0 ? string.Join(", ", _projects) : "нет проектов";
        return base.GetInfo() + $", Компания: {Company}, Должность: {Position}, " +
               $"Стаж: {Experience} лет, Зарплата: {Salary:C}, Проекты: {projectsStr}";
    }
    
    public override void Work()
    {
        Console.WriteLine($"{Name} работает в {Company} над проектами: {(_projects.Count > 0 ? string.Join(", ", _projects) : "новые проекты")}");
    }
    
    public override void Study()
    {
        Console.WriteLine($"{Name} повышает квалификацию на курсах.");
    }
    
    public override void Relax()
    {
        Console.WriteLine($"{Name} отдыхает после рабочего дня.");
    }
}

// ========== КЛАСС TEACHER (наследует от Employee) ==========
public class Teacher : Employee, ITeacher
{
    private string _subject;
    private string _academicDegree;
    private List<Student> _students = new List<Student>();
    
    public string Subject => _subject;
    public string AcademicDegree => _academicDegree;
    public int StudentsCount => _students.Count;
    
    public Teacher(string name, int age, string gender, string company, decimal salary, 
                   string position, int experience, string subject, string academicDegree) 
        : base(name, age, gender, company, salary, position, experience)
    {
        _subject = subject;
        _academicDegree = academicDegree;
    }
    
    // Реализация ITeacher
    public void Teach()
    {
        Console.WriteLine($"{Name} преподает {Subject} студентам.");
    }
    
    public void GradeStudent(Student student, int grade)
    {
        if (_students.Contains(student))
        {
            student.AddGrade(grade);
            Console.WriteLine($"Преподаватель {Name} выставил оценку {grade} студенту {student.Name} по предмету {Subject}");
        }
        else
        {
            Console.WriteLine($"Студент {student.Name} не учится у преподавателя {Name}");
        }
    }
    
    public void ConductExam()
    {
        Console.WriteLine($"\n{Name} проводит экзамен по предмету {Subject}");
        Random rand = new Random();
        foreach (var student in _students)
        {
            int grade = rand.Next(3, 6);
            GradeStudent(student, grade);
        }
    }
    
    public void AddStudent(Student student)
    {
        if (!_students.Contains(student))
        {
            _students.Add(student);
            Console.WriteLine($"Студент {student.Name} добавлен к преподавателю {Name}");
        }
    }
    
    // Переопределение методов
    public override void SayHello()
    {
        Console.WriteLine($"Здравствуйте! Я преподаватель {Name}, {AcademicDegree}, веду предмет '{Subject}'.");
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Предмет: {Subject}, Ученая степень: {AcademicDegree}, Студентов: {StudentsCount}";
    }
    
    public override void Work()
    {
        base.Work();
        Teach();
    }
    
    public override void Study()
    {
        Console.WriteLine($"{Name} повышает квалификацию по предмету {Subject}.");
    }
}

// ========== КЛАСС PROFESSOR (множественное наследование через интерфейсы) ==========
public class Professor : Teacher, IResearcher, IMentor, IConferenceSpeaker
{
    private string _researchField;
    private string _laboratoryName;
    private List<string> _publications = new List<string>();
    private List<Student> _mentoredStudents = new List<Student>();
    private List<string> _conferences = new List<string>();
    
    public string ResearchField => _researchField;
    public string LaboratoryName => _laboratoryName;
    
    // IResearcher
    public int PublicationsCount => _publications.Count;
    
    // IMentor
    public int StudentsMentored => _mentoredStudents.Count;
    
    // IConferenceSpeaker
    public int ConferencesAttended => _conferences.Count;
    
    public Professor(string name, int age, string gender, string company, decimal salary,
                     string position, int experience, string subject, string academicDegree,
                     string researchField, string laboratoryName) 
        : base(name, age, gender, company, salary, position, experience, subject, academicDegree)
    {
        _researchField = researchField;
        _laboratoryName = laboratoryName;
    }
    
    // Реализация IResearcher
    public void DoResearch()
    {
        Console.WriteLine($"Профессор {Name} проводит исследования в области {ResearchField} в лаборатории {LaboratoryName}");
    }
    
    public void PublishArticle(string title)
    {
        _publications.Add(title);
        Console.WriteLine($"Профессор {Name} опубликовал статью: '{title}' (Всего: {PublicationsCount})");
    }
    
    // Реализация IMentor
    public void MentorStudent(Student student)
    {
        if (!_mentoredStudents.Contains(student))
        {
            _mentoredStudents.Add(student);
            Console.WriteLine($"Профессор {Name} взял шефство над студентом {student.Name}");
        }
    }
    
    public void GiveAdvice()
    {
        Console.WriteLine($"Профессор {Name} дает ценный совет по научной карьере студентам");
    }
    
    // Реализация IConferenceSpeaker
    public void SpeakAtConference(string conferenceName)
    {
        _conferences.Add(conferenceName);
        Console.WriteLine($"Профессор {Name} выступает на конференции: {conferenceName}");
    }
    
    public void PresentResearch()
    {
        Console.WriteLine($"Профессор {Name} представляет результаты исследований в {ResearchField}");
    }
    
    // Переопределение методов
    public override void SayHello()
    {
        Console.WriteLine($"Здравствуйте! Я профессор {Name}, {AcademicDegree}, заведующий лабораторией {LaboratoryName}.");
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Лаборатория: {LaboratoryName}, Область исследований: {ResearchField}, " +
               $"Публикаций: {PublicationsCount}, Студентов под наставничеством: {StudentsMentored}, " +
               $"Конференций: {ConferencesAttended}";
    }
    
    public override void Work()
    {
        base.Work();
        DoResearch();
    }
}

// ========== СЕРВИСНЫЕ КЛАССЫ ДЛЯ УПРАВЛЕНИЯ ЗАВИСИМОСТЯМИ ==========

// Сервис для работы с людьми
public class PersonService
{
    private readonly IPerson _person;
    
    public PersonService(IPerson person)
    {
        _person = person;
    }
    
    public void DemonstratePerson()
    {
        Console.WriteLine($"\n--- Демонстрация: {_person.GetInfo()} ---");
        _person.SayHello();
        _person.Work();
        _person.Study();
        _person.Relax();
    }
}

// Сервис для работы со студентами
public class StudentService
{
    private readonly IStudent _student;
    private readonly string _studentName;
    
    public StudentService(IStudent student, string studentName)
    {
        _student = student;
        _studentName = studentName;
    }
    
    public void ShowAcademicPerformance()
    {
        Console.WriteLine($"\n--- Успеваемость студента {_studentName} ---");
        Console.WriteLine($"Университет: {_student.University}");
        Console.WriteLine($"Курс: {_student.Course}");
        Console.WriteLine($"Средний балл: {_student.GetAverageGrade():F2}");
        _student.Study();
        _student.TakeExam();
    }
    
    public void AddGrades(params int[] grades)
    {
        foreach (var grade in grades)
        {
            _student.AddGrade(grade);
        }
    }
}

// Сервис для работы с работниками
public class WorkerService
{
    private readonly IWorker _worker;
    private readonly string _workerName;
    
    public WorkerService(IWorker worker, string workerName)
    {
        _worker = worker;
        _workerName = workerName;
    }
    
    public void ShowWorkInfo()
    {
        Console.WriteLine($"\n--- Информация о работнике {_workerName} ---");
        Console.WriteLine($"Компания: {_worker.Company}");
        Console.WriteLine($"Зарплата: {_worker.Salary:C}");
        _worker.Work();
        _worker.TakeBreak();
    }
    
    public void GiveBonus(decimal amount)
    {
        _worker.ReceiveBonus(amount);
    }
}

// Сервис для работы с преподавателями
public class TeacherService
{
    private readonly ITeacher _teacher;
    private readonly string _teacherName;
    
    public TeacherService(ITeacher teacher, string teacherName)
    {
        _teacher = teacher;
        _teacherName = teacherName;
    }
    
    public void DemonstrateTeaching()
    {
        Console.WriteLine($"\n--- Преподавательская деятельность {_teacherName} ---");
        Console.WriteLine($"Предмет: {_teacher.Subject}");
        Console.WriteLine($"Ученая степень: {_teacher.AcademicDegree}");
        _teacher.Teach();
        _teacher.ConductExam();
    }
}

// Сервис для работы с исследователями
public class ResearchService
{
    private readonly IResearcher _researcher;
    private readonly string _researcherName;
    
    public ResearchService(IResearcher researcher, string researcherName)
    {
        _researcher = researcher;
        _researcherName = researcherName;
    }
    
    public void ShowResearchActivity()
    {
        Console.WriteLine($"\n--- Научная деятельность {_researcherName} ---");
        Console.WriteLine($"Количество публикаций: {_researcher.PublicationsCount}");
        _researcher.DoResearch();
        _researcher.PublishArticle("Новое исследование в области науки");
    }
}

// Университет с управлением зависимостями
public class University
{
    private readonly string _name;
    private readonly List<IPerson> _members;
    
    public University(string name, List<IPerson> members)
    {
        _name = name;
        _members = members;
    }
    
    public void ShowAllMembers()
    {
        Console.WriteLine($"\n=== Университет '{_name}' ===");
        Console.WriteLine($"Всего членов: {_members.Count}");
        
        foreach (var member in _members)
        {
            Console.WriteLine($"\n{member.GetInfo()}");
            member.SayHello();
        }
    }
    
    public void ConductClasses()
    {
        Console.WriteLine($"\n=== Проведение занятий в университете '{_name}' ===");
        foreach (var member in _members)
        {
            if (member is ITeacher teacher)
            {
                teacher.Teach();
            }
            else if (member is IStudent student)
            {
                student.Study();
            }
        }
    }
    
    public void OrganizeConference()
    {
        Console.WriteLine($"\n=== Организация конференции в университете '{_name}' ===");
        foreach (var member in _members)
        {
            if (member is IConferenceSpeaker speaker)
            {
                speaker.SpeakAtConference($"Научная конференция {_name}");
                speaker.PresentResearch();
            }
        }
    }
}

// ========== ДЕМОНСТРАЦИЯ ЯВНОЙ РЕАЛИЗАЦИИ ИНТЕРФЕЙСОВ ==========
class ExplicitImplementationDemo
{
    public static void Run()
    {
        Console.WriteLine("========== ЯВНАЯ РЕАЛИЗАЦИЯ ИНТЕРФЕЙСОВ ==========\n");
        
        // Создание объектов
        Student student = new Student("Анна Смирнова", 20, "Женский", "МГУ", 3, "Программирование", "anna@mail.ru");
        Employee employee = new Employee("Пётр Сидоров", 35, "Мужской", "ООО Ромашка", 75000, "Разработчик", 5);
        Teacher teacher = new Teacher("Елена Иванова", 45, "Женский", "МГУ", 120000, "Доцент", 15, "Математика", "Кандидат наук");
        Professor professor = new Professor("Сергей Николаев", 50, "Мужской", "МГУ", 200000, "Профессор", 25,
                                            "Программирование", "Доктор наук", "Искусственный интеллект", "Лаборатория ИИ");
        
        // Демонстрация работы через разные интерфейсы
        Console.WriteLine("=== 1. Через интерфейс IPerson ===\n");
        
        IPerson personStudent = student;
        personStudent.SayHello();
        personStudent.Work();
        personStudent.Study();
        
        Console.WriteLine("\n=== 2. Через интерфейс IStudent ===\n");
        
        IStudent istudent = student;
        istudent.AddGrade(5);
        istudent.AddGrade(4);
        istudent.AddGrade(5);
        istudent.Study();
        istudent.TakeExam();
        Console.WriteLine($"Средний балл: {istudent.GetAverageGrade():F2}");
        
        Console.WriteLine("\n=== 3. Через интерфейс IWorker ===\n");
        
        IWorker iworker = employee;
        iworker.Work();
        iworker.TakeBreak();
        iworker.ReceiveBonus(15000);
        
        Console.WriteLine("\n=== 4. Через интерфейс ITeacher ===\n");
        
        ITeacher iteacher = teacher;
        iteacher.Teach();
        iteacher.GradeStudent(student, 5);
        
        Console.WriteLine("\n=== 5. Через интерфейсы IResearcher, IMentor, IConferenceSpeaker ===\n");
        
        IResearcher researcher = professor;
        researcher.DoResearch();
        researcher.PublishArticle("Новые методы машинного обучения");
        
        IMentor mentor = professor;
        mentor.MentorStudent(student);
        mentor.GiveAdvice();
        
        IConferenceSpeaker speaker = professor;
        speaker.SpeakAtConference("Международная конференция по ИИ");
        speaker.PresentResearch();
    }
}

// ========== ДЕМОНСТРАЦИЯ УПРАВЛЕНИЯ ЗАВИСИМОСТЯМИ ==========
class DependencyInjectionDemo
{
    public static void Run()
    {
        Console.WriteLine("\n========== УПРАВЛЕНИЕ ЗАВИСИМОСТЯМИ (Dependency Injection) ==========\n");
        
        // Создание объектов
        Student student = new Student("Анна Смирнова", 20, "Женский", "МГУ", 3, "Программирование");
        Student student2 = new Student("Олег Новиков", 22, "Мужской", "МГУ", 4, "Искусственный интеллект");
        Student student3 = new Student("Елена Попова", 19, "Женский", "МГУ", 2, "Биоинформатика");
        
        Employee employee = new Employee("Пётр Сидоров", 35, "Мужской", "ООО Ромашка", 75000, "Разработчик", 5);
        
        Teacher teacher = new Teacher("Елена Иванова", 45, "Женский", "МГУ", 120000, "Доцент", 15, "Математика", "Кандидат наук");
        
        Professor professor = new Professor("Сергей Николаев", 50, "Мужской", "МГУ", 200000, "Профессор", 25,
                                            "Программирование", "Доктор наук", "Искусственный интеллект", "Лаборатория ИИ");
        
        // Внедрение зависимостей через сервисы
        Console.WriteLine("=== 1. PersonService (работа с IPerson) ===\n");
        
        PersonService studentService = new PersonService(student);
        studentService.DemonstratePerson();
        
        PersonService teacherService = new PersonService(teacher);
        teacherService.DemonstratePerson();
        
        PersonService professorService = new PersonService(professor);
        professorService.DemonstratePerson();
        
        Console.WriteLine("\n=== 2. StudentService (работа с IStudent) ===\n");
        
        StudentService stdService = new StudentService(student, student.Name);
        stdService.AddGrades(5, 4, 5, 5, 4);
        stdService.ShowAcademicPerformance();
        
        Console.WriteLine("\n=== 3. WorkerService (работа с IWorker) ===\n");
        
        WorkerService workerService = new WorkerService(employee, employee.Name);
        workerService.ShowWorkInfo();
        workerService.GiveBonus(15000);
        
        Console.WriteLine("\n=== 4. TeacherService (работа с ITeacher) ===\n");
        
        TeacherService teacherSvc = new TeacherService(teacher, teacher.Name);
        teacherSvc.DemonstrateTeaching();
        
        Console.WriteLine("\n=== 5. ResearchService (работа с IResearcher) ===\n");
        
        ResearchService researchService = new ResearchService(professor, professor.Name);
        researchService.ShowResearchActivity();
        
        // Университет с внедрением зависимостей
        Console.WriteLine("\n=== 6. University (коллекция IPerson) ===\n");
        
        List<IPerson> universityMembers = new List<IPerson>
        {
            student,
            student2,
            student3,
            teacher,
            professor,
            employee
        };
        
        University msu = new University("МГУ им. Ломоносова", universityMembers);
        msu.ShowAllMembers();
        msu.ConductClasses();
        msu.OrganizeConference();
    }
}

// ========== ПОЛНАЯ ДЕМОНСТРАЦИЯ ==========

Console.WriteLine("==================== ИНДИВИДУАЛЬНЫЙ ПРОЕКТ ====================");
Console.WriteLine("Тема: Явная реализация интерфейсов и управление зависимостями");
Console.WriteLine("===============================================================\n");

// Демонстрация явной реализации интерфейсов
ExplicitImplementationDemo.Run();

// Демонстрация управления зависимостями
DependencyInjectionDemo.Run();

// Итоговая статистика
Console.WriteLine("\n==================== ИТОГОВАЯ СТАТИСТИКА ====================");
Person.DisplayTotalPeople();

Console.WriteLine("\n==================== ПРОЕКТ ЗАВЕРШЕН ====================");
```
