```

using System;
using System.Collections.Generic;

// ========== БАЗОВЫЙ КЛАСС PERSON ==========
public class Person
{
    // Приватные поля для инкапсуляции
    private string _name;
    private int _age;
    private string _gender;
    
    // Свойства с геттерами и сеттерами
    public string Name
    {
        get { return _name; }
        set 
        { 
            if (!string.IsNullOrWhiteSpace(value))
                _name = value;
            else
                throw new ArgumentException("Имя не может быть пустым!");
        }
    }
    
    public int Age
    {
        get { return _age; }
        set 
        { 
            if (value >= 0 && value <= 120)
                _age = value;
            else
                throw new ArgumentOutOfRangeException("Возраст должен быть от 0 до 120 лет!");
        }
    }
    
    public string Gender
    {
        get { return _gender; }
        set 
        { 
            if (value == "Мужской" || value == "Женский" || value == "Другой")
                _gender = value;
            else
                throw new ArgumentException("Пол должен быть 'Мужской', 'Женский' или 'Другой'!");
        }
    }
    
    // Статический атрибут
    public static int TotalPeople { get; private set; } = 0;
    protected static List<Person> allPeople = new List<Person>();
    
    // Конструктор
    public Person(string name, int age, string gender)
    {
        Name = name;
        Age = age;
        Gender = gender;
        TotalPeople++;
        allPeople.Add(this);
    }
    
    // Виртуальные методы
    public virtual string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Пол: {Gender}";
    }
    
    public virtual void SayHello()
    {
        Console.WriteLine($"Привет! Меня зовут {Name}.");
    }
    
    // Статический метод
    public static void DisplayTotalPeople()
    {
        Console.WriteLine($"\nВсего создано людей: {TotalPeople}");
    }
}

// ========== ПРОСТОЕ НАСЛЕДОВАНИЕ ==========
// Student наследует от Person
public class Student : Person
{
    private string _university;
    private int _course;
    private List<int> _grades = new List<int>();
    
    public string University
    {
        get { return _university; }
        set 
        { 
            if (!string.IsNullOrWhiteSpace(value))
                _university = value;
            else
                throw new ArgumentException("Название университета не может быть пустым!");
        }
    }
    
    public int Course
    {
        get { return _course; }
        set 
        { 
            if (value >= 1 && value <= 6)
                _course = value;
            else
                throw new ArgumentOutOfRangeException("Курс должен быть от 1 до 6!");
        }
    }
    
    public Student(string name, int age, string gender, string university, int course) 
        : base(name, age, gender)
    {
        University = university;
        Course = course;
    }
    
    public void AddGrade(int grade)
    {
        if (grade >= 2 && grade <= 5)
        {
            _grades.Add(grade);
            Console.WriteLine($"Студенту {Name} добавлена оценка: {grade}");
        }
        else
        {
            Console.WriteLine("Оценка должна быть от 2 до 5!");
        }
    }
    
    public double GetAverageGrade()
    {
        if (_grades.Count == 0) return 0;
        double sum = 0;
        foreach (var g in _grades) sum += g;
        return sum / _grades.Count;
    }
    
    public void ShowGrades()
    {
        if (_grades.Count > 0)
        {
            Console.Write($"Оценки {Name}: ");
            for (int i = 0; i < _grades.Count; i++)
            {
                Console.Write(_grades[i]);
                if (i < _grades.Count - 1) Console.Write(", ");
            }
            Console.WriteLine($" (Средний: {GetAverageGrade():F2})");
        }
        else
        {
            Console.WriteLine($"У {Name} пока нет оценок");
        }
    }
    
    public override void SayHello()
    {
        Console.WriteLine($"Привет! Я студент {Name}, учусь на {Course}-м курсе в {University}.");
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Университет: {University}, Курс: {Course}, Средний балл: {GetAverageGrade():F2}";
    }
    
    public void Study()
    {
        Console.WriteLine($"{Name} сейчас учится в {University}.");
    }
}

// Employee наследует от Person
public class Employee : Person
{
    private string _company;
    private decimal _salary;
    private List<string> _projects = new List<string>();
    
    public string Company
    {
        get { return _company; }
        set 
        { 
            if (!string.IsNullOrWhiteSpace(value))
                _company = value;
            else
                throw new ArgumentException("Название компании не может быть пустым!");
        }
    }
    
    public decimal Salary
    {
        get { return _salary; }
        set 
        { 
            if (value >= 0)
                _salary = value;
            else
                throw new ArgumentOutOfRangeException("Зарплата не может быть отрицательной!");
        }
    }
    
    public Employee(string name, int age, string gender, string company, decimal salary) 
        : base(name, age, gender)
    {
        Company = company;
        Salary = salary;
    }
    
    public void AddProject(string project)
    {
        _projects.Add(project);
        Console.WriteLine($"{Name} добавлен в проект: {project}");
    }
    
    public void RaiseSalary(decimal amount)
    {
        if (amount > 0)
        {
            Salary += amount;
            Console.WriteLine($"Зарплата {Name} повышена до {Salary:C}");
        }
    }
    
    public override string GetInfo()
    {
        string projectsStr = _projects.Count > 0 ? string.Join(", ", _projects) : "нет проектов";
        return base.GetInfo() + $", Компания: {Company}, Зарплата: {Salary:C}, Проекты: {projectsStr}";
    }
    
    public void Work()
    {
        Console.WriteLine($"{Name} сейчас работает в {Company}.");
    }
}

// ========== СЛОЖНОЕ (МНОГОУРОВНЕВОЕ) НАСЛЕДОВАНИЕ ==========
// Teacher наследует от Employee (Employee -> Person)
public class Teacher : Employee
{
    private string _subject;
    private int _experience;
    private List<Student> _students = new List<Student>();
    private string _academicDegree; // Ученая степень
    
    public string Subject
    {
        get { return _subject; }
        set 
        { 
            if (!string.IsNullOrWhiteSpace(value))
                _subject = value;
            else
                throw new ArgumentException("Название предмета не может быть пустым!");
        }
    }
    
    public int Experience
    {
        get { return _experience; }
        set 
        { 
            if (value >= 0 && value <= 70)
                _experience = value;
            else
                throw new ArgumentOutOfRangeException("Стаж должен быть от 0 до 70 лет!");
        }
    }
    
    public string AcademicDegree
    {
        get { return _academicDegree; }
        set { _academicDegree = value; }
    }
    
    public Teacher(string name, int age, string gender, string company, decimal salary,
                   string subject, int experience, string academicDegree) 
        : base(name, age, gender, company, salary)
    {
        Subject = subject;
        Experience = experience;
        AcademicDegree = academicDegree;
    }
    
    public void AddStudent(Student student)
    {
        if (!_students.Contains(student))
        {
            _students.Add(student);
            Console.WriteLine($"Студент {student.Name} добавлен к преподавателю {Name}");
        }
    }
    
    public void GradeStudent(Student student, int grade)
    {
        if (_students.Contains(student))
        {
            student.AddGrade(grade);
            Console.WriteLine($"Преподаватель {Name} выставил оценку {grade} студенту {student.Name}");
        }
        else
        {
            Console.WriteLine($"Студент {student.Name} не учится у преподавателя {Name}");
        }
    }
    
    public override void SayHello()
    {
        Console.WriteLine($"Здравствуйте! Я преподаватель {Name}, {AcademicDegree}, веду предмет '{Subject}'.");
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Предмет: {Subject}, Стаж: {Experience} лет, Степень: {AcademicDegree}, Студентов: {_students.Count}";
    }
    
    public void Teach()
    {
        Console.WriteLine($"{Name} преподаёт {Subject}.");
    }
    
    public void ConductExam()
    {
        Console.WriteLine($"{Name} проводит экзамен по предмету {Subject}");
        foreach (var student in _students)
        {
            Random rand = new Random();
            int grade = rand.Next(3, 6);
            GradeStudent(student, grade);
        }
    }
}

// UniversityDean наследует от Teacher (еще один уровень сложного наследования)
public class UniversityDean : Teacher
{
    private string _universityName;
    private int _yearsAsDean;
    private List<string> _initiatives = new List<string>();
    
    public string UniversityName
    {
        get { return _universityName; }
        set { _universityName = value; }
    }
    
    public int YearsAsDean
    {
        get { return _yearsAsDean; }
        set { _yearsAsDean = value; }
    }
    
    public UniversityDean(string name, int age, string gender, string company, decimal salary,
                          string subject, int experience, string academicDegree,
                          string universityName, int yearsAsDean)
        : base(name, age, gender, company, salary, subject, experience, academicDegree)
    {
        UniversityName = universityName;
        YearsAsDean = yearsAsDean;
    }
    
    public void AddInitiative(string initiative)
    {
        _initiatives.Add(initiative);
        Console.WriteLine($"Декан {Name} предложил инициативу: {initiative}");
    }
    
    public void ManageFaculty()
    {
        Console.WriteLine($"Декан {Name} управляет факультетом в {UniversityName} ({YearsAsDean} лет на посту)");
    }
    
    public override void SayHello()
    {
        Console.WriteLine($"Здравствуйте! Я декан {Name}, {AcademicDegree}, управляю факультетом в {UniversityName}.");
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Должность: Декан {UniversityName}, Лет на посту: {YearsAsDean}";
    }
}

// ========== ИНТЕРФЕЙСЫ ДЛЯ МНОЖЕСТВЕННОГО НАСЛЕДОВАНИЯ ==========
public interface IResearcher
{
    int PublicationsCount { get; set; }
    void DoResearch();
    void PublishArticle(string title);
}

public interface IMentor
{
    int StudentsMentored { get; set; }
    void MentorStudent(Student student);
    void GiveAdvice();
}

public interface IConferenceSpeaker
{
    int ConferencesAttended { get; set; }
    void SpeakAtConference(string conferenceName);
    void PresentResearch();
}

public interface IProjectManager
{
    int ProjectsManaged { get; set; }
    void ManageProject(string projectName);
    void AllocateResources();
}

// ========== МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ ЧЕРЕЗ ИНТЕРФЕЙСЫ ==========
// Professor наследует от Teacher и реализует несколько интерфейсов
public class Professor : Teacher, IResearcher, IMentor, IConferenceSpeaker
{
    // IResearcher
    public int PublicationsCount { get; set; }
    private List<string> _publications = new List<string>();
    
    // IMentor
    public int StudentsMentored { get; set; }
    private List<Student> _mentoredStudents = new List<Student>();
    
    // IConferenceSpeaker
    public int ConferencesAttended { get; set; }
    private List<string> _conferences = new List<string>();
    
    // Собственные свойства
    public string ResearchField { get; set; }
    public string LaboratoryName { get; set; }
    
    public Professor(string name, int age, string gender, string company, decimal salary,
                     string subject, int experience, string academicDegree,
                     string researchField, string laboratoryName, int publicationsCount)
        : base(name, age, gender, company, salary, subject, experience, academicDegree)
    {
        ResearchField = researchField;
        LaboratoryName = laboratoryName;
        PublicationsCount = publicationsCount;
        StudentsMentored = 0;
        ConferencesAttended = 0;
    }
    
    // Реализация IResearcher
    public void DoResearch()
    {
        Console.WriteLine($"Профессор {Name} проводит исследования в области {ResearchField}");
    }
    
    public void PublishArticle(string title)
    {
        _publications.Add(title);
        PublicationsCount++;
        Console.WriteLine($"Профессор {Name} опубликовал статью: '{title}'");
        Console.WriteLine($"Всего публикаций: {PublicationsCount}");
    }
    
    // Реализация IMentor
    public void MentorStudent(Student student)
    {
        if (!_mentoredStudents.Contains(student))
        {
            _mentoredStudents.Add(student);
            StudentsMentored++;
            Console.WriteLine($"Профессор {Name} взял шефство над студентом {student.Name}");
        }
        else
        {
            Console.WriteLine($"Профессор {Name} уже наставляет студента {student.Name}");
        }
    }
    
    public void GiveAdvice()
    {
        Console.WriteLine($"Профессор {Name} дает ценный совет по научной карьере");
    }
    
    // Реализация IConferenceSpeaker
    public void SpeakAtConference(string conferenceName)
    {
        _conferences.Add(conferenceName);
        ConferencesAttended++;
        Console.WriteLine($"Профессор {Name} выступает на конференции: {conferenceName}");
    }
    
    public void PresentResearch()
    {
        Console.WriteLine($"Профессор {Name} представляет результаты исследований в {ResearchField}");
    }
    
    // Переопределение методов
    public override void SayHello()
    {
        Console.WriteLine($"Здравствуйте! Я профессор {Name}, {AcademicDegree}, заведующий лабораторией {LaboratoryName}.");
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Лаборатория: {LaboratoryName}, Научная область: {ResearchField}, " +
               $"Публикаций: {PublicationsCount}, Студентов под наставничеством: {StudentsMentored}, " +
               $"Конференций: {ConferencesAttended}";
    }
    
    public void ConductScientificConference()
    {
        Console.WriteLine($"Профессор {Name} организует научную конференцию по {ResearchField}");
    }
}

// GraduateStudent наследует от Student и реализует IResearcher
public class GraduateStudent : Student, IResearcher
{
    // IResearcher
    public int PublicationsCount { get; set; }
    private List<string> _publications = new List<string>();
    
    // Собственные свойства
    public string ThesisTopic { get; set; }
    public Professor ScientificAdvisor { get; set; }
    public string ResearchArea { get; set; }
    
    public GraduateStudent(string name, int age, string gender, string university, int course,
                           string thesisTopic, string researchArea, Professor advisor)
        : base(name, age, gender, university, course)
    {
        ThesisTopic = thesisTopic;
        ResearchArea = researchArea;
        ScientificAdvisor = advisor;
        PublicationsCount = 0;
    }
    
    // Реализация IResearcher
    public void DoResearch()
    {
        Console.WriteLine($"Аспирант {Name} проводит исследования по теме '{ThesisTopic}'");
    }
    
    public void PublishArticle(string title)
    {
        _publications.Add(title);
        PublicationsCount++;
        Console.WriteLine($"Аспирант {Name} опубликовал статью: '{title}'");
    }
    
    public void DefendThesis()
    {
        Console.WriteLine($"Аспирант {Name} защищает диссертацию по теме '{ThesisTopic}'");
        if (PublicationsCount >= 3)
        {
            Console.WriteLine($"Поздравляем! {Name} успешно защитил диссертацию! (Публикаций: {PublicationsCount})");
        }
        else
        {
            Console.WriteLine($"Недостаточно публикаций для защиты. Необходимо: {3 - PublicationsCount} статьи");
        }
    }
    
    public override void SayHello()
    {
        Console.WriteLine($"Здравствуйте! Я аспирант {Name}, работаю над диссертацией по теме '{ThesisTopic}'.");
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Тема диссертации: {ThesisTopic}, Научный руководитель: {ScientificAdvisor.Name}, " +
               $"Публикаций: {PublicationsCount}";
    }
}

// ResearchAssistant наследует от Employee и реализует IResearcher и IProjectManager
public class ResearchAssistant : Employee, IResearcher, IProjectManager
{
    // IResearcher
    public int PublicationsCount { get; set; }
    private List<string> _publications = new List<string>();
    
    // IProjectManager
    public int ProjectsManaged { get; set; }
    private List<string> _projectsManaged = new List<string>();
    
    // Собственные свойства
    public string ResearchLab { get; set; }
    public string Specialization { get; set; }
    
    public ResearchAssistant(string name, int age, string gender, string company, decimal salary,
                             string researchLab, string specialization)
        : base(name, age, gender, company, salary)
    {
        ResearchLab = researchLab;
        Specialization = specialization;
        PublicationsCount = 0;
        ProjectsManaged = 0;
    }
    
    // Реализация IResearcher
    public void DoResearch()
    {
        Console.WriteLine($"Научный сотрудник {Name} проводит исследования в лаборатории {ResearchLab}");
    }
    
    public void PublishArticle(string title)
    {
        _publications.Add(title);
        PublicationsCount++;
        Console.WriteLine($"Научный сотрудник {Name} опубликовал статью: '{title}'");
    }
    
    // Реализация IProjectManager
    public void ManageProject(string projectName)
    {
        _projectsManaged.Add(projectName);
        ProjectsManaged++;
        Console.WriteLine($"Научный сотрудник {Name} управляет проектом: {projectName}");
    }
    
    public void AllocateResources()
    {
        Console.WriteLine($"Научный сотрудник {Name} распределяет ресурсы в лаборатории {ResearchLab}");
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Лаборатория: {ResearchLab}, Специализация: {Specialization}, " +
               $"Публикаций: {PublicationsCount}, Проектов под управлением: {ProjectsManaged}";
    }
}

// ========== ДЕМОНСТРАЦИЯ ==========
Console.WriteLine("========== ДЕМОНСТРАЦИЯ ВСЕХ ТИПОВ НАСЛЕДОВАНИЯ ==========\n");

// ========== 1. ПРОСТОЕ НАСЛЕДОВАНИЕ ==========
Console.WriteLine("=== 1. ПРОСТОЕ НАСЛЕДОВАНИЕ ===\n");

Student student = new Student("Анна Смирнова", 20, "Женский", "МГУ", 3);
Employee employee = new Employee("Пётр Сидоров", 35, "Мужской", "ООО Ромашка", 75000);

Console.WriteLine("--- Студент (наследует Person) ---");
student.SayHello();
student.AddGrade(5);
student.AddGrade(4);
student.ShowGrades();

Console.WriteLine("\n--- Сотрудник (наследует Person) ---");
employee.SayHello();
employee.AddProject("Разработка ПО");
employee.RaiseSalary(15000);
Console.WriteLine(employee.GetInfo());

// ========== 2. СЛОЖНОЕ (МНОГОУРОВНЕВОЕ) НАСЛЕДОВАНИЕ ==========
Console.WriteLine("\n=== 2. СЛОЖНОЕ (МНОГОУРОВНЕВОЕ) НАСЛЕДОВАНИЕ ===\n");

Console.WriteLine("--- Уровень 1: Person → Employee → Teacher ---");
Teacher teacher = new Teacher("Елена Иванова", 45, "Женский", "МГУ", 85000,
                                "Математика", 15, "Кандидат наук");
teacher.SayHello();
teacher.AddStudent(student);
teacher.GradeStudent(student, 5);
teacher.ConductExam();

Console.WriteLine("\n--- Уровень 2: Person → Employee → Teacher → UniversityDean ---");
UniversityDean dean = new UniversityDean("Дмитрий Петров", 55, "Мужской", "МГУ", 120000,
                                            "Физика", 25, "Доктор наук", "МГУ", 8);
dean.SayHello();
dean.AddInitiative("Открыть новую лабораторию");
dean.AddInitiative("Увеличить стипендии");
dean.ManageFaculty();
Console.WriteLine(dean.GetInfo());

// ========== 3. МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ (ЧЕРЕЗ ИНТЕРФЕЙСЫ) ==========
Console.WriteLine("\n=== 3. МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ (ЧЕРЕЗ ИНТЕРФЕЙСЫ) ===\n");

Console.WriteLine("--- Профессор (Teacher + IResearcher + IMentor + IConferenceSpeaker) ---");
Professor professor = new Professor("Сергей Николаев", 50, "Мужской", "МГУ", 150000,
                                    "Программирование", 20, "Доктор наук",
                                    "Искусственный интеллект", "Лаборатория ИИ", 45);
professor.SayHello();
professor.DoResearch();
professor.PublishArticle("Нейронные сети нового поколения");
professor.MentorStudent(student);
professor.SpeakAtConference("International AI Conference 2024");
professor.PresentResearch();
Console.WriteLine(professor.GetInfo());

Console.WriteLine("\n--- Аспирант (Student + IResearcher) ---");
GraduateStudent gradStudent = new GraduateStudent("Иван Козлов", 25, "Мужской", "МГУ", 2,
                                                    "Глубокое обучение в компьютерном зрении",
                                                    "Искусственный интеллект", professor);
gradStudent.SayHello();
gradStudent.DoResearch();
gradStudent.PublishArticle("Сверточные сети для распознавания образов");
gradStudent.PublishArticle("Оптимизация нейросетей");
gradStudent.DefendThesis();
Console.WriteLine(gradStudent.GetInfo());

Console.WriteLine("\n--- Научный сотрудник (Employee + IResearcher + IProjectManager) ---");
ResearchAssistant researchAssistant = new ResearchAssistant("Мария Воронова", 32, "Женский",
                                                                "НИИ Искусственного интеллекта", 95000,
                                                                "Лаборатория ИИ", "Машинное обучение");
researchAssistant.SayHello();
researchAssistant.DoResearch();
researchAssistant.PublishArticle("Методы оптимизации");
researchAssistant.ManageProject("Разработка системы рекомендаций");
researchAssistant.AllocateResources();
Console.WriteLine(researchAssistant.GetInfo());

// ========== 4. ПОЛИМОРФИЗМ С РАЗНЫМИ ТИПАМИ НАСЛЕДОВАНИЯ ==========
Console.WriteLine("\n=== 4. ПОЛИМОРФИЗМ С РАЗНЫМИ ТИПАМИ НАСЛЕДОВАНИЯ ===\n");

Console.WriteLine("--- Массив объектов Person ---");
Person[] people = new Person[]
{
    student,
    employee,
    teacher,
    dean,
    professor,
    gradStudent,
    researchAssistant
};

foreach (var p in people)
{
    Console.WriteLine(p.GetInfo());
    p.SayHello();
    Console.WriteLine();
}

Console.WriteLine("--- Работа через интерфейсы (множественное наследование) ---");
IResearcher[] researchers = new IResearcher[]
{
    professor,
    gradStudent,
    researchAssistant
};

foreach (var researcher in researchers)
{
    researcher.DoResearch();
    researcher.PublishArticle($"Статья от {researcher.GetType().Name}");
    Console.WriteLine();
}

Console.WriteLine("--- Взаимодействие объектов ---");
professor.ConductScientificConference();
professor.MentorStudent(gradStudent);
gradStudent.DefendThesis();

// ========== 5. СТАТИСТИКА ==========
Console.WriteLine("\n=== 5. СТАТИСТИКА ===\n");
Person.DisplayTotalPeople();
Console.WriteLine($"\nПрофессор {professor.Name}: {professor.PublicationsCount} публикаций, " +
                    $"{professor.StudentsMentored} студентов под наставничеством, " +
                    $"{professor.ConferencesAttended} конференций");

Console.WriteLine("\n========== ДЕМОНСТРАЦИЯ ЗАВЕРШЕНА ==========");
```
