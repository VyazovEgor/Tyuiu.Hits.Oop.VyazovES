<h1 style="color:DodgerBlue">Сложное наследование</h1>

Сложное наследование в C# отличается от простого тем, что оно позволяет создавать более глубокие иерархии классов, где один класс может наследоваться от другого, который уже является производным классом. Это позволяет еще больше абстрагировать общие свойства и методы, делая код более модульным и легким для поддержки.

Давайте рассмотрим пример сложного наследования, используя классы `Vehicle`, `Car`, и `Motorcycle`, а также введем новый класс `SportsCar`, который будет наследоваться от `Car`.

### Шаг 1: Базовый класс Vehicle

Начнем с определения базового класса `Vehicle`, как и в предыдущем примере.

```csharp
public class Vehicle
{
    public string Make { get; set; }
    public string Model { get; set; }

    public Vehicle(string make, string model)
    {
        Make = make;
        Model = model;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Производитель: {Make}, Модель: {Model}");
    }
}
```

### Шаг 2: Производный класс Car

Теперь определим класс `Car`, который наследуется от `Vehicle`.

```csharp
public class Car : Vehicle
{
    public int Doors { get; set; }

    public Car(string make, string model, int doors) : base(make, model)
    {
        Doors = doors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Двери: {Doors}");
    }
}
```

### Шаг 3: Производный класс SportsCar

Введем класс `SportsCar`, который наследуется от `Car`, демонстрируя сложное наследование.

```csharp
public class SportsCar : Car
{
    public bool IsConvertible { get; set; }

    public SportsCar(string make, string model, int doors, bool isConvertible) 
        : base(make, model, doors)
    {
        IsConvertible = isConvertible;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Автомобиль с откидным верхом: {IsConvertible}");
    }
}
```

### Шаг 4: Производный класс Motorcycle

Определим класс `Motorcycle`, который также наследуется от `Vehicle`.

```csharp
public class Motorcycle : Vehicle
{
    public string EngineType { get; set; }

    public Motorcycle(string make, string model, string engineType) : base(make, model)
    {
        EngineType = engineType;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Тип двигателя: {EngineType}");
    }
}
```

### Шаг 5: Использование классов

Теперь мы можем создать экземпляры наших классов и использовать их.

```csharp
Car myCar = new Car("Toyota", "Corolla", 4);
myCar.DisplayInfo();

SportsCar mySportsCar = new SportsCar("Ferrari", "488 Spider", 2, true);
mySportsCar.DisplayInfo();

Motorcycle myMotorcycle = new Motorcycle("Ducati", "Panigale V4", "V4");
myMotorcycle.DisplayInfo();
```

### Объяснение кода

- **Базовый класс `Vehicle`** предоставляет основные свойства и методы для всех транспортных средств.
- **Класс `Car`** наследует от `Vehicle` и добавляет свойство `Doors`.
- **Класс `SportsCar`** наследует от `Car`, демонстрируя сложное наследование, и добавляет свойство `IsConvertible`.
- **Класс `Motorcycle`** также наследует от `Vehicle`, но имеет свое собственное свойство `EngineType`.

### Отличие сложного наследования от простого

- **Глубина иерархии**: Сложное наследование позволяет создавать более глубокие иерархии классов, где классы могут наследоваться друг от друга на несколько уровней вниз.
- **Переиспользование кода**: Сложное наследование увеличивает возможности для переиспользования кода, поскольку классы могут наследовать не только непосредственно от базового класса, но и от других производных классов.
- **Флексибельность**: Сложное наследование обеспечивает большую гибкость в проектировании программ, позволяя разработчикам точно определять отношения между классами и их роли в системе.

Таким образом, сложное наследование в C# представляет собой мощный инструмент для создания хорошо структурированных и масштабируемых приложений, позволяя эффективно организовывать код и избегать дублирования.

````
using System;

// ========== УРОВЕНЬ 1: Базовый класс ==========
public class Vehicle
{
    // Базовые свойства
    public string Make { get; set; }
    public string Model { get; set; }
    public int Year { get; set; }
    
    // Статический счетчик
    public static int TotalVehicles { get; private set; } = 0;
    
    // Конструктор
    public Vehicle(string make, string model, int year)
    {
        Make = make;
        Model = model;
        Year = year;
        TotalVehicles++;
    }
    
    // Виртуальный метод
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Производитель: {Make}, Модель: {Model}, Год: {Year}");
    }
    
    public virtual void StartEngine()
    {
        Console.WriteLine("Двигатель запущен");
    }
}

// ========== УРОВЕНЬ 2: Наземные транспортные средства ==========
public class LandVehicle : Vehicle
{
    // Новые свойства для наземного транспорта
    public int Wheels { get; set; }
    public string DriveType { get; set; } // полный, передний, задний
    
    public LandVehicle(string make, string model, int year, int wheels, string driveType) 
        : base(make, model, year)
    {
        Wheels = wheels;
        DriveType = driveType;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Колеса: {Wheels}, Привод: {DriveType}");
    }
    
    public virtual void Drive()
    {
        Console.WriteLine("Наземное транспортное средство движется");
    }
}

// ========== УРОВЕНЬ 3: Автомобили (наследуют от LandVehicle) ==========
public class Car : LandVehicle
{
    // Специфичные для автомобилей свойства
    public int Doors { get; set; }
    public string BodyType { get; set; }
    public int Horsepower { get; set; }
    
    public Car(string make, string model, int year, int wheels, string driveType, 
               int doors, string bodyType, int horsepower) 
        : base(make, model, year, wheels, driveType)
    {
        Doors = doors;
        BodyType = bodyType;
        Horsepower = horsepower;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Двери: {Doors}, Тип кузова: {BodyType}, Л.с.: {Horsepower}");
    }
    
    public override void Drive()
    {
        Console.WriteLine($"Автомобиль {Make} {Model} едет по дороге");
    }
    
    public override void StartEngine()
    {
        Console.WriteLine($"Двигатель автомобиля {Make} {Model} запущен (VRRROOOOM!)");
    }
    
    // Специфичный метод
    public void OpenTrunk()
    {
        Console.WriteLine("Багажник открыт");
    }
}

// ========== УРОВЕНЬ 3: Спортивные автомобили (наследуют от Car) ==========
public class SportsCar : Car
{
    // Специфичные для спорткаров свойства
    public double AccelerationTo100 { get; set; } // секунды
    public int MaxSpeed { get; set; }
    public bool HasSpoiler { get; set; }
    
    public SportsCar(string make, string model, int year, int wheels, string driveType,
                     int doors, string bodyType, int horsepower, 
                     double accelerationTo100, int maxSpeed, bool hasSpoiler)
        : base(make, model, year, wheels, driveType, doors, bodyType, horsepower)
    {
        AccelerationTo100 = accelerationTo100;
        MaxSpeed = maxSpeed;
        HasSpoiler = hasSpoiler;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"0-100 км/ч: {AccelerationTo100} сек, Макс. скорость: {MaxSpeed} км/ч, Антикрыло: {(HasSpoiler ? "Да" : "Нет")}");
    }
    
    public override void StartEngine()
    {
        Console.WriteLine($"Спортивный автомобиль {Make} {Model} ревет: BRRRAAAP! BRRRAAAP!");
    }
    
    public override void Drive()
    {
        Console.WriteLine($"Спортивный автомобиль {Make} {Model} мчится с максимальной скоростью {MaxSpeed} км/ч!");
    }
    
    // Специфичный метод
    public void ActivateSportMode()
    {
        Console.WriteLine("Спортивный режим активирован! Подвеска стала жестче, отклик педали газа улучшен!");
    }
    
    public void Drift()
    {
        Console.WriteLine($"Спортивный автомобиль {Make} {Model} входит в управляемый занос!");
    }
}

// ========== УРОВЕНЬ 3: Электромобили (наследуют от Car) ==========
public class ElectricCar : Car
{
    // Специфичные для электромобилей свойства
    public int BatteryCapacity { get; set; } // кВт/ч
    public int Range { get; set; } // км
    public int ChargingTime { get; set; } // минут до 80%
    
    public ElectricCar(string make, string model, int year, int wheels, string driveType,
                       int doors, string bodyType, int horsepower,
                       int batteryCapacity, int range, int chargingTime)
        : base(make, model, year, wheels, driveType, doors, bodyType, horsepower)
    {
        BatteryCapacity = batteryCapacity;
        Range = range;
        ChargingTime = chargingTime;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Батарея: {BatteryCapacity} кВт/ч, Запас хода: {Range} км, Зарядка: {ChargingTime} мин");
    }
    
    public override void StartEngine()
    {
        Console.WriteLine($"Электромобиль {Make} {Model} бесшумно активирован (ZZZzzz)");
    }
    
    public override void Drive()
    {
        Console.WriteLine($"Электромобиль {Make} {Model} тихо движется с запасом хода {Range} км");
    }
    
    // Специфичный метод
    public void Charge()
    {
        Console.WriteLine($"Зарядка электромобиля {Make} {Model}... {ChargingTime} минут до 80%");
    }
    
    public void RegenerateEnergy()
    {
        Console.WriteLine("Рекуперативное торможение: энергия возвращается в батарею!");
    }
}

// ========== УРОВЕНЬ 2: Мотоциклы (наследуют от LandVehicle) ==========
public class Motorcycle : LandVehicle
{
    // Специфичные для мотоциклов свойства
    public string EngineType { get; set; }
    public bool HasSidecar { get; set; }
    
    public Motorcycle(string make, string model, int year, int wheels, string driveType,
                      string engineType, bool hasSidecar)
        : base(make, model, year, wheels, driveType)
    {
        EngineType = engineType;
        HasSidecar = hasSidecar;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Тип двигателя: {EngineType}, Коляска: {(HasSidecar ? "Да" : "Нет")}");
    }
    
    public override void Drive()
    {
        Console.WriteLine($"Мотоцикл {Make} {Model} мчится по трассе");
    }
    
    public virtual void Wheelie()
    {
        Console.WriteLine($"Мотоцикл {Make} {Model} делает вили! (езда на заднем колесе)");
    }
}

// ========== УРОВЕНЬ 3: Спортивные мотоциклы (наследуют от Motorcycle) ==========
public class SportBike : Motorcycle
{
    public int MaxRPM { get; set; }
    public bool HasQuickShifter { get; set; }
    
    public SportBike(string make, string model, int year, int wheels, string driveType,
                     string engineType, bool hasSidecar, int maxRPM, bool hasQuickShifter)
        : base(make, model, year, wheels, driveType, engineType, hasSidecar)
    {
        MaxRPM = maxRPM;
        HasQuickShifter = hasQuickShifter;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Макс. оборотов: {MaxRPM} об/мин, Квикшифтер: {(HasQuickShifter ? "Да" : "Нет")}");
    }
    
    public override void Wheelie()
    {
        Console.WriteLine($"Спортбайк {Make} {Model} выполняет высокий вили на {MaxRPM} об/мин!");
    }
    
    public void Lean()
    {
        Console.WriteLine($"Спортбайк {Make} {Model} наклоняется в повороте под острым углом");
    }
}

// ========== Демонстрация сложного наследования для техники ==========
Console.WriteLine("========== СЛОЖНОЕ НАСЛЕДОВАНИЕ: ТЕХНИКА ==========\n");

// Уровень 1: Vehicle
Console.WriteLine("--- Уровень 1: Базовый Vehicle ---");
Vehicle vehicle = new Vehicle("Generic", "Vehicle", 2020);
vehicle.DisplayInfo();
vehicle.StartEngine();

Console.WriteLine("\n--- Уровень 2: LandVehicle ---");
LandVehicle landVehicle = new LandVehicle("Generic", "LandVehicle", 2021, 4, "Полный");
landVehicle.DisplayInfo();
landVehicle.Drive();

Console.WriteLine("\n--- Уровень 3: Car ---");
Car car = new Car("Toyota", "Camry", 2022, 4, "Передний", 4, "Седан", 203);
car.DisplayInfo();
car.StartEngine();
car.Drive();
car.OpenTrunk();

Console.WriteLine("\n--- Уровень 4: SportsCar (наследует от Car) ---");
SportsCar sportsCar = new SportsCar("Ferrari", "F8 Tributo", 2023, 4, "Задний", 
                                    2, "Купе", 710, 2.9, 340, true);
sportsCar.DisplayInfo();
sportsCar.StartEngine();
sportsCar.Drive();
sportsCar.ActivateSportMode();
sportsCar.Drift();

Console.WriteLine("\n--- Уровень 4: ElectricCar (наследует от Car) ---");
ElectricCar electricCar = new ElectricCar("Tesla", "Model S Plaid", 2023, 4, "Полный",
                                            4, "Лифтбек", 1020, 100, 650, 30);
electricCar.DisplayInfo();
electricCar.StartEngine();
electricCar.Drive();
electricCar.Charge();
electricCar.RegenerateEnergy();

Console.WriteLine("\n--- Уровень 3: Motorcycle ---");
Motorcycle motorcycle = new Motorcycle("Harley-Davidson", "Street Glide", 2022, 2, "Задний",
                                        "V-Twin", false);
motorcycle.DisplayInfo();
motorcycle.Drive();
motorcycle.Wheelie();

Console.WriteLine("\n--- Уровень 4: SportBike (наследует от Motorcycle) ---");
SportBike sportBike = new SportBike("Yamaha", "R1", 2023, 2, "Задний",
                                    "Inline-4", false, 14000, true);
sportBike.DisplayInfo();
sportBike.Drive();
sportBike.Wheelie();
sportBike.Lean();

// Демонстрация полиморфизма в многоуровневой иерархии
Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА В ИЕРАРХИИ ===");
Vehicle[] vehicles = new Vehicle[]
{
    new Car("Honda", "Accord", 2022, 4, "Передний", 4, "Седан", 192),
    new SportsCar("Porsche", "911 GT3", 2023, 4, "Задний", 2, "Купе", 502, 3.2, 318, true),
    new ElectricCar("Rivian", "R1T", 2023, 4, "Полный", 4, "Пикап", 835, 135, 505, 40),
    new Motorcycle("Ducati", "Panigale V4", 2023, 2, "Задний", "V4", false),
    new SportBike("Kawasaki", "Ninja H2R", 2023, 2, "Задний", "Supercharged", false, 15000, true)
};

foreach (var v in vehicles)
{
    Console.WriteLine();
    v.DisplayInfo();
    v.StartEngine();
}
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также сложное наследование.

````
using System;

// ========== УРОВЕНЬ 1: Базовый класс Animal ==========
public class Animal
{
    // Базовые свойства
    public string Name { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }
    
    // Статический счетчик
    public static int TotalAnimals { get; private set; } = 0;
    
    // Конструктор
    public Animal(string name, int age, double weight)
    {
        Name = name;
        Age = age;
        Weight = weight;
        TotalAnimals++;
    }
    
    // Виртуальные методы
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Имя: {Name}, Возраст: {Age} лет, Вес: {Weight} кг");
    }
    
    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} издает звук");
    }
    
    public virtual void Eat()
    {
        Console.WriteLine($"{Name} ест");
    }
}

// ========== УРОВЕНЬ 2: Млекопитающие (наследуют от Animal) ==========
public class Mammal : Animal
{
    // Свойства млекопитающих
    public bool HasFur { get; set; }
    public int GestationPeriod { get; set; } // дней
    public string Diet { get; set; } // травоядный, хищник, всеядный
    
    public Mammal(string name, int age, double weight, bool hasFur, int gestationPeriod, string diet)
        : base(name, age, weight)
    {
        HasFur = hasFur;
        GestationPeriod = gestationPeriod;
        Diet = diet;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Шерсть: {(HasFur ? "Да" : "Нет")}, Беременность: {GestationPeriod} дней, Рацион: {Diet}");
    }
    
    public virtual void GiveBirth()
    {
        Console.WriteLine($"{Name} рожает детеныша");
    }
    
    public virtual void FeedMilk()
    {
        Console.WriteLine($"{Name} кормит детеныша молоком");
    }
}

// ========== УРОВЕНЬ 3: Собачьи (наследуют от Mammal) ==========
public class Canine : Mammal
{
    // Свойства собачьих
    public string PackBehavior { get; set; } // стайное/одиночное
    public double HowlVolume { get; set; } // уровень воя в дБ
    
    public Canine(string name, int age, double weight, bool hasFur, int gestationPeriod, string diet,
                  string packBehavior, double howlVolume)
        : base(name, age, weight, hasFur, gestationPeriod, diet)
    {
        PackBehavior = packBehavior;
        HowlVolume = howlVolume;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Поведение: {PackBehavior}, Громкость воя: {HowlVolume} дБ");
    }
    
    public virtual void Howl()
    {
        Console.WriteLine($"{Name} воет с громкостью {HowlVolume} дБ: Ауууууу!");
    }
    
    public virtual void Hunt()
    {
        Console.WriteLine($"{Name} охотится стаей");
    }
}

// ========== УРОВЕНЬ 4: Волки (наследуют от Canine) ==========
public class Wolf : Canine
{
    // Специфичные для волков свойства
    public int PackSize { get; set; }
    public string TerritorySize { get; set; } // км²
    
    public Wolf(string name, int age, double weight, bool hasFur, int gestationPeriod, string diet,
                string packBehavior, double howlVolume, int packSize, string territorySize)
        : base(name, age, weight, hasFur, gestationPeriod, diet, packBehavior, howlVolume)
    {
        PackSize = packSize;
        TerritorySize = territorySize;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Размер стаи: {PackSize} особей, Территория: {TerritorySize} км²");
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"Волк {Name} издает протяжный вой, созывая стаю!");
    }
    
    public override void Hunt()
    {
        Console.WriteLine($"Стая волков из {PackSize} особей во главе с {Name} охотится на территории {TerritorySize} км²");
    }
    
    public void LeadPack()
    {
        Console.WriteLine($"Вожак стаи {Name} ведет охоту");
    }
}

// ========== УРОВЕНЬ 4: Собаки (наследуют от Canine) ==========
public class Dog : Canine
{
    // Специфичные для собак свойства
    public string Breed { get; set; }
    public bool IsTrained { get; set; }
    public string FavoriteToy { get; set; }
    
    public Dog(string name, int age, double weight, bool hasFur, int gestationPeriod, string diet,
               string packBehavior, double howlVolume, string breed, bool isTrained, string favoriteToy)
        : base(name, age, weight, hasFur, gestationPeriod, diet, packBehavior, howlVolume)
    {
        Breed = breed;
        IsTrained = isTrained;
        FavoriteToy = favoriteToy;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}, Дрессирована: {(IsTrained ? "Да" : "Нет")}, Любимая игрушка: {FavoriteToy}");
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"Собака {Name} лает: Гав-гав!");
    }
    
    public override void Hunt()
    {
        Console.WriteLine($"Собака {Name} играет с {FavoriteToy}");
    }
    
    public void PerformCommand(string command)
    {
        if (IsTrained)
            Console.WriteLine($"Собака {Name} выполняет команду: {command}");
        else
            Console.WriteLine($"Собака {Name} еще не дрессирована");
    }
    
    public void Fetch()
    {
        Console.WriteLine($"Собака {Name} приносит {FavoriteToy}");
    }
}

// ========== УРОВЕНЬ 3: Кошачьи (наследуют от Mammal) ==========
public class Feline : Mammal
{
    // Свойства кошачьих
    public bool HasRetractableClaws { get; set; }
    public double NightVision { get; set; } // уровень ночного зрения
    
    public Feline(string name, int age, double weight, bool hasFur, int gestationPeriod, string diet,
                  bool hasRetractableClaws, double nightVision)
        : base(name, age, weight, hasFur, gestationPeriod, diet)
    {
        HasRetractableClaws = hasRetractableClaws;
        NightVision = nightVision;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Втягивающиеся когти: {(HasRetractableClaws ? "Да" : "Нет")}, Ночное зрение: {NightVision}/10");
    }
    
    public virtual void Stalk()
    {
        Console.WriteLine($"{Name} крадется незаметно");
    }
    
    public virtual void Pounce()
    {
        Console.WriteLine($"{Name} прыгает на добычу!");
    }
}

// ========== УРОВЕНЬ 4: Львы (наследуют от Feline) ==========
public class Lion : Feline
{
    // Специфичные для львов свойства
    public bool HasMane { get; set; }
    public int PrideSize { get; set; }
    
    public Lion(string name, int age, double weight, bool hasFur, int gestationPeriod, string diet,
                bool hasRetractableClaws, double nightVision, bool hasMane, int prideSize)
        : base(name, age, weight, hasFur, gestationPeriod, diet, hasRetractableClaws, nightVision)
    {
        HasMane = hasMane;
        PrideSize = prideSize;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Грива: {(HasMane ? "Есть" : "Нет")}, Размер прайда: {PrideSize} особей");
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"Лев {Name} рычит: РРРРРРРР!");
    }
    
    public override void Stalk()
    {
        Console.WriteLine($"Лев {Name} крадется к добыче, используя высокую траву");
    }
    
    public void Roar()
    {
        Console.WriteLine($"Львиный рык {Name} слышен на 8 км!");
    }
}

// ========== УРОВЕНЬ 4: Домашние кошки (наследуют от Feline) ==========
public class HouseCat : Feline
{
    // Специфичные для домашних кошек свойства
    public string Color { get; set; }
    public bool IsIndoor { get; set; }
    public int LivesLeft { get; set; }
    
    public HouseCat(string name, int age, double weight, bool hasFur, int gestationPeriod, string diet,
                    bool hasRetractableClaws, double nightVision, string color, bool isIndoor, int livesLeft)
        : base(name, age, weight, hasFur, gestationPeriod, diet, hasRetractableClaws, nightVision)
    {
        Color = color;
        IsIndoor = isIndoor;
        LivesLeft = livesLeft;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Окрас: {Color}, Домашняя: {(IsIndoor ? "Да" : "Нет")}, Жизней осталось: {LivesLeft}");
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"Кошка {Name} мяукает: Мяу-мяу!");
    }
    
    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет");
    }
    
    public void Scratch()
    {
        if (HasRetractableClaws)
            Console.WriteLine($"{Name} точит когти");
    }
}

// ========== УРОВЕНЬ 2: Птицы (наследуют от Animal) ==========
public class Bird : Animal
{
    // Свойства птиц
    public double WingSpan { get; set; }
    public bool CanFly { get; set; }
    public string FeatherColor { get; set; }
    
    public Bird(string name, int age, double weight, double wingSpan, bool canFly, string featherColor)
        : base(name, age, weight)
    {
        WingSpan = wingSpan;
        CanFly = canFly;
        FeatherColor = featherColor;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Размах крыльев: {WingSpan} м, Летает: {(CanFly ? "Да" : "Нет")}, Цвет перьев: {FeatherColor}");
    }
    
    public virtual void Fly()
    {
        if (CanFly)
            Console.WriteLine($"{Name} летит с размахом крыльев {WingSpan} м");
        else
            Console.WriteLine($"{Name} не может летать");
    }
    
    public virtual void BuildNest()
    {
        Console.WriteLine($"{Name} строит гнездо");
    }
}

// ========== УРОВЕНЬ 3: Хищные птицы (наследуют от Bird) ==========
public class BirdOfPrey : Bird
{
    // Свойства хищных птиц
    public double TalonStrength { get; set; }
    public double VisionAcuity { get; set; } // острота зрения
    
    public BirdOfPrey(string name, int age, double weight, double wingSpan, bool canFly, string featherColor,
                      double talonStrength, double visionAcuity)
        : base(name, age, weight, wingSpan, canFly, featherColor)
    {
        TalonStrength = talonStrength;
        VisionAcuity = visionAcuity;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Сила когтей: {TalonStrength}/10, Острота зрения: {VisionAcuity}/10");
    }
    
    public virtual void Hunt()
    {
        Console.WriteLine($"{Name} высматривает добычу с высоты");
    }
    
    public virtual void Dive()
    {
        Console.WriteLine($"{Name} пикирует на добычу со скоростью!");
    }
}

// ========== УРОВЕНЬ 4: Орлы (наследуют от BirdOfPrey) ==========
public class Eagle : BirdOfPrey
{
    // Специфичные для орлов свойства
    public double Altitude { get; set; } // высота полета
    public string PreyType { get; set; }
    
    public Eagle(string name, int age, double weight, double wingSpan, bool canFly, string featherColor,
                 double talonStrength, double visionAcuity, double altitude, string preyType)
        : base(name, age, weight, wingSpan, canFly, featherColor, talonStrength, visionAcuity)
    {
        Altitude = altitude;
        PreyType = preyType;
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Высота полета: {Altitude} м, Основная добыча: {PreyType}");
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"Орел {Name} издает пронзительный клич!");
    }
    
    public override void Dive()
    {
        Console.WriteLine($"Орел {Name} пикирует на {PreyType} с высоты {Altitude} м!");
    }
    
    public void Soar()
    {
        Console.WriteLine($"Орел {Name} парит в небе на высоте {Altitude} м");
    }
}

// ========== Демонстрация сложного наследования для животных ==========
class AnimalHierarchyDemo
{
    public static void Run()
    {
        Console.WriteLine("========== СЛОЖНОЕ НАСЛЕДОВАНИЕ: ЖИВОТНЫЕ ==========\n");
        
        // Уровень 1: Animal
        Console.WriteLine("--- Уровень 1: Базовый Animal ---");
        Animal animal = new Animal("Generic Animal", 5, 50);
        animal.DisplayInfo();
        animal.MakeSound();
        
        Console.WriteLine("\n--- Уровень 2: Mammal ---");
        Mammal mammal = new Mammal("Generic Mammal", 3, 30, true, 90, "Всеядный");
        mammal.DisplayInfo();
        mammal.GiveBirth();
        mammal.FeedMilk();
        
        Console.WriteLine("\n--- Уровень 3: Canine ---");
        Canine canine = new Canine("Generic Canine", 4, 25, true, 63, "Хищник", "Стайное", 85);
        canine.DisplayInfo();
        canine.Howl();
        canine.Hunt();
        
        Console.WriteLine("\n--- Уровень 4: Wolf (наследует от Canine) ---");
        Wolf wolf = new Wolf("Альфа", 6, 65, true, 63, "Хищник", "Стайное", 95, 8, "500");
        wolf.DisplayInfo();
        wolf.MakeSound();
        wolf.Hunt();
        wolf.LeadPack();
        
        Console.WriteLine("\n--- Уровень 4: Dog (наследует от Canine) ---");
        Dog dog = new Dog("Рекс", 3, 30, true, 63, "Всеядный", "Социальное", 70, "Овчарка", true, "Мяч");
        dog.DisplayInfo();
        dog.MakeSound();
        dog.Hunt();
        dog.PerformCommand("Сидеть");
        dog.Fetch();
        
        Console.WriteLine("\n--- Уровень 3: Feline ---");
        Feline feline = new Feline("Generic Feline", 5, 40, true, 65, "Хищник", true, 9.5);
        feline.DisplayInfo();
        feline.Stalk();
        feline.Pounce();
        
        Console.WriteLine("\n--- Уровень 4: Lion (наследует от Feline) ---");
        Lion lion = new Lion("Симба", 8, 190, true, 110, "Хищник", true, 9.8, true, 12);
        lion.DisplayInfo();
        lion.MakeSound();
        lion.Stalk();
        lion.Roar();
        
        Console.WriteLine("\n--- Уровень 4: HouseCat (наследует от Feline) ---");
        HouseCat cat = new HouseCat("Муся", 2, 4.5, true, 63, "Хищник", true, 9.2, "Серый", true, 9);
        cat.DisplayInfo();
        cat.MakeSound();
        cat.Purr();
        cat.Scratch();
        
        Console.WriteLine("\n--- Уровень 2: Bird ---");
        Bird bird = new Bird("Generic Bird", 1, 0.5, 0.5, true, "Серый");
        bird.DisplayInfo();
        bird.Fly();
        bird.BuildNest();
        
        Console.WriteLine("\n--- Уровень 3: BirdOfPrey ---");
        BirdOfPrey birdOfPrey = new BirdOfPrey("Generic Raptor", 3, 2, 1.5, true, "Коричневый", 8, 9);
        birdOfPrey.DisplayInfo();
        birdOfPrey.Hunt();
        birdOfPrey.Dive();
        
        Console.WriteLine("\n--- Уровень 4: Eagle (наследует от BirdOfPrey) ---");
        Eagle eagle = new Eagle("Скай", 5, 6, 2.3, true, "Темно-коричневый", 9.5, 9.8, 3000, "Зайцы и рыба");
        eagle.DisplayInfo();
        eagle.MakeSound();
        eagle.Dive();
        eagle.Soar();
        
        // Демонстрация полиморфизма в многоуровневой иерархии
        Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА В ИЕРАРХИИ ===");
        Animal[] animals = new Animal[]
        {
            new Wolf("Белый", 4, 55, true, 63, "Хищник", "Стайное", 92, 6, "400"),
            new Dog("Шарик", 2, 20, true, 63, "Всеядный", "Социальное", 65, "Дворняга", false, "Кость"),
            new Lion("Нала", 7, 170, true, 110, "Хищник", true, 9.6, false, 15),
            new HouseCat("Васька", 4, 5, true, 63, "Хищник", true, 9.0, "Черный", false, 7),
            new Eagle("Круз", 8, 7, 2.5, true, "Золотистый", 9.8, 9.9, 3500, "Змеи")
        };
        
        foreach (var a in animals)
        {
            Console.WriteLine();
            a.DisplayInfo();
            a.MakeSound();
        }
        
        Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ СПЕЦИФИЧНЫХ МЕТОДОВ ===");
        foreach (var a in animals)
        {
            if (a is Wolf w) w.LeadPack();
            else if (a is Dog d) d.Fetch();
            else if (a is Lion l) l.Roar();
            else if (a is HouseCat c) c.Purr();
            else if (a is Eagle e) e.Soar();
        }
    }
}

// ========== ЗАПУСК ВСЕХ ДЕМОНСТРАЦИЙ ==========

Console.WriteLine("\n" + new string('=', 60) + "\n");
AnimalHierarchyDemo.Run();
````
