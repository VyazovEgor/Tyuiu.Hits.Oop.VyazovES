<h1 style="color:DodgerBlue">Простое наследование</h1>


Простое наследование в C# позволяет создавать новые классы на основе существующих, наследуя их свойства и методы. Это мощный механизм для повторного использования кода и организации программ на основе иерархии классов. Давайте рассмотрим пример наследования с классами `Vehicle`, `Car` и `Motorcycle`.

### Шаг 1: Определение базового класса

Сначала определим базовый класс `Vehicle`. Этот класс будет содержать общие свойства и методы, которые могут быть у всех транспортных средств.

```csharp
public class Vehicle
{
    // Свойство для хранения марки транспортного средства
    public string Make { get; set; }

    // Свойство для хранения модели транспортного средства
    public string Model { get; set; }

    // Конструктор класса Vehicle
    public Vehicle(string make, string model)
    {
        Make = make;
        Model = model;
    }

    // Метод для отображения информации о транспортном средстве
    public void DisplayInfo()
    {
        Console.WriteLine($"Производитель: {Make}, Модель: {Model}");
    }
}
```

### Шаг 2: Создание производных классов

Теперь создадим два производных класса `Car` и `Motorcycle`, которые наследуют от базового класса `Vehicle`.

#### Класс Car

Класс `Car` добавит дополнительные свойства, специфичные для автомобилей, например, количество дверей.

```csharp
public class Car : Vehicle
{
    // Свойство для хранения количества дверей
    public int Doors { get; set; }

    // Конструктор класса Car
    public Car(string make, string model, int doors) : base(make, model)
    {
        Doors = doors;
    }

    // Переопределенный метод DisplayInfo для отображения информации о автомобиле
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Двери: {Doors}");
    }
}
```

#### Класс Motorcycle

Аналогично, класс `Motorcycle` может добавить свои специфические свойства, например, тип двигателя.

```csharp
public class Motorcycle : Vehicle
{
    // Свойство для хранения типа двигателя
    public string EngineType { get; set; }

    // Конструктор класса Motorcycle
    public Motorcycle(string make, string model, string engineType) : base(make, model)
    {
        EngineType = engineType;
    }

    // Переопределенный метод DisplayInfo для отображения информации о мотоцикле
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Вид двигателя: {EngineType}");
    }
}
```

### Шаг 3: Использование классов

Теперь мы можем создать экземпляры наших классов и использовать их.

```csharp
// Создаем объект класса Car
Car myCar = new Car("Toyota", "Corolla", 4);
myCar.DisplayInfo(); // Выводит информацию о автомобиле

// Создаем объект класса Motorcycle
Motorcycle myMotorcycle = new Motorcycle("Honda", "CBR600RR", "Inline Four");
myMotorcycle.DisplayInfo(); // Выводит информацию о мотоцикле


```

### Объяснение кода

- **Базовый класс `Vehicle`**: Определяет общую структуру для всех транспортных средств, включая свойства `Make` и `Model`, а также метод `DisplayInfo()` для вывода информации.
- **Производные классы `Car` и `Motorcycle`**: Наследуют от `Vehicle` и добавляют собственные свойства (`Doors` для `Car` и `EngineType` для `Motorcycle`). Также переопределяют метод `DisplayInfo()` для вывода дополнительной информации.
- **Использование классов**: В `Main` создаются экземпляры `Car` и `Motorcycle`, и вызывается метод `DisplayInfo()` для каждого из них, демонстрируя различие в информации, которую они выводят.

Этот пример показывает, как можно использовать простое наследование в C# для создания иерархии классов, где каждый класс расширяет функциональность предка, сохраняя при этом возможность индивидуализации своих свойств и поведения.

````
using System;

// Базовый класс Vehicle
public class Vehicle
{
    // Свойство для хранения марки транспортного средства
    public string Make { get; set; }

    // Свойство для хранения модели транспортного средства
    public string Model { get; set; }

    // Конструктор класса Vehicle
    public Vehicle(string make, string model)
    {
        Make = make;
        Model = model;
    }

    // Виртуальный метод для отображения информации о транспортном средстве
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Производитель: {Make}, Модель: {Model}");
    }
}

// Производный класс Car (наследует от Vehicle)
public class Car : Vehicle
{
    // Свойство для хранения количества дверей
    public int Doors { get; set; }
    
    // Свойство для хранения типа кузова
    public string BodyType { get; set; }

    // Конструктор класса Car
    public Car(string make, string model, int doors, string bodyType) : base(make, model)
    {
        Doors = doors;
        BodyType = bodyType;
    }

    // Переопределенный метод DisplayInfo для отображения информации о автомобиле
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Количество дверей: {Doors}");
        Console.WriteLine($"Тип кузова: {BodyType}");
    }
    
    // Специфичный метод для автомобиля
    public void Drive()
    {
        Console.WriteLine($"Автомобиль {Make} {Model} едет по дороге");
    }
}

// Производный класс Motorcycle (наследует от Vehicle)
public class Motorcycle : Vehicle
{
    // Свойство для хранения типа двигателя
    public string EngineType { get; set; }
    
    // Свойство для хранения типа мотоцикла (спортивный, круизер и т.д.)
    public string MotorcycleType { get; set; }

    // Конструктор класса Motorcycle
    public Motorcycle(string make, string model, string engineType, string motorcycleType) 
        : base(make, model)
    {
        EngineType = engineType;
        MotorcycleType = motorcycleType;
    }

    // Переопределенный метод DisplayInfo для отображения информации о мотоцикле
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Тип двигателя: {EngineType}");
        Console.WriteLine($"Тип мотоцикла: {MotorcycleType}");
    }
    
    // Специфичный метод для мотоцикла
    public void Ride()
    {
        Console.WriteLine($"Мотоцикл {Make} {Model} мчится по трассе");
    }
}

// Третий производный класс Truck (Грузовик) для дополнительной демонстрации
public class Truck : Vehicle
{
    // Свойство для хранения грузоподъемности
    public double LoadCapacity { get; set; }
    
    // Свойство для хранения количества осей
    public int Axles { get; set; }

    // Конструктор класса Truck
    public Truck(string make, string model, double loadCapacity, int axles) 
        : base(make, model)
    {
        LoadCapacity = loadCapacity;
        Axles = axles;
    }

    // Переопределенный метод DisplayInfo
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Грузоподъемность: {LoadCapacity} тонн");
        Console.WriteLine($"Количество осей: {Axles}");
    }
    
    // Специфичный метод для грузовика
    public void LoadCargo()
    {
        Console.WriteLine($"Грузовик {Make} {Model} загружен на {LoadCapacity} тонн");
    }
}

// Демонстрация работы классов
Console.WriteLine("=== ДЕМОНСТРАЦИЯ ПРОСТОГО НАСЛЕДОВАНИЯ (ТРАНСПОРТНЫЕ СРЕДСТВА) ===\n");
        
// Создаем объект класса Car
Console.WriteLine("--- Автомобиль ---");
Car myCar = new Car("Toyota", "Corolla", 4, "Седан");
myCar.DisplayInfo();
myCar.Drive();

Console.WriteLine("\n--- Мотоцикл ---");
// Создаем объект класса Motorcycle
Motorcycle myMotorcycle = new Motorcycle("Honda", "CBR600RR", "Inline Four", "Спортивный");
myMotorcycle.DisplayInfo();
myMotorcycle.Ride();

Console.WriteLine("\n--- Грузовик ---");
// Создаем объект класса Truck
Truck myTruck = new Truck("Volvo", "FH16", 25.5, 3);
myTruck.DisplayInfo();
myTruck.LoadCargo();

Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА ===");

// Массив объектов базового класса
Vehicle[] vehicles = new Vehicle[]
{
    new Car("BMW", "X5", 5, "Внедорожник"),
    new Motorcycle("Yamaha", "R1", "V4", "Спортивный"),
    new Truck("MAN", "TGX", 30, 4),
    new Car("Ford", "Focus", 4, "Хэтчбек")
};

// Полиморфный вызов метода DisplayInfo
foreach (var vehicle in vehicles)
{
    Console.WriteLine();
    vehicle.DisplayInfo();
}
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также простое наследование.

````
using System;

// Базовый класс Animal
public class Animal
{
    // Свойства для хранения имени и возраста животного
    public string Name { get; set; }
    public int Age { get; set; }
    
    // Свойство для хранения вида животного
    public string Species { get; set; }

    // Конструктор класса Animal
    public Animal(string name, int age, string species)
    {
        Name = name;
        Age = age;
        Species = species;
    }

    // Виртуальный метод для отображения информации о животном
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Имя: {Name}, Возраст: {Age} лет, Вид: {Species}");
    }
    
    // Виртуальный метод для издания звука
    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} издает звук");
    }
}

// Производный класс Dog (Собака)
public class Dog : Animal
{
    // Свойство для хранения породы
    public string Breed { get; set; }
    
    // Свойство для хранения уровня дрессировки
    public bool IsTrained { get; set; }

    // Конструктор класса Dog
    public Dog(string name, int age, string breed, bool isTrained) 
        : base(name, age, "Собака")
    {
        Breed = breed;
        IsTrained = isTrained;
    }

    // Переопределенный метод DisplayInfo
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}");
        Console.WriteLine($"Дрессирована: {(IsTrained ? "Да" : "Нет")}");
    }
    
    // Переопределенный метод MakeSound
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} лает: Гав-гав!");
    }
    
    // Специфичный метод для собаки
    public void Bark()
    {
        Console.WriteLine($"{Name} громко лает!");
    }
    
    // Метод для выполнения команды
    public void PerformCommand(string command)
    {
        if (IsTrained)
            Console.WriteLine($"{Name} выполняет команду: {command}");
        else
            Console.WriteLine($"{Name} еще не дрессирован(а)");
    }
}

// Производный класс Cat (Кошка)
public class Cat : Animal
{
    // Свойство для хранения окраса
    public string Color { get; set; }
    
    // Свойство для хранения домашняя или уличная
    public bool IsIndoor { get; set; }

    // Конструктор класса Cat
    public Cat(string name, int age, string color, bool isIndoor) 
        : base(name, age, "Кошка")
    {
        Color = color;
        IsIndoor = isIndoor;
    }

    // Переопределенный метод DisplayInfo
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Окрас: {Color}");
        Console.WriteLine($"Домашняя: {(IsIndoor ? "Да" : "Нет")}");
    }
    
    // Переопределенный метод MakeSound
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} мяукает: Мяу-мяу!");
    }
    
    // Специфичный метод для кошки
    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет");
    }
    
    // Метод для охоты
    public void Hunt()
    {
        if (!IsIndoor)
            Console.WriteLine($"{Name} охотится на мышей");
        else
            Console.WriteLine($"{Name} домашняя кошка и не охотится");
    }
}

// Производный класс Bird (Птица)
public class Bird : Animal
{
    // Свойство для хранения размаха крыльев
    public double WingSpan { get; set; }
    
    // Свойство для хранения может ли летать
    public bool CanFly { get; set; }

    // Конструктор класса Bird
    public Bird(string name, int age, double wingSpan, bool canFly) 
        : base(name, age, "Птица")
    {
        WingSpan = wingSpan;
        CanFly = canFly;
    }

    // Переопределенный метод DisplayInfo
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Размах крыльев: {WingSpan} м");
        Console.WriteLine($"Умеет летать: {(CanFly ? "Да" : "Нет")}");
    }
    
    // Переопределенный метод MakeSound
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} чирикает: Чик-чирик!");
    }
    
    // Специфичный метод для птицы
    public void Fly()
    {
        if (CanFly)
            Console.WriteLine($"{Name} летит с размахом крыльев {WingSpan} м");
        else
            Console.WriteLine($"{Name} не может летать");
    }
    
    // Метод для строительства гнезда
    public void BuildNest()
    {
        Console.WriteLine($"{Name} строит гнездо");
    }
}

// Производный класс Fish (Рыба) - четвертый класс для демонстрации
public class Fish : Animal
{
    // Свойство для хранения типа воды (пресная/соленая)
    public string WaterType { get; set; }
    
    // Свойство для хранения глубины обитания
    public double Depth { get; set; }

    // Конструктор класса Fish
    public Fish(string name, int age, string waterType, double depth) 
        : base(name, age, "Рыба")
    {
        WaterType = waterType;
        Depth = depth;
    }

    // Переопределенный метод DisplayInfo
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Тип воды: {WaterType}");
        Console.WriteLine($"Глубина обитания: {Depth} м");
    }
    
    // Переопределенный метод MakeSound
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} издает булькающие звуки: Буль-буль!");
    }
    
    // Специфичный метод для рыбы
    public void Swim()
    {
        Console.WriteLine($"{Name} плавает в {WaterType} воде на глубине {Depth} м");
    }
}

// Демонстрация работы классов

Console.WriteLine("=== ДЕМОНСТРАЦИЯ ПРОСТОГО НАСЛЕДОВАНИЯ (ЖИВОТНЫЕ) ===\n");

// Создаем объект класса Dog
Console.WriteLine("--- Собака ---");
Dog myDog = new Dog("Рекс", 3, "Овчарка", true);
myDog.DisplayInfo();
myDog.MakeSound();
myDog.Bark();
myDog.PerformCommand("Сидеть");

Console.WriteLine("\n--- Кошка ---");
// Создаем объект класса Cat
Cat myCat = new Cat("Муся", 2, "Серый", true);
myCat.DisplayInfo();
myCat.MakeSound();
myCat.Purr();
myCat.Hunt();

Console.WriteLine("\n--- Птица ---");
// Создаем объект класса Bird
Bird myBird = new Bird("Кеша", 1, 0.3, true);
myBird.DisplayInfo();
myBird.MakeSound();
myBird.Fly();
myBird.BuildNest();

Console.WriteLine("\n--- Рыба ---");
// Создаем объект класса Fish
Fish myFish = new Fish("Немо", 2, "Соленая", 15);
myFish.DisplayInfo();
myFish.MakeSound();
myFish.Swim();

Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА ===");

// Массив объектов базового класса
Animal[] animals = new Animal[]
{
    new Dog("Шарик", 4, "Дворняга", false),
    new Cat("Васька", 5, "Черный", false),
    new Bird("Пингвин", 3, 0.5, false),
    new Fish("Золотая рыбка", 1, "Пресная", 0.5),
    new Dog("Лайка", 2, "Хаски", true)
};

// Полиморфный вызов методов DisplayInfo и MakeSound
foreach (var animal in animals)
{
    Console.WriteLine();
    animal.DisplayInfo();
    animal.MakeSound();
}

Console.WriteLine("\n=== ДОПОЛНИТЕЛЬНАЯ ДЕМОНСТРАЦИЯ ===");

// Демонстрация специфичных методов через приведение типов
foreach (var animal in animals)
{
    if (animal is Dog dog)
    {
        dog.Bark();
    }
    else if (animal is Cat cat)
    {
        cat.Purr();
    }
    else if (animal is Bird bird)
    {
        bird.Fly();
    }
    else if (animal is Fish fish)
    {
        fish.Swim();
    }
}
````
