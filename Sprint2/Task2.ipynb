<h1 style="color:DodgerBlue">Множественное наследование</h1>

Множественное наследование — это концепция, при которой класс может наследовать свойства и методы сразу от нескольких базовых классов. 

Множественное наследование — это особенность некоторых языков программирования, которая позволяет классу наследовать свойства и методы от двух или более суперклассов. Несмотря на то, что множественное наследование может привести к увеличению гибкости и повторного использования кода, оно также может усложнить архитектуру программы и привести к проблемам, таким как "проблема ромба". Вот некоторые языки программирования, которые поддерживают множественное наследование:

1. **C++**: Один из наиболее известных языков, поддерживающих множественное наследование. В C++, класс может наследовать свойства и методы от нескольких базовых классов, что позволяет создавать сложные иерархии наследования.

2. **Python**: Поддерживает множественное наследование, что позволяет классу наследовать от нескольких базовых классов. Python использует алгоритм C3 для разрешения порядка методов при множественном наследовании, чтобы избежать неоднозначности.

3. **Perl и Perl 6 (Raku)**: Оба языка поддерживают множественное наследование. В Perl это достигается через механизм "исходных пакетов" (source packages), а в Raku — через более явные средства.

4. **Common Lisp (через CLOS)**: Common Lisp Object System (CLOS) поддерживает множественное наследование, позволяя объектам быть экземплярами нескольких классов одновременно.

5. **Eiffel**: Язык программирования Eiffel также поддерживает множественное наследование, предлагая строгие типы и контрактное программирование.

6. **Tcl**: Скриптовый язык Tcl поддерживает множественное наследование через систему объектов, введенную в Tcl 8.6.

7. **Ada**: Язык программирования Ada поддерживает множественное наследование через механизм интерфейсов и классов.

Множественное наследование может быть мощным инструментом для создания гибких и модульных систем, но его использование требует тщательного планирования и понимания потенциальных проблем, таких как "проблема ромба". 

"Проблема ромба" — это термин в объектно-ориентированном программировании, который описывает ситуацию, возникающую при множественном наследовании, когда один класс наследуется от двух или более базовых классов, которые сами по себе наследуются от общего базового класса. Это создает конфигурацию в форме ромба и может привести к неоднозначности и сложностям в определении порядка вызова методов или доступа к свойствам

В некоторых языках, например в Java и C#, вместо прямой поддержки множественного наследования используются интерфейсы или другие механизмы для достижения подобного функционала.


Давайте рассмотрим пример в C# с использованием классов `Vehicle`, `Car`, и `Motorcycle`, а также введем интерфейсы для демонстрации множественного наследования через интерфейсы.

### Шаг 1: Определение интерфейсов

Сначала определим два интерфейса, которые будут представлять различные аспекты транспортных средств.


```csharp

public interface IVehicle
{
    string Make { get; set; }
    string Model { get; set; }
    void DisplayInfo();
}

public interface IMotorized
{
    string EngineType { get; set; }
}

```

### Шаг 2: Реализация интерфейсов в классах

Теперь определим классы `Car` и `Motorcycle`, которые реализуют эти интерфейсы.

#### Класс Car

```csharp
public class Car : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    public int Doors { get; set; }

    public Car(string make, string model, int doors)
    {
        Make = make;
        Model = model;
        Doors = doors;
    }

    public void DisplayInfo()
    {
        Console.WriteLine($"Автомобиль: Производитель: {Make}, Модель: {Model}, Двери: {Doors}");
    }
}
```

#### Класс Motorcycle

```csharp
public class Motorcycle : IVehicle, IMotorized
{
    public string Make { get; set; }
    public string Model { get; set; }
    public string EngineType { get; set; }

    public Motorcycle(string make, string model, string engineType)
    {
        Make = make;
        Model = model;
        EngineType = engineType;
    }

    public void DisplayInfo()
    {
        Console.WriteLine($"Мотоцикл: Производитель: {Make}, Модель: {Model}, Тип двигетеля: {EngineType}");
    }
}
```

### Шаг 3: Использование классов

Теперь мы можем создать экземпляры наших классов и использовать их.

```csharp

        Car myCar = new Car("Toyota", "Corolla", 4);
        myCar.DisplayInfo();

        Motorcycle myMotorcycle = new Motorcycle("Honda", "CBR600RR", "Inline Four");
        myMotorcycle.DisplayInfo();
```

### Объяснение кода

- **Интерфейсы `IVehicle` и `IMotorized`** определяют контракты для классов, которые хотят представлять транспортные средства или моторизированные объекты соответственно.
- **Класс `Car`** реализует интерфейс `IVehicle`, предоставляя информацию о марке, модели и количестве дверей автомобиля.
- **Класс `Motorcycle`** реализует оба интерфейса (`IVehicle` и `IMotorized`), что демонстрирует множественное наследование через интерфейсы в C#. Это позволяет мотоциклу иметь свойства как транспортного средства, так и моторизированного объекта.

### Отличие множественного наследования от сложного

- **Поддержка языком**: В C# непосредственное множественное наследование классов не поддерживается, но можно использовать интерфейсы для достижения подобного поведения.
- **Гибкость**: Интерфейсы позволяют классам реализовывать множество контрактов, что обеспечивает большую гибкость в проектировании программ.
- **Расширяемость**: Использование интерфейсов облегчает добавление новых функций и поведений без изменения существующего кода, поскольку новые интерфейсы могут быть реализованы существующими классами без их модификации.

Таким образом, хотя C# не поддерживает множественное наследование напрямую для классов, использование интерфейсов предоставляет мощный механизм для реализации подобного поведения, обеспечивая гибкость и расширяемость приложений.

Множественное наследование в C# позволяет классу наследовать свойства и методы от нескольких базовых классов. Однако, стоит отметить, что прямое множественное наследование (когда класс наследуется от двух или более классов) не поддерживается в C#. Вместо этого, для достижения подобного поведения используется интерфейс или абстрактный класс как общий базовый класс, а затем реализация или наследование от других классов для расширения функциональности.

Давайте рассмотрим пример с использованием абстрактного класса для демонстрации множественного наследования. Предположим, у нас есть абстрактный класс `Vehicle`, который представляет собой транспортное средство, и два интерфейса: `Car` и `Motorcycle`. Класс `Car` и `Motorcycle` будут реализовывать эти интерфейсы, предоставляя свою специфику.

### Шаг 1: Определение абстрактного класса

```csharp
// Абстрактный класс Vehicle, представляющий транспортное средство
public abstract class Vehicle
{
    // Свойство Speed для определения скорости транспортного средства
    public int Speed { get; set; }

    // Конструктор по умолчанию
    protected Vehicle()
    {
        Speed = 0;
    }

    // Метод Move, который будет переопределен в производных классах
    public virtual void Move()
    {
        Console.WriteLine("Транспортное средство движется.");
    }
}
```

### Шаг 2: Определение интерфейсов

```csharp
// Интерфейс Car, определяющий специфику автомобиля
public interface ICar
{
    void Honk(); // Метод для сигнала гона
}

// Интерфейс Motorcycle, определяющий специфику мотоцикла
public interface IMotorcycle
{
    void Wheelie(); // Метод для выполнения трюка "колесо"
}
```

### Шаг 3: Реализация классов

Теперь мы можем создать классы `Car` и `Motorcycle`, которые реализуют интерфейсы `ICar` и `IMotorcycle` соответственно, и наследуются от `Vehicle`.

```csharp
// Класс Car, реализующий интерфейс ICar и наследующий от Vehicle
public class Car : Vehicle, ICar
{
    public Car() : base()
    {
    }

    public void Honk()
    {
        Console.WriteLine("Автомобиль издает звук сигнала клаксона");
    }

    // Переопределение метода Move
    public override void Move()
    {
        base.Move();
        Console.WriteLine("Автомобиль движется со скоростью " + Speed);
    }
}

// Класс Motorcycle, реализующий интерфейс IMotorcycle и наследующий от Vehicle
public class Motorcycle : Vehicle, IMotorcycle
{
    public Motorcycle() : base()
    {
    }

    public void Wheelie()
    {
        Console.WriteLine("Мотоциклист выполняет трюк 'колесо'");
    }

    // Переопределение метода Move
    public override void Move()
    {
        base.Move();
        Console.WriteLine("Мотоцикл движется со скоростью " + Speed);
    }
}
```

### Объяснение кода

- **Абстрактный класс `Vehicle`** служит базовым классом для всех транспортных средств. Он содержит общую логику и свойства, такие как скорость (`Speed`). Метод `Move()` является виртуальным, что позволяет его переопределять в производных классах.
  
- **Интерфейсы `ICar` и `IMotorcycle`** определяют специфические для каждого типа транспортного средства действия, которые могут быть выполнены этим типом транспортного средства.

- **Классы `Car` и `Motorcycle`** реализуют соответствующие интерфейсы и наследуют от `Vehicle`, тем самым объединяя в себе общие и специфические характеристики транспортных средств.

Этот подход позволяет достичь многого, что можно было бы сделать при прямом множественном наследовании, но с большей гибкостью и в соответствии с принципами объектно-ориентированного программирования в C#.

## Разница между абстракными классами и интерфейсами
Абстрактные классы и интерфейсы в C# используются для определения контрактов и шаблонов поведения, но они имеют ключевые различия, которые влияют на их использование и возможности:

### 1. Наследование vs. Реализация

- **Абстрактные классы** могут быть наследованы другими классами. Это позволяет создавать иерархии классов, где подклассы наследуют свойства и методы от родительского абстрактного класса.
- **Интерфейсы** должны быть реализованы классами. Класс может реализовать несколько интерфейсов, что позволяет использовать множественное наследование по контрактам.

### 2. Членские переменные

- **Абстрактные классы** могут содержать членские переменные (поля), включая приватные и защищенные. Это позволяет хранить состояние в абстрактном классе.
- **Интерфейсы** не могут содержать членские переменные. Все члены интерфейса должны быть явно помечены как `abstract`, если они не реализованы.

### 3. Реализация по умолчанию и операторы

- **С C# 8.0 и выше**, интерфейсы могут содержать реализации по умолчанию для своих члена. Это позволяет предоставлять стандартную реализацию, которую классы могут использовать без необходимости повторной реализации.
- Абстрактные классы также могут содержать реализации по умолчанию для своих методов, но это менее распространено, поскольку абстрактные классы обычно используются как основа для наследования.

### 4. Приватные члены

- **Абстрактные классы** могут иметь приватные члены, доступ к которым ограничен самим классом и его подклассами.
- **Интерфейсы** не могут иметь приватных членов, так как они предназначены только для определения контракта, который должен быть реализован классами.

### 5. Использование

- **Абстрактные классы** часто используются для создания иерархий классов, где базовый класс определяет общую логику и свойства, а подклассы расширяют или модифицируют эту логику.
- **Интерфейсы** идеально подходят для определения контрактов между классами, особенно когда необходимо обеспечить высокую степень независимости и гибкости в дизайне системы.

### Заключение

Выбор между абстрактным классом и интерфейсом зависит от конкретных требований проекта. Если вам нужна иерархия классов с общими свойствами и методами, используйте абстрактный класс. Если вы хотите определить контракты, которые могут быть реализованы несколькими классами, используйте интерфейсы.

````
using System;

// ========== ИНТЕРФЕЙСЫ ДЛЯ РАЗЛИЧНЫХ ХАРАКТЕРИСТИК ==========

// Интерфейс для транспортных средств с двигателем
public interface IEnginePowered
{
    int Horsepower { get; set; }
    double EngineDisplacement { get; set; } // Изменено на double для объема двигателя
    void StartEngine();
    void StopEngine();
    void RevEngine();
}

// Интерфейс для электрических транспортных средств
public interface IElectricVehicle
{
    int BatteryCapacity { get; set; }
    int Range { get; set; }
    void Charge();
    void CheckBattery();
}

// Интерфейс для транспортных средств с возможностью полета
public interface IFlyable
{
    int MaxAltitude { get; set; }
    void TakeOff();
    void Land();
    void Fly();
}

// Интерфейс для водных транспортных средств
public interface IWaterVehicle
{
    double Displacement { get; set; }
    void Sail();
    void Anchor();
}

// Интерфейс для гоночных транспортных средств
public interface IRaceCapable
{
    double AccelerationTo100 { get; set; }
    int MaxSpeed { get; set; }
    void ActivateSportMode();
    void Drift();
}

// Интерфейс для транспортных средств с автопилотом
public interface IAutopilot
{
    int AutopilotLevel { get; set; }
    void EnableAutopilot();
    void DisableAutopilot();
    void SelfPark();
}

// Интерфейс для транспортных средств с возможностью буксировки
public interface ITowCapable
{
    double TowingCapacity { get; set; }
    void Tow(Vehicle vehicle);
}

// ========== БАЗОВЫЙ КЛАСС ==========
public class Vehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    public int Year { get; set; }
    public string Color { get; set; }
    public static int TotalVehicles { get; private set; } = 0;
    
    public Vehicle(string make, string model, int year, string color)
    {
        Make = make;
        Model = model;
        Year = year;
        Color = color;
        TotalVehicles++;
    }
    
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Производитель: {Make}, Модель: {Model}, Год: {Year}, Цвет: {Color}");
    }
}

// ========== КЛАССЫ С МНОЖЕСТВЕННЫМ НАСЛЕДОВАНИЕМ (ИНТЕРФЕЙСЫ) ==========

// 1. Электромобиль-спорткар (наследует: IEnginePowered, IElectricVehicle, IRaceCapable)
public class ElectricSportsCar : Vehicle, IEnginePowered, IElectricVehicle, IRaceCapable
{
    // IEnginePowered
    public int Horsepower { get; set; }
    public double EngineDisplacement { get; set; }
    
    // IElectricVehicle
    public int BatteryCapacity { get; set; }
    public int Range { get; set; }
    
    // IRaceCapable
    public double AccelerationTo100 { get; set; }
    public int MaxSpeed { get; set; }
    
    // Собственные свойства
    public bool HasLudicrousMode { get; set; }
    
    public ElectricSportsCar(string make, string model, int year, string color, 
                             int horsepower, int batteryCapacity, int range, 
                             double accelerationTo100, int maxSpeed, bool hasLudicrousMode)
        : base(make, model, year, color)
    {
        Horsepower = horsepower;
        BatteryCapacity = batteryCapacity;
        Range = range;
        AccelerationTo100 = accelerationTo100;
        MaxSpeed = maxSpeed;
        HasLudicrousMode = hasLudicrousMode;
        EngineDisplacement = 0; // Электромобили не имеют объема двигателя
    }
    
    // Реализация IEnginePowered
    public void StartEngine()
    {
        Console.WriteLine($"Электромотор {Make} {Model} активирован (тихий гул)");
    }
    
    public void StopEngine()
    {
        Console.WriteLine($"Электромотор {Make} {Model} деактивирован");
    }
    
    public void RevEngine()
    {
        Console.WriteLine($"Электромотор издает футуристический звук: WOOOOSH!");
    }
    
    // Реализация IElectricVehicle
    public void Charge()
    {
        Console.WriteLine($"Зарядка {Make} {Model}... {BatteryCapacity} кВт/ч, запас хода {Range} км");
    }
    
    public void CheckBattery()
    {
        Console.WriteLine($"Уровень заряда батареи: 85%");
    }
    
    // Реализация IRaceCapable
    public void ActivateSportMode()
    {
        Console.WriteLine($"Спортивный режим активирован! Разгон 0-100 за {AccelerationTo100} сек");
        if (HasLudicrousMode)
            Console.WriteLine("LUDICROUS MODE ACTIVATED! МАКСИМАЛЬНОЕ УСКОРЕНИЕ!");
    }
    
    public void Drift()
    {
        Console.WriteLine($"Электроспорткар {Make} {Model} дрифтует с мгновенным крутящим моментом!");
    }
    
    // Переопределение DisplayInfo
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Мощность: {Horsepower} л.с., Батарея: {BatteryCapacity} кВт/ч");
        Console.WriteLine($"Запас хода: {Range} км, 0-100: {AccelerationTo100} сек, Макс. скорость: {MaxSpeed} км/ч");
        Console.WriteLine($"Ludicrous Mode: {(HasLudicrousMode ? "Да" : "Нет")}");
    }
}

// 2. Летающий автомобиль (наследует: IEnginePowered, IFlyable)
public class FlyingCar : Vehicle, IEnginePowered, IFlyable
{
    // IEnginePowered
    public int Horsepower { get; set; }
    public double EngineDisplacement { get; set; }
    
    // IFlyable
    public int MaxAltitude { get; set; }
    
    // Собственные свойства
    public double WingSpan { get; set; }
    public bool IsFlying { get; private set; } = false;
    
    public FlyingCar(string make, string model, int year, string color,
                     int horsepower, double engineDisplacement, int maxAltitude, double wingSpan)
        : base(make, model, year, color)
    {
        Horsepower = horsepower;
        EngineDisplacement = engineDisplacement;
        MaxAltitude = maxAltitude;
        WingSpan = wingSpan;
    }
    
    // Реализация IEnginePowered
    public void StartEngine()
    {
        Console.WriteLine($"Двигатель {Make} {Model} запущен! Можно ехать или лететь");
    }
    
    public void StopEngine()
    {
        Console.WriteLine($"Двигатель {Make} {Model} остановлен");
        if (IsFlying)
        {
            Console.WriteLine("ВНИМАНИЕ! НЕЛЬЗЯ ОСТАНАВЛИВАТЬ ДВИГАТЕЛЬ В ПОЛЕТЕ!");
        }
    }
    
    public void RevEngine()
    {
        Console.WriteLine($"Двигатель ревет: VRRROOOOOM!");
    }
    
    // Реализация IFlyable
    public void TakeOff()
    {
        IsFlying = true;
        Console.WriteLine($"{Make} {Model} взлетает! Крылья раскрыты, размах {WingSpan} м");
    }
    
    public void Land()
    {
        IsFlying = false;
        Console.WriteLine($"{Make} {Model} совершает посадку");
    }
    
    public void Fly()
    {
        if (IsFlying)
            Console.WriteLine($"{Make} {Model} летит на высоте {MaxAltitude} м");
        else
            Console.WriteLine($"Сначала нужно взлететь! Вызовите TakeOff()");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Мощность: {Horsepower} л.с., Объем: {EngineDisplacement} л");
        Console.WriteLine($"Макс. высота полета: {MaxAltitude} м, Размах крыльев: {WingSpan} м");
        Console.WriteLine($"Режим: {(IsFlying ? "ПОЛЕТ" : "НА ЗЕМЛЕ")}");
    }
}

// 3. Амфибия (наследует: IEnginePowered, IWaterVehicle)
public class AmphibiousVehicle : Vehicle, IEnginePowered, IWaterVehicle
{
    // IEnginePowered
    public int Horsepower { get; set; }
    public double EngineDisplacement { get; set; }
    
    // IWaterVehicle
    public double Displacement { get; set; }
    
    // Собственные свойства
    public bool OnWater { get; private set; } = false;
    public string PropellerType { get; set; }
    
    public AmphibiousVehicle(string make, string model, int year, string color,
                             int horsepower, double engineDisplacement, double displacement, string propellerType)
        : base(make, model, year, color)
    {
        Horsepower = horsepower;
        EngineDisplacement = engineDisplacement;
        Displacement = displacement;
        PropellerType = propellerType;
    }
    
    // Реализация IEnginePowered
    public void StartEngine()
    {
        Console.WriteLine($"Двигатель {Make} {Model} запущен");
    }
    
    public void StopEngine()
    {
        Console.WriteLine($"Двигатель {Make} {Model} остановлен");
    }
    
    public void RevEngine()
    {
        Console.WriteLine($"VRRROOOOOM! Амфибия готова к движению!");
    }
    
    // Реализация IWaterVehicle
    public void Sail()
    {
        OnWater = true;
        Console.WriteLine($"{Make} {Model} спущен на воду! Винт {PropellerType} вращается");
    }
    
    public void Anchor()
    {
        OnWater = false;
        Console.WriteLine($"{Make} {Model} бросил якорь");
    }
    
    // Специфичные методы
    public void DriveOnLand()
    {
        OnWater = false;
        Console.WriteLine($"{Make} {Model} выезжает на сушу");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Мощность: {Horsepower} л.с., Объем: {EngineDisplacement} л");
        Console.WriteLine($"Водоизмещение: {Displacement} т, Тип винта: {PropellerType}");
        Console.WriteLine($"Режим: {(OnWater ? "НА ВОДЕ" : "НА СУШЕ")}");
    }
}

// 4. Гоночный грузовик (наследует: IEnginePowered, IRaceCapable, ITowCapable)
public class RacingTruck : Vehicle, IEnginePowered, IRaceCapable, ITowCapable
{
    // IEnginePowered
    public int Horsepower { get; set; }
    public double EngineDisplacement { get; set; }
    
    // IRaceCapable
    public double AccelerationTo100 { get; set; }
    public int MaxSpeed { get; set; }
    
    // ITowCapable
    public double TowingCapacity { get; set; }
    
    // Собственные свойства
    public double LoadCapacity { get; set; }
    public int Axles { get; set; }
    
    public RacingTruck(string make, string model, int year, string color,
                       int horsepower, double engineDisplacement, double accelerationTo100, 
                       int maxSpeed, double towingCapacity, double loadCapacity, int axles)
        : base(make, model, year, color)
    {
        Horsepower = horsepower;
        EngineDisplacement = engineDisplacement;
        AccelerationTo100 = accelerationTo100;
        MaxSpeed = maxSpeed;
        TowingCapacity = towingCapacity;
        LoadCapacity = loadCapacity;
        Axles = axles;
    }
    
    // Реализация IEnginePowered
    public void StartEngine()
    {
        Console.WriteLine($"Дизельный двигатель {Make} {Model} зарычал: BRRRRMMM!");
    }
    
    public void StopEngine()
    {
        Console.WriteLine($"Двигатель {Make} {Model} заглушен");
    }
    
    public void RevEngine()
    {
        Console.WriteLine($"Турбодизель ревет с глушителя с двойным выхлопом!");
    }
    
    // Реализация IRaceCapable
    public void ActivateSportMode()
    {
        Console.WriteLine($"Гоночный режим активирован! Разгон 0-100 за {AccelerationTo100} сек");
        Console.WriteLine($"Максимальная скорость {MaxSpeed} км/ч для 40-тонного грузовика!");
    }
    
    public void Drift()
    {
        Console.WriteLine($"Гоночный грузовик {Make} {Model} заходит в дрифт! НЕВЕРОЯТНО!");
    }
    
    // Реализация ITowCapable
    public void Tow(Vehicle vehicle)
    {
        Console.WriteLine($"Грузовик {Make} {Model} буксирует {vehicle.Make} {vehicle.Model}");
        Console.WriteLine($"Грузоподъемность буксировки: {TowingCapacity} т");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Мощность: {Horsepower} л.с., Объем: {EngineDisplacement} л");
        Console.WriteLine($"0-100: {AccelerationTo100} сек, Макс. скорость: {MaxSpeed} км/ч");
        Console.WriteLine($"Грузоподъемность: {LoadCapacity} т, Оси: {Axles}, Буксировка: {TowingCapacity} т");
    }
}

// ========== ДЕМОНСТРАЦИЯ МНОЖЕСТВЕННОГО НАСЛЕДОВАНИЯ ДЛЯ ТЕХНИКИ ==========
class VehicleMultipleInheritanceDemo
{
    public static void Run()
    {
        Console.WriteLine("========== МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ: ТЕХНИКА ==========\n");
        
        // 1. Электромобиль-спорткар (3 интерфейса)
        Console.WriteLine("--- Электромобиль-спорткар (IEnginePowered + IElectricVehicle + IRaceCapable) ---");
        ElectricSportsCar tesla = new ElectricSportsCar("Tesla", "Roadster 2.0", 2024, "Красный",
                                                         1200, 200, 1000, 1.9, 400, true);
        tesla.DisplayInfo();
        tesla.StartEngine();
        tesla.ActivateSportMode();
        tesla.Charge();
        tesla.Drift();
        Console.WriteLine();
        
        // 2. Летающий автомобиль (2 интерфейса)
        Console.WriteLine("--- Летающий автомобиль (IEnginePowered + IFlyable) ---");
        FlyingCar flyingCar = new FlyingCar("AeroMobil", "AM 4.0", 2023, "Серебристый",
                                            300, 2.0, 3000, 8.2);
        flyingCar.DisplayInfo();
        flyingCar.StartEngine();
        flyingCar.TakeOff();
        flyingCar.Fly();
        flyingCar.Land();
        Console.WriteLine();
        
        // 3. Амфибия (2 интерфейса)
        Console.WriteLine("--- Амфибия (IEnginePowered + IWaterVehicle) ---");
        AmphibiousVehicle amphibian = new AmphibiousVehicle("Gibbs", "Humdinga", 2022, "Синий",
                                                             350, 3.6, 2.5, "Водометный");
        amphibian.DisplayInfo();
        amphibian.StartEngine();
        amphibian.Sail();
        amphibian.DriveOnLand();
        Console.WriteLine();
        
        // 4. Гоночный грузовик (3 интерфейса)
        Console.WriteLine("--- Гоночный грузовик (IEnginePowered + IRaceCapable + ITowCapable) ---");
        RacingTruck racingTruck = new RacingTruck("Volvo", "Iron Knight", 2023, "Оранжевый",
                                                  2400, 12.8, 4.6, 276, 200, 40, 3);
        racingTruck.DisplayInfo();
        racingTruck.StartEngine();
        racingTruck.ActivateSportMode();
        
        Vehicle towedCar = new Vehicle("Smart", "Fortwo", 2022, "Белый");
        racingTruck.Tow(towedCar);
        racingTruck.Drift();
        
        Console.WriteLine("\n=== ПОЛИМОРФИЗМ ЧЕРЕЗ ИНТЕРФЕЙСЫ ===");
        
        // Работа через интерфейс IEnginePowered
        Console.WriteLine("\n--- Все транспортные средства с двигателем ---");
        IEnginePowered[] engineVehicles = new IEnginePowered[]
        {
            tesla,
            flyingCar,
            amphibian,
            racingTruck
        };
        
        foreach (var ev in engineVehicles)
        {
            ev.StartEngine();
            ev.RevEngine();
        }
        
        // Работа через интерфейс IRaceCapable
        Console.WriteLine("\n--- Гоночные транспортные средства ---");
        IRaceCapable[] raceVehicles = new IRaceCapable[]
        {
            tesla,
            racingTruck
        };
        
        foreach (var rv in raceVehicles)
        {
            rv.ActivateSportMode();
            rv.Drift();
        }
    }
}

// Запуск демонстрации
VehicleMultipleInheritanceDemo.Run();
````

````
using System;
using System.Collections.Generic;

// ========== ИНТЕРФЕЙСЫ ДЛЯ РАЗЛИЧНЫХ СПОСОБНОСТЕЙ ЖИВОТНЫХ ==========

// Интерфейс для летающих животных
public interface IFlyer
{
    double WingSpan { get; set; }
    int MaxFlightAltitude { get; set; }
    void Fly();
    void Land();
    void Soar();
}

// Интерфейс для плавающих животных
public interface ISwimmer
{
    double MaxDepth { get; set; }
    double SwimSpeed { get; set; }
    void Swim();
    void Dive();
    void Surface();
}

// Интерфейс для бегающих животных
public interface IRunner
{
    double MaxSpeed { get; set; }
    double RunDistance { get; set; }
    void Run();
    void Sprint();
    void Jog();
}

// Интерфейс для хищников
public interface IPredator
{
    string PreyType { get; set; }
    double HuntingSuccessRate { get; set; }
    void Hunt();
    void Stalk();
    void Pounce();
}

// Интерфейс для травоядных
public interface IHerbivore
{
    string DietType { get; set; }
    double DailyFoodIntake { get; set; }
    void Graze();
    void Digest();
}

// Интерфейс для домашних животных
public interface IPet
{
    string OwnerName { get; set; }
    bool IsTrained { get; set; }
    void Play();
    void PerformTrick();
    void ShowAffection();
}

// Интерфейс для ночных животных
public interface INocturnal
{
    double NightVisionLevel { get; set; }
    void HuntAtNight();
    void SleepDuringDay();
}

// Интерфейс для социальных животных
public interface ISocial
{
    int GroupSize { get; set; }
    void Communicate();
    void Cooperate();
    void ProtectGroup();
}

// ========== БАЗОВЫЙ КЛАСС ==========
public class Animal
{
    public string Name { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }
    public static int TotalAnimals { get; private set; } = 0;
    
    public Animal(string name, int age, double weight)
    {
        Name = name;
        Age = age;
        Weight = weight;
        TotalAnimals++;
    }
    
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Имя: {Name}, Возраст: {Age} лет, Вес: {Weight} кг");
    }
    
    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} издает звук");
    }
    
    public virtual void Eat()
    {
        Console.WriteLine($"{Name} ест");
    }
}

// ========== КЛАССЫ С МНОЖЕСТВЕННЫМ НАСЛЕДОВАНИЕМ (ИНТЕРФЕЙСЫ) ==========

// 1. Утка (наследует: IFlyer, ISwimmer, IRunner)
public class Duck : Animal, IFlyer, ISwimmer, IRunner
{
    // IFlyer
    public double WingSpan { get; set; }
    public int MaxFlightAltitude { get; set; }
    
    // ISwimmer
    public double MaxDepth { get; set; }
    public double SwimSpeed { get; set; }
    
    // IRunner
    public double MaxSpeed { get; set; }
    public double RunDistance { get; set; }
    
    // Собственные свойства
    public string FeatherColor { get; set; }
    public bool IsMigratory { get; set; }
    
    public Duck(string name, int age, double weight, double wingSpan, int maxFlightAltitude,
                double maxDepth, double swimSpeed, double maxSpeed, double runDistance,
                string featherColor, bool isMigratory)
        : base(name, age, weight)
    {
        WingSpan = wingSpan;
        MaxFlightAltitude = maxFlightAltitude;
        MaxDepth = maxDepth;
        SwimSpeed = swimSpeed;
        MaxSpeed = maxSpeed;
        RunDistance = runDistance;
        FeatherColor = featherColor;
        IsMigratory = isMigratory;
    }
    
    // Реализация IFlyer
    public void Fly()
    {
        Console.WriteLine($"Утка {Name} летит на высоте {MaxFlightAltitude} м, размах крыльев {WingSpan} м");
    }
    
    public void Land()
    {
        Console.WriteLine($"Утка {Name} приземляется на воду");
    }
    
    public void Soar()
    {
        Console.WriteLine($"Утка {Name} парит в воздухе");
    }
    
    // Реализация ISwimmer
    public void Swim()
    {
        Console.WriteLine($"Утка {Name} плывет со скоростью {SwimSpeed} км/ч");
    }
    
    public void Dive()
    {
        Console.WriteLine($"Утка {Name} ныряет на глубину {MaxDepth} м");
    }
    
    public void Surface()
    {
        Console.WriteLine($"Утка {Name} всплывает на поверхность");
    }
    
    // Реализация IRunner
    public void Run()
    {
        Console.WriteLine($"Утка {Name} бежит со скоростью {MaxSpeed} км/ч");
    }
    
    public void Sprint()
    {
        Console.WriteLine($"Утка {Name} ускоряется до {MaxSpeed * 1.2} км/ч!");
    }
    
    public void Jog()
    {
        Console.WriteLine($"Утка {Name} переваливается с лапы на лапу");
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"Утка {Name} крякает: Кря-кря!");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Цвет перьев: {FeatherColor}, Перелетная: {(IsMigratory ? "Да" : "Нет")}");
        Console.WriteLine($"Летает: размах {WingSpan} м, высота {MaxFlightAltitude} м");
        Console.WriteLine($"Плавает: глубина {MaxDepth} м, скорость {SwimSpeed} км/ч");
        Console.WriteLine($"Бегает: скорость {MaxSpeed} км/ч, дистанция {RunDistance} км");
    }
}

// 2. Летучая мышь (наследует: IFlyer, IPredator, INocturnal)
public class Bat : Animal, IFlyer, IPredator, INocturnal
{
    // IFlyer
    public double WingSpan { get; set; }
    public int MaxFlightAltitude { get; set; }
    
    // IPredator
    public string PreyType { get; set; }
    public double HuntingSuccessRate { get; set; }
    
    // INocturnal
    public double NightVisionLevel { get; set; }
    
    // Собственные свойства
    public double EcholocationRange { get; set; }
    
    public Bat(string name, int age, double weight, double wingSpan, int maxFlightAltitude,
               string preyType, double huntingSuccessRate, double nightVisionLevel, double echolocationRange)
        : base(name, age, weight)
    {
        WingSpan = wingSpan;
        MaxFlightAltitude = maxFlightAltitude;
        PreyType = preyType;
        HuntingSuccessRate = huntingSuccessRate;
        NightVisionLevel = nightVisionLevel;
        EcholocationRange = echolocationRange;
    }
    
    // Реализация IFlyer
    public void Fly()
    {
        Console.WriteLine($"Летучая мышь {Name} летит, используя эхолокацию на {EcholocationRange} м");
    }
    
    public void Land()
    {
        Console.WriteLine($"Летучая мышь {Name} приземляется вверх ногами");
    }
    
    public void Soar()
    {
        Console.WriteLine($"Летучая мышь {Name} планирует между деревьями");
    }
    
    // Реализация IPredator
    public void Hunt()
    {
        Console.WriteLine($"Летучая мышь {Name} охотится на {PreyType} с успешностью {HuntingSuccessRate}%");
    }
    
    public void Stalk()
    {
        Console.WriteLine($"Летучая мышь {Name} бесшумно приближается к добыче");
    }
    
    public void Pounce()
    {
        Console.WriteLine($"Летучая мышь {Name} пикирует на добычу!");
    }
    
    // Реализация INocturnal
    public void HuntAtNight()
    {
        Console.WriteLine($"Летучая мышь {Name} охотится ночью, используя ночное зрение {NightVisionLevel}/10");
    }
    
    public void SleepDuringDay()
    {
        Console.WriteLine($"Летучая мышь {Name} спит днем в пещере");
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"Летучая мышь {Name} издает ультразвук: (не слышно человеку)");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Размах крыльев: {WingSpan} м, Высота полета: {MaxFlightAltitude} м");
        Console.WriteLine($"Эхолокация: {EcholocationRange} м, Ночное зрение: {NightVisionLevel}/10");
        Console.WriteLine($"Хищник: охотится на {PreyType}, успешность {HuntingSuccessRate}%");
    }
}

// 3. Морской котик (наследует: ISwimmer, IRunner, IPredator)
public class Seal : Animal, ISwimmer, IRunner, IPredator
{
    // ISwimmer
    public double MaxDepth { get; set; }
    public double SwimSpeed { get; set; }
    
    // IRunner
    public double MaxSpeed { get; set; }
    public double RunDistance { get; set; }
    
    // IPredator
    public string PreyType { get; set; }
    public double HuntingSuccessRate { get; set; }
    
    // Собственные свойства
    public double BlubberThickness { get; set; }
    
    public Seal(string name, int age, double weight, double maxDepth, double swimSpeed,
                double maxSpeed, double runDistance, string preyType, 
                double huntingSuccessRate, double blubberThickness)
        : base(name, age, weight)
    {
        MaxDepth = maxDepth;
        SwimSpeed = swimSpeed;
        MaxSpeed = maxSpeed;
        RunDistance = runDistance;
        PreyType = preyType;
        HuntingSuccessRate = huntingSuccessRate;
        BlubberThickness = blubberThickness;
    }
    
    // Реализация ISwimmer
    public void Swim()
    {
        Console.WriteLine($"Морской котик {Name} плывет со скоростью {SwimSpeed} км/ч");
    }
    
    public void Dive()
    {
        Console.WriteLine($"Морской котик {Name} ныряет на глубину {MaxDepth} м");
    }
    
    public void Surface()
    {
        Console.WriteLine($"Морской котик {Name} выныривает на поверхность");
    }
    
    // Реализация IRunner
    public void Run()
    {
        Console.WriteLine($"Морской котик {Name} неуклюже бежит по суше со скоростью {MaxSpeed} км/ч");
    }
    
    public void Sprint()
    {
        Console.WriteLine($"Морской котик {Name} ускоряется на короткой дистанции!");
    }
    
    public void Jog()
    {
        Console.WriteLine($"Морской котик {Name} переваливается с боку на бок");
    }
    
    // Реализация IPredator
    public void Hunt()
    {
        Console.WriteLine($"Морской котик {Name} охотится на {PreyType} в воде");
    }
    
    public void Stalk()
    {
        Console.WriteLine($"Морской котик {Name} незаметно подплывает к добыче");
    }
    
    public void Pounce()
    {
        Console.WriteLine($"Морской котик {Name} хватает добычу своими ластами!");
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"Морской котик {Name} лает: Ар-ар-ар!");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Толщина жира: {BlubberThickness} см");
        Console.WriteLine($"В воде: глубина {MaxDepth} м, скорость {SwimSpeed} км/ч");
        Console.WriteLine($"На суше: скорость {MaxSpeed} км/ч, дистанция {RunDistance} м");
        Console.WriteLine($"Хищник: охотится на {PreyType}, успешность {HuntingSuccessRate}%");
    }
}

// 4. Собака (наследует: IRunner, IPredator, IPet, ISocial)
public class Dog : Animal, IRunner, IPredator, IPet, ISocial
{
    // IRunner
    public double MaxSpeed { get; set; }
    public double RunDistance { get; set; }
    
    // IPredator
    public string PreyType { get; set; }
    public double HuntingSuccessRate { get; set; }
    
    // IPet
    public string OwnerName { get; set; }
    public bool IsTrained { get; set; }
    
    // ISocial
    public int GroupSize { get; set; }
    
    // Собственные свойства
    public string Breed { get; set; }
    public string FavoriteToy { get; set; }
    
    public Dog(string name, int age, double weight, double maxSpeed, double runDistance,
               string preyType, double huntingSuccessRate, string ownerName, 
               bool isTrained, int groupSize, string breed, string favoriteToy)
        : base(name, age, weight)
    {
        MaxSpeed = maxSpeed;
        RunDistance = runDistance;
        PreyType = preyType;
        HuntingSuccessRate = huntingSuccessRate;
        OwnerName = ownerName;
        IsTrained = isTrained;
        GroupSize = groupSize;
        Breed = breed;
        FavoriteToy = favoriteToy;
    }
    
    // Реализация IRunner
    public void Run()
    {
        Console.WriteLine($"Собака {Name} бежит со скоростью {MaxSpeed} км/ч");
    }
    
    public void Sprint()
    {
        Console.WriteLine($"Собака {Name} мчится со скоростью {MaxSpeed * 1.3} км/ч!");
    }
    
    public void Jog()
    {
        Console.WriteLine($"Собака {Name} трусцой бежит рядом с хозяином");
    }
    
    // Реализация IPredator
    public void Hunt()
    {
        Console.WriteLine($"Собака {Name} охотится с успешностью {HuntingSuccessRate}%");
    }
    
    public void Stalk()
    {
        Console.WriteLine($"Собака {Name} крадется к добыче");
    }
    
    public void Pounce()
    {
        Console.WriteLine($"Собака {Name} прыгает на {FavoriteToy}");
    }
    
    // Реализация IPet
    public void Play()
    {
        Console.WriteLine($"Собака {Name} играет с {FavoriteToy}");
    }
    
    public void PerformTrick()
    {
        if (IsTrained)
            Console.WriteLine($"Собака {Name} выполняет трюк по команде!");
        else
            Console.WriteLine($"Собака {Name} еще не обучена трюкам");
    }
    
    public void ShowAffection()
    {
        Console.WriteLine($"Собака {Name} виляет хвостом и лижет руку");
    }
    
    // Реализация ISocial
    public void Communicate()
    {
        Console.WriteLine($"Собака {Name} лает, чтобы общаться с другими собаками");
    }
    
    public void Cooperate()
    {
        Console.WriteLine($"Собака {Name} сотрудничает со стаей из {GroupSize} особей");
    }
    
    public void ProtectGroup()
    {
        Console.WriteLine($"Собака {Name} защищает {OwnerName} и семью");
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"Собака {Name} лает: Гав-гав!");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}, Любимая игрушка: {FavoriteToy}");
        Console.WriteLine($"Скорость: {MaxSpeed} км/ч, Дистанция: {RunDistance} км");
        Console.WriteLine($"Хозяин: {OwnerName}, Дрессирована: {(IsTrained ? "Да" : "Нет")}");
        Console.WriteLine($"Социальная: живет в группе из {GroupSize} особей");
    }
}

// 5. Попугай (наследует: IFlyer, IPet, ISocial)
public class Parrot : Animal, IFlyer, IPet, ISocial
{
    // IFlyer
    public double WingSpan { get; set; }
    public int MaxFlightAltitude { get; set; }
    
    // IPet
    public string OwnerName { get; set; }
    public bool IsTrained { get; set; }
    
    // ISocial
    public int GroupSize { get; set; }
    
    // Собственные свойства
    public string FeatherColor { get; set; }
    public List<string> Vocabulary { get; set; } = new List<string>();
    
    public Parrot(string name, int age, double weight, double wingSpan, int maxFlightAltitude,
                  string ownerName, bool isTrained, int groupSize, string featherColor)
        : base(name, age, weight)
    {
        WingSpan = wingSpan;
        MaxFlightAltitude = maxFlightAltitude;
        OwnerName = ownerName;
        IsTrained = isTrained;
        GroupSize = groupSize;
        FeatherColor = featherColor;
    }
    
    // Реализация IFlyer
    public void Fly()
    {
        Console.WriteLine($"Попугай {Name} летит, переливаясь {FeatherColor} перьями");
    }
    
    public void Land()
    {
        Console.WriteLine($"Попугай {Name} приземляется на плечо хозяину");
    }
    
    public void Soar()
    {
        Console.WriteLine($"Попугай {Name} парит в небе");
    }
    
    // Реализация IPet
    public void Play()
    {
        Console.WriteLine($"Попугай {Name} играет с зеркальцем");
    }
    
    public void PerformTrick()
    {
        if (IsTrained && Vocabulary.Count > 0)
            Console.WriteLine($"Попугай {Name} говорит: {Vocabulary[0]}");
        else
            Console.WriteLine($"Попугай {Name} машет крыльями");
    }
    
    public void ShowAffection()
    {
        Console.WriteLine($"Попугай {Name} трется клювом о руку");
    }
    
    // Реализация ISocial
    public void Communicate()
    {
        Console.WriteLine($"Попугай {Name} общается с другими птицами");
    }
    
    public void Cooperate()
    {
        Console.WriteLine($"Попугай {Name} ухаживает за другими птицами");
    }
    
    public void ProtectGroup()
    {
        Console.WriteLine($"Попугай {Name} предупреждает стаю об опасности громким криком");
    }
    
    public void LearnWord(string word)
    {
        Vocabulary.Add(word);
        Console.WriteLine($"Попугай {Name} выучил слово: {word}");
    }
    
    public override void MakeSound()
    {
        if (Vocabulary.Count > 0)
            Console.WriteLine($"Попугай {Name} говорит: {string.Join(", ", Vocabulary)}");
        else
            Console.WriteLine($"Попугай {Name} чирикает");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Цвет: {FeatherColor}, Размах крыльев: {WingSpan} м");
        Console.WriteLine($"Хозяин: {OwnerName}, Дрессирован: {(IsTrained ? "Да" : "Нет")}");
        Console.WriteLine($"Словарный запас: {Vocabulary.Count} слов");
    }
}

// ========== ДЕМОНСТРАЦИЯ МНОЖЕСТВЕННОГО НАСЛЕДОВАНИЯ ДЛЯ ЖИВОТНЫХ ==========
class AnimalMultipleInheritanceDemo
{
    public static void Run()
    {
        Console.WriteLine("========== МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ: ЖИВОТНЫЕ ==========\n");
        
        // 1. Утка (3 интерфейса: IFlyer, ISwimmer, IRunner)
        Console.WriteLine("--- Утка (IFlyer + ISwimmer + IRunner) ---");
        Duck duck = new Duck("Дональд", 3, 1.2, 0.8, 200, 5, 15, 25, 10,
                              "Зеленый", true);
        duck.DisplayInfo();
        duck.MakeSound();
        duck.Fly();
        duck.Swim();
        duck.Run();
        Console.WriteLine();
        
        // 2. Летучая мышь (3 интерфейса: IFlyer, IPredator, INocturnal)
        Console.WriteLine("--- Летучая мышь (IFlyer + IPredator + INocturnal) ---");
        Bat bat = new Bat("Бэтти", 2, 0.05, 0.5, 1000, "Насекомые", 85, 9.5, 50);
        bat.DisplayInfo();
        bat.MakeSound();
        bat.Fly();
        bat.Hunt();
        bat.HuntAtNight();
        Console.WriteLine();
        
        // 3. Морской котик (3 интерфейса: ISwimmer, IRunner, IPredator)
        Console.WriteLine("--- Морской котик (ISwimmer + IRunner + IPredator) ---");
        Seal seal = new Seal("Саймон", 5, 150, 200, 35, 15, 100, "Рыба", 75, 10);
        seal.DisplayInfo();
        seal.MakeSound();
        seal.Swim();
        seal.Dive();
        seal.Run();
        seal.Hunt();
        Console.WriteLine();
        
        // 4. Собака (4 интерфейса: IRunner, IPredator, IPet, ISocial)
        Console.WriteLine("--- Собака (IRunner + IPredator + IPet + ISocial) ---");
        Dog dog = new Dog("Рекс", 3, 30, 45, 10000, "Мяч", 80, "Иван", true, 1, "Овчарка", "Мяч");
        dog.DisplayInfo();
        dog.MakeSound();
        dog.Run();
        dog.Sprint();
        dog.Play();
        dog.PerformTrick();
        dog.Communicate();
        Console.WriteLine();
        
        // 5. Попугай (3 интерфейса: IFlyer, IPet, ISocial)
        Console.WriteLine("--- Попугай (IFlyer + IPet + ISocial) ---");
        Parrot parrot = new Parrot("Кеша", 2, 0.4, 0.3, 1000, "Мария", true, 5, "Ярко-зеленый");
        parrot.LearnWord("Привет");
        parrot.LearnWord("Кеша хороший");
        parrot.DisplayInfo();
        parrot.MakeSound();
        parrot.Fly();
        parrot.Play();
        parrot.PerformTrick();
        parrot.Communicate();
        
        Console.WriteLine("\n=== ПОЛИМОРФИЗМ ЧЕРЕЗ ИНТЕРФЕЙСЫ ===");
        
        // Работа через интерфейс IFlyer
        Console.WriteLine("\n--- Все летающие животные ---");
        IFlyer[] flyers = new IFlyer[]
        {
            duck,
            bat,
            parrot
        };
        
        foreach (var flyer in flyers)
        {
            flyer.Fly();
            flyer.Land();
        }
        
        // Работа через интерфейс ISwimmer
        Console.WriteLine("\n--- Все плавающие животные ---");
        ISwimmer[] swimmers = new ISwimmer[]
        {
            duck,
            seal
        };
        
        foreach (var swimmer in swimmers)
        {
            swimmer.Swim();
            swimmer.Dive();
        }
        
        // Работа через интерфейс IRunner
        Console.WriteLine("\n--- Все бегающие животные ---");
        IRunner[] runners = new IRunner[]
        {
            duck,
            seal,
            dog
        };
        
        foreach (var runner in runners)
        {
            runner.Run();
        }
        
        // Работа через интерфейс IPet
        Console.WriteLine("\n--- Все домашние животные ---");
        IPet[] pets = new IPet[]
        {
            dog,
            parrot
        };
        
        foreach (var pet in pets)
        {
            pet.Play();
            pet.ShowAffection();
        }
        
        // Работа через интерфейс IPredator
        Console.WriteLine("\n--- Все хищники ---");
        IPredator[] predators = new IPredator[]
        {
            bat,
            seal,
            dog
        };
        
        foreach (var predator in predators)
        {
            predator.Hunt();
        }
    }
}

// ========== ЗАПУСК ВСЕХ ДЕМОНСТРАЦИЙ ==========
Console.WriteLine("\n" + new string('=', 80) + "\n");
AnimalMultipleInheritanceDemo.Run();
````
